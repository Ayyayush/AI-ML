```python
# Creating dataframe
""" 
📘 Note:
DataFrame banane ke multiple tarike h Pandas me.
1️⃣ Using lists (2D list)
2️⃣ Using dictionary (columns as keys)
3️⃣ Using CSV file
4️⃣ Using NumPy array
DataFrame basically ek 2D labeled data structure hota h (like Excel sheet).
"""

import numpy as np
import pandas as pd


# -------------------------------------------------------------------
# 🧩 METHOD - 1 : Using Lists
# -------------------------------------------------------------------

# yaha hum ek 2D list bna rahe h jisme har inner list ek row represent karti h
student_data = [
    [100, 80, 70],
    [90, 70, 7],
    [120, 100, 14],
    [80, 50, 2]
]

# columns ka naam dene ke liye 'columns' parameter use karte h
df1 = pd.DataFrame(student_data, columns=['iq', 'marks', 'package'])

print("DataFrame created using list:\n", df1, "\n")


# -------------------------------------------------------------------
# 🧩 METHOD - 2 : Using Dictionary
# -------------------------------------------------------------------
# ye approach zyada readable hoti h kyunki columns ke naam khud define kar sakte ho

student_dict = {
    'iq': [100, 90, 120, 80],
    'marks': [80, 70, 100, 50],
    'package': [70, 7, 14, 2]
}

# DataFrame banana direct possible h
df2 = pd.DataFrame(student_dict)

print("DataFrame created using dictionary:\n", df2, "\n")

# agar chahe to DataFrame ko variable me store kar skte ho
student = pd.DataFrame(student_dict)
print("Stored DataFrame (student):\n", student, "\n")




# -------------------------------------------------------------------
# 🧩 METHOD - 3 : Using CSV file
# -------------------------------------------------------------------
"""
read_csv() method ka use tab hota h jab data kisi CSV file me store ho.
For example: agar ek file h 'students.csv' naam se, to usse read krne ke liye:
"""

# using read_csv
movies = pd.read_csv('content/movies.csv')
print(movies.head())

ipl = pd.read_csv('content/ipl-matches.csv')
print(ipl.head())

# -------------------------------------------------------------------
# 🧩 METHOD - 4 : Using NumPy Array
# -------------------------------------------------------------------

# NumPy array bhi use kr sakte h DataFrame banane ke liye
arr = np.array([[100, 80, 70],
                [90, 70, 7],
                [120, 100, 14],
                [80, 50, 2]])

df4 = pd.DataFrame(arr, columns=['iq', 'marks', 'package'])
print("DataFrame created using NumPy array:\n", df4, "\n")

# -------------------------------------------------------------------
# ✅ Summary:
# -------------------------------------------------------------------
"""
List → Basic data se directly DataFrame
Dict → Zyada readable (column-based)
CSV → Real-world data import karne ke liye best
NumPy → Mathematical data se DataFrame banana ho to
"""
```

```python
# 🧠 DataFrame Attributes and Methods
"""
Ye section me hum Pandas DataFrame ke important attributes aur methods dekhenge:
1️⃣ shape
2️⃣ dtypes
3️⃣ index
4️⃣ columns
5️⃣ values
6️⃣ head() and tail()
7️⃣ sample()
8️⃣ info()
9️⃣ describe()
🔟 isnull()
1️⃣1️⃣ duplicated()
1️⃣2️⃣ rename()
"""

import pandas as pd

# -------------------------------------------------------------------
# 🔹 Create a simple DataFrame for demonstration
# -------------------------------------------------------------------
data = {
    'Name': ['Aman', 'Riya', 'Kabir', 'Neha', 'Sohan'],
    'Age': [21, 23, 22, None, 24],          # Neha ke liye Age missing (NaN)
    'Marks': [85, 90, 75, 88, 92]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df, "\n")


# -------------------------------------------------------------------
# 🧩 1. shape → rows aur columns count batata h
# -------------------------------------------------------------------
print("Shape of DataFrame:", df.shape)         # (5, 3)
print()


# -------------------------------------------------------------------
# 🧩 2. dtypes → har column ka datatype batata h
# -------------------------------------------------------------------
print("Data types of each column:\n", df.dtypes, "\n")


# -------------------------------------------------------------------
# 🧩 3. index → DataFrame ke row indexes (labels) dikhata h
# -------------------------------------------------------------------
print("Index labels:\n", df.index, "\n")       # default index (0 to n-1)
# agar custom index dena ho to: df.index = ['a','b','c','d','e']


# -------------------------------------------------------------------
# 🧩 4. columns → DataFrame ke column names dikhata h
# -------------------------------------------------------------------
print("Column names:\n", df.columns, "\n")


# -------------------------------------------------------------------
# 🧩 5. values → sirf raw numpy array return karta h bina labels ke
# -------------------------------------------------------------------
print("All values in DataFrame (as NumPy array):\n", df.values, "\n")


# -------------------------------------------------------------------
# 🧩 6. head() & tail() → starting aur ending rows dekhne ke liye
# -------------------------------------------------------------------
print("First 3 rows using head():\n", df.head(3), "\n")
print("Last 2 rows using tail():\n", df.tail(2), "\n")


# -------------------------------------------------------------------
# 🧩 7. sample() → random row(s) dekhne ke liye
# -------------------------------------------------------------------
print("Random 2 rows using sample():\n", df.sample(2), "\n")


# -------------------------------------------------------------------
# 🧩 8. info() → dataframe ka summary deta h (datatype, non-null values etc.)
# -------------------------------------------------------------------
print("DataFrame Info:")
df.info()
print()


# -------------------------------------------------------------------
# 🧩 9. describe() → numerical columns ka statistical summary deta h
# -------------------------------------------------------------------
print("Statistical Summary using describe():\n", df.describe(), "\n")


# -------------------------------------------------------------------
# 🧩 10. isnull() → missing (NaN) values ko detect karta h
# -------------------------------------------------------------------
print("Check for missing values:\n", df.isnull(), "\n")

# total missing count per column
print("Total null values in each column:\n", df.isnull().sum(), "\n")


# -------------------------------------------------------------------
# 🧩 11. duplicated() → duplicate rows detect karta h
# -------------------------------------------------------------------
# ek duplicate row add karte h demo ke liye
df.loc[len(df)] = ['Aman', 21, 85]          # duplicate of 1st row
print("After adding a duplicate row:\n", df, "\n")

print("Duplicate row check (True means duplicate):\n", df.duplicated(), "\n")

# agar duplicate remove karna ho:
# df = df.drop_duplicates()


# -------------------------------------------------------------------
# 🧩 12. rename() → columns ya indexes ke naam change karne ke liye
# -------------------------------------------------------------------
print("Renaming column 'Marks' → 'Score':\n")
df_renamed = df.rename(columns={'Marks': 'Score'})
print(df_renamed, "\n")


# -------------------------------------------------------------------
# ✅ Summary:
# -------------------------------------------------------------------
"""
shape        → (rows, columns)
dtypes       → column ke data types
index        → row labels
columns      → column names
values       → raw data (as NumPy array)
head(), tail() → starting/ending rows
sample()     → random row(s)
info()       → structure + datatypes + non-null count
describe()   → summary stats for numerical columns
isnull()     → missing values check
duplicated() → duplicate rows detect
rename()     → column/index name change
"""
```

# 📊 Comparison Chart: Pandas Series vs DataFrame

---

## 🔹 Core Comparison

| Feature              | Series                                  | DataFrame                               |
|----------------------|------------------------------------------|------------------------------------------|
| Definition           | One-dimensional labeled array             | Two-dimensional labeled data structure   |
| Dimensionality       | 1D                                       | 2D                                       |
| Data type            | Homogeneous (same data type)              | Each column can have different dtype     |
| Structure            | Single column of data                     | Table (rows × columns)                   |
| Index                | Single index                              | Row index + column labels                |
| Can hold multiple columns | ❌ No                               | ✅ Yes                                   |
| Think of as          | Single column / vector                   | Excel sheet / SQL table                  |
| Underlying structure | NumPy array                               | Collection of Series                     |

---

## 🔹 Functional Perspective (Same Task, Different Style)

| Task                  | Series Style                  | DataFrame Style                 |
|-----------------------|-------------------------------|----------------------------------|
| Create data           | `pd.Series(data)`             | `pd.DataFrame(data)`             |
| Access data           | `s[index]`                    | `df[column]`                     |
| Mean                  | `s.mean()`                    | `df.mean()`                      |
| Sum                   | `s.sum()`                     | `df.sum()`                       |
| Filtering             | `s[s > 10]`                   | `df[df['col'] > 10]`             |
| Describe              | `s.describe()`                | `df.describe()`                  |
| Shape                 | `s.shape`                     | `df.shape`                       |

---

## 🔹 Use-Case Comparison

| Scenario                     | Use Series | Use DataFrame |
|------------------------------|------------|---------------|
| Single variable analysis     | ✅ Yes     | ❌ Not ideal  |
| Tabular data analysis        | ❌ No      | ✅ Yes        |
| Time series (single metric)  | ✅ Yes     | ❌ / ⚠️       |
| Multiple features analysis   | ❌ No      | ✅ Yes        |
| ML dataset preparation       | ❌ No      | ✅ Yes        |

---

## 🧠 Easy Mental Model

- **Series**  
  → Ek column  
  → Index + values  

- **DataFrame**  
  → Multiple columns  
  → Rows + columns (table)

---

## 🎯 Interview One-Liner

> A Pandas Series is a one-dimensional labeled array, while a DataFrame is a two-dimensional labeled data structure composed of multiple Series.

---

## 📌 Summary

- Series → 1D, single column  
- DataFrame → 2D, table-like  
- DataFrame = collection of Series  
- Most real-world datasets → DataFrame

```python
# movies with rating higher than 8 and votes>10000
movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)].shape[0]
```

```python
# ------------------------------------------------------------
# Step 1: Create mask for "Action" genre using split + lambda
# ------------------------------------------------------------
# genres column contains strings like: "Action|Adventure|Sci-Fi"
# str.split('|') → converts string into list
# lambda x: 'Action' in x → checks if "Action" exists in that list
mask1 = movies['genres'].str.split('|').apply(lambda x: 'Action' in x)

# ------------------------------------------------------------
# Step 2: Create mask for IMDb rating greater than 7.5
# ------------------------------------------------------------
mask2 = movies['imdb_rating'] > 7.5

# ------------------------------------------------------------
# Step 3: Combine both masks using AND condition
# ------------------------------------------------------------
# Filters movies which are:
# - Action genre
# - IMDb rating > 7.5
movies[mask1 & mask2]
```

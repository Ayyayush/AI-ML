````python
# ===================== IMPORTS =====================
import os
import sys
from groq import Groq
from dotenv import load_dotenv


# ===================== LOAD ENV =====================
load_dotenv()                                        # Loads .env file
API_KEY = os.getenv("GROQ_API_KEY")

if not API_KEY:
    print("❌ GROQ_API_KEY not found")
    sys.exit()


# ===================== MAIN APP CLASS =====================
class NLPApp:

    def __init__(self):
        """
        Constructor:
        - Initializes in-memory user database
        - Starts the first menu
        """
        self.__database = {}                          # {email: [name, password]}
        self.client = Groq(api_key=API_KEY)           # Groq client
        self.model = "llama-3.3-70b-versatile"
        self.__first_menu()


    # ===================== FIRST MENU =====================
    def __first_menu(self):
        first_input = input("""
        👋 Welcome to Emotion Analyzer!
        How would you like to proceed?

        1. Not a member? Register
        2. Already a member? Login
        3. Exit

        👉 Your choice: """)

        if first_input == '1':
            self.__register()
        elif first_input == '2':
            self.__login()
        else:
            print("👋 Exiting Emotion Analyzer. Bye!")
            sys.exit()


    # ===================== REGISTER =====================
    def __register(self):
        name = input("👤 Enter your name: ")
        email = input("📧 Enter your email: ")
        password = input("🔑 Enter your password: ")

        if email in self.__database:
            print("⚠️ Email already registered. Please login.")
            self.__first_menu()
        else:
            self.__database[email] = [name, password]
            print("✅ Registration successful! Please login.")
            self.__login()


    # ===================== LOGIN =====================
    def __login(self):
        email = input("📧 Enter your email: ")
        password = input("🔑 Enter your password: ")

        if email in self.__database and self.__database[email][1] == password:
            print(f"✅ Welcome, {self.__database[email][0]}!")
            self.__second_menu()
        else:
            print("❌ Invalid email or password.")
            self.__first_menu()


    # ===================== SECOND MENU =====================
    def __second_menu(self):
        second_input = input("""
        🧠 Emotion Analyzer Dashboard

        1. Named Entity Recognition (NER)
        2. Language Detection
        3. Emotion (Sentiment) Analysis
        4. Logout

        👉 Your choice: """)

        if second_input == '1':
            self.ner()
        elif second_input == '2':
            self.language_detection()
        elif second_input == '3':
            self.sentiment_analysis()
        elif second_input == '4':
            print("👋 Logged out successfully!")
            self.__first_menu()
        else:
            print("⚠️ Invalid choice.")
            self.__second_menu()


    # ===================== NER =====================
    def ner(self):
        text = input("📝 Enter text for NER: ")

        prompt = f"""
        Extract named entities from the text below.
        Categorize them into Person, Organization, Location.

        Text:
        "{text}"
        """

        response = self.client.chat.completions.create(
            model=self.model,
            messages=[{"role": "user", "content": prompt}],
            temperature=0
        )

        print("📌 Entities Found:")
        print(response.choices[0].message.content.strip())

        self.__second_menu()


    # ===================== LANGUAGE DETECTION =====================
    def language_detection(self):
        text = input("🌐 Enter text for language detection: ")

        prompt = f"""
        Detect the language of the following text.
        Respond with only the language name.

        Text:
        "{text}"
        """

        response = self.client.chat.completions.create(
            model=self.model,
            messages=[{"role": "user", "content": prompt}],
            temperature=0
        )

        print("🌍 Detected Language:")
        print(response.choices[0].message.content.strip())

        self.__second_menu()


    # ===================== SENTIMENT / EMOTION =====================
    def sentiment_analysis(self):
        text = input("💬 Enter text for emotion analysis: ")

        prompt = f"""
        Classify the emotion of the following text strictly as:
        Positive, Neutral, or Negative.

        Text:
        "{text}"

        Respond with only one word.
        """

        response = self.client.chat.completions.create(
            model=self.model,
            messages=[{"role": "user", "content": prompt}],
            temperature=0
        )

        print("📊 Emotion Result:")
        print(response.choices[0].message.content.strip())

        self.__second_menu()


# ===================== APP START =====================
if __name__ == "__main__":
    NLPApp()
````

# 🧩 What is Flask

ye basically ek module ke libray samjho 
**Flask** is a **lightweight, open-source web framework** written in **Python**.  
It is designed to make web development simple, flexible, and easy to scale. Flask is often called a **“micro-framework”** because it doesn’t include tools or libraries by default (like database layers or form validation) — but it’s easily extendable through **extensions**.

---

## 🚀 Key Features of Flask

- **Lightweight & Simple:** Minimal setup required; perfect for beginners and small projects.  
- **Modular & Extensible:** Add only the components you need using extensions (e.g., Flask-SQLAlchemy for databases).  
- **Built-in Development Server:** Comes with a debugger and development server.  
- **Jinja2 Templating Engine:** Write dynamic web pages easily using HTML templates.  
- **RESTful Request Handling:** Ideal for building APIs.  
- **Compatible with WSGI:** Works with any WSGI-compliant web server.

---

## 🧠 Why Use Flask

- Easy to learn and get started with.  
- Flexible — you decide the project structure.  
- Great for prototyping and small to medium web apps.  
- Large community and rich third-party extensions.

---

````python
# ============================================================
# SIMPLE FLASK APPLICATION (FINAL VERSION)
# ============================================================

# Import the Flask class from the flask package
# Flask is a lightweight web framework used to build web applications
from flask import Flask


# ------------------------------------------------------------
# CREATE FLASK APPLICATION OBJECT
# ------------------------------------------------------------
# Flask(__name__) creates the Flask app instance
# __name__ helps Flask determine:
# - where the application is located
# - where to find templates and static files
app = Flask(__name__)


# ------------------------------------------------------------
# ROUTE: HOME PAGE
# ------------------------------------------------------------
# @app.route('/') is a decorator
# It tells Flask:
# "When the user visits the root URL '/', execute the function below"
@app.route('/')
def home():
    """
    This function is called when a user visits:
    http://127.0.0.1:5000/

    The returned value is sent back to the browser
    as the HTTP response.
    """
    return "Hello, Flask!"


# ------------------------------------------------------------
# ROUTE: ABOUT PAGE (EXTRA EXAMPLE)
# ------------------------------------------------------------
# This route demonstrates how to create multiple endpoints
@app.route('/about')
def about():
    """
    This function runs when the user visits:
    http://127.0.0.1:5000/about
    """
    return "This is a simple Flask application."


# ------------------------------------------------------------
# APPLICATION ENTRY POINT
# ------------------------------------------------------------
# This condition ensures that the Flask app runs only when
# this file is executed directly.
#
# If this file is imported into another Python file,
# the Flask server will NOT start automatically.
if __name__ == '__main__':

    # Start the Flask development server
    #
    # debug=True enables:
    # 1. Automatic server reload on code changes
    # 2. Detailed error messages in the browser
    #
    # ⚠️ Never use debug=True in production
    app.run(debug=True)
````

````python
from flask import Flask

"""
*-------------------------------------------------------------*
*  STEP 1: IMPORTING FLASK                                    *
*-------------------------------------------------------------*
*  'Flask' ek class hai jo Flask framework ka core part hai.  *
*  Is class ka ek object banana padta hai taaki hum ek Flask   *
*  application bana saken.                                     *
*                                                             *
*  Agar 'ModuleNotFoundError: No module named flask' aaye,     *
*  iska matlab hai Flask install nahi hai.                     *
*  Uske liye terminal me likho:                                *
*      pip install flask                                       *
*-------------------------------------------------------------*
"""

# Flask class ka ek object create kiya
# __name__ ek special Python variable hai jo current file ka name rakhta hai.
# Jab Flask app create hoti hai, usse yeh batana padta hai ki resources
# (like templates, static files, etc.) kis location par milenge.
app = Flask(__name__)

"""
*-------------------------------------------------------------*
*  STEP 2: CREATING A ROUTE                                   *
*-------------------------------------------------------------*
*  @app.route('/') ek "decorator" hai jo URL path ko ek        *
*  specific function se map karta hai.                         *
*                                                             *
*  Jab koi user browser me http://127.0.0.1:5000/ likhega,     *
*  tab Flask yeh dekhega ki '/' route kis function se linked    *
*  hai, aur us function ka output web page par show karega.    *
*-------------------------------------------------------------*
"""

@app.route('/')
def index():  
    """
    *---------------------------------------------------------*
    *  STEP 3: DEFINING A VIEW FUNCTION                       *
    *---------------------------------------------------------*
    *  'index()' ek normal Python function hai jo response     *
    *  return karta hai. Flask is response ko HTTP Response     *
    *  me convert karke browser ko bhejta hai.                 *
    *                                                         *
    *  Is example me humne simple text "Hello, Ayush" return   *
    *  kiya hai, jo browser me show hoga.                      *
    *---------------------------------------------------------*
    """
    return " <h1 style='color:green' >Hello, Ayush</h1>"


"""
*-------------------------------------------------------------*
*  STEP 4: RUNNING THE FLASK APPLICATION                      *
*-------------------------------------------------------------*
*  Normally, Flask app ko start karne ke liye likha jata hai: *
*       if __name__ == '__main__':                            *
*           app.run(debug=True)                               *
*                                                             *
*  Lekin jab hum Flask ko JUPYTER NOTEBOOK ke andar chalate   *
*  hain, to 'app.run()' ek BLOCKING CALL hota hai. Matlab:    *
*  - Jupyter ka cell hamesha "Running" state me atak jata hai. *
*  - Notebook me control wapas nahi aata.                     *
*                                                             *
*  Is problem ko solve karne ke liye, hum Flask ko ek          *
*  BACKGROUND THREAD me chalate hain.                         *
*                                                             *
*  Ye thread Flask server ko alag se run karega, jisse         *
*  notebook responsive rahega aur aap dusre cells bhi run      *
*  kar sakte ho.                                               *
*-------------------------------------------------------------*
"""

import sys  # 'sys.modules' se check karenge ki hum ipykernel (Jupyter) me hain ya nahi.

if 'ipykernel' in sys.modules:
    """
    *---------------------------------------------------------*
    *  STEP 4A: RUNNING INSIDE JUPYTER NOTEBOOK               *
    *---------------------------------------------------------*
    *  Agar 'ipykernel' module loaded hai, iska matlab code    *
    *  Jupyter ya IPython environment me run ho raha hai.      *
    *                                                         *
    *  Yahan hum 'Thread' ka use karenge taaki Flask server     *
    *  background me chale.                                   *
    *---------------------------------------------------------*
    """
    from threading import Thread

    def _run():
        """
        *-----------------------------------------------------*
        *  Ye internal function Flask app ko run karta hai.   *
        *  use_reloader=False lagane se Flask ka internal     *
        *  reloader disable ho jata hai, warna Jupyter me     *
        *  multiple servers ek saath start ho jate.           *
        *-----------------------------------------------------*
        """
        app.run(debug=True, use_reloader=False)

    # Flask ko background thread me start kar rahe hain.
    Thread(target=_run, daemon=True).start()

    print('✅ Flask app started in background: http://127.0.0.1:5000')
    print('💡 Tip: Open the link above in your browser to see the result.')

else:
    """
    *---------------------------------------------------------*
    *  STEP 4B: RUNNING AS A NORMAL PYTHON SCRIPT             *
    *---------------------------------------------------------*
    *  Agar hum Jupyter me nahi hain, to normal way se        *
    *  Flask server run karenge.                              *
    *                                                         *
    *  debug=True lagane ka fayda:                            *
    *  - Code change karne par server auto-reload hota hai.   *
    *  - Error messages me detailed traceback milta hai.      *
    *---------------------------------------------------------*
    """
    if __name__ == '__main__':
        app.run(debug=True)
````

# 🧩 Jinja Template in HTML (Flask)  

Jinja (ya **Jinja2**) ek **templating engine** hai jo Flask ke saath use hota hai  
taaki hum **HTML ke andar Python-like logic** likh sakein.

Simple words mein:  
👉 **Jinja = HTML + Dynamic Data**

---

## ❓ Need of Jinja Template

Normal HTML:
- Static hoti hai  
- Same data baar-baar likhna padta hai  

Jinja use karne se:
- HTML **dynamic** ho jaati hai  
- Backend (Flask) ka data directly HTML mein use hota hai  

👉 Flask ke saath dynamic websites banane ke liye **Jinja zaroori** hai.

---

## 🌟 Advantages of Jinja

- HTML ke andar **Python variables** use kar sakte ho  
- **Loops (`for`)** aur **conditions (`if`)** likh sakte ho  
- Code duplication kam hota hai  
- Clean aur readable templates  
- Reusable layouts (base template concept)  
- Secure (auto-escaping XSS attacks se bachata hai)

---

## 🧠 Jinja vs Normal HTML

| Normal HTML | Jinja Template |
|-----------|----------------|
| Static | Dynamic |
| Repetition zyada | Reusable |
| No logic | Logic allowed |
| Backend se direct connect nahi | Backend data use hota hai |

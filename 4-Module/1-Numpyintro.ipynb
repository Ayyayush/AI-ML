# Numpy In Python
Numpy is the fundamental package for scientific programming in Python.  
It is a Python Library that provides a multidimensional array.  
At the core of the Numpy Package is the **ndarray object**.  
This encapsulates **n-dimensional arrays** of **homogenous data types**.  

NumPy is widely used for:
- Mathematical operations on arrays.
- Linear algebra computations.
- Fourier transforms.
- Random number generation.
- Integration with C/C++ and Fortran code.

It provides powerful **vectorization capabilities**, making operations faster compared to traditional Python lists.  
Internally, NumPy arrays are implemented in C, which leads to **efficient memory management** and **optimized performance**.


# Characteristics

1. **Homogeneous Data**  
   All elements in a NumPy array must be of the same data type (e.g., int, float, complex).  
   This allows for efficient storage and computation.

2. **Multidimensional**  
   NumPy supports 1D, 2D, and n-dimensional arrays (matrices, tensors, etc.), enabling complex mathematical operations easily.

3. **Fast and Memory Efficient**  
   NumPy arrays are implemented in C, making them significantly faster than standard Python lists for numerical tasks.

4. **Supports Broadcasting**  
   NumPy allows arithmetic operations between arrays of different shapes and sizes through a feature called **broadcasting**.

5. **Mathematical and Statistical Functions**  
   It provides a rich set of mathematical, logical, and statistical functions for array processing.

6. **Vectorization**  
   Operations can be performed on entire arrays without writing explicit loops, improving performance and readability.

7. **Integration**  
   NumPy integrates seamlessly with other Python libraries like Pandas, SciPy, Matplotlib, and TensorFlow.

8. **Slicing and Indexing**  
   Similar to lists, NumPy arrays support slicing, fancy indexing, and boolean indexing for flexible data access.

9. **Mutable**  
   NumPy arrays are mutable, meaning their elements can be modified after creation.

10. **Data Sharing**  
    Slicing an array creates a **view**, not a copy — meaning changes in one reflect in the other, saving memory.

## NumPy Array Attributes (Hinglish)

NumPy array ke **attributes** hume array ke baare mein important information dete hain  
jaise → shape, size, data type, dimensions, etc.

---

## 1. `ndim` (Number of Dimensions)

Batata hai array **kitne dimensions** ka hai.

- 1D → vector
- 2D → matrix
- 3D+ → tensor

👉 Simple words: array ki depth

---

## 2. `shape`

Batata hai array ka **structure / size in each dimension**.

- 1D → (elements,)
- 2D → (rows, columns)
- 3D → (blocks, rows, columns)

👉 Layout samajhne ke liye use hota hai

---

## 3. `size`

Array ke **total elements** batata hai.

Formula:
- `size = rows × columns × ...`

👉 Kitne total values hain array mein

---

## 4. `dtype` (Data Type)

Batata hai array ke elements ka **data type**.

Examples:
- `int32`
- `float64`
- `bool`

👉 NumPy arrays **same type ka data** store karte hain (homogeneous)

---

## 5. `itemsize`

Batata hai **ek element kitni memory** (bytes) le raha hai.

Example:
- `int32` → 4 bytes
- `float64` → 8 bytes

👉 Memory optimization samajhne ke kaam aata hai

---

## 6. `nbytes`

Batata hai array ne **total kitni memory** li hai.

Formula:
- `nbytes = size × itemsize`

👉 Pure array ki memory usage

---

## 7. `T` (Transpose)

Rows ko columns aur columns ko rows bana deta hai.

- Sirf **2D ya higher** arrays ke liye meaningful

👉 Matrix operations mein bahut use hota hai

---

## Changing Datatype in Python (Type Conversion)

Python mein **datatype change karne ko type conversion** kehte hain.  
Iska matlab hota hai ek datatype ki value ko kisi aur datatype mein convert karna.

---

## Types of Datatype Conversion

1. **Implicit Type Conversion**
2. **Explicit Type Conversion (Type Casting)**

---

## 1. Implicit Type Conversion

Python **automatically** datatype change karta hai  
jab data loss ka risk nahi hota.

### Example

- `int + float → float`
- Python khud convert kar deta hai

👉 Safe conversion hota hai

---

## 2. Explicit Type Conversion (Type Casting)

Jab hum **khud manually datatype change** karte hain using built-in functions.

Common functions:
- `int()`
- `float()`
- `str()`
- `bool()`
- `list()`
- `tuple()`
- `set()`

---

## Common Type Casting Examples

### 1. Convert to `int`

- `"10"` → `10`
- `3.7` → `3` (decimal part remove ho jaata hai)

⚠️ `"10.5"` directly `int()` mein error dega

---

### 2. Convert to `float`

- `5` → `5.0`
- `"3.14"` → `3.14`

---

### 3. Convert to `str`

- `10` → `"10"`
- `3.5` → `"3.5"`

👉 Printing ya concatenation ke liye useful

---

### 4. Convert to `bool`

Rules:
- `0`, `0.0`, `""`, `None` → `False`
- Baaki sab → `True`

Examples:
- `bool(0)` → False
- `bool(10)` → True
- `bool("")` → False

---

### 5. Convert Between Collections

- `list()` → list banata hai
- `tuple()` → tuple banata hai
- `set()` → unique elements

Example idea:
- List → Set (duplicates remove ho jaate hain)

---

## Why Datatype Conversion is Needed?

- User input string hota hai
- Mathematical operations ke liye
- Data cleaning
- Performance optimization

---

## Common Mistakes

- Invalid string ko number mein convert karna
- Data loss (float → int)
- Wrong boolean assumptions

---

## Summary

- Datatype conversion = type change
- Implicit → Python karta hai
- Explicit → Programmer karta hai
- Built-in functions se easy conversion possible



````python
# ============================================================
# ARRAY OPERATIONS ON NUMPY ARRAY
# ============================================================

# NumPy is a powerful library for numerical computations
import numpy as np


# ------------------------------------------------------------
# 1. CREATING NUMPY ARRAYS
# ------------------------------------------------------------

# Creating a 1D NumPy array
arr = np.array([1, 2, 3, 4, 5])
print("Original Array:", arr)


# ------------------------------------------------------------
# 2. BASIC ARITHMETIC OPERATIONS (ELEMENT-WISE)
# ------------------------------------------------------------

# Addition (adds 2 to every element)
print("Add 2:", arr + 2)

# Subtraction
print("Subtract 1:", arr - 1)

# Multiplication
print("Multiply by 3:", arr * 3)

# Division
print("Divide by 2:", arr / 2)


# ------------------------------------------------------------
# 3. ARRAY TO ARRAY OPERATIONS
# ------------------------------------------------------------

# Creating another array of same size
arr2 = np.array([10, 20, 30, 40, 50])
print("Second Array:", arr2)

# Element-wise addition
print("Array + Array:", arr + arr2)

# Element-wise multiplication
print("Array * Array:", arr * arr2)


# ------------------------------------------------------------
# 4. MATHEMATICAL FUNCTIONS
# ------------------------------------------------------------

print("Sum:", np.sum(arr))          # Sum of all elements
print("Mean:", np.mean(arr))        # Average
print("Maximum:", np.max(arr))      # Largest element
print("Minimum:", np.min(arr))      # Smallest element
print("Square Root:", np.sqrt(arr)) # Square root of each element


# ------------------------------------------------------------
# 5. INDEXING AND SLICING
# ------------------------------------------------------------

print("First element:", arr[0])     # Access first element
print("Last element:", arr[-1])     # Access last element

print("Slice (1 to 3):", arr[1:4])  # Slicing (index 1 to 3)


# ------------------------------------------------------------
# 6. CONDITIONAL OPERATIONS
# ------------------------------------------------------------

# Creating boolean array
print("arr > 3:", arr > 3)

# Filtering elements based on condition
print("Elements greater than 3:", arr[arr > 3])


# ------------------------------------------------------------
# 7. SHAPE, SIZE AND DIMENSIONS
# ------------------------------------------------------------

print("Shape:", arr.shape)          # Shape of array
print("Size:", arr.size)            # Total number of elements
print("Dimensions:", arr.ndim)      # Number of dimensions


# ------------------------------------------------------------
# 8. RESHAPING ARRAYS
# ------------------------------------------------------------

# Reshaping 1D array into 2D array (5 rows, 1 column)
reshaped_arr = arr.reshape(5, 1)
print("Reshaped Array:\n", reshaped_arr)


# ------------------------------------------------------------
# 9. TYPE CONVERSION
# ------------------------------------------------------------

# Convert array elements to float
float_arr = arr.astype(float)
print("Float Array:", float_arr)


# ------------------------------------------------------------
# 10. COPY VS VIEW
# ------------------------------------------------------------

# View (changes affect original array)
view_arr = arr.view()
view_arr[0] = 100
print("After modifying view:", arr)

# Copy (changes do NOT affect original array)
copy_arr = arr.copy()
copy_arr[1] = 200
print("After modifying copy:", arr)


# ------------------------------------------------------------
# 11. COMPARISON WITH PYTHON LIST
# ------------------------------------------------------------

# Python list
lst = [1, 2, 3, 4, 5]

# List comprehension (slower)
lst_result = [i * 2 for i in lst]

# NumPy operation (faster)
np_result = arr * 2

print("List result:", lst_result)
print("NumPy result:", np_result)


# ------------------------------------------------------------
# END OF FILE
# ------------------------------------------------------------
````

# Slicing in Python

Slicing ka matlab hota hai — **sequence (like list, tuple, string, ya NumPy array)** ke ek portion ko extract karna.  
Python me slicing ek powerful feature hai jo **index-based range** ke through kaam karta hai.

---

## 🔹 Basic Syntax

```python
sequence[start : stop : step]


````python
# Original array
arr = [10, 20, 30, 40, 50, 60]

# Basic slicing
print(arr[1:4])        # Output: [20, 30, 40]

# Slicing with omitted start or stop
print(arr[:4])          # Output: [10, 20, 30, 40] → start omitted, defaults to 0
print(arr[2:])          # Output: [30, 40, 50, 60] → stop omitted, goes till end

# Slicing with step
arr = [10, 20, 30, 40, 50, 60, 70, 80]
print(arr[1:7:2])       # Output: [20, 40, 60] → start=1, stop=7, step=2

# Slicing with negative indices
arr = [10, 20, 30, 40, 50, 60]
print(arr[-4:-1])       # Output: [30, 40, 50] → using negative indices

# String slicing
s = "PYTHON"
print(s[1:4])           # Output: "YTH" → slice from index 1 to 3
print(s[::-1])          # Output: "NOHTYP" → reverse string

# Reversing an array
arr = [10, 20, 30, 40, 50, 60]
print(arr[::-1])        # Output: [60, 50, 40, 30, 20, 10] → reverse array
````

````python
# ============================================================
# VARIOUS WAYS OF CREATING AND USING ARRAYS IN PYTHON
# ============================================================

# ------------------------------------------------------------
# 1️⃣ NUMPY ARRAYS
# ------------------------------------------------------------

import numpy as np


# -------------------------------
# 1.1 Creating array using np.array()
# -------------------------------

arr1 = np.array([10, 20, 30, 40, 50])
print("np.array():", arr1)


# -------------------------------
# 1.2 Creating array of zeros
# -------------------------------

zeros_arr = np.zeros(5)
print("np.zeros():", zeros_arr)


# -------------------------------
# 1.3 Creating array of ones
# -------------------------------

ones_arr = np.ones(5)
print("np.ones():", ones_arr)


# -------------------------------
# 1.4 Creating array using arange()
# -------------------------------
# Similar to Python range but returns NumPy array

range_arr = np.arange(1, 11)
print("np.arange():", range_arr)


# -------------------------------
# 1.5 Creating array using linspace()
# -------------------------------
# Creates evenly spaced numbers

lin_arr = np.linspace(1, 10, 5)
print("np.linspace():", lin_arr)


# -------------------------------
# 1.6 Creating array using reshape()
# -------------------------------

reshape_arr = np.array([1, 2, 3, 4, 5, 6]).reshape(2, 3)
print("np.array().reshape():\n", reshape_arr)


# ------------------------------------------------------------
# 2️⃣ RANDOM ARRAYS USING np.random
# ------------------------------------------------------------

# Random float values between 0 and 1
rand_arr = np.random.rand(5)
print("np.random.rand():", rand_arr)

# Random integers between 1 and 10
rand_int_arr = np.random.randint(1, 10, 5)
print("np.random.randint():", rand_int_arr)


# ------------------------------------------------------------
# 3️⃣ PYTHON BUILT-IN array MODULE
# ------------------------------------------------------------
# This is NOT NumPy
# It stores elements of SAME data type only

from array import *

# 'i' → integer type array
arr = array('i', [10, 20, 30, 40, 50])

print("\nPython array module:")
print("Array:", arr)

# Accessing elements
print("First element:", arr[0])     # 10
print("Third element:", arr[2])     # 30
print("Last element:", arr[4])      # 50


# ------------------------------------------------------------
# 4️⃣ COMPARISON: NumPy array vs array module
# ------------------------------------------------------------

print("\nComparison:")
print("NumPy supports vector operations →", arr1 * 2)
print("array module does NOT support this directly")
````

````python
# ============================================================
# NUMPY ARRAY SHAPE OPERATIONS
# 1. Reshape
# 2. Transpose
# 3. Ravel
# ============================================================

import numpy as np


# ------------------------------------------------------------
# ORIGINAL ARRAY
# ------------------------------------------------------------

# Creating a 1D NumPy array
arr = np.array([1, 2, 3, 4, 5, 6])
print("Original Array:")
print(arr)
print("Shape:", arr.shape)


# ------------------------------------------------------------
# 1️⃣ RESHAPE
# ------------------------------------------------------------
# reshape() changes the shape of the array WITHOUT changing data
# Total number of elements must remain same

reshaped_arr = arr.reshape(2, 3)   # 2 rows, 3 columns
print("\nAfter Reshape (2 x 3):")
print(reshaped_arr)
print("Shape:", reshaped_arr.shape)

# Another reshape
reshaped_arr2 = arr.reshape(3, 2)
print("\nAfter Reshape (3 x 2):")
print(reshaped_arr2)


# ------------------------------------------------------------
# 2️⃣ TRANSPOSE
# ------------------------------------------------------------
# transpose() swaps rows and columns
# Works mainly on 2D (or higher) arrays

transposed_arr = reshaped_arr.T    # OR np.transpose(reshaped_arr)

print("\nAfter Transpose:")
print(transposed_arr)
print("Shape:", transposed_arr.shape)

# Explanation:
# Original shape  : (2, 3)
# Transposed shape: (3, 2)


# ------------------------------------------------------------
# 3️⃣ RAVEL
# ------------------------------------------------------------
# ravel() converts any N-dimensional array into a 1D array
# It returns a VIEW (not a copy) whenever possible

ravel_arr = reshaped_arr.ravel()

print("\nAfter Ravel:")
print(ravel_arr)
print("Shape:", ravel_arr.shape)


# ------------------------------------------------------------
# EXTRA: RAVEL vs FLATTEN (IMPORTANT)
# ------------------------------------------------------------

# flatten() also converts to 1D but returns a COPY
flatten_arr = reshaped_arr.flatten()

# Modify ravel result
ravel_arr[0] = 100

print("\nAfter modifying ravel:")
print("Reshaped Array:")
print(reshaped_arr)     # AFFECTED

print("\nFlatten Array:")
print(flatten_arr)      # NOT affected
````


## Stacking in NumPy (Hinglish)

**Stacking** ka matlab hota hai **multiple arrays ko combine karna** ek naya array banane ke liye.

NumPy mein mainly 3 types ki stacking hoti hai:
- Vertical Stacking
- Horizontal Stacking
- Depth Stacking

---

## 1. Vertical Stacking (`vstack`)

- Arrays ko **row-wise** stack karta hai
- Matlab ek array ke neeche dusra array

👉 Condition: columns same hone chahiye


## 2. Horizontal Stacking (`hstack`)

- Arrays ko **column-wise** stack karta hai
- Matlab side-by-side combine karta hai

👉 Condition: rows same hone chahiye

---

## 3. Depth Stacking (`dstack`)

- Arrays ko **third dimension** mein stack karta hai
- Mostly image processing mein use hota hai

**Conceptual Example:**



````python
import numpy as np

print("=========== STACKING ===========")

# --------------------------------
# Base arrays
# --------------------------------
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Array a:", a)
print("Array b:", b)

# --------------------------------
# 1. np.stack()  (adds new axis)
# --------------------------------
print("\nnp.stack(axis=0):")
print(np.stack((a, b), axis=0))
# Shape -> (2, 3)

print("\nnp.stack(axis=1):")
print(np.stack((a, b), axis=1))
# Shape -> (3, 2)

# --------------------------------
# 2. np.vstack()  (row-wise stacking)
# --------------------------------
print("\nnp.vstack():")
print(np.vstack((a, b)))
# Shape -> (2, 3)

# --------------------------------
# 3. np.hstack()  (column-wise stacking)
# --------------------------------
print("\nnp.hstack():")
print(np.hstack((a, b)))
# Shape -> (6,)

# --------------------------------
# 4. np.concatenate()
# --------------------------------
print("\nnp.concatenate(axis=0):")
print(np.concatenate((a, b), axis=0))


print("\n=========== SPLITTING ===========")

# --------------------------------
# Base array for splitting
# --------------------------------
arr = np.array([1, 2, 3, 4, 5, 6])
print("\nOriginal array:", arr)

# --------------------------------
# 1. np.split()  (equal split)
# --------------------------------
print("\nnp.split into 3 parts:")
print(np.split(arr, 3))

# --------------------------------
# 2. np.split()  (unequal split using indices)
# --------------------------------
print("\nnp.split using indices [2, 4]:")
print(np.split(arr, [2, 4]))


# --------------------------------
# 2D array for hsplit / vsplit
# --------------------------------
arr2d = np.array([[1, 2, 3, 4],
                  [5, 6, 7, 8]])

print("\n2D array:\n", arr2d)

# --------------------------------
# 3. np.hsplit()  (column-wise split)
# --------------------------------
print("\nnp.hsplit into 2 parts:")
print(np.hsplit(arr2d, 2))

# --------------------------------
# 4. np.vsplit()  (row-wise split)
# --------------------------------
print("\nnp.vsplit into 2 parts:")
print(np.vsplit(arr2d, 2))
````

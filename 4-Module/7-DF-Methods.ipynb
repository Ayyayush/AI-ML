```python
ipl = pd.read_csv('content/ipl-matches.csv')
```

```python
# ------------------------------------------------------------
# Step 1: Check which MatchNumber values are NOT digits
# ------------------------------------------------------------
# str.isdigit() → True if value contains only digits (e.g. "12")
# ~ (tilde) → negates the condition
# Result: True for matches like "Final", "Qualifier", "Eliminator"
mask = ~ipl['MatchNumber'].str.isdigit()
mask
# ------------------------------------------------------------
# Step 2: Filter such matches and select Player_of_Match column
# ------------------------------------------------------------
# This keeps only non-numeric MatchNumber rows
players = ipl[mask]['Player_of_Match']
players

# ------------------------------------------------------------
# Step 3: Count frequency of Player of the Match awards
# ------------------------------------------------------------
# value_counts() → counts how many times each player appears
players.value_counts()
```

```python
# find which player has won most potm -> in finals and qualifiers
ipl[~ipl['MatchNumber'].str.isdigit()]['Player_of_Match'].value_counts()
```

```python
# Toss decision plot
ipl['TossDecision'].value_counts().plot(kind='pie')
```

```python
# how many matches each team has played

# method-1
# ------------------------------------------------------------
# Team1 column:
# Counts how many times each team appeared as Team1
# ------------------------------------------------------------
team1_matches = ipl['Team1'].value_counts()

# ------------------------------------------------------------
# Team2 column:
# Counts how many times each team appeared as Team2
# ------------------------------------------------------------
team2_matches = ipl['Team2'].value_counts()

# ------------------------------------------------------------
# Total matches played by each team
# ------------------------------------------------------------
# Adding both counts because:
# Every match has exactly one Team1 and one Team2
# So a team's total matches = appearances in Team1 + Team2
total_matches = team1_matches + team2_matches

# ------------------------------------------------------------
# Sort teams by total matches played (highest to lowest)
# ------------------------------------------------------------
total_matches.sort_values(ascending=False)




# method -2 
(ipl['Team2'].value_counts() + ipl['Team1'].value_counts()).sort_values(ascending=False)
```

```python
import pandas as pd

# ------------------------------------------------------------
# Create a DataFrame containing two points per row
# ------------------------------------------------------------
# Each cell stores a tuple representing (x, y) coordinates
points_df = pd.DataFrame(
    {
        '1st point': [(3, 4), (-6, 5), (0, 0), (-10, 1), (4, 5)],
        '2nd point': [(-3, 4), (0, 0), (2, 2), (10, 10), (1, 1)]
    }
)

points_df


# ------------------------------------------------------------
# Function to calculate Euclidean distance between two points
# ------------------------------------------------------------
# Formula:
# distance = √[(x2 - x1)^2 + (y2 - y1)^2]
def euclidean(row):
    
    # Extract first point (x1, y1)
    pt_A = row['1st point']
    
    # Extract second point (x2, y2)
    pt_B = row['2nd point']
    
    # Apply Euclidean distance formula
    return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] - pt_B[1])**2)**0.5
```

```python
# ------------------------------------------------------------
# apply(..., axis=1) → function runs on each ROW
# ------------------------------------------------------------
# A new column 'distance' is created permanently
# chahe toh bina naya column add kare bhi kr skte the 
points_df['distance'] = points_df.apply(euclidean, axis=1)

points_df
```

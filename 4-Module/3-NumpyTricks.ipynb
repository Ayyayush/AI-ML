# Definition

## np.sort  
Sorts the elements of an array along a specified axis and returns a sorted copy of the array.

## np.append  
Appends values to the end of an array and returns a new array (does not modify the original array).

## np.concatenate  
Joins two or more arrays along an existing axis into a single array.

## np.unique  
Returns the unique elements of an array by removing all duplicate values.

## np.expand_dims  
Expands the shape of an array by adding a new axis at the specified position.

## np.where  
`np.where` is used in **three main ways** to locate or select data based on a condition.

### 1️⃣ Condition Only  
Returns the **indices** where the condition is `True`.

> Used to find positions of elements satisfying a condition.

### 2️⃣ Condition with Two Arrays  
Selects values from one array if condition is `True`, otherwise from another array.

> Works like a **vectorized if-else**.

### 3️⃣ Condition on Multi-dimensional Arrays  
Returns indices for each dimension where the condition is `True`.

> Helpful in filtering rows/columns in 2D or higher arrays.

---

## np.argmax  
Returns the index of the **maximum value** along a specified axis in an array.

---

## np.argmin  
Returns the index of the **minimum value** along a specified axis in an array.


```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Sort a 1D array in ascending order
# -----------------------------------------------------------
arr1 = np.array([5, 2, 9, 1, 7])
result1 = np.sort(arr1)
print("Example 1 → Basic 1D sort:\n", result1)
# Output: [1 2 5 7 9]


# -----------------------------------------------------------
# 🧩 Example 2: Sort a 1D array in descending order
# -----------------------------------------------------------
arr2 = np.array([5, 2, 9, 1, 7])
result2 = np.sort(arr2)[::-1]
print("\nExample 2 → Descending order sort:\n", result2)
# Output: [9 7 5 2 1]


# -----------------------------------------------------------
# 🧩 Example 3: Sort a 2D array row-wise (axis=1)
# -----------------------------------------------------------
arr3 = np.array([[9, 2, 7],
                 [4, 6, 1]])
result3 = np.sort(arr3, axis=1)
print("\nExample 3 → Row-wise sort (axis=1):\n", result3)
# Output:
# [[2 7 9]
#  [1 4 6]]


# -----------------------------------------------------------
# 🧩 Example 4: Sort a 2D array column-wise (axis=0)
# -----------------------------------------------------------
arr4 = np.array([[9, 2, 7],
                 [4, 6, 1]])
result4 = np.sort(arr4, axis=0)
print("\nExample 4 → Column-wise sort (axis=0):\n", result4)
# Output:
# [[4 2 1]
#  [9 6 7]]


# -----------------------------------------------------------
# 🧩 Example 5: Sort a multi-dimensional array (flattened)
# -----------------------------------------------------------
arr5 = np.array([[3, 2], [8, 5]])
result5 = np.sort(arr5, axis=None)
print("\nExample 5 → Flatten and sort (axis=None):\n", result5)
# Output: [2 3 5 8]


# -----------------------------------------------------------
# 🧩 Example 6: Use different sorting algorithms via 'kind'
# -----------------------------------------------------------
arr6 = np.array([10, 40, 30, 20, 10])

# Quicksort (default)
quick_sorted = np.sort(arr6, kind='quicksort')

# Mergesort
merge_sorted = np.sort(arr6, kind='mergesort')

# Heapsort
heap_sorted = np.sort(arr6, kind='heapsort')

# Stable sort (preserves order of equal elements)
stable_sorted = np.sort(arr6, kind='stable')

print("\nExample 6 → Sorting with different algorithms:")
print("Quicksort:", quick_sorted)
print("Mergesort:", merge_sorted)
print("Heapsort:", heap_sorted)
print("Stable sort:", stable_sorted)
# Output: All will be [10 10 20 30 40] but stable maintains relative order


# -----------------------------------------------------------
# 🧩 Example 7: Sort structured arrays using 'order'
# -----------------------------------------------------------
data = np.array([
    ('Charlie', 20),
    ('Alice', 25),
    ('Bob', 22)
], dtype=[('name', 'U10'), ('age', 'i4')])

result7 = np.sort(data, order='age')
print("\nExample 7 → Structured array sorted by field 'age':\n", result7)
# Output:
# [('Charlie', 20) ('Bob', 22) ('Alice', 25)]


# -----------------------------------------------------------
# 🧩 Example 8: Sorting complex numbers by magnitude
# -----------------------------------------------------------
arr8 = np.array([3+4j, 1+2j, 5+1j])
result8 = np.sort(np.abs(arr8))  # sort by absolute value
print("\nExample 8 → Sort complex numbers by magnitude:\n", result8)
# Output: [2.23606798 4.12310563 5.09901951]


# -----------------------------------------------------------
# 🧩 Example 9: Using argsort() to get sorted indices
# -----------------------------------------------------------
arr9 = np.array([50, 10, 40, 30])
indices = np.argsort(arr9)
sorted_arr = arr9[indices]
print("\nExample 9 → Using np.argsort():")
print("Indices that sort the array:", indices)
print("Sorted array:", sorted_arr)
# Output:
# Indices: [1 3 2 0]
# Sorted array: [10 30 40 50]


# -----------------------------------------------------------
# 🧩 Example 10: Partial sorting using np.partition()
# -----------------------------------------------------------
arr10 = np.array([10, 3, 7, 5, 9])
result10 = np.partition(arr10, 2)  # elements before index 2 are smallest 2 elements
print("\nExample 10 → Partial sort with np.partition():\n", result10)
# Output (order before pivot not guaranteed): [5 3 7 10 9]
```

## 3️⃣ Key Difference: `kind` vs `order`

| Feature        | `kind`                         | `order`                         |
|---------------|--------------------------------|---------------------------------|
| Used for      | Sorting algorithm               | Field / column selection        |
| Works on      | Normal NumPy arrays             | Structured (record) arrays      |
| Controls      | **How** sorting is performed    | **Which column** is sorted      |
| Example       | `quicksort`, `mergesort`        | `age`, `name`                   |

### One-Line Summary
- **`kind`** decides *kaunsa sorting algorithm use hoga*  
- **`order`** decides *kis field/column ke basis par sorting hogi*


```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Append a single value to a 1D array
# -----------------------------------------------------------
arr1 = np.array([10, 20, 30])
result1 = np.append(arr1, 200)
print("Example 1 → Append single element:\n", result1)
# Output: [ 10  20  30 200]


# -----------------------------------------------------------
# 🧩 Example 2: Append multiple values to a 1D array
# -----------------------------------------------------------
arr2 = np.array([1, 2, 3])
result2 = np.append(arr2, [4, 5, 6])
print("\nExample 2 → Append multiple elements:\n", result2)
# Output: [1 2 3 4 5 6]


# -----------------------------------------------------------
# 🧩 Example 3: Append row to a 2D array (axis not specified → flattened)
# -----------------------------------------------------------
arr3 = np.array([[1, 2], [3, 4]])
result3 = np.append(arr3, [5, 6])
print("\nExample 3 → Append without axis (array flattened):\n", result3)
# Output: [1 2 3 4 5 6]


# -----------------------------------------------------------
# 🧩 Example 4: Append row to a 2D array (axis=0 → add new row)
# -----------------------------------------------------------
arr4 = np.array([[1, 2], [3, 4]])
result4 = np.append(arr4, [[5, 6]], axis=0)
print("\nExample 4 → Append new row (axis=0):\n", result4)
# Output:
# [[1 2]
#  [3 4]
#  [5 6]]


# -----------------------------------------------------------
# 🧩 Example 5: Append column to a 2D array (axis=1 → add new column)
# -----------------------------------------------------------
arr5 = np.array([[1, 2], [3, 4]])
new_col = np.array([[5], [6]])
result5 = np.append(arr5, new_col, axis=1)
print("\nExample 5 → Append new column (axis=1):\n", result5)
# Output:
# [[1 2 5]
#  [3 4 6]]


# -----------------------------------------------------------
# 🧩 Example 6: Append arrays of equal shape (row-wise)
# -----------------------------------------------------------
a = np.array([[10, 20], [30, 40]])
b = np.array([[50, 60]])
result6 = np.append(a, b, axis=0)
print("\nExample 6 → Append 2D arrays row-wise:\n", result6)
# Output:
# [[10 20]
#  [30 40]
#  [50 60]]


# -----------------------------------------------------------
# 🧩 Example 7: Append arrays of equal shape (column-wise)
# -----------------------------------------------------------
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
result7 = np.append(a, b, axis=1)
print("\nExample 7 → Append 2D arrays column-wise:\n", result7)
# Output:
# [[1 2 5 6]
#  [3 4 7 8]]


# -----------------------------------------------------------
# 🧩 Example 8: Append multiple 1D arrays together
# -----------------------------------------------------------
arrs = [np.array([1, 2]), np.array([3, 4]), np.array([5])]
merged = np.append(arrs[0], [arrs[1], arrs[2]])
print("\nExample 8 → Append multiple arrays:\n", merged)
# Output: [1 2 3 4 5]
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Concatenate two 1D arrays
# -----------------------------------------------------------
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
result1 = np.concatenate((arr1, arr2))
print("Example 1 → Concatenate 1D arrays:\n", result1)
# Output: [1 2 3 4 5 6]


# -----------------------------------------------------------
# 🧩 Example 2: Concatenate more than two 1D arrays
# -----------------------------------------------------------
arr3 = np.array([7, 8])
result2 = np.concatenate((arr1, arr2, arr3))
print("\nExample 2 → Concatenate multiple 1D arrays:\n", result2)
# Output: [1 2 3 4 5 6 7 8]


# -----------------------------------------------------------
# 🧩 Example 3: Concatenate 2D arrays along rows (axis=0)
# -----------------------------------------------------------
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6]])
result3 = np.concatenate((a, b), axis=0)
print("\nExample 3 → Concatenate 2D arrays row-wise (axis=0):\n", result3)
# Output:
# [[1 2]
#  [3 4]
#  [5 6]]


# -----------------------------------------------------------
# 🧩 Example 4: Concatenate 2D arrays along columns (axis=1)
# -----------------------------------------------------------
c = np.array([[7],
              [8],
              [9]])
result4 = np.concatenate((result3, c), axis=1)
print("\nExample 4 → Concatenate 2D arrays column-wise (axis=1):\n", result4)
# Output:
# [[1 2 7]
#  [3 4 8]
#  [5 6 9]]


# -----------------------------------------------------------
# 🧩 Example 5: Concatenate 2D arrays (flattened if axis=None)
# -----------------------------------------------------------
arrA = np.array([[10, 20], [30, 40]])
arrB = np.array([[50, 60]])
result5 = np.concatenate((arrA.flatten(), arrB.flatten()))
print("\nExample 5 → Flatten both and concatenate:\n", result5)
# Output: [10 20 30 40 50 60]


# -----------------------------------------------------------
# 🧩 Example 6: Concatenate 3D arrays (axis=0)
# -----------------------------------------------------------
arr6a = np.array([[[1, 2]], [[3, 4]]])   # shape (2,1,2)
arr6b = np.array([[[5, 6]]])             # shape (1,1,2)
result6 = np.concatenate((arr6a, arr6b), axis=0)
print("\nExample 6 → Concatenate 3D arrays (axis=0):\n", result6)
# Output:
# [[[1 2]]
#  [[3 4]]
#  [[5 6]]]


# -----------------------------------------------------------
# 🧩 Example 7: Concatenate with mismatched shapes (❌ Error)
# -----------------------------------------------------------
try:
    x = np.array([[1, 2, 3]])
    y = np.array([[4, 5]])
    np.concatenate((x, y), axis=1)
except ValueError as e:
    print("\nExample 7 → Mismatched shapes error:")
    print(e)
# Output: ValueError (dimensions along concatenation axis must match)


# -----------------------------------------------------------
# 🧩 Example 8: Concatenate using np.r_ and np.c_ shortcuts
# -----------------------------------------------------------
r_concat = np.r_[arr1, arr2]      # row-wise concat (for 1D)
c_concat = np.c_[a, b.T]          # column-wise concat (for 2D)
print("\nExample 8 → Using np.r_ and np.c_ shortcuts:")
print("np.r_:", r_concat)
print("np.c_:\n", c_concat)
# Output:
# np.r_: [1 2 3 4 5 6]
# np.c_:
# [[1 2 5]
#  [3 4 6]]


# -----------------------------------------------------------
# 🧩 Example 9: Concatenate using different axes on 3D arrays
# -----------------------------------------------------------
X = np.arange(8).reshape(2, 2, 2)
Y = np.arange(8, 16).reshape(2, 2, 2)

concat_axis0 = np.concatenate((X, Y), axis=0)
concat_axis1 = np.concatenate((X, Y), axis=1)
concat_axis2 = np.concatenate((X, Y), axis=2)

print("\nExample 9 → Concatenate 3D arrays along different axes:")
print("Axis=0 shape:", concat_axis0.shape)
print("Axis=1 shape:", concat_axis1.shape)
print("Axis=2 shape:", concat_axis2.shape)
# Output:
# Axis=0 shape: (4, 2, 2)
# Axis=1 shape: (2, 4, 2)
# Axis=2 shape: (2, 2, 4)
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Get unique elements from a 1D array
# -----------------------------------------------------------
arr1 = np.array([1, 2, 2, 3, 4, 4, 4, 5])
result1 = np.unique(arr1)
print("Example 1 → Unique elements from 1D array:\n", result1)
# Output: [1 2 3 4 5]


# -----------------------------------------------------------
# 🧩 Example 2: Get unique elements from a 2D array (flattened)
# -----------------------------------------------------------
arr2 = np.array([[1, 2, 2],
                 [3, 4, 1]])
result2 = np.unique(arr2)
print("\nExample 2 → Unique elements from 2D array (flattened):\n", result2)
# Output: [1 2 3 4]


# -----------------------------------------------------------
# 🧩 Example 3: Get unique elements along a specific axis (axis=0)
# -----------------------------------------------------------
arr3 = np.array([[1, 2],
                 [1, 2],
                 [3, 4]])
result3 = np.unique(arr3, axis=0)
print("\nExample 3 → Unique rows (axis=0):\n", result3)
# Output:
# [[1 2]
#  [3 4]]


# -----------------------------------------------------------
# 🧩 Example 4: Get unique elements along a specific axis (axis=1)
# -----------------------------------------------------------
arr4 = np.array([[1, 1, 2],
                 [3, 3, 4]])
result4 = np.unique(arr4, axis=1)
print("\nExample 4 → Unique columns (axis=1):\n", result4)
# Output:
# [[1 2]
#  [3 4]]


# -----------------------------------------------------------
# 🧩 Example 5: Return both unique elements and their indices
# -----------------------------------------------------------
arr5 = np.array([3, 1, 2, 3, 2, 1, 4])
unique_vals, indices = np.unique(arr5, return_index=True)
print("\nExample 5 → Unique elements with first occurrence indices:")
print("Unique values:", unique_vals)
print("Indices:", indices)
# Output:
# Unique values: [1 2 3 4]
# Indices: [1 2 0 6]


# -----------------------------------------------------------
# 🧩 Example 6: Return unique elements and counts
# -----------------------------------------------------------
arr6 = np.array([1, 2, 2, 3, 3, 3, 4])
unique_vals, counts = np.unique(arr6, return_counts=True)
print("\nExample 6 → Unique elements with their counts:")
print("Unique values:", unique_vals)
print("Counts:", counts)
# Output:
# Unique values: [1 2 3 4]
# Counts: [1 2 3 1]


# -----------------------------------------------------------
# 🧩 Example 7: Return inverse indices (to reconstruct original array)
# -----------------------------------------------------------
arr7 = np.array([10, 20, 10, 30])
unique_vals, inverse = np.unique(arr7, return_inverse=True)
reconstructed = unique_vals[inverse]
print("\nExample 7 → Unique values with inverse mapping:")
print("Unique values:", unique_vals)
print("Inverse indices:", inverse)
print("Reconstructed array:", reconstructed)
# Output:
# Unique values: [10 20 30]
# Inverse indices: [0 1 0 2]
# Reconstructed array: [10 20 10 30]


# -----------------------------------------------------------
# 🧩 Example 8: Return both counts and indices together
# -----------------------------------------------------------
arr8 = np.array(['apple', 'banana', 'apple', 'cherry', 'banana'])
unique_vals, indices, counts = np.unique(arr8, return_index=True, return_counts=True)
print("\nExample 8 → Unique strings with index & count:")
print("Unique values:", unique_vals)
print("Indices:", indices)
print("Counts:", counts)
# Output:
# Unique values: ['apple' 'banana' 'cherry']
# Indices: [0 1 3]
# Counts: [2 2 1]


# -----------------------------------------------------------
# 🧩 Example 9: Sorting behavior (np.unique always sorts results)
# -----------------------------------------------------------
arr9 = np.array([40, 10, 30, 10])
result9 = np.unique(arr9)
print("\nExample 9 → np.unique() always returns sorted unique values:\n", result9)
# Output: [10 30 40]


# -----------------------------------------------------------
# 🧩 Example 10: Working with boolean or mixed data
# -----------------------------------------------------------
arr10 = np.array([True, False, True, False, True])
result10 = np.unique(arr10)
print("\nExample 10 → Unique elements from boolean array:\n", result10)
# Output: [False  True]
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Add a new axis to a 1D array (axis=0)
# -----------------------------------------------------------
arr1 = np.array([1, 2, 3, 4])
expanded1 = np.expand_dims(arr1, axis=0)
print("Example 1 → Add new axis at position 0:\n", expanded1)
print("Shape:", expanded1.shape)
# Output:
# [[1 2 3 4]]
# Shape: (1, 4)


# -----------------------------------------------------------
# 🧩 Example 2: Add a new axis to a 1D array (axis=1)
# -----------------------------------------------------------
expanded2 = np.expand_dims(arr1, axis=1)
print("\nExample 2 → Add new axis at position 1:\n", expanded2)
print("Shape:", expanded2.shape)
# Output:
# [[1]
#  [2]
#  [3]
#  [4]]
# Shape: (4, 1)


# -----------------------------------------------------------
# 🧩 Example 3: Add new axes to a 2D array (axis=0)
# -----------------------------------------------------------
arr2 = np.array([[10, 20], [30, 40]])
expanded3 = np.expand_dims(arr2, axis=0)
print("\nExample 3 → Add new axis at position 0:\n", expanded3)
print("Shape:", expanded3.shape)
# Output:
# [[[10 20]
#   [30 40]]]
# Shape: (1, 2, 2)


# -----------------------------------------------------------
# 🧩 Example 4: Add new axis at the end (axis=-1)
# -----------------------------------------------------------
expanded4 = np.expand_dims(arr2, axis=-1)
print("\nExample 4 → Add new axis at last position:\n", expanded4)
print("Shape:", expanded4.shape)
# Output:
# [[[10]
#   [20]]
#  [[30]
#   [40]]]
# Shape: (2, 2, 1)


# -----------------------------------------------------------
# 🧩 Example 5: Expand multiple dimensions using chaining
# -----------------------------------------------------------
arr3 = np.array([1, 2, 3])
expanded5 = np.expand_dims(np.expand_dims(arr3, axis=0), axis=-1)
print("\nExample 5 → Add multiple new axes (like 3D):\n", expanded5)
print("Shape:", expanded5.shape)
# Output:
# [[[1]
#   [2]
#   [3]]]
# Shape: (1, 3, 1)


# -----------------------------------------------------------
# 🧩 Example 6: Equivalent operation using np.newaxis
# -----------------------------------------------------------
arr4 = np.array([10, 20, 30])
newaxis_ver = arr4[np.newaxis, :]
print("\nExample 6 → Using np.newaxis (same as expand_dims axis=0):\n", newaxis_ver)
print("Shape:", newaxis_ver.shape)
# Output:
# [[10 20 30]]
# Shape: (1, 3)


# -----------------------------------------------------------
# 🧩 Example 7: Add axis before and after using np.newaxis
# -----------------------------------------------------------
expanded7 = arr4[:, np.newaxis]
print("\nExample 7 → Using np.newaxis to add column dimension:\n", expanded7)
print("Shape:", expanded7.shape)
# Output:
# [[10]
#  [20]
#  [30]]
# Shape: (3, 1)


# -----------------------------------------------------------
# 🧩 Example 8: Add axis in middle of a 3D array
# -----------------------------------------------------------
arr5 = np.arange(12).reshape(2, 3, 2)
expanded8 = np.expand_dims(arr5, axis=2)
print("\nExample 8 → Add axis in middle of a 3D array:\n", expanded8.shape)
# Output:
# Original shape: (2, 3, 2)
# New shape: (2, 3, 1, 2)


# -----------------------------------------------------------
# 🧩 Example 9: Compare np.expand_dims vs reshape()
# -----------------------------------------------------------
arr6 = np.array([1, 2, 3, 4])
expanded9 = np.expand_dims(arr6, axis=0)
reshaped9 = arr6.reshape(1, 4)
print("\nExample 9 → expand_dims vs reshape (same result):")
print("expand_dims shape:", expanded9.shape)
print("reshape shape:", reshaped9.shape)
# Output:
# Both: (1, 4)


# -----------------------------------------------------------
# 🧩 Example 10: Expand for broadcasting
# -----------------------------------------------------------
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
a_exp = np.expand_dims(a, axis=1)  # shape (3,1)
result = a_exp + b                 # broadcast addition
print("\nExample 10 → Using expand_dims for broadcasting:")
print("a_exp shape:", a_exp.shape)
print("b shape:", b.shape)
print("Result:\n", result)
# Output:
# a_exp shape: (3, 1)
# b shape: (3,)
# Result:
# [[11 21 31]
#  [12 22 32]
#  [13 23 33]]
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Basic conditional usage (1D array)
# -----------------------------------------------------------
arr = np.array([10, 20, 30, 40, 50])
indices = np.where(arr > 25)
print("Example 1 → Indices where arr > 25:\n", indices)
print("Values:", arr[indices])
# Output:
# (array([2, 3, 4]),)
# Values: [30 40 50]


# -----------------------------------------------------------
# 🧩 Example 2: Using np.where with condition, x, and y (element-wise)
# -----------------------------------------------------------
arr2 = np.array([1, 2, 3, 4, 5])
result = np.where(arr2 % 2 == 0, 'Even', 'Odd')
print("\nExample 2 → Replace elements based on condition:\n", result)
# Output: ['Odd' 'Even' 'Odd' 'Even' 'Odd']


# -----------------------------------------------------------
# 🧩 Example 3: Multiple conditions using logical operators
# -----------------------------------------------------------
arr3 = np.array([5, 10, 15, 20, 25, 30])
cond = (arr3 > 10) & (arr3 < 25)
indices = np.where(cond)
print("\nExample 3 → Indices where 10 < arr < 25:\n", indices)
print("Values:", arr3[indices])
# Output:
# (array([2, 3, 4]),)
# Values: [15 20 25]


# -----------------------------------------------------------
# 🧩 Example 4: np.where with 2D array condition
# -----------------------------------------------------------
arr4 = np.array([[10, 5, 8],
                 [3, 12, 7],
                 [15, 2, 9]])
indices = np.where(arr4 > 8)
print("\nExample 4 → Indices where arr4 > 8:\n", indices)
print("Values:", arr4[indices])
# Output:
# Indices: (array([0, 1, 2]), array([0, 1, 2]))
# Values: [10 12 15  9]


# -----------------------------------------------------------
# 🧩 Example 5: Using np.where to replace values (2D array)
# -----------------------------------------------------------
arr5 = np.array([[1, 2, 3],
                 [4, 5, 6]])
new_arr5 = np.where(arr5 % 2 == 0, arr5, -1)
print("\nExample 5 → Replace odd numbers with -1:\n", new_arr5)
# Output:
# [[-1  2 -1]
#  [ 4 -1  6]]


# -----------------------------------------------------------
# 🧩 Example 6: Return different arrays based on condition
# -----------------------------------------------------------
x = np.array([100, 200, 300, 400])
y = np.array([10, 20, 30, 40])
result = np.where(x > 250, x, y)
print("\nExample 6 → Pick values from x or y based on condition:\n", result)
# Output: [ 10  20 300 400]


# -----------------------------------------------------------
# 🧩 Example 7: Using np.where() for NaN replacement
# -----------------------------------------------------------
arr7 = np.array([10, np.nan, 20, np.nan, 30])
cleaned = np.where(np.isnan(arr7), 0, arr7)
print("\nExample 7 → Replace NaN with 0:\n", cleaned)
# Output: [10.  0. 20.  0. 30.]


# -----------------------------------------------------------
# 🧩 Example 8: Using np.where() for thresholding
# -----------------------------------------------------------
arr8 = np.array([3, 7, 12, 18, 5])
thresholded = np.where(arr8 > 10, 10, arr8)
print("\nExample 8 → Cap all values >10 to 10:\n", thresholded)
# Output: [ 3  7 10 10  5]


# -----------------------------------------------------------
# 🧩 Example 9: Using np.where() with images (3D array example)
# -----------------------------------------------------------
img = np.array([[[255, 255, 255], [100, 100, 100]],
                [[50, 50, 50], [200, 200, 200]]])
# Turn bright pixels (sum > 400) into black [0,0,0]
darkened = np.where(img.sum(axis=2, keepdims=True) > 400, 0, img)
print("\nExample 9 → Replace bright pixels with black:\n", darkened)
# Output: [[[  0   0   0]
#            [100 100 100]]
#           [[ 50  50  50]
#            [  0   0   0]]]


# -----------------------------------------------------------
# 🧩 Example 10: Using np.where() for coordinate extraction
# -----------------------------------------------------------
grid = np.array([[0, 1, 0],
                 [1, 1, 0],
                 [0, 0, 1]])
coords = np.where(grid == 1)
print("\nExample 10 → Coordinates where value == 1:")
for r, c in zip(coords[0], coords[1]):
    print(f"({r}, {c})")
# Output:
# (0, 1)
# (1, 0)
# (1, 1)
# (2, 2)
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Basic 1D array usage
# -----------------------------------------------------------
arr = np.array([10, 25, 5, 42, 18])
index = np.argmax(arr)
print("Example 1 → Index of max element:", index)
print("Max value:", arr[index])
# Output:
# Index: 3
# Value: 42


# -----------------------------------------------------------
# 🧩 Example 2: 2D array – flatten by default
# -----------------------------------------------------------
arr2 = np.array([[3, 7, 2],
                 [9, 1, 4]])
index2 = np.argmax(arr2)
print("\nExample 2 → Flattened array max index:", index2)
print("Max value:", arr2.flatten()[index2])
# Output:
# Index: 3 (corresponds to value 9)


# -----------------------------------------------------------
# 🧩 Example 3: Along specific axis (axis=0 → column-wise)
# -----------------------------------------------------------
arr3 = np.array([[5, 8, 12],
                 [20, 3, 9],
                 [7, 15, 10]])
col_indices = np.argmax(arr3, axis=0)
print("\nExample 3 → Column-wise argmax:", col_indices)
# Output: [1 2 0]
# Explanation:
# For col1: max=20 (row1)
# For col2: max=15 (row2)
# For col3: max=12 (row0)


# -----------------------------------------------------------
# 🧩 Example 4: Along rows (axis=1 → row-wise)
# -----------------------------------------------------------
row_indices = np.argmax(arr3, axis=1)
print("\nExample 4 → Row-wise argmax:", row_indices)
# Output: [2 0 1]
# Each value shows index of max in that row.


# -----------------------------------------------------------
# 🧩 Example 5: Using np.unravel_index() to get 2D position
# -----------------------------------------------------------
arr4 = np.array([[2, 5, 9],
                 [4, 11, 6]])
flat_index = np.argmax(arr4)
pos = np.unravel_index(flat_index, arr4.shape)
print("\nExample 5 → 2D position of max element:", pos)
print("Max value:", arr4[pos])
# Output:
# Index: (1, 1)
# Value: 11


# -----------------------------------------------------------
# 🧩 Example 6: np.argmax() on boolean array
# -----------------------------------------------------------
bool_arr = np.array([False, False, True, True])
first_true = np.argmax(bool_arr)
print("\nExample 6 → First True index:", first_true)
# Output: 2
# Note: argmax returns first maximum (True=1 > False=0)


# -----------------------------------------------------------
# 🧩 Example 7: With NaN values (careful!)
# -----------------------------------------------------------
arr7 = np.array([np.nan, 10, 5, np.nan, 20])
index7 = np.argmax(arr7)
print("\nExample 7 → Handling NaN values:", index7)
print("Max value:", arr7[index7])
# Output:
# Index: 4
# NaN is ignored in comparison (treated as invalid)


# -----------------------------------------------------------
# 🧩 Example 8: np.argmax() on 3D array
# -----------------------------------------------------------
arr8 = np.array([[[1, 5],
                  [7, 3]],
                 [[9, 2],
                  [4, 6]]])
index8 = np.argmax(arr8)
coords = np.unravel_index(index8, arr8.shape)
print("\nExample 8 → 3D array argmax position:", coords)
print("Max value:", arr8[coords])
# Output:
# Position: (1, 0, 0)
# Value: 9


# -----------------------------------------------------------
# 🧩 Example 9: Argmax along last axis
# -----------------------------------------------------------
arr9 = np.array([[1, 4, 9],
                 [8, 3, 2]])
res9 = np.argmax(arr9, axis=-1)
print("\nExample 9 → Argmax along last axis:", res9)
# Output: [2 0]
# Row 1: max=9 at index 2
# Row 2: max=8 at index 0


# -----------------------------------------------------------
# 🧩 Example 10: Compare with np.max()
# -----------------------------------------------------------
arr10 = np.array([2, 4, 8, 1])
print("\nExample 10 → Difference between np.max() and np.argmax():")
print("np.max(arr10)  =", np.max(arr10))
print("np.argmax(arr10) =", np.argmax(arr10))
# Output:
# Max value = 8
# Argmax index = 2
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Basic 1D array cumulative sum
# -----------------------------------------------------------
arr = np.array([1, 2, 3, 4, 5])
cumulative = np.cumsum(arr)
print("Example 1 → Cumulative sum of arr:\n", cumulative)
# Output: [ 1  3  6 10 15]
# Explanation:
# Each element = sum of all previous elements including itself


# -----------------------------------------------------------
# 🧩 Example 2: Start from a different initial array
# -----------------------------------------------------------
arr2 = np.array([10, -2, 7, 5])
cum2 = np.cumsum(arr2)
print("\nExample 2 → Handles positive & negative values:\n", cum2)
# Output: [10  8 15 20]


# -----------------------------------------------------------
# 🧩 Example 3: Using np.cumsum on 2D array (flattened by default)
# -----------------------------------------------------------
arr3 = np.array([[1, 2, 3],
                 [4, 5, 6]])
cum3 = np.cumsum(arr3)
print("\nExample 3 → Flattened cumulative sum:\n", cum3)
# Output: [ 1  3  6 10 15 21]


# -----------------------------------------------------------
# 🧩 Example 4: Cumulative sum along axis=0 (column-wise)
# -----------------------------------------------------------
cum4 = np.cumsum(arr3, axis=0)
print("\nExample 4 → Column-wise cumulative sum (axis=0):\n", cum4)
# Output:
# [[1 2 3]
#  [5 7 9]]


# -----------------------------------------------------------
# 🧩 Example 5: Cumulative sum along axis=1 (row-wise)
# -----------------------------------------------------------
cum5 = np.cumsum(arr3, axis=1)
print("\nExample 5 → Row-wise cumulative sum (axis=1):\n", cum5)
# Output:
# [[ 1  3  6]
#  [ 4  9 15]]


# -----------------------------------------------------------
# 🧩 Example 6: Cumulative sum of boolean array
# -----------------------------------------------------------
bool_arr = np.array([True, False, True, True])
cum6 = np.cumsum(bool_arr)
print("\nExample 6 → Boolean array cumulative sum:\n", cum6)
# Output: [1 1 2 3]
# Explanation:
# True = 1, False = 0


# -----------------------------------------------------------
# 🧩 Example 7: Cumulative sum on 3D array
# -----------------------------------------------------------
arr7 = np.arange(1, 9).reshape(2, 2, 2)
cum7 = np.cumsum(arr7, axis=2)
print("\nExample 7 → 3D array cumulative sum (axis=2):\n", cum7)
# Output:
# [[[ 1  3]
#   [ 3  7]]
#  [[ 5 11]
#   [ 7 15]]]


# -----------------------------------------------------------
# 🧩 Example 8: Compare np.cumsum() with running total using loop
# -----------------------------------------------------------
arr8 = np.array([2, 4, 6, 8])
running_total = []
total = 0
for val in arr8:
    total += val
    running_total.append(total)
print("\nExample 8 → Manual running total:", running_total)
print("np.cumsum() equivalent:", np.cumsum(arr8))
# Output: both [2 6 12 20]


# -----------------------------------------------------------
# 🧩 Example 9: Use np.cumsum() for moving sum threshold
# -----------------------------------------------------------
arr9 = np.array([3, 1, 4, 2, 5])
cum9 = np.cumsum(arr9)
threshold_index = np.where(cum9 > 7)[0][0]
print("\nExample 9 → First index where cumulative sum > 7:", threshold_index)
# Output: 2 (since 3+1+4 = 8 > 7)


# -----------------------------------------------------------
# 🧩 Example 10: Using np.cumsum() with dtype specification
# -----------------------------------------------------------
arr10 = np.array([1000, 2000, 3000], dtype=np.int16)
cum10 = np.cumsum(arr10, dtype=np.int64)
print("\nExample 10 → Prevent integer overflow with dtype:\n", cum10)
# Output: [1000 3000 6000]
# Note: dtype=int64 avoids overflow for large integer sums
```

# 📐 Some Mathematical & Common NumPy Functions (Definitions)

---

## 🔢 Mathematical / Statistical

### percentile  
Returns the value below which a given percentage of data falls.

### median  
Returns the middle value of the data after sorting (50th percentile).

### histogram  
Computes the frequency distribution of numerical data by dividing it into bins.

### corrcoef (correlation coefficient)  
Returns the correlation matrix showing the linear relationship between variables.

---

## 🧰 Common NumPy Utility Functions

### np.isin  
Checks whether each element of an array is present in another array and returns a boolean array.

### np.clip  
Limits the values in an array to a specified minimum and maximum range.

### np.flip  
Reverses the order of elements in an array along a specified axis.

### np.put  
Replaces specified indices of an array with given values.

---

## 🔁 Some More Common NumPy Functions

### np.mean  
Returns the average (mean) of array elements.

### np.sum  
Returns the sum of all elements in the array.

### np.min  
Returns the minimum value in the array.

### np.max  
Returns the maximum value in the array.

### np.std  
Returns the standard deviation of the array.

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Basic percentile of 1D array
# -----------------------------------------------------------
arr = np.array([10, 20, 30, 40, 50])
p25 = np.percentile(arr, 25)
print("Example 1 → 25th percentile:", p25)
# Output: 20
# Explanation: 25% of data lies below 20


# -----------------------------------------------------------
# 🧩 Example 2: 50th percentile (median)
# -----------------------------------------------------------
p50 = np.percentile(arr, 50)
print("\nExample 2 → 50th percentile (median):", p50)
# Output: 30


# -----------------------------------------------------------
# 🧩 Example 3: 75th percentile
# -----------------------------------------------------------
p75 = np.percentile(arr, 75)
print("\nExample 3 → 75th percentile:", p75)
# Output: 40


# -----------------------------------------------------------
# 🧩 Example 4: Multiple percentiles at once
# -----------------------------------------------------------
p_values = np.percentile(arr, [25, 50, 75])
print("\nExample 4 → Multiple percentiles [25,50,75]:", p_values)
# Output: [20 30 40]


# -----------------------------------------------------------
# 🧩 Example 5: 2D array, flatten by default
# -----------------------------------------------------------
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])
p50_2d = np.percentile(arr2, 50)
print("\nExample 5 → 50th percentile of 2D array (flattened):", p50_2d)
# Output: 35.0


# -----------------------------------------------------------
# 🧩 Example 6: Percentile along axis=0 (column-wise)
# -----------------------------------------------------------
p50_col = np.percentile(arr2, 50, axis=0)
print("\nExample 6 → Column-wise 50th percentile:\n", p50_col)
# Output: [25. 35. 45.]


# -----------------------------------------------------------
# 🧩 Example 7: Percentile along axis=1 (row-wise)
# -----------------------------------------------------------
p50_row = np.percentile(arr2, 50, axis=1)
print("\nExample 7 → Row-wise 50th percentile:\n", p50_row)
# Output: [20. 50.]


# -----------------------------------------------------------
# 🧩 Example 8: Interpolation methods
# -----------------------------------------------------------
arr3 = np.array([1, 2, 3, 4])
p50_linear = np.percentile(arr3, 50, interpolation='linear')   # default
p50_midpoint = np.percentile(arr3, 50, interpolation='midpoint')
print("\nExample 8 → Different interpolation methods:")
print("Linear:", p50_linear)
print("Midpoint:", p50_midpoint)
# Output:
# Linear: 2.5
# Midpoint: 2.5


# -----------------------------------------------------------
# 🧩 Example 9: Percentile with non-uniform data
# -----------------------------------------------------------
arr4 = np.array([3, 7, 8, 5, 12, 14, 21, 13, 18])
p90 = np.percentile(arr4, 90)
print("\nExample 9 → 90th percentile:", p90)
# Output: 20.0


# -----------------------------------------------------------
# 🧩 Example 10: Using percentile for outlier detection
# -----------------------------------------------------------
arr5 = np.array([10, 12, 13, 15, 19, 25, 100])
lower = np.percentile(arr5, 5)
upper = np.percentile(arr5, 95)
print("\nExample 10 → 5th and 95th percentiles for outlier detection:")
print("Lower bound:", lower)
print("Upper bound:", upper)
# Output:
# Lower bound: 10.3
# Upper bound: 81.0
# Use values outside this range as potential outliers
```

```python
import numpy as np

# -----------------------------------------------------------
# 🧩 Example 1: Correlation coefficient of two 1D arrays
# -----------------------------------------------------------
x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 3, 2, 1])
corr1 = np.corrcoef(x, y)
print("Example 1 → Correlation matrix:\n", corr1)
# Output:
# [[ 1. -1.]
#  [-1.  1.]]
# Explanation: Perfect negative correlation


# -----------------------------------------------------------
# 🧩 Example 2: Perfect positive correlation
# -----------------------------------------------------------
y2 = np.array([2, 4, 6, 8, 10])
corr2 = np.corrcoef(x, y2)
print("\nExample 2 → Perfect positive correlation:\n", corr2)
# Output:
# [[1. 1.]
#  [1. 1.]]


# -----------------------------------------------------------
# 🧩 Example 3: 1D array with itself
# -----------------------------------------------------------
corr3 = np.corrcoef(x)
print("\nExample 3 → Correlation of array with itself:\n", corr3)
# Output: 1 (perfect correlation)


# -----------------------------------------------------------
# 🧩 Example 4: Correlation coefficient of 2D array (rows treated as variables)
# -----------------------------------------------------------
data = np.array([[1, 2, 3, 4, 5],
                 [5, 4, 3, 2, 1],
                 [2, 3, 4, 5, 6]])
corr4 = np.corrcoef(data)
print("\nExample 4 → Correlation matrix of 2D array:\n", corr4)


# -----------------------------------------------------------
# 🧩 Example 5: Correlation along rows (rowvar=True, default)
# -----------------------------------------------------------
corr5 = np.corrcoef(data, rowvar=True)
print("\nExample 5 → Row-wise correlation (default):\n", corr5)


# -----------------------------------------------------------
# 🧩 Example 6: Correlation along columns (rowvar=False)
# -----------------------------------------------------------
corr6 = np.corrcoef(data, rowvar=False)
print("\nExample 6 → Column-wise correlation (rowvar=False):\n", corr6)


# -----------------------------------------------------------
# 🧩 Example 7: Correlation with random data
# -----------------------------------------------------------
np.random.seed(0)
x_rand = np.random.rand(5)
y_rand = np.random.rand(5)
corr7 = np.corrcoef(x_rand, y_rand)
print("\nExample 7 → Correlation with random arrays:\n", corr7)


# -----------------------------------------------------------
# 🧩 Example 8: Extracting correlation coefficient
# -----------------------------------------------------------
x8 = np.array([1, 2, 3])
y8 = np.array([2, 4, 6])
corr_matrix = np.corrcoef(x8, y8)
corr_coeff = corr_matrix[0, 1]
print("\nExample 8 → Correlation coefficient:", corr_coeff)
# Output: 1.0 (perfect positive correlation)


# -----------------------------------------------------------
# 🧩 Example 9: Negative correlation example
# -----------------------------------------------------------
x9 = np.array([1, 2, 3, 4, 5])
y9 = np.array([10, 8, 6, 4, 2])
corr9 = np.corrcoef(x9, y9)[0,1]
print("\nExample 9 → Negative correlation coefficient:", corr9)
# Output: -1.0 (perfect negative correlation)


# -----------------------------------------------------------
# 🧩 Example 10: Correlation for multiple variables
# -----------------------------------------------------------
data10 = np.array([[1, 2, 3, 4],
                   [2, 3, 4, 5],
                   [4, 3, 2, 1]])
corr10 = np.corrcoef(data10)
print("\nExample 10 → Correlation matrix for multiple variables:\n", corr10)
```

# 🧠 NumPy Set Functions

**NumPy Set Functions** ka use arrays par **set-like operations** perform karne ke liye hota hai  
jaise union, intersection, difference, etc.

👉 Ye functions **unique elements** ke concept par kaam karte hain (duplicates ignore hote hain).

---

## 🔹 Why Set Functions?

- Duplicate values remove karna  
- Common elements find karna  
- Difference nikalna  
- Membership check karna  

Data cleaning & analysis mein kaafi useful hote hain.

---

## 🔢 Common NumPy Set Functions

### 1️⃣ np.unique  
Array ke **unique elements** return karta hai.

> Duplicates remove kar deta hai.

---

### 2️⃣ np.union1d  
Do arrays ka **union** return karta hai.

> Dono arrays ke saare unique elements.

---

### 3️⃣ np.intersect1d  
Do arrays ke **common elements** return karta hai.

> Sirf shared values.

---

### 4️⃣ np.setdiff1d  
Pehle array ke wo elements return karta hai  
jo **dusre array mein nahi** hote.

> A − B operation.

---

### 5️⃣ np.setxor1d  
Dono arrays ke **unique but non-common** elements return karta hai.

> Common values hata deta hai.

---

### 6️⃣ np.isin  
Check karta hai ki array ka har element  
dusre array mein present hai ya nahi.

> Output: Boolean array (`True / False`).

---

## 🧠 Quick Comparison Table

| Function | Purpose |
|-------|--------|
| `unique` | Remove duplicates |
| `union1d` | All unique elements |
| `intersect1d` | Common elements |
| `setdiff1d` | Difference |
| `setxor1d` | Non-common elements |
| `isin` | Membership check |

---

## ⚠️ Important Notes

- Ye functions **1D arrays** par kaam karte hain
- Output hamesha **sorted** hota hai
- Duplicates automatically remove ho jaate hain


```python
# 🧠 NumPy Set Functions

import numpy as np

# Create two sample arrays
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([4, 5, 6, 7, 8])

# 🔹 1️⃣ np.union1d(arr1, arr2)
# Returns the union of two arrays (unique elements from both)
print("Union:", np.union1d(arr1, arr2))
# Output → [1 2 3 4 5 6 7 8]

# 🔹 2️⃣ np.intersect1d(arr1, arr2)
# Returns common elements between both arrays
print("Intersection:", np.intersect1d(arr1, arr2))
# Output → [4 5]

# 🔹 3️⃣ np.setdiff1d(arr1, arr2)
# Elements present in arr1 but NOT in arr2
print("Difference (arr1 - arr2):", np.setdiff1d(arr1, arr2))
# Output → [1 2 3]

# 🔹 4️⃣ np.setxor1d(arr1, arr2)
# Elements present in either arr1 or arr2, but NOT in both (symmetric difference)
print("Symmetric Difference:", np.setxor1d(arr1, arr2))
# Output → [1 2 3 6 7 8]

# 🔹 5️⃣ np.in1d(arr1, arr2)
# Returns a Boolean array showing which elements of arr1 are in arr2
print("in1d:", np.in1d(arr1, arr2))
# Output → [False False False  True  True]

# 🔹 6️⃣ np.unique(arr)
# Returns sorted unique elements of an array
print("Unique elements:", np.unique(arr1))
# Output → [1 2 3 4 5]
```

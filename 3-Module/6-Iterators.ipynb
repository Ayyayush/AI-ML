## What is an Iteration

Iteration is a general term for taking each item of something, one after another. Any time you use a loop, explicit or implicit, to go over a group of items, that is iteration.

```python
# Example
num = [1,2,3]

for i in num:
    print(i)
```

## What is Iterator

An Iterator is an object that allows the programmer to traverse through a sequence of data without having to store the entire data in the memory

```python
# Creating a list using list comprehension
# This stores ALL numbers from 1 to 9999 in memory
L = [x for x in range(1, 10000)]

# Uncommenting this loop would print double of each element
# for i in L:
#     print(i * 2)

import sys

# getsizeof() returns memory size in bytes
# Dividing by 64 just scales the value for easier comparison
print(sys.getsizeof(L) / 64)


# Creating a range object
# This DOES NOT store all numbers in memory
# It only stores start, stop, and step values
x = range(1, 10000000000)

# Uncommenting this loop would generate values one by one
# for i in x:
#     print(i * 2)

# Range object takes very little memory
print(sys.getsizeof(x) / 64)
```

## What is `range` in Python?

`range` is a **built-in function** in Python used to generate a sequence of numbers.

- It returns a **range object**, not a list.
- Numbers are generated **on demand** (lazy evaluation).
- Commonly used in loops like `for`.

### Syntax

range(start, stop, step)

- `start` → starting number (default = 0)
- `stop` → ending number (**excluded**)
- `step` → gap between numbers (default = 1)

### Example

range(0, 5) → generates  
0, 1, 2, 3, 4  

⚠️ `5` is **not included**

## What is Iterable

An **Iterable** is an object that can be looped over (iterated) one element at a time.

- It represents a collection of elements.
- It does **not** store iteration state by itself.
- It can generate an **Iterator** when passed to the `iter()` method.
- An iterable must implement the `__iter__()` method.

### Key Points

- Iterable → gives an Iterator
- Used in loops like `for`
- Common examples: list, tuple, string, set, dictionary

### Relationship Between Iterable and Iterator

- Iterable: **provides** an iterator
- Iterator: **does the actual iteration**

When an iterable is passed to the `iter()` function, it returns an iterator object that can be used to access elements one by one.

### Examples of Iterables

- List  
- Tuple  
- String  
- Set  
- Dictionary  
- Range object

In simple terms, **if an object can be used in a loop, it is an iterable**.


```python
# Creating a list
L = [1, 2, 3]

# type of L
# A list is an ITERABLE
type(L)


# iter(L) converts the iterable into an iterator
# An iterator remembers its current position
type(iter(L))

# iter(L)  ---> iterator object
```

## Understanding How `for` Loop Works in Python

A `for` loop is used to **iterate over an iterable object** and execute a block of code for each element.

---

## Basic Idea

- Python first takes an **iterable**
- It converts it into an **iterator**
- Then it fetches elements **one by one**
- The loop stops when elements are exhausted

---

## General Syntax

for variable in iterable:
    statement(s)

- `iterable` → list, tuple, string, range, etc.
- `variable` → holds one element at a time
- `statement(s)` → executed for each element

---

## Step-by-Step Working of `for` Loop

1. Python calls `iter(iterable)`  
   → gets an iterator object

2. Python calls `next(iterator)`  
   → gets the first value

3. Value is assigned to the loop variable

4. Loop body is executed

5. Steps 2–4 repeat until  
   → `StopIteration` exception occurs

6. Loop ends automatically

---

## Flow Representation

Iterable  
↓  
Iterator created  
↓  
Get next value  
↓  
Execute loop body  
↓  
Repeat until no values left  

---

## Example Explanation (Conceptual)

If iterable has values: `1, 2, 3`

Loop execution:
- Iteration 1 → variable = 1
- Iteration 2 → variable = 2
- Iteration 3 → variable = 3
- No more values → loop stops

---

## Important Points

- `for` loop **does not need index**
- It works directly on elements
- Safer and cleaner than manual indexing
- Automatically handles loop termination


```python
num = [1,2,3]

for i in num:
    print(i)
```

```python
# Creating a list (ITERABLE)
num = [1, 2, 3]

# Step 1: Fetching the iterator from the iterable
iter_num = iter(num)

# Step 2: Fetching elements one by one using next()
next(iter_num)     # returns 1
next(iter_num)     # returns 2
next(iter_num)     # returns 3
next(iter_num)     # ❌ StopIteration exception
```

## Making our own for loop

```python
def mera_khudka_for_loop(iterable):
    
    # Step 1: Convert the iterable into an iterator
    iterator = iter(iterable)
    
    # Step 2: Infinite loop (same idea used internally by Python)
    while True:
        
        try:
            # Step 3: Fetch the next element from the iterator
            print(next(iterator))
        
        except StopIteration:
            # Step 4: StopIteration means no elements are left
            # Break the loop just like Python does internally
            break
        
a = [1,2,3]
b = range(1,11)
c = (1,2,3)
d = {1,2,3}
e = {0:1,1:1}

mera_khudka_for_loop(a)
mera_khudka_for_loop(b)
mera_khudka_for_loop(c)
mera_khudka_for_loop(e)
```

## A confusing point

```python
# Creating a list (ITERABLE)
num = [1, 2, 3]

# Creating an iterator from the iterable
iter_obj = iter(num)

# Printing memory address (id) of the iterator
print(id(iter_obj), 'Address of iterator 1')


# Calling iter() on an ITERATOR
# IMPORTANT: iter(iterator) returns the SAME iterator
iter_obj2 = iter(iter_obj)

# Printing memory address again
print(id(iter_obj2), 'Address of iterator 2')
```

## Let's create our own range() function

```python
class mera_range:
    
    def __init__(self,start,end):
        self.start = start
        self.end = end
        
    def __iter__(self):
        return mera_range_iterator(self)
    
class mera_range_iterator:
    
    def __init__(self,iterable_obj):
        self.iterable = iterable_obj
    
    def __iter__(self):
        return self
    
    def __next__(self):
        
        if self.iterable.start >= self.iterable.end:
            raise StopIteration
            
        current = self.iterable.start
        self.iterable.start+=1
        return current
```

```python
class mera_range:
    
    def __init__(self, start, end):
        self.start = start        # starting value
        self.end = end            # ending value (exclusive)
        
    def __iter__(self):
        # Each call to iter() returns a NEW iterator
        return mera_range_iterator(self.start, self.end)


class mera_range_iterator:
    
    def __init__(self, start, end):
        self.current = start      # iterator maintains its OWN state
        self.end = end
    
    def __iter__(self):
        return self               # iterator returns itself
    
    def __next__(self):
        
        # Stop condition
        if self.current >= self.end:
            raise StopIteration
        
        value = self.current     # store current value
        self.current += 1        # move to next value
        return value
```

There are 2 stages where error may happen in a program

- During compilation -> Syntax Error
- During execution -> Exceptions

### Syntax Error

- Something in the program is not written according to the program grammar.
- Error is raised by the interpreter/compiler
- You can solve it by rectifying the program


```python
# ❌ EXAMPLES OF SYNTAX ERRORS IN PYTHON
# ------------------------------------

# 1. Missing parentheses in print (Python 3)
print 'hello world'          # ❌ SyntaxError: print must use parentheses


# 2. Missing colon after if condition
if x > 10                    # ❌ SyntaxError: colon ':' is required
    print("Greater")


# 3. Incorrect indentation
if True:
print("Hello")               # ❌ SyntaxError: expected an indented block


# 4. Missing closing parenthesis
print("Hello"                # ❌ SyntaxError: ')' is missing


# 5. Missing closing quote
name = "Aayu                 # ❌ SyntaxError: string literal not closed


# 6. Using assignment '=' instead of comparison '=='
if x = 5:                    # ❌ SyntaxError: '=' cannot be used in condition
    print("Yes")


# 7. Incorrect for loop syntax (missing colon)
for i in range(5)            # ❌ SyntaxError: colon ':' is required
    print(i)


# 8. Incorrect function definition
def add(a, b)                # ❌ SyntaxError: missing colon ':'
    return a + b


# 9. Incorrect list syntax (missing closing bracket)
numbers = [1, 2, 3, 4        # ❌ SyntaxError: ']' is missing


# 10. Using reserved keyword as variable name
class = 10                   # ❌ SyntaxError: 'class' is a keyword


# 11. Extra closing parenthesis
print("Hello"))              # ❌ SyntaxError: unmatched ')'


# 12. Incorrect else syntax
if x > 5:
    print("Hi")
else                          # ❌ SyntaxError: colon ':' is required
    print("Hello")
```

### Other examples of syntax error

- Leaving symbols like colon,brackets
- Misspelling a keyword
- Incorrect indentation
- empty if/else/loops/class/functions

```python
# ❌ EXAMPLES OF IndexError IN PYTHON
# ---------------------------------
# IndexError is raised when we try to access a position
# that does NOT exist in a sequence (list, tuple, string, etc.)

# 1. Accessing an index that is out of range in a list
L = [1, 2, 3]
L[100]                      # ❌ IndexError: list index out of range


# 2. Accessing negative index beyond range
L = [10, 20, 30]
L[-5]                       # ❌ IndexError: negative index out of range


# 3. Accessing index in an empty list
L = []
L[0]                        # ❌ IndexError: list index out of range


# 4. Accessing index beyond string length
name = "python"
name[10]                    # ❌ IndexError: string index out of range


# 5. Accessing index beyond tuple length
T = (1, 2, 3)
T[5]                        # ❌ IndexError: tuple index out of range


# 6. Using loop with incorrect range
nums = [1, 2, 3]
for i in range(5):          # ❌ range goes beyond list length
    print(nums[i])          # ❌ IndexError when i = 3


# 7. Accessing nested list with wrong index
matrix = [[1, 2], [3, 4]]
matrix[2][0]                # ❌ IndexError: outer list index out of range


# 8. Accessing character index after slicing
text = "hello"
part = text[:2]             # part = "he"
part[5]                     # ❌ IndexError: string index out of range


# 9. Accessing list element using incorrect user input index
data = [10, 20, 30]
index = 3
data[index]                 # ❌ IndexError: valid indices are 0,1,2
```

```python
# ❌ EXAMPLES OF ModuleNotFoundError IN PYTHON
# ------------------------------------------
# ModuleNotFoundError is raised when Python
# cannot find the specified module.

# 1. Importing a module that does NOT exist
import mathi                  # ❌ ModuleNotFoundError: No module named 'mathi'
math.floor(5.3)


# 2. Misspelled standard library module
import numppy                 # ❌ ModuleNotFoundError: No module named 'numppy'


# 3. Importing a third-party module that is NOT installed
import pandasx                # ❌ ModuleNotFoundError: No module named 'pandasx'


# 4. Importing a module from a wrong path
from mymodule import test     # ❌ ModuleNotFoundError: 'mymodule' not found


# 5. Importing file that is not in the current directory
import helper                 # ❌ ModuleNotFoundError if helper.py is not present


# 6. Incorrect package name while importing submodule
import sklearns.linear_model  # ❌ ModuleNotFoundError: No module named 'sklearns'


# 7. Virtual environment not activated
import django                 # ❌ ModuleNotFoundError if django not installed in current venv


# 8. Importing module with wrong case (Python is case-sensitive)
import Math                   # ❌ ModuleNotFoundError: No module named 'Math'
```

```python
# ❌ EXAMPLES OF AttributeError IN PYTHON
# -------------------------------------
# AttributeError occurs when we try to access
# an attribute or method that does NOT exist
# for that object type.

# 1. Calling string method on a list
L = [1, 2, 3]
L.upper()                    # ❌ AttributeError: 'list' object has no attribute 'upper'


# 2. Calling list method on a string
name = "python"
name.append("3")             # ❌ AttributeError: 'str' object has no attribute 'append'


# 3. Calling dictionary method on a list
nums = [1, 2, 3]
nums.keys()                  # ❌ AttributeError: 'list' object has no attribute 'keys'


# 4. Calling string method on an integer
x = 10
x.lower()                    # ❌ AttributeError: 'int' object has no attribute 'lower'


# 5. Calling tuple method on a list
data = [1, 2, 3]
data.counts(2)               # ❌ AttributeError: 'list' object has no attribute 'counts'


# 6. Accessing non-existing attribute of an object
class Student:
    def __init__(self, name):
        self.name = name

s = Student("Aayu")
s.age                         # ❌ AttributeError: 'Student' object has no attribute 'age'


# 7. Misspelled method name
text = "hello"
text.uppper()                # ❌ AttributeError: 'str' object has no attribute 'uppper'


# 8. Calling method on NoneType
result = None
result.append(10)            # ❌ AttributeError: 'NoneType' object has no attribute 'append'
```

### Exceptions

Exceptions are **runtime errors** that occur while a program is executing.  
They usually happen due to **unexpected situations** that the program cannot handle normally.

- Exceptions are **raised by the Python runtime**
- They interrupt the **normal flow** of the program
- They must be handled **on the fly** to prevent program crashes

---

### Examples of Exceptions

- **Memory Overflow** – Program tries to use more memory than available
- **Division by Zero** – A number is divided by `0` (logical error)
- **Database Error** – Connection failure, query issues, or missing data
- **File Not Found** – Trying to access a file that does not exist
- **Type Error** – Invalid operation between incompatible data types

---

### Why Is It Important to Handle Exceptions?

Handling exceptions is important because:

- Prevents the program from **crashing abruptly**
- Helps maintain **program stability**
- Allows execution of remaining code even after an error
- Improves **user experience** by showing meaningful error messages
- Makes debugging and maintenance **easier**
- Essential for **real-world applications** like web apps, databases, and APIs

---

### How to Handle Exceptions

Python provides structured ways to handle exceptions so that errors can be managed gracefully instead of stopping execution.

The most common mechanism is:

- **Try–Except Block**

---

### Try Block

- Contains the code that **may cause an exception**
- Python **monitors this block** for runtime errors
- If no exception occurs, execution continues normally

---

### Except Block (Catch Block)

- Executes **only when an exception occurs**
- Catches the error raised in the try block
- Prevents program termination
- Can handle **specific exceptions** or **general exceptions**

---

### Key Points to Remember

- Not all errors can be avoided, but **all exceptions should be handled**
- Exception handling makes programs **robust and reliable**
- Always handle exceptions where failure is **expected or possible**
- Proper exception handling is a **core requirement** in production-level code

```python
# Bank class definition
class Bank:

    def __init__(self, balance):
        self.balance = balance            # Stores initial account balance


    def withdraw(self, amount):
        # Check if withdrawal amount is negative
        if amount < 0:
            raise Exception('amount cannot be -ve')   # Raising custom exception

        # Check if sufficient balance is available
        if self.balance < amount:
            raise Exception('paise nai hai re tere paas')  # Insufficient balance exception

        # Deduct amount from balance if all checks pass
        self.balance = self.balance - amount


# Creating an object of Bank class with initial balance
obj = Bank(10000)

# Exception handling block
try:
    obj.withdraw(15000)        # Risky code: withdrawal more than balance

except Exception as e:
    print(e)                   # Prints the error message raised in withdraw()

else:
    print(obj.balance)         # Executes only if NO exception occurs
```

## Creating Custom Exceptions

Custom exceptions are **user-defined error types** created to represent **specific problems** in an application.

They are useful when:
- Built-in exceptions are **not descriptive enough**
- You want **clear, meaningful error messages**
- You are building **large or real-world applications**
- You want better **error categorization and debugging**

### Why Create Custom Exceptions?

- Makes code **more readable and maintainable**
- Helps identify **exact cause of failure**
- Improves **debugging**
- Allows handling **application-specific errors**
- Common in frameworks, APIs, and enterprise-level software

### Key Points About Custom Exceptions

- Custom exceptions are usually created by **extending the base `Exception` class**
- They behave like normal exceptions
- They can carry **custom messages or data**
- They follow the same **try–except handling mechanism**

### Real-World Examples

- InvalidAgeError
- InsufficientBalanceError
- UnauthorizedAccessError
- DatabaseConnectionError
- FileFormatNotSupportedError

---

## Exception Hierarchy in Python

Python exceptions follow a **hierarchical tree structure**.  
This hierarchy helps Python decide **which exception to catch first**.

### Top-Level Exception Classes

- **BaseException**
  - Root of all exceptions
  - Rarely handled directly

### Important Built-in Exception Classes

- **Exception**
  - Base class for most runtime exceptions
  - Recommended parent for custom exceptions

- **ArithmeticError**
  - Errors related to numeric calculations  
  - Examples: division errors, overflow

- **ZeroDivisionError**
  - Raised when division by zero occurs

- **TypeError**
  - Raised when an operation is applied to an invalid type

- **ValueError**
  - Raised when a function receives a valid type but invalid value

- **IndexError**
  - Raised when accessing an invalid index in a sequence

- **KeyError**
  - Raised when a key is not found in a dictionary

- **FileNotFoundError**
  - Raised when a file operation fails due to missing file

- **ImportError**
  - Raised when an import statement fails

---

### Why Exception Hierarchy Matters

- Helps catch **specific errors before general ones**
- Prevents **unintended exception handling**
- Makes exception handling **structured and predictable**
- Encourages best practice:  
  **Catch specific exceptions before generic ones**

---

### Best Practices

- Always create custom exceptions by extending **Exception**
- Avoid catching **BaseException** directly
- Handle **specific exceptions first**
- Use custom exceptions for **business logic errors**
- Keep exception names **clear and meaningful**


```python
# Custom Exception class created by inheriting from Exception
class MyException(Exception):

    def __init__(self, message):
        print(message)              # Prints the custom error message when exception is raised


# Bank class definition
class Bank:

    def __init__(self, balance):
        self.balance = balance      # Stores initial account balance


    def withdraw(self, amount):
        # Check if withdrawal amount is negative
        if amount < 0:
            raise MyException('amount cannot be -ve')     # Raising custom exception

        # Check if balance is insufficient
        if self.balance < amount:
            raise MyException('paise nai hai tere paas')  # Raising custom exception

        # Deduct amount if all validations pass
        self.balance = self.balance - amount


# Creating Bank object with initial balance
obj = Bank(10000)

# Exception handling block
try:
    obj.withdraw(5000)             # Risky operation (may raise MyException)

except MyException as e:
    pass                           # Exception is caught but ignored (no action taken)

else:
    print(obj.balance)             # Executes only if NO exception occurs
```

```python
# Custom exception class for security-related errors
class SecurityError(Exception):

    def __init__(self, message):
        print(message)                  # Prints security warning message

    def logout(self):
        print('logout')                 # Custom method to perform logout action


# Google class to simulate login system
class Google:

    def __init__(self, name, email, password, device):
        self.name = name                # Stores user name
        self.email = email              # Stores registered email
        self.password = password        # Stores registered password
        self.device = device            # Stores registered device


    def login(self, email, password, device):
        # Check if login attempt is from a different device
        if device != self.device:
            raise SecurityError('bhai teri to lag gayi')   # Raise custom security exception

        # Check email and password correctness
        if email == self.email and password == self.password:
            print('welcome')            # Successful login
        else:
            print('login error')        # Invalid credentials


# Creating Google object with registered credentials and device
obj = Google('nitish', 'nitish@gmail.com', '1234', 'android')


# Exception handling block
try:
    obj.login('nitish@gmail.com', '1234', 'windows')   # Risky login attempt (wrong device)

except SecurityError as e:
    e.logout()                     # Calls logout() method of SecurityError

else:
    print(obj.name)                # Executes only if NO exception occurs

finally:
    print('database connection closed')   # Always executes (cleanup code)
```

## 🔹 How `open()` Works Internally

### 1️⃣ File Location
- Python first checks if the file exists in the given path.
- If the file exists → it is opened.
- If the file does not exist:
  - Read mode (`r`) → error occurs.
  - Write (`w`) or Append (`a`) mode → new file is created.

---

### 2️⃣ Mode-Based Permissions
The mode decides **what you are allowed to do** with the file.

| Mode | Description |
|-----|-------------|
| `r` | Read only |
| `w` | Write (overwrites existing data) |
| `a` | Append (adds data at the end) |
| `r+` | Read and Write |
| `w+` | Write and Read |
| `a+` | Append and Read |

---

### 3️⃣ File Pointer (Cursor)
- A pointer is set to track where reading or writing will happen.
- In read mode → pointer starts at the beginning.
- In write mode → pointer starts at the beginning and old data is erased.
- In append mode → pointer starts at the end of the file.

---

### 4️⃣ File Object Creation
- `open()` returns a **file object**.
- This object is used to interact with the file.

---

## 🔹 File Object Capabilities
A file object allows:
- Reading data from a file
- Writing data into a file
- Closing the file safely

---

## 🔹 Write Mode (`w`) Behavior
- Creates a new file if it does not exist.
- Deletes existing content if the file already exists.
- New data replaces old data completely.

---

## 🔹 Writing Multiple Lines
- Files do not move to the next line automatically.
- New lines must be added manually.
- Each new write continues from the current pointer position.

---

## 🔹 Importance of Closing a File
Closing a file is very important because:
- Data is safely saved.
- System memory is released.
- File corruption is prevented.
- No further operations are allowed after closing.

---

## 🔹 Best Practice
Using automatic file handling is recommended because:
- The file closes automatically.
- Reduces chances of errors.
- Makes code cleaner and safer.

---

## 🔹 Common Mistakes
- Forgetting to close the file.
- Using write mode when append was required.
- Trying to read from a closed file.
- Expecting automatic new lines while writing.

```python
# ✅ Reading from files using read() and readline()

# ---------------------------------------------
# 📌 Method 1: Using read()
# ---------------------------------------------
"""
read() method:
- Puri file ka content ek hi baar mein read karta hai.
- Isme flexibility hoti hai: aap chahe toh kuch specific characters bhi read kar sakte ho.
"""

f = open('sample.txt', 'r')             # File ko read mode mein open kar rahe hain

s = f.read()                            # Pura file content ek sath read karega
print(s)                                # Print the complete content


# s = f.read(10)                          # Ab try kar rahe hain sirf next 10 characters read karna
# print(s)                                # Lekin file already end tak read ho chuki hai, toh kuch return nahi hoga

f.close()                               # File close karna zaroori hai

```

```python

# ---------------------------------------------
# 📌 Method 2: Using readline()
# ---------------------------------------------
"""
readline() method:
- File se ek ek line read karta hai.
- Har baar call karne pe agla line return karta hai.
"""

f = open('sample.txt', 'r')             # File ko read mode mein dobara open kar rahe hain

s = f.readline()                        # Pehli line read karega
print(s,end=' ')

s = f.readline()                        # Dusri line read karega
print(s,end=' ')

f.close()                               # File close kar dena best practice hai
```

```python
# first method
f = open('sample.txt', 'r')             # File ko read mode mein open kiya

while True:
    data=f.readline()
    if data=='':
        break
    else:
        print(data,end='')
        
f.close()




# second method
f = open('sample.txt', 'r')             # File ko read mode mein open kiya

line = f.readline()                     # Pehli line read karo
while line:                             # Jab tak line empty na ho (EOF nahi aaya)
    print(line.strip())                 # print karna, strip() se \n hata rahe
    line = f.readline()                 # agla line read karo

f.close()
```

```python
# File write kar rahe ho using `with` block
with open('sample.txt','w') as f:
    f.write('selmon bhai')              # ✅ File ke andar "selmon bhai" likh diya
```

```python
# File read kar rahe ho using `with` block
with open ('sample.txt','r') as f:
    print(f.read(10))                     # ✅ File ka pura content read karke print kar diya
    
```

```python
# --------------------------------------------
# ✍️ Step 1: Write some sample data to a file
# --------------------------------------------
with open('sample.txt', 'w') as f:
    f.write("Selmon bhai is best hero in Bollywood.")

# --------------------------------------------
# 📖 Step 2: Read file content 10 characters at a time
# --------------------------------------------
with open('sample.txt', 'r') as f:
    print(f.read(10))          # 🔹 First 10 characters: 'Selmon bha'
    print(f.read(10))          # 🔹 Next 10 characters: 'i is best '
    print(f.read(10))          # 🔹 Next 10 characters: 'hero in Bo'
    print(f.read(10))          # 🔹 Next 10 characters: 'llywood.'  (may be less than 10 if end reached)

# ---------------------------------------------------
# 🧠 Explanation & Benefits of using read(n) like this
# ---------------------------------------------------
"""
✅ Memory Efficient:
   - File agar bahut badi ho (e.g. 5GB logs), toh ek sath load karna memory crash kar sakta hai.
   - read(n) se hum file ko parts mein load kar sakte hain.

✅ Stream Processing:
   - Real-time processing jaise video/audio ya logs ka analysis mein useful hota hai.

✅ Partial Read:
   - Jab humein sirf kuch specific bytes ya characters chahiye hote hain (e.g. first 100 chars).

✅ Better Control:
   - read(n) se hum exactly decide karte hain kitna read karna hai.
   - Aap `seek()` ke saath combine karke kahin se bhi reading start kar sakte ho.

🧠 Under the Hood:
   - Har `read(n)` call ke baad file pointer us position ke aage move ho jaata hai.
   - Toh next `read()` usi point se aage ka data read karega.
"""
```

```python
# --------------------------------------------
# 🛠 Step 1: Create a big file with 1000 lines
# --------------------------------------------

big_L = ["hello world\n" for i in range(1000)]   # \n added to simulate new lines

with open('big.txt', 'w') as f:
    f.writelines(big_L)                          # Writes all 1000 lines to 'big.txt'
    
# --------------------------------------------
# 📖 Step 2: Read the big file line by line
# --------------------------------------------

with open('big.txt', 'r') as f:
    for line in f:                   # Reads one line at a time, memory-efficient
        print(line.strip())         # .strip() removes trailing newline

```

```python
# second method of loading 
# --------------------------------------------
# 📖 Read a file in fixed-size chunks
# --------------------------------------------

with open('sample.txt', 'r') as f:                   # File ko read mode mein open karo
    chunk_size = 100                              # Har baar 100 characters read karenge

    chunk = f.read(chunk_size)                    # Pehla chunk read karo
    while len(chunk) > 0:                         # Jab tak chunk khali na ho
        print(chunk, end='')                      # Chunk ko print karo bina extra newline ke
        chunk = f.read(chunk_size)                # Next chunk read karo

```

```python
# --------------------------------------------
# 📖 TELL & SEEK Mechanism in File Handling
# --------------------------------------------

"""
📌 tell() ek mechanism hai jo yeh batata hai ki file pointer (cursor) abhi kis position par hai.
Yani, ab tak kitne characters read ho chuke hain.
"""

"""
📌 seek(offset) ka use karke hum file pointer ko kisi bhi specific position par bhej sakte hain.
Yeh cursor ko move karta hai, jaise rewind karna ya aage jump karna.
"""

with open('sample.txt', 'r') as f:                  # File ko read mode mein open karo
    print(f.read(10))                               # Pehle 10 characters read karo
    print(f.tell())                                 # Cursor ab kis position par hai, yeh print karo

    f.seek(0)                                       # Cursor ko wapas starting (0th position) pe le jao
    print(f.read(10))                               # Dobara pehle 10 characters read karo
    
    print(f.tell())
```

```python
# SEEK AND WRITE 
with open('sample.txt', 'w') as f:                  # File ko read mode mein open karo
    f.write('Hello')
    f.seek(0)
    f.write('X')
    
```

# Problems with working in text mode 


1. ❌ Cannot handle binary files like images, audio, or video files.
2. ❌ Not suitable for storing non-string data types such as int, float, list, tuples, etc.
3. ❌ Automatic encoding/decoding (e.g., UTF-8) can sometimes corrupt binary data.
4. ❌ Text mode may introduce platform-dependent newline conversions (e.g., \n vs \r\n).
5. ❌ File size may increase due to encoding differences or newline conversions.


# adavantages of working in binary mode 

1. ✅ Can handle all types of binary files like images, PDFs, audio, video, etc.
2. ✅ No encoding/decoding done automatically — raw bytes are read/written.
3. ✅ Platform-independent — no automatic newline conversion.
4. ✅ More efficient for large files — reads/writes in raw format.
5. ✅ Supports custom serialization formats like pickle, struct, etc.



```python
# incorrect way
# ❌ Incorrect way (Text mode) — will give error for binary files like images
with open('image.png', 'r') as f:                       # Text mode ❌
    f.read()                                            # ❌ Gives error: 'UnicodeDecodeError'
```

```python

# ✅ Correct way (Binary mode)
with open('image.png', 'rb') as f:                      # Open original file in read-binary mode
    with open('copy_image.png', 'wb') as wf:            # Open new file in write-binary mode
        wf.write(f.read())                              # Copy the binary content
```

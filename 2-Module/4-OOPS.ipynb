## Abstraction in Python

**Abstraction** is an **Object-Oriented Programming (OOP)** concept that focuses on **hiding internal implementation details** and showing **only the essential features** to the user.

In simple terms:
> *“What an object does” is shown,  
> “How it does it” is hidden.*

---

## Why Abstraction Is Needed

Abstraction is important because it:

- Reduces **complexity**
- Hides **unnecessary implementation details**
- Improves **code readability**
- Makes systems **easier to maintain**
- Enhances **security**
- Helps in **large-scale application design**

---

## Real-Life Example

- **Car**
  - You know how to drive (steering, brake, accelerator)
  - You don’t need to know how the engine works internally

- **ATM Machine**
  - You see options like withdraw, deposit, balance check
  - Internal banking logic is hidden

---

## Abstraction in Python

Python supports abstraction mainly through:

- **Abstract Classes**
- **Abstract Methods**

These ensure that **child classes implement required behavior**.

---

## Abstract Class

An **abstract class** is a class that:

- Cannot be instantiated directly
- Acts as a **blueprint** for other classes
- May contain abstract methods and normal methods

Purpose:
- To define **what methods must exist**, not how they work

---

## Abstract Method

An **abstract method** is a method that:

- Has **no implementation** in the abstract class
- Must be **implemented by child classes**
- Enforces a **common interface**

---

## Key Characteristics of Abstraction

- Focuses on **design**, not implementation
- Promotes **loose coupling**
- Ensures **method consistency** across subclasses
- Helps follow **interface-based programming**

---

## Advantages of Abstraction

- Hides complex logic
- Makes code more flexible
- Simplifies testing
- Improves scalability
- Encourages clean architecture

---

## Disadvantages of Abstraction

- Can increase **initial complexity**
- Requires proper planning
- Overuse may make code harder to understand

---

## Abstraction vs Encapsulation

| Abstraction | Encapsulation |
|------------|---------------|
| Hides implementation | Hides data |
| Focuses on **what** | Focuses on **how** |
| Design-level concept | Implementation-level concept |
| Achieved using abstract classes | Achieved using access modifiers |

---

## When to Use Abstraction

- When building **frameworks**
- When designing **large systems**
- When multiple classes share common behavior
- When enforcing **method rules** across classes

---

## Summary

- Abstraction hides **unnecessary details**
- Shows only **essential functionality**
- Helps manage **complex systems**
- Core principle of **clean OOP design**


```python
from abc import ABC, abstractmethod        # Importing ABC (Abstract Base Class) and abstractmethod decorator

class BankApp(ABC):                        # Parent abstract class representing a generic bank app
    
    def databse(self):                     # Concrete method (already implemented)
        print('connected to database')     # Simulates database connection
        
    @abstractmethod
    def security(self):                    # Abstract method (must be implemented by child classes)
        pass                               # No implementation here


class MobileApp(BankApp):                  # Child class inheriting from BankApp
    
    def mobile_login(self):                # Method specific to MobileApp
        print('login into mobile')          # Simulates mobile login
        
    def security(self):                    # Implementation of abstract method from BankApp
        print('mobile security')            # Defines mobile-specific security logic
        

mob = MobileApp()                          # Creating object of MobileApp class
```

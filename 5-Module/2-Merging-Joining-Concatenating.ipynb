👉 **Concatenation** in Pandas ka use **DataFrames ya Series ko stack / join**
karne ke liye hota hai (row-wise ya column-wise).

---

## 🔹 Basic Syntax
```python
pd.concat([df1, df2])


````python
# pd.concat
# df.concat
# ignore_index
# df.append
# mullitindex -> fetch using iloc
# concat dataframes horizontally
````

````python
# ------------------------------------------------------------
# Concatenating two DataFrames using pandas
# ------------------------------------------------------------

regs1 = pd.concat([nov, dec])                 # concatenate nov and dec DataFrames (original index retained)

regs2 = pd.concat(                            # concatenate nov and dec DataFrames
    [nov, dec],
    ignore_index=True                         # reset index after concatenation
)

regs                                          # display the DataFrame (assuming regs already exists)
````

````python
# ------------------------------------------------------------
# Appending one DataFrame (dec) to another (nov)
# ------------------------------------------------------------

nov.append(dec, ignore_index=True)            # append dec rows to nov and reset index

# DataFrame.append() is DEPRECATED in recent versions of pandas
# alternative 
pd.concat([nov, dec], ignore_index=True)
````

````python
# ------------------------------------------------------------
# Creating a MultiIndex DataFrame using pd.concat
# ------------------------------------------------------------

multi = pd.concat(                            # concatenate DataFrames with hierarchical indexing
    [nov, dec],                               # list of DataFrames to concatenate
    keys=['Nov', 'Dec']                       # keys become outer level of MultiIndex
)
# keys mein list pass karenge aur naam de skte h one for each data frame

multi

# ------------------------------------------------------------
# Accessing a specific row from the MultiIndex DataFrame
# ------------------------------------------------------------

multi.loc[('Dec', 4)]                         # access row with index 4 from 'Dec' DataFrame
````

# Joining in Pandas

👉 **Joining** in Pandas ka use **DataFrames ko index ke basis par**
combine karne ke liye hota hai.

---

## 🔹 Basic Syntax
```python
df1.join(df2)


````python
regs
````

````python
# ------------------------------------------------------------
# Performing INNER JOIN using pandas merge
# ------------------------------------------------------------

students.merge(                               # merge students DataFrame
    regs,                                    # with regs DataFrame
    how='inner',                             # INNER JOIN → only matching records
    on='student_id'                          # common key column
)
````

````python
# ------------------------------------------------------------
# Performing LEFT JOIN using pandas merge
# ------------------------------------------------------------

courses.merge(                               # merge courses DataFrame
    regs,                                   # with regs DataFrame
    how='left',                             # LEFT JOIN → all rows from left DataFrame
    on='course_id'                          # common key column
)
````

````python
# ------------------------------------------------------------
# Creating a temporary DataFrame for RIGHT JOIN example
# ------------------------------------------------------------

temp_df = pd.DataFrame({
    'student_id': [26, 27, 28],              # student IDs
    'name': ['Nitish', 'Ankit', 'Rahul'],     # student names
    'partner': [28, 26, 17]                   # partner student IDs
})


# ------------------------------------------------------------
# Appending temporary DataFrame to existing students DataFrame
# ------------------------------------------------------------

students = pd.concat(                        # concatenate DataFrames
    [students, temp_df],                     # existing + new student records
    ignore_index=True                        # reset index after concatenation
)


students.tail()
````

````python
students.merge(regs,how='right',on='student_id')
````

````python
regs.merge(students,how='left',on='student_id')
````

````python
# outer join
students.merge(regs,how='outer',on='student_id').tail(10)
````

# Questions

````python
# ------------------------------------------------------------
# 1. Finding total revenue generated
# ------------------------------------------------------------

total = regs.merge(                           # merge registrations with courses DataFrame
    courses,                                 # courses DataFrame (contains price)
    how='inner',                             # INNER JOIN to match valid course_ids
    on='course_id'                           # join on course_id
)['price'].sum()                             # sum of course prices = total revenue

total                                        # display total revenue
````

````python
# ------------------------------------------------------------
# 2. Finding month-by-month revenue
# ------------------------------------------------------------

temp_df = pd.concat(                          # concatenate Nov and Dec registration DataFrames
    [nov, dec],                               # list of monthly DataFrames
    keys=['Nov', 'Dec']                       # keys represent month names
).reset_index()                              # reset index to convert MultiIndex into columns


temp_df

temp_df.merge(                                # merge with courses DataFrame to get course prices
    courses,                                 # courses DataFrame
    on='course_id'                           # join on course_id
).groupby(                                   # group data month-wise
    'level_0'                                # level_0 contains month names (Nov / Dec)
)['price'].sum()                              # sum prices to calculate revenue per month

````

````python
# ------------------------------------------------------------
# 3. Printing the registration table
# Columns required → name, course, price
# ------------------------------------------------------------

regs.merge(                                  # merge registrations with students
    students,                                # students DataFrame
    on='student_id'                          # join on student_id
).merge(                                     # merge result with courses
    courses,                                 # courses DataFrame
    on='course_id'                           # join on course_id
)[
    ['name', 'course_name', 'price']         # select required columns
]
````

````python
# ------------------------------------------------------------
# 4. Plotting bar chart for revenue per course
# ------------------------------------------------------------

regs.merge(                                  # merge registrations with courses
    courses,                                # courses DataFrame (contains course price)
    on='course_id'                           # join on course_id
).groupby(                                   # group data course-wise
    'course_name'                            # group by course name
)['price'].sum().plot(                       # sum price = revenue per course
    kind='bar'                               # bar chart
)
````

````python
# ------------------------------------------------------------
# 5. Finding students who enrolled in both months
# ------------------------------------------------------------

common_student_id = np.intersect1d(           # find common elements between two arrays
    nov['student_id'],                        # student IDs from November
    dec['student_id']                         # student IDs from December
)

common_student_id                             # display common student IDs
````

````python
# ------------------------------------------------------------
# 6. Finding courses that got NO enrollments
# ------------------------------------------------------------

# courses['course_id']  → all available course IDs
# regs['course_id']     → course IDs that have enrollments

course_id_list = np.setdiff1d(                # find course IDs with no enrollments
    courses['course_id'],                     # all course IDs
    regs['course_id']                         # enrolled course IDs
)

courses[                                     # filter courses DataFrame
    courses['course_id'].isin(course_id_list)# select only non-enrolled courses
]
````

````python
# ------------------------------------------------------------
# 7. Finding students who did NOT enroll into any courses
# ------------------------------------------------------------

student_id_list = np.setdiff1d(               # find student IDs with no enrollments
    students['student_id'],                   # all student IDs
    regs['student_id']                        # enrolled student IDs
)

students[                                    # filter students DataFrame
    students['student_id'].isin(student_id_list)
].shape[0]                                   # count of students with no enrollments


# ------------------------------------------------------------
# Calculating percentage of students not enrolled
# ------------------------------------------------------------

(10 / 28) * 100                               # percentage calculation
````

````python
# ------------------------------------------------------------
# 8. Printing student name → partner name for all enrolled students
# Using SELF JOIN on students table
# ------------------------------------------------------------

students.merge(                              # self join on students DataFrame
    students,                                # same DataFrame used twice
    how='inner',                             # INNER JOIN
    left_on='partner',                       # partner ID from left table
    right_on='student_id'                    # student ID from right table
)[
    ['name_x', 'name_y']                     # name_x → student, name_y → partner
]
````

````python
# ------------------------------------------------------------
# 9. Finding top 3 students with highest number of enrollments
# ------------------------------------------------------------

regs.merge(                                  # merge registrations with students
    students,                                # students DataFrame
    on='student_id'                          # join on student_id
).groupby(                                   # group by student
    ['student_id', 'name']                   # student ID and name
)['name'].count().sort_values(                # count enrollments per student
    ascending=False                          # sort in descending order
).head(3)                                    # pick top 3 students
````

````python
# ------------------------------------------------------------
# 10. Finding top 3 students who spent the most money on courses
# ------------------------------------------------------------

regs.merge(                                  # merge registrations with students
    students,                                # students DataFrame
    on='student_id'                          # join on student_id
).merge(                                     # merge result with courses
    courses,                                 # courses DataFrame (contains price)
    on='course_id'                           # join on course_id
).groupby(                                   # group by student
    ['student_id', 'name']                   # student ID and name
)['price'].sum().sort_values(                # total amount spent by each student
    ascending=False                          # sort highest spenders first
).head(3)                                    # select top 3 students
````

````python
# Alternate syntax for merge
# students.merge(regs)

pd.merge(students,regs,how='inner',on='student_id')
````

# find top 3 studiums with highest sixes/match ratio

````python


temp_df = delivery.merge(matches,left_on='match_id',right_on='id')
temp_df
six_df = temp_df[temp_df['batsman_runs'] == 6]   
````

````python
# stadium -> sixes
num_sixes = six_df.groupby('venue')['venue'].count()
num_sixes

num_matches = matches['venue'].value_counts()
num_matches

(num_sixes/num_matches).sort_values(ascending=False).head(3)
````

````python
# ------------------------------------------------------------
# Finding Orange Cap holder (highest run scorer) for each season
# ------------------------------------------------------------

temp_df.groupby(                              # group data
    ['season', 'batsman']                     # season-wise and batsman-wise
)['batsman_runs'].sum().reset_index(          # total runs scored by each batsman in each season
).sort_values(                                # sort based on runs
    'batsman_runs',
    ascending=False                           # highest runs first
).drop_duplicates(                            # keep only top scorer per season
    subset=['season'],                        # one record per season
    keep='first'                              # highest run scorer stays
).sort_values(                                # final sorting
    'season'                                  # season-wise order
)
````

````python
# ------------------------------------------------------------
# Calculating total runs scored by each batsman in each season
# ------------------------------------------------------------

temp_df.groupby(                              # group the data
    ['season', 'batsman']                     # season-wise and batsman-wise grouping
)['batsman_runs'].sum().reset_index(          # sum total runs and convert index to columns
).sort_values(                                # sort the result
    'batsman_runs',                           # based on total runs scored
    ascending=False                           # highest run scorers first
)
````

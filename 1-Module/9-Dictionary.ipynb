/*
 # Dictionary in Python

 A dictionary is a collection of key-value pairs used to store data.
 It is similar to a map where each value is accessed using a unique key.
 Unlike other data types that store only values, dictionaries store key-value mappings.

--------------------------------------------

 Syntax ::
 dict = {"key" : "value"}

 Example ::
 student = {"name":"Ayush", "age":20}

--------------------------------------------



 ### Characteristics

 • Unordered collection (order preserved from Python 3.7+)
 • Keys must be unique
 • Keys must be immutable (string, number, tuple)
 • Values can be of any data type
 • Mutable data structure (modifiable after creation)
 • Accessed using key instead of index



--------------------------------------------

 ### Useful Dictionary Functions

 • dict.keys()     returns list of all keys
 • dict.values()   returns list of all values
 • dict.items()    returns list of (key, value) pairs
 • dict.get(key)   safely access key without error
 • len(dict)       gives number of elements
 • dict.pop(key)   removes an item by key

--------------------------------------------

 ### Nested Dictionary

 • Dictionary inside a dictionary
 • Useful for storing structured data like student records,
   JSON data, database-like structures etc.

--------------------------------------------

 ### When to Use a Dictionary

 • When data requires a unique identifier for each value
 • When fast lookup or mapping is required
 • Best for structured and real-world data storage

--------------------------------------------

*/


```python
# Creating Dictionary
# An empty dictionary can be created using {}
d = {}

# 1-D Dictionary
# Simple dictionary with key-value pairs
d1 = {'name': 'ayush'}

# Dictionary with Mixed Keys
# Keys can be immutable types like numbers, strings, or tuples
# Values can be any data type
d2 = {(1, 2, 3): 1, 'hello': "world"}

# 2-D Dictionary
# Dictionary containing another dictionary as values
# Useful for nested or structured data
d3 = {
    'student1': {'name': 'Ayush', 'age': 20},
    'student2': {'name': 'Riya', 'age': 19}
}

# Using Sequence and dict() function
# Dictionaries can be created from sequences like lists or tuples of key-value pairs
seq_dict = dict([('name', 'Ayush'), ('age', 20)])

# Duplicate Keys
# Keys must be unique. If duplicate keys are provided, the last value will overwrite previous ones
dup_dict = {'a': 1, 'a': 2}  # 'a' will have value 2

# Mutable Items as Keys
# Keys must be immutable
# Mutable items like lists or sets cannot be used as dictionary keys
# Example: invalid_keys = {[1,2,3]: "value"}  # This will raise TypeError
```

```python
# Accessing Items in a Dictionary

# Example dictionary
student = {
    'name': 'Ayush',
    'age': 20,
    'course': 'B.Tech'
}

# 1. Accessing a value using key
print(student['name'])  # Output: Ayush
print(student['age'])   # Output: 20

# 2. Using get() method (safer way)
print(student.get('name'))       # Output: Ayush
print(student.get('city'))       # Output: None (key doesn't exist)
print(student.get('city', 'Delhi'))  # Output: Delhi (default value)

# 3. Accessing Nested Dictionary
student_info = {
    'student1': {'name': 'Ayush', 'age': 20},
    'student2': {'name': 'Riya', 'age': 19}
}

print(student_info['student1']['name'])  # Output: Ayush
print(student_info['student2']['age'])   # Output: 19

# 4. Iterating through dictionary
# Access all keys
for key in student:
    print(key)
# Output:
# name
# age
# course

# Access all values
for value in student.values():
    print(value)
# Output:
# Ayush
# 20
# B.Tech

# Access both keys and values
for key, value in student.items():
    print(key, ":", value)
# Output:
# name : Ayush
# age : 20
# course : B.Tech
```

```python
# Adding Key-Value Pair in Dictionary

# Example dictionary
student = {
    'name': 'Ayush',
    'age': 20,
    'course': 'B.Tech'
}

# 1. Adding a new key-value pair
student['city'] = 'Delhi'
print(student)
# Output: {'name': 'Ayush', 'age': 20, 'course': 'B.Tech', 'city': 'Delhi'}

# 2. Updating an existing key's value
student['age'] = 21
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech', 'city': 'Delhi'}

# 3. Adding multiple key-value pairs using update() method
student.update({'roll_no': 101, 'semester': 5})
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech', 'city': 'Delhi', 'roll_no': 101, 'semester': 5}

# 4. Adding a nested dictionary
student['marks'] = {'Math': 95, 'Physics': 90}
print(student)
# Output includes nested dictionary: 'marks': {'Math': 95, 'Physics': 90}

# Notes:
# - Direct assignment adds a new key or updates an existing one.
# - update() method can add multiple key-value pairs at once.
# - Nested dictionaries allow structured storage for complex data.
```

```python
# Removing Key-Value Pair in Dictionary

# Example dictionary
student = {
    'name': 'Ayush',
    'age': 21,
    'course': 'B.Tech',
    'city': 'Delhi',
    'roll_no': 101
}

# 1. Using pop()
# Removes the item with the specified key and returns its value
age_value = student.pop('age')
print("Removed Age:", age_value)
print(student)
# Output: {'name': 'Ayush', 'course': 'B.Tech', 'city': 'Delhi', 'roll_no': 101}

# 2. Using popitem()
# Removes the last inserted key-value pair (Python 3.7+)
last_item = student.popitem()
print("Removed Last Item:", last_item)
print(student)
# Output: {'name': 'Ayush', 'course': 'B.Tech', 'city': 'Delhi'}

# 3. Using del
# Deletes the specified key or the whole dictionary
del student['city']
print(student)
# Output: {'name': 'Ayush', 'course': 'B.Tech'}

# Deleting the whole dictionary
# del student
# print(student)  # This would raise NameError

# 4. Using clear()
# Removes all items from the dictionary but keeps the dictionary object
student.clear()
print(student)
# Output: {}
```

```python
# Editing Key-Value Pair in Dictionary

# Example dictionary
student = {
    'name': 'Ayush',
    'age': 20,
    'course': 'B.Tech',
    'city': 'Delhi'
}

# 1. Updating the value of an existing key directly
student['age'] = 21
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech', 'city': 'Delhi'}

# 2. Updating multiple values using update() method
student.update({'course': 'B.Tech (CSE)', 'city': 'Noida'})
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech (CSE)', 'city': 'Noida'}

# 3. Editing values inside nested dictionary
student['marks'] = {'Math': 95, 'Physics': 90}
# Update Physics marks
student['marks']['Physics'] = 92
print(student)
# Output includes updated nested value: 'marks': {'Math': 95, 'Physics': 92}

# Notes:
# - Direct assignment changes the value for a single key.
# - update() is useful to modify multiple keys at once.
# - Nested dictionaries require chaining keys to edit inner values.
```

```python
# Dictionary Operations

# Example dictionary
student = {
    'name': 'Ayush',
    'age': 21,
    'course': 'B.Tech',
    'city': 'Noida'
}

# 1. Membership Testing
# Check if a key exists in the dictionary using 'in' keyword
print('name' in student)   # Output: True
print('marks' in student)  # Output: False

# Check if a key does NOT exist using 'not in'
print('marks' not in student)  # Output: True

# Note: Membership operators check keys, not values

# 2. Iteration over Dictionary

# a) Iterate through keys
for key in student:
    print(key)
# Output:
# name
# age
# course
# city

# b) Iterate through values
for value in student.values():
    print(value)
# Output:
# Ayush
# 21
# B.Tech
# Noida

# c) Iterate through key-value pairs
for key, value in student.items():
    print(key, ":", value)
# Output:
# name : Ayush
# age : 21
# course : B.Tech
# city : Noida

# Notes:
# - Iteration can be done using keys(), values(), or items() methods.
# - Membership checks only the keys in the dictionary.
```

```python
# Dictionary Functions / Methods

# Example dictionary
student = {
    'name': 'Ayush',
    'age': 21,
    'course': 'B.Tech',
    'city': 'Noida'
}

# 1. keys()
# Returns all keys in the dictionary
print(student.keys())
# Output: dict_keys(['name', 'age', 'course', 'city'])

# 2. values()
# Returns all values in the dictionary
print(student.values())
# Output: dict_values(['Ayush', 21, 'B.Tech', 'Noida'])

# 3. items()
# Returns all key-value pairs as tuples
print(student.items())
# Output: dict_items([('name', 'Ayush'), ('age', 21), ('course', 'B.Tech'), ('city', 'Noida')])

# 4. get()
# Returns the value for the specified key safely
print(student.get('name'))      # Output: Ayush
print(student.get('marks'))     # Output: None
print(student.get('marks', 0))  # Output: 0 (default value)

# 5. pop()
# Removes the specified key and returns its value
removed_value = student.pop('city')
print("Removed:", removed_value)
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech'}

# 6. popitem()
# Removes the last inserted key-value pair
last_item = student.popitem()
print("Removed last item:", last_item)
print(student)

# 7. clear()
# Removes all items from the dictionary
student.clear()
print(student)  # Output: {}

# 8. update()
# Adds or updates multiple key-value pairs
student.update({'name': 'Ayush', 'age': 22})
print(student)
# Output: {'name': 'Ayush', 'age': 22}

# 9. copy()
# Creates a shallow copy of the dictionary
student_copy = student.copy()
print(student_copy)
# Output: {'name': 'Ayush', 'age': 22}

# 10. len()
# Returns the number of items in the dictionary
print(len(student))  # Output: 2

# 11. max() and min()
# max() and min() work on dictionary keys
print(max(student))  # Output: 'name'  (alphabetically largest key)
print(min(student))  # Output: 'age'   (alphabetically smallest key)

# 12. sorted()
# Returns sorted list of keys
print(sorted(student))  # Output: ['age', 'name']

# Notes:
# - keys(), values(), items() are useful for iteration
# - get() avoids KeyError
# - pop() and popitem() remove items
# - update() can add or change multiple keys at once
# - copy() creates a new dictionary without affecting the original
# - len() gives the number of key-value pairs
# - max() and min() return the largest and smallest keys alphabetically
# - sorted() returns a sorted list of dictionary keys
```

```python
# Dictionary Comprehension
# A concise way to create dictionaries using expressions inside {}

# 1. Simple Dictionary Comprehension
# Create a dictionary with numbers as keys and their squares as values
squares = {x: x**2 for x in range(1, 6)}
print(squares)
# Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# 2. Dictionary Comprehension with Condition
# Include only even numbers as keys
even_squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)
# Output: {2: 4, 4: 16, 6: 36, 8: 64, 10: 100}

# 3. Transforming Existing Dictionary
# Double the values of an existing dictionary
original = {'a': 1, 'b': 2, 'c': 3}
doubled = {k: v*2 for k, v in original.items()}
print(doubled)
# Output: {'a': 2, 'b': 4, 'c': 6}

# 4. Swapping keys and values
# Reverse keys and values of a dictionary
original = {'x': 10, 'y': 20, 'z': 30}
swapped = {v: k for k, v in original.items()}
print(swapped)
# Output: {10: 'x', 20: 'y', 30: 'z'}

# Notes:
# - Dictionary comprehension is a compact and readable way to create dictionaries.
# - Can include conditions and transformations.
# - Useful for mapping, filtering, and modifying dictionaries efficiently.
 
```

```python
# Using zip() to Create Dictionary

# Example 1: Simple zip of two lists
keys = ['name', 'age', 'course']
values = ['Ayush', 21, 'B.Tech']

# Create dictionary by zipping keys and values
student = dict(zip(keys, values))
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech'}

# Example 2: Zip with unequal length lists
keys = ['name', 'age', 'course', 'city']
values = ['Ayush', 21, 'B.Tech']
# Extra keys are ignored
student = dict(zip(keys, values))
print(student)
# Output: {'name': 'Ayush', 'age': 21, 'course': 'B.Tech'}

# Example 3: Zip with tuples
keys = ('x', 'y', 'z')
values = (10, 20, 30)
coord = dict(zip(keys, values))
print(coord)
# Output: {'x': 10, 'y': 20, 'z': 30}

# Notes:
# - zip() pairs elements from two sequences (keys and values)
# - dict(zip(keys, values)) creates a dictionary from those pairs
# - If lengths differ, the extra elements are ignored
# - Works with lists, tuples, or any iterable sequences
```

```python
# Nested Dictionary Comprehension
# Creating dictionaries inside dictionaries using comprehension

# Example 1: Create a nested dictionary with numbers and their squares and cubes
nested_dict = {x: {'square': x**2, 'cube': x**3} for x in range(1, 6)}
print(nested_dict)
# Output: 
# {
#  1: {'square': 1, 'cube': 1}, 
#  2: {'square': 4, 'cube': 8}, 
#  3: {'square': 9, 'cube': 27}, 
#  4: {'square': 16, 'cube': 64}, 
#  5: {'square': 25, 'cube': 125}
# }

# Example 2: Nested comprehension with condition
# Only include even numbers as keys
nested_even = {x: {'square': x**2, 'cube': x**3} for x in range(1, 11) if x % 2 == 0}
print(nested_even)
# Output: 
# {
#  2: {'square': 4, 'cube': 8}, 
#  4: {'square': 16, 'cube': 64}, 
#  6: {'square': 36, 'cube': 216}, 
#  8: {'square': 64, 'cube': 512}, 
#  10: {'square': 100, 'cube': 1000}
# }

# Example 3: Nested dictionary from two lists
keys = ['Alice', 'Bob']
subjects = ['Math', 'Physics']
marks = [[95, 90], [88, 92]]

student_marks = {k: {sub: mark for sub, mark in zip(subjects, m)} for k, m in zip(keys, marks)}
print(student_marks)
# Output:
# {'Alice': {'Math': 95, 'Physics': 90}, 'Bob': {'Math': 88, 'Physics': 92}}

# Notes:
# - Nested comprehension allows creation of dictionaries with complex structures
# - Can include conditions to filter keys or values
# - Very useful for mapping multidimensional or hierarchical data
# - Works well with zip() to pair data from multiple sequences
```

# Operators in Python

Operators wo symbols hote hain jo values par operations perform karte hain.  
Python mein different types ke operators hote hain jinka use mathematical, logical, comparison, assignment, etc. tasks ke liye hota h.

---

## 🔢 1. Arithmetic Operators

Ye basic mathematical operations ke liye use hote hain:

| Operator | Meaning         | Example (`a = 10`, `b = 3`) | Output |
|----------|------------------|-----------------------------|--------|
| `+`      | Addition         | `a + b`                     | `13`   |
| `-`      | Subtraction      | `a - b`                     | `7`    |
| `*`      | Multiplication   | `a * b`                     | `30`   |
| `/`      | Division         | `a / b`                     | `3.33` |
| `//`     | Floor Division   | `a // b`                    | `3`    |
| `%`      | Modulus          | `a % b`                     | `1`    |
| `**`     | Exponentiation   | `a ** b`                    | `1000` |

---

## 🟰 2. Assignment Operators

Inka use kisi value ko variable mein assign karne ke liye hota h:

| Operator | Meaning           | Example     |
|----------|--------------------|-------------|
| `=`      | Assign              | `x = 5`     |
| `+=`     | x = x + y           | `x += 2`    |
| `-=`     | x = x - y           | `x -= 3`    |
| `*=`     | x = x * y           | `x *= 4`    |
| `/=`     | x = x / y           | `x /= 2`    |
| `//=`    | x = x // y          | `x //= 3`   |
| `%=`     | x = x % y           | `x %= 2`    |
| `**=`    | x = x ** y          | `x **= 3`   |

---

## 🤝 3. Comparison (Relational) Operators

Ye check karte hain do values ke beech ka relation:

| Operator | Meaning             | Example    | Output |
|----------|----------------------|------------|--------|
| `==`     | Equal to             | `a == b`   | False  |
| `!=`     | Not equal to         | `a != b`   | True   |
| `>`      | Greater than         | `a > b`    | True   |
| `<`      | Less than            | `a < b`    | False  |
| `>=`     | Greater or equal     | `a >= b`   | True   |
| `<=`     | Less or equal        | `a <= b`   | False  |

---

## 🧠 4. Logical Operators

Ye mainly conditions mein use hote hain:

| Operator | Meaning                    | Example               |
|----------|-----------------------------|------------------------|
| `and`    | True if both conditions true | `a > 5 and b < 5`      |
| `or`     | True if any condition true   | `a > 5 or b > 10`      |
| `not`    | Reverses condition           | `not(a > 5)`           |

---

## 🧾 5. Identity Operators

Ye check karte hain ki do variables memory mein same object ko point kar rahe hain ya nahi:

| Operator | Meaning            | Example         |
|----------|---------------------|------------------|
| `is`     | True if same object | `x is y`         |
| `is not` | True if not same    | `x is not y`     |

---

## 📦 6. Membership Operators

Ye check karte hain ki koi value kisi sequence (list, tuple, string) mein hai ya nahi:

| Operator | Meaning               | Example          |
|----------|------------------------|-------------------|
| `in`     | True if present        | `3 in [1,2,3]`    |
| `not in` | True if not present    | `4 not in [1,2]`  |

---

## 🧱 7. Bitwise Operators

Ye binary level par operations perform karte hain:

| Operator | Meaning       | Example     |
|----------|----------------|-------------|
| `&`      | AND            | `a & b`     |
| `|`      | OR             | `a | b`     |
| `^`      | XOR            | `a ^ b`     |
| `~`      | NOT            | `~a`        |
| `<<`     | Left Shift     | `a << 2`    |
| `>>`     | Right Shift    | `a >> 2`    |

---

Python mein har type ka operator hota hai jiska use specific condition ya logic ko likhne mein hota h.  
Operators help karte hain values ke beech **interaction ya decision making** karne mein.

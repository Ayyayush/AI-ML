# Data Types in Python

# 1. Numeric Types:
#    - int     → integers jaise 1, 2, -5
#    - float   → decimal numbers jaise 3.14, -0.99
#    - complex → complex numbers jaise 2 + 3j

# 2. Sequence Types:
#    - str     → strings (text) jaise "hello", 'world'
#    - list    → mutable sequence jaise [1, 2, 3]

# ============================================
# User Input in Python - All in One Example
# ============================================

# 1) Basic Input (always string)
name = input("Enter your name: ")                           # input by default string deta hai
print("Hello,", name)                                       


# 2) Numeric Input (type conversion)
age = int(input("Enter your age: "))                        # string -> int
height = float(input("Enter your height: "))                # string -> float
print("Age:", age, "Height:", height)


# 3) Taking Multiple Inputs
# Example: user enters -> 10 20
a, b = input("Enter two numbers (space separated): ").split()     # split se separate values
print("a =", a, "b =", b)                                         # abhi ye string hain

# Convert directly to int using map
x, y = map(int, input("Enter two numbers again: ").split())       # map(int, split input)
print("Sum =", x + y)


# 4) Taking List Input
# Example: user enters -> 1 2 3 4 5
nums = list(map(int, input("Enter multiple numbers (space separated): ").split()))
print("List =", nums)


# 5) Input with Separator
# Example: user enters -> apple,banana,mango
data = input("Enter comma-separated values: ").split(",")   # split by comma
print("Data =", data)


# Type conversion in Python

# Implicit Type Conversion
x = 10        # int
y = 2.5       # float

z = x + y     # int + float -> float
print(z)      # 12.5
print(type(z))  # <class 'float'>





# Explicit Type Conversion
a = "100"
b = int(a)          # string → int
c = float(a)        # string → float
d = str(200)        # int → string
e = list("hello")   # string → list of characters

print(b, type(b))   # 100 <class 'int'>
print(c, type(c))   # 100.0 <class 'float'>
print(d, type(d))   # '200' <class 'str'>
print(e, type(e))   # ['h', 'e', 'l', 'l', 'o'] <class 'list'>

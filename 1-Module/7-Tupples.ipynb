# Tuples

# A tuple in Python is similar to a list.
# The main difference is that tuples are immutable (cannot be changed after creation),
# while lists are mutable (can be modified after creation).

# In short: Tuple = Immutable List


# Characteristics of Tuple:
# 1. Ordered → The elements in a tuple have a defined order, and that order will not change.
# 2. Immutable → Once created, you cannot add, remove, or change any element.
# 3. Allow duplicate values → Tuples can contain duplicate items.
# 4. Can store different data types → Like integers, strings, floats, and even lists inside them.
# 5. Tuples are defined using parentheses ( ), whereas lists use square brackets [ ].



# Advantages:
# ✅ Faster than lists (because they are immutable)
# ✅ Can be used as keys in dictionaries (lists cannot)
# ✅ Protect data from modification (useful for fixed data)
# ✅ Require less memory compared to lists


# Disadvantages:
# ❌ Cannot be modified (no append(), remove(), or pop())
# ❌ Cannot sort or reverse directly
# ❌ Limited flexibility compared to lists
# ❌ Less suitable for dynamic data



# Creating a tuple
my_tuple = (10, 20, 30, 40)
print(my_tuple)


# Single element tuple (Note: add comma)
single = (5,)
print(type(single))          # <class 'tuple'>


# Tuple with mixed data types
mixed_tuple = (1, "Python", 3.14, True)
print(mixed_tuple)

# Tuple with duplicate values
dup_tuple = (1, 2, 2, 3, 3, 3)
print(dup_tuple)

# Accessing elements
print(my_tuple[1])           # Access element at index 1

# Slicing
print(my_tuple[1:3])         # Get elements from index 1 to 2

# Nested tuple
nested_tuple = (1, (2, 3), (4, 5))
print(nested_tuple[1][1])    # Access nested element (3)

# Tuple unpacking
a, b, c = (5, 10, 15)
print(a, b, c)

# Tuple length
print(len(my_tuple))

# Tuple concatenation
new_tuple = my_tuple + (50, 60)
print(new_tuple)

# Iterating over a tuple
for item in my_tuple:
    print(item)



# Tuple Functions

# Python mein tuples ke saath kuch built-in functions use kiye ja sakte hain
# even though tuples immutable hote hain (elements change nahi kar sakte),
# lekin hum unhe analyze ya use karne ke liye kuch functions laga sakte hain.


# 1️⃣ len() → Returns the number of elements in the tuple
tup = (10, 20, 30, 40)
print(len(tup))                        # Output: 4


# 2️⃣ max() → Returns the maximum element (only works if all elements are comparable)
num_tup = (4, 7, 1, 9, 3)
print(max(num_tup))                    # Output: 9


# 3️⃣ min() → Returns the minimum element
print(min(num_tup))                    # Output: 1


# 4️⃣ sum() → Returns the sum of all numeric elements in a tuple
print(sum(num_tup))                    # Output: 24


# 5️⃣ tuple() → Converts another data type into a tuple
my_list = [1, 2, 3, 4]
converted = tuple(my_list)
print(converted)                       # Output: (1, 2, 3, 4)


# 6️⃣ count() → Returns how many times a particular element appears in a tuple
dup_tup = (2, 4, 2, 6, 2, 8)
print(dup_tup.count(2))                # Output: 3


# 7️⃣ index() → Returns the index of the first occurrence of an element
letters = ('a', 'b', 'c', 'b', 'd')
print(letters.index('b'))              # Output: 1


# 8️⃣ sorted() → Returns a sorted list (does not modify the tuple)
nums = (5, 2, 8, 1, 3)
print(sorted(nums))                    # Output: [1, 2, 3, 5, 8]


# 9️⃣ any() → Returns True if any element of the tuple is True
bool_tup = (0, False, 5, 0)
print(any(bool_tup))                   # Output: True (because 5 is True)


# 🔟 all() → Returns True only if all elements are True (non-zero or non-empty)
bool_tup2 = (1, 2, 3)
print(all(bool_tup2))                  # Output: True


# 11️⃣ enumerate() → Returns both index and value while looping
for index, value in enumerate(('x', 'y', 'z')):
    print(index, value)
# Output:
# 0 x
# 1 y
# 2 z


# 12️⃣ reversed() → Returns an iterator that accesses the tuple in reverse order
rev_tup = (1, 2, 3, 4)
print(tuple(reversed(rev_tup)))        # Output: (4, 3, 2, 1)


# 13️⃣ zip() → Combine multiple tuples element-wise
tup1 = (1, 2, 3)
tup2 = ('a', 'b', 'c')
zipped = tuple(zip(tup1, tup2))
print(zipped)                          # Output: ((1, 'a'), (2, 'b'), (3, 'c'))


# 14️⃣ in keyword → To check if an element exists in tuple
print(2 in tup1)                       # Output: True
print(9 in tup1)                       # Output: False



# Difference Between List and Tuple

# Lists and Tuples dono hi sequence data types hain in Python,
# lekin dono ke beech kuch major differences hote hain:

# 🔹 LIST vs TUPLE COMPARISON TABLE 🔹

# | Feature / Property              | List                              | Tuple                              |
# |----------------------------------|-----------------------------------|------------------------------------|
# | Definition                      | Ordered collection of elements    | Ordered collection of elements     |
# | Syntax                          | Defined using [ ] (square brackets) | Defined using ( ) (parentheses)    |
# | Mutability                      | Mutable (can be changed)          | Immutable (cannot be changed)      |
# | Methods                         | Many (append, remove, pop, etc.)  | Very few (count, index)            |
# | Iteration Speed                 | Slower                            | Faster                             |
# | Memory Usage                    | More memory                       | Less memory                        |
# | Can be used as Dictionary Key   | ❌ No                              | ✅ Yes                             |
# | Suitable for                    | Data that changes frequently      | Fixed data (constant set of values)|
# | Example                         | [1, 2, 3]                         | (1, 2, 3)                          |
# | Nesting                         | Allowed                           | Allowed                            |
# | Performance                     | Slightly slower                   | Slightly faster                    |
# | Conversion                      | tuple(list_name)                  | list(tuple_name)                   |


# 🔸 Example Demonstration:

# List Example
my_list = [10, 20, 30]
my_list.append(40)                     # You can modify a list
print("List:", my_list)                # Output: [10, 20, 30, 40]

# Tuple Example
my_tuple = (10, 20, 30)
# my_tuple.append(40)                  # ❌ This will cause an error (tuples are immutable)
print("Tuple:", my_tuple)              # Output: (10, 20, 30)


# ✅ Summary:
# - Use LIST when you need to modify, add, or remove items frequently.
# - Use TUPLE when your data is fixed and should not change during program execution.

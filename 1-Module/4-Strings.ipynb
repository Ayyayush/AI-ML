# -----------------------------------------------------
# Arithmetic Operations  
# -----------------------------------------------------
# +  (Concatenation) → 2 strings ko jodta hai
# *  (Repetition)    → String ko multiple times repeat karta hai

s1 = "Hello "
s2 = "World"

print(s1 + s2)             # Output: Hello World   (Concatenation)
print(s1 * 3)              # Output: Hello Hello Hello   (Repetition)

# Note: Strings ko divide (-, /) nahi kar sakte, sirf + aur * allowed hai


# -----------------------------------------------------
# Relational Operations  
# -----------------------------------------------------
# Relational operators (==, !=, <, >, <=, >=) se hum strings compare karte hain
# Comparison lexicographically hota hai (dictionary order ke hisaab se)

a = "apple"
b = "banana"

print(a == b)              # False  (kyunki dono same nahi hain)
print(a != b)              # True
print(a < b)               # True   (kyunki 'a' alphabetically 'b' se pehle aata hai)
print(a > b)               # False


# -----------------------------------------------------
# Logical Operations  
# -----------------------------------------------------
# Logical operators (and, or, not) strings ke truth values pe kaam karte hain
# Non-empty string → True, Empty string → False

x = "Hello"
y = ""

print(x and y)             # Output: ''   (kyunki y False hai)
print(x or y)              # Output: 'Hello'   (kyunki x True hai)
print(not x)               # Output: False   (kyunki x empty nahi hai)


# -----------------------------------------------------

for ch in s:

# -----------------------------------------------------
# Membership Operator  
# -----------------------------------------------------
# 'in' aur 'not in' operators check karte hain ki koi substring string ke andar hai ya nahi

sentence = "Learning Python is fun"

print("Python" in sentence)        # Output: True   (kyunki 'Python' present hai)
print("Java" not in sentence)      # Output: True   (kyunki 'Java' present nahi hai)


# -----------------------------------------------------
# COMMON FUNCTIONS IN STRING  
# -----------------------------------------------------

s = "   Hello Python World   "

# -----------------------------------------------------
# 1. len()
# -----------------------------------------------------
# String ki length (characters ki total count) batata hai
print(len(s))                    # Output: 23


# -----------------------------------------------------
# 2. upper()
# -----------------------------------------------------
# String ke saare characters ko uppercase mein convert karta hai
print(s.upper())                 # Output: "   HELLO PYTHON WORLD   "


# -----------------------------------------------------
# 3. lower()
# -----------------------------------------------------
# String ke saare characters ko lowercase mein convert karta hai
print(s.lower())                 # Output: "   hello python world   "


# -----------------------------------------------------
# 4. strip(), lstrip(), rstrip()
# -----------------------------------------------------
# strip() → leading aur trailing spaces hataata hai
# lstrip() → sirf left side ke spaces hatata hai
# rstrip() → sirf right side ke spaces hatata hai
print(s.strip())                 # Output: "Hello Python World"
print(s.lstrip())                # Output: "Hello Python World   "
print(s.rstrip())                # Output: "   Hello Python World"


# -----------------------------------------------------
# 5. replace(old, new)
# -----------------------------------------------------
# String ke andar kisi word ko doosre word se replace karta hai
print(s.replace("Python", "Java"))   # Output: "   Hello Java World   "


# -----------------------------------------------------
# 6. split()
# -----------------------------------------------------
# String ko list mein todta hai (by default spaces ke basis pe)
words = s.split()
print(words)                    # Output: ['Hello', 'Python', 'World']


# -----------------------------------------------------
# 7. join()
# -----------------------------------------------------
# List ke elements ko ek single string mein join karta hai
joined = "-".join(words)
print(joined)                   # Output: "Hello-Python-World"


# -----------------------------------------------------
# 8. find() aur index()
# -----------------------------------------------------
# find() → substring ka index return karta hai (agar milta hai)
# index() → same kaam karta hai, par agar substring na mile toh error deta hai

# print(s.index("Java"))        # Error: ValueError (kyunki 'Java' nahi hai)


# -----------------------------------------------------
# 9. count()
# -----------------------------------------------------

# -----------------------------------------------------
# 13. max() aur min() Functions  
# -----------------------------------------------------
# max() → string ke andar sabse "maximum" (lexicographically bada) character return karta hai  
# min() → string ke andar sabse "minimum" (lexicographically chhota) character return karta hai  
# ASCII values ke base par compare karta hai  

text = "Python"
print(max(text))                # Output: y  (kyunki 'y' ka ASCII sabse bada hai)

# -----------------------------------------------------
# 14. format() Function  
# -----------------------------------------------------
# format() function dynamic string formatting ke liye use hota hai  
# Curly braces {} placeholders ke jaise kaam karte hain  

name = "Treddy"
age = 21
language = "Python"

# Example 1: Positional formatting
print("My name is {} and I am {} years old.".format(name, age))
# Output: My name is Treddy and I am 21 years old.

# Example 2: Indexed formatting
print("I love {2}, my name is {0}, and I am {1}.".format(name, age, language))
# Output: I love Python, my name is Treddy, and I am 21.

# Example 3: Named formatting
print("Name: {n}, Age: {a}, Language: {l}".format(n=name, a=age, l=language))
# Output: Name: Treddy, Age: 21, Language: Python

# Example 4: Float formatting
pi = 3.1415926535
print("Value of PI: {:.2f}".format(pi))   # Output: Value of PI: 3.14

```python
# -----------------------------------------------------
# PRACTICE PROBLEMS ON STRINGS  
# -----------------------------------------------------

# -----------------------------------------------------
# 1️⃣ Find length of string without using len() function  
# -----------------------------------------------------
s = "Hello Python"
count = 0
for ch in s:
    count += 1
print("Length of string:", count)        # Output: 12




# -----------------------------------------------------
# 2️⃣ Extract username from an email ID  
# -----------------------------------------------------
email = "treddy123@gmail.com"
username = ""
for ch in email:
    if ch == "@":                        # jaise hi '@' mile loop break karo
        break
    username += ch
print("Username is:", username)          # Output: treddy123




# -----------------------------------------------------
# 3️⃣ Remove a particular character from a string  
# -----------------------------------------------------
s = "programming"
ch_remove = 'm'
result = ""
for ch in s:
    if ch != ch_remove:                  # agar character remove wala nahi hai toh add kar do
        result += ch
print("After removing character:", result)   # Output: prograing




# -----------------------------------------------------
# 4️⃣ Check whether a given string is palindrome or not  
# -----------------------------------------------------
s = 'abba'
rev = s[::-1]                            # reverse string using slicing
if s == rev:
    print(s, "is a palindrome")          # Output: abba is a palindrome
else:
    print(s, "is not a palindrome")





# -----------------------------------------------------
# 5️⃣ Count number of words in string without using split()  
# -----------------------------------------------------
s = "Python is very easy"
count = 1                                # 1 word by default (kyunki last word ke baad space nahi hoti)
for ch in s:
    if ch == " ":                        # jab bhi space mile, ek word complete hua
        count += 1
print("Total words:", count)             # Output: 4




# -----------------------------------------------------
# 6️⃣ Convert a string to title case without using title()  
# -----------------------------------------------------
s = "hello python world"
result = ""
make_upper = True                # har new word ka pehla letter capital banana h


for ch in s:
    if make_upper and ch != " ":
        result += ch.upper()
        make_upper = False
    else:
        result += ch.lower()
    if ch == " ":
        make_upper = True
print("Title Case:", result)             # Output: Hello Python World





# -----------------------------------------------------
# 7️⃣ Convert an integer into string  
# -----------------------------------------------------


num = 12345
temp = num
result = ""                        # yahan hum string build karenge

while temp > 0:
    digit = temp % 10              # last digit nikala
    char = chr(digit + 48)         # ASCII 48 = '0', so 49='1', 50='2'... 
    result = char + result         # har new digit ko beginning mein jodo
    temp = temp // 10              # last digit hata do

print("Converted string:", result)  # Output: "12345"
print(type(result))                 # Output: <class 'str'>



num = 12345
string_num = str(num)                    # type conversion integer → string
print("Converted string:", string_num)   # Output: "12345"
print(type(string_num))                  # Output: <class 'str'>
```

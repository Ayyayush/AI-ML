# ============================================
# 🧠 Sets in Python
# ============================================

# 🔹 Definition:
# Set ek aisa collection hota hai jisme multiple unique elements store kiye ja sakte hain.
# Duplicate elements automatically remove ho jaate hain.
# Sets are unordered, mutable, and unindexed.

# Example:
my_set = {1, 2, 3, 3, 4}
print(my_set)                         # Output: {1, 2, 3, 4}  → duplicate 3 removed


# ============================================
# 🔹 Characteristics of Sets
# ============================================

# 1️⃣ Unordered → Elements ka order fix nahi hota.
# 2️⃣ Unindexed → Elements ko index se access nahi kar sakte (e.g., my_set[0] ❌).
# 3️⃣ Mutable → Hum set mein naya element add ya remove kar sakte hain.
# 4️⃣ No Duplicates → Automatically duplicates ko remove kar deta hai.
# 5️⃣ Heterogeneous → Different data types store kar sakta hai (int, str, tuple etc.)
# 6️⃣ Hash-based → Internally hash table ke through elements store karta hai, isi wajah se fast hota hai.


# Example:
set1 = {1, "apple", (2, 3)}
print(set1)                           # Output: {1, 'apple', (2, 3)}


# ============================================
# 🔹 Advantages of Sets
# ============================================

# ✅ Fast Membership Testing:
# Checking if element exists (using 'in') is faster than list because of hashing.

# ✅ Automatic Duplicate Removal:
# Set automatically removes duplicate values — perfect for unique data storage.

# ✅ Useful Mathematical Operations:
# Supports operations like union, intersection, difference, symmetric difference etc.

# ✅ Highly Efficient:
# Operations like add, remove, and search are on average O(1).


# ============================================
# 🔹 Disadvantages of Sets
# ============================================

# ❌ Unordered → Cannot access elements by index.
# ❌ Unhashable Elements Restriction → Cannot contain mutable items like lists or dictionaries.
# ❌ Non-reversible → Once created, order cannot be predicted or reversed.
# ❌ No Duplicate Preservation → If duplicate order matters (like list), set won’t help.




# ============================================
# 🔹 Difference Between Sets and Lists
# ============================================

# | Feature / Property        | Set                              | List                            |
# |----------------------------|----------------------------------|---------------------------------|
# | Syntax                    | { } curly braces                 | [ ] square brackets             |
# | Order                     | Unordered                        | Ordered                         |
# | Indexing / Slicing        | Not supported                    | Supported                       |
# | Duplicates                | Not allowed                      | Allowed                         |
# | Mutable                   | Yes                              | Yes                             |
# | Speed (Search/Add)        | Faster (uses hashing)             | Slower (linear search)          |
# | Use Case                  | When unique, unordered data needed| When order matters              |
# | Example                   | {1,2,3}                          | [1,2,3]                         |




# ============================================
# 🔹 Difference Between Sets and Tuples
# ============================================

# | Feature / Property        | Set                              | Tuple                           |
# |----------------------------|----------------------------------|---------------------------------|
# | Syntax                    | { } curly braces                 | ( ) parentheses                 |
# | Mutability                | Mutable                          | Immutable                       |
# | Order                     | Unordered                        | Ordered                         |
# | Duplicates                | Not allowed                      | Allowed                         |
# | Indexing / Slicing        | Not supported                    | Supported                       |
# | Speed                     | Faster for search (hashing)      | Depends on traversal            |
# | Use Case                  | Unique, unordered collections     | Fixed, ordered data collection  |
# | Example                   | {1, 2, 3}                        | (1, 2, 3)                       |



# Creating Sets 

# 👉 A Set in Python is created using curly braces {} or by using the set() constructor.

# ------------------------------
# 1️⃣ Creating a Set using {}
fruits = {"apple", "banana", "cherry"}
print(fruits)                             # Output: {'banana', 'cherry', 'apple'}

# ------------------------------
# 2️⃣ Creating a Set using set() constructor
numbers = set([1, 2, 3, 4, 4, 2])
print(numbers)                            # Output: {1, 2, 3, 4}  → Duplicates removed automatically

# ------------------------------
# 3️⃣ Creating an Empty Set
empty_set = set()                         # ✅ Correct way
print(type(empty_set))                    # Output: <class 'set'>

# ❌ Wrong Way → empty_set = {} creates an empty dictionary, not a set
wrong_set = {}
print(type(wrong_set))                    # Output: <class 'dict'>

# ------------------------------
# 4️⃣ Creating a Set from a String
chars = set("campusx")
print(chars)                              # Output: {'p', 'a', 'x', 's', 'u', 'c', 'm'} (order may vary)

# ------------------------------
# 5️⃣ Creating a Set using Tuple
tuple_set = set((10, 20, 30, 40))
print(tuple_set)                          # Output: {40, 10, 20, 30}

# ------------------------------
# 6️⃣ Creating a Set using Mixed Data Types
mixed = {"apple", 10, True, (1, 2)}
print(mixed)                              # ✅ Works fine (because tuple and bool are immutable)

# ------------------------------
# 7️⃣ Invalid Example (Immutable Rule)
# invalid = {["apple", "banana"], 10}     # ❌ Error: list is mutable, cannot be in a set


# Accessing Items in a Set

# ⚡ Important:
# Sets are *unordered* collections → So elements do NOT have indexes like lists or tuples.
# That means we cannot access elements using indexing or slicing.
# But we can still access items by looping or checking membership.

# ------------------------------
# 1️⃣ Looping through Set Elements
fruits = {"apple", "banana", "cherry"}

for item in fruits:
    print(item)                      # Output order may vary because sets are unordered

# ------------------------------
# 2️⃣ Checking if an Element Exists
print("apple" in fruits)             # ✅ True → 'apple' exists in set
print("mango" in fruits)             # ❌ False → 'mango' not in set

# ------------------------------
# 3️⃣ Checking if an Element Does NOT Exist
print("mango" not in fruits)         # ✅ True → 'mango' not found in set

# ------------------------------
# 4️⃣ Converting Set to List/Tuple to Access by Index
# 👉 Since sets are unordered, we first convert them to a list or tuple for index-based access
fruits_list = list(fruits)
print(fruits_list[0])                # Accessing element by index after conversion

# ------------------------------
# 5️⃣ Iterating with Enumeration
for index, item in enumerate(fruits):
    print(index, item)               # Gives an index number (not actual order)

# ------------------------------
# ⚠️ NOTE:
# - Direct indexing like fruits[0] ❌ will raise an error
# - Order of elements in a set can change anytime


# Adding Items in a Set

# ⚡ Reminder:
# - Sets are *mutable* → you can add or remove elements after creation.
# - But elements themselves must be *immutable* (no lists or dicts allowed).

# ------------------------------
# 1️⃣ Using add() Method
# 👉 Used to add a single element to the set.

fruits = {"apple", "banana", "cherry"}
fruits.add("mango")                        # Adds one new element
print(fruits)                              # Output: {'apple', 'banana', 'cherry', 'mango'} (order may vary)

# If you try to add an existing element, it won’t give an error; it simply ignores it.
fruits.add("apple")
print(fruits)                              # 'apple' won't be added again (no duplicates allowed)

# ------------------------------
# 2️⃣ Using update() Method
# 👉 Used to add *multiple* elements at once from any iterable (list, tuple, set, etc.)

fruits.update(["kiwi", "grapes"])
print(fruits)                              # Output: {'apple', 'banana', 'cherry', 'mango', 'kiwi', 'grapes'}

# You can also add elements from another set or tuple
more_fruits = {"papaya", "orange"}
fruits.update(more_fruits)
print(fruits)                      # Output includes both previous and new fruits

# ------------------------------
# 3️⃣ Adding Elements from a String
# Every character of string is treated as a separate element
chars = set()
chars.update("data")
print(chars)                               # Output: {'a', 't', 'd'}

# ------------------------------
# ⚠️ Note:
# - add() → single element only
# - update() → multiple elements (from iterable)
# - Adding duplicates has no effect (set automatically removes duplicates)


# Deleting Items in a Set

# ⚡ Reminder:
# - Sets are mutable, so we can remove or clear elements anytime.
# - But since sets are unordered, we cannot remove by index (like lists).

fruits = {"apple", "banana", "cherry", "mango", "grapes"}

# ------------------------------
# 1️⃣ remove() Method
# 👉 Removes a specific element from the set.
# ⚠️ If the element does NOT exist, it raises a KeyError.

fruits.remove("banana")
print(fruits)                             # Output: {'apple', 'cherry', 'mango', 'grapes'}

# fruits.remove("kiwi")                   # ❌ Will give KeyError: 'kiwi' (because not present)

# ------------------------------
# 2️⃣ discard() Method
# 👉 Removes a specific element, BUT does NOT give error if the element doesn't exist.

fruits.discard("apple")                   # Removes apple safely
fruits.discard("kiwi")                    # No error even though 'kiwi' not in set
print(fruits)                             # Output: {'cherry', 'mango', 'grapes'}

# ------------------------------
# 3️⃣ pop() Method
# 👉 Removes and returns a random element (since sets are unordered).
# Useful when you just want to remove something but don't care which.

removed_item = fruits.pop()
print("Removed:", removed_item)           # Output: any random element
print(fruits)                             # Remaining elements after pop()

# ------------------------------
# 4️⃣ clear() Method
# 👉 Removes all elements from the set (empties it completely)

fruits.clear()
print(fruits)                             # Output: set()

# ------------------------------
# 5️⃣ del Keyword
# 👉 Deletes the entire set object from memory.

numbers = {1, 2, 3}
del numbers                               # The variable itself is deleted
# print(numbers)                          # ❌ Will give NameError: name 'numbers' is not defined

# ------------------------------
# ⚡ Summary:
# remove(x)  → deletes 'x' (error if not found)
# discard(x) → deletes 'x' (no error if not found)
# pop()      → removes random element
# clear()    → removes all elements
# del        → deletes entire set variable


# Set Operations in Python

# ⚡ Reminder:
# Sets are very powerful for mathematical operations like
# Union, Intersection, Difference, and Symmetric Difference.

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# ------------------------------
# 1️⃣ UNION → Combines all unique elements from both sets
# Syntax: A.union(B) or A | B

union_set = A.union(B)
print("Union:", union_set)                       # Output: {1, 2, 3, 4, 5, 6, 7, 8}

# Using operator
print("Union (|):", A | B)                       # Same result

# ------------------------------
# 2️⃣ INTERSECTION → Common elements in both sets
# Syntax: A.intersection(B) or A & B

intersect_set = A.intersection(B)
print("Intersection:", intersect_set)            # Output: {4, 5}

# Using operator
print("Intersection (&):", A & B)

# ------------------------------
# 3️⃣ DIFFERENCE → Elements in A but not in B
# Syntax: A.difference(B) or A - B

diff_set = A.difference(B)
print("Difference (A - B):", diff_set)           # Output: {1, 2, 3}

# Using operator
print("Difference (-):", A - B)

# Similarly → B - A gives elements in B not in A
print("Difference (B - A):", B - A)              # Output: {8, 6, 7}

# ------------------------------
# 4️⃣ SYMMETRIC DIFFERENCE → Elements that are in A or B but not both
# Syntax: A.symmetric_difference(B) or A ^ B

sym_diff = A.symmetric_difference(B)
print("Symmetric Difference:", sym_diff)         # Output: {1, 2, 3, 6, 7, 8}

# Using operator
print("Symmetric Difference (^):", A ^ B)

# ------------------------------
# 5️⃣ SUBSET & SUPERSET Checking
# Subset → Checks if all elements of A are in B
print(A.issubset(B))                             # False → A is not subset of B

# Superset → Checks if A contains all elements of B
print(A.issuperset(B))                           # False → A doesn’t contain all of B

# ------------------------------
# 6️⃣ DISJOINT CHECKING
# Checks whether two sets have no common elements
print(A.isdisjoint(B))                           # False → (4, 5 are common)

# ------------------------------
# ⚡ Summary Table:
# Operation                 | Method                     | Operator
# --------------------------|----------------------------|-----------
# Union                     | A.union(B)                 | A | B
# Intersection              | A.intersection(B)          | A & B
# Difference                | A.difference(B)            | A - B
# Symmetric Difference       | A.symmetric_difference(B)  | A ^ B
# Subset                    | A.issubset(B)              | —
# Superset                  | A.issuperset(B)            | —
# Disjoint                  | A.isdisjoint(B)            | —


# Set Functions in Python

# ⚡ Reminder:
# Sets ke upar hum kuch built-in functions use kar sakte hain 
# jo numerical aur iterable operations perform karte hain.

numbers = {10, 20, 30, 40, 50}

# ------------------------------
# 1️⃣ len()
# 👉 Returns total number of elements in the set

print("Length:", len(numbers))                      # Output: 5

# ------------------------------
# 2️⃣ max()
# 👉 Returns the largest element from the set

print("Maximum:", max(numbers))                     # Output: 50

# ------------------------------
# 3️⃣ min()
# 👉 Returns the smallest element from the set

print("Minimum:", min(numbers))                     # Output: 10

# ------------------------------
# 4️⃣ sum()
# 👉 Returns the sum of all numerical elements in the set

print("Sum:", sum(numbers))                         # Output: 150

# ------------------------------
# 5️⃣ sorted()
# 👉 Returns a sorted list of set elements (ascending by default)

print("Sorted:", sorted(numbers))                   # Output: [10, 20, 30, 40, 50]
print("Sorted (descending):", sorted(numbers, reverse=True))   # Output: [50, 40, 30, 20, 10]

# ------------------------------
# 6️⃣ any()
# 👉 Returns True if any element in set is True (non-zero / non-empty)

sample_set = {0, 0, 0, 5}
print("Any True?:", any(sample_set))                # True (since 5 is non-zero)

# ------------------------------
# 7️⃣ all()
# 👉 Returns True only if all elements are True (non-zero / non-empty)

print("All True?:", all(sample_set))                # False (since there are 0s)

# ------------------------------
# 8️⃣ copy()
# 👉 Returns a shallow copy of the set

copy_set = numbers.copy()
print("Copy of set:", copy_set)                     # Output: {10, 20, 30, 40, 50}

# ------------------------------
# 9️⃣ clear()
# 👉 Removes all elements from the set (makes it empty)

copy_set.clear()
print("After clear:", copy_set)                     # Output: set()

# ------------------------------
# ⚡ Summary Table:
# Function      | Description
# --------------|-------------------------------
# len()         | Counts elements in set
# max()         | Finds largest value
# min()         | Finds smallest value
# sum()         | Adds all numbers
# sorted()      | Returns sorted list of elements
# any()         | True if any element is True
# all()         | True if all elements are True
# copy()        | Copies set
# clear()       | Removes all elements


# FrozenSet in Python

# ⚡ Definition:
# A frozenset is an *immutable* version of a set.
# Once created, you cannot add, remove, or modify its elements.
# It is hashable — meaning you can use it as a key inside a dictionary or as an element inside another set.

# ------------------------------
# 1️⃣ Creating a FrozenSet
normal_set = {1, 2, 3, 4}
frozen = frozenset(normal_set)

print("Normal Set:", normal_set)
print("Frozen Set:", frozen)

# Output:
# Normal Set: {1, 2, 3, 4}
# Frozen Set: frozenset({1, 2, 3, 4})

# ------------------------------
# 2️⃣ Creating FrozenSet Directly from Iterable
f1 = frozenset([10, 20, 30])
f2 = frozenset(("a", "b", "c"))
f3 = frozenset("campusx")

print(f1)                                     # frozenset({10, 20, 30})
print(f2)                                     # frozenset({'a', 'b', 'c'})
print(f3)                                     # frozenset({'x', 'p', 'a', 's', 'u', 'c', 'm'}) (order may vary)

# ------------------------------
# 3️⃣ Performing Set Operations on FrozenSets
# ✅ You can still do union, intersection, difference, etc. (but they return new frozensets)

A = frozenset({1, 2, 3, 4})
B = frozenset({3, 4, 5, 6})

print("Union:", A | B)                        # frozenset({1, 2, 3, 4, 5, 6})
print("Intersection:", A & B)                 # frozenset({3, 4})
print("Difference:", A - B)                   # frozenset({1, 2})
print("Symmetric Difference:", A ^ B)         # frozenset({1, 2, 5, 6})

# ------------------------------
# 4️⃣ Immutable Nature (❌ Modification Not Allowed)
# A.add(10)                                  # ❌ Error → AttributeError: 'frozenset' object has no attribute 'add'
# A.remove(2)                                # ❌ Error → same reason

# ------------------------------
# 5️⃣ Using FrozenSet as Dictionary Key (✅ Allowed)
data = {frozenset({1, 2, 3}): "Group A", frozenset({4, 5}): "Group B"}
print(data)                                  # Output: {frozenset({1, 2, 3}): 'Group A', frozenset({4, 5}): 'Group B'}

# ------------------------------
# ⚡ Key Differences Between set() and frozenset()
# ------------------------------------------------
# Feature               | set() (Mutable)               | frozenset() (Immutable)
# -----------------------|------------------------------|-----------------------------
# Mutable?              | Yes                          | No
# Can be dictionary key? | No                           | Yes
# Supports add/remove?   | Yes                          | No
# Hashable?              | No                           | Yes
# Syntax Example         | {1, 2, 3} or set([1, 2, 3])  | frozenset([1, 2, 3])


# Set Comprehension in Python

# ⚡ Definition:
# Set Comprehension ek short aur elegant way hai set banane ka, using loops and conditions — 
# just like list comprehension, but it creates a *set* instead of a list.

# Syntax:
# {expression for item in iterable if condition}

# ------------------------------
# 1️⃣ Basic Example
squares = {x**2 for x in range(6)}
print("Squares:", squares)                         # Output: {0, 1, 4, 9, 16, 25}

# ------------------------------
# 2️⃣ Set Comprehension with Condition
even_squares = {x**2 for x in range(10) if x % 2 == 0}
print("Even Squares:", even_squares)               # Output: {0, 64, 4, 16, 36}

# ------------------------------
# 3️⃣ Set Comprehension from a String
chars = {ch for ch in "campusx"}
print("Unique Characters:", chars)                 # Output: {'p', 'a', 'x', 's', 'u', 'c', 'm'} (order may vary)

# ------------------------------
# 4️⃣ Filtering and Transforming Together
names = ["Alice", "Bob", "Ankit", "Babli", "Arjun"]
A_names = {name for name in names if name.startswith("A")}
print("Names starting with A:", A_names)           # Output: {'Ankit', 'Arjun', 'Alice'}

# ------------------------------
# 5️⃣ Nested Set Comprehension Example
pairs = {(x, y) for x in range(3) for y in range(3) if x != y}
print("Pairs (x, y) where x != y:", pairs)
# Output: {(0, 1), (1, 0), (2, 0), (0, 2), (2, 1), (1, 2)}

# ------------------------------
# 6️⃣ Removing Duplicates Automatically
nums = [1, 2, 2, 3, 4, 4, 5]
unique_squares = {x**2 for x in nums}
print("Unique Squares:", unique_squares)           # Output: {1, 4, 9, 16, 25}

# ------------------------------
# ⚡ Advantages of Set Comprehension:
# ✅ More concise and readable
# ✅ Automatically removes duplicates
# ✅ Fast for simple filtering and transformation

# ------------------------------
# ⚠️ Note:
# - Result is always an *unordered* set
# - No duplicates allowed
# - Output order can change anytime

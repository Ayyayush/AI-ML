## 🔹 Types of Arguments in Python

### 1️⃣ Positional Arguments
- Order important hota hai  
- Values position ke hisaab se assign hoti hain

> Jo value pehle likhi, wo pehle parameter me jayegi.

---

### 2️⃣ Keyword Arguments
- Parameter ka naam use karke value dete hain  
- Order important nahi hota  

👉 Hinglish:
> Naam likh ke value dena = keyword argument.

---

### 3️⃣ Default Arguments
- Parameter ke liye pehle se value set hoti hai  
- Agar user value na de, default use hoti hai  

👉 Hinglish:

---

### 4️⃣ Variable Length Arguments
- Jab arguments ki number fixed na ho

---

## 🔹 `*args` kya hota hai?

### 📌 Definition
`*args` is a **special keyword** used to pass a **variable number of non-keyword arguments** to a function.

### 📌 Key Points
- Arguments **tuple** ke form me milte hain  
- Order maintained hota hai  
- Name `args` convention hai (`*` important hai)

👉 Hinglish:
> `*args` se hum multiple values ek saath pass kar sakte hain.

---

## 🔹 `**kwargs` kya hota hai?

### 📌 Definition
`**kwargs` is a **special keyword** used to pass a **variable number of keyword arguments** to a function.

### 📌 Key Points
- Arguments **dictionary** ke form me milte hain  
- Key-value pair hota hai  
- Name `kwargs` convention hai (`**` important hai)

👉 Hinglish:
> `**kwargs` tab use hota hai jab arguments naam ke saath aaye.

---

## 🔹 Difference between `*args` and `**kwargs`

| Feature | `*args` | `**kwargs` |
|------|-------|---------|
| Type | Tuple | Dictionary |
| Arguments | Non-keyword | Keyword |
| Order | Important | Not important |
| Use | Multiple values | Multiple named values |

---

## 🔹 Important Rules

- `*args` hamesha **positional arguments ke baad** aata hai  
- `**kwargs` hamesha **last me** aata hai  
- Dono optional hote hain  

👉 Hinglish:
> Sequence follow karna zaroori hai warna error aayega.

---

## 🔹 When to Use `*args` and `**kwargs`?

### Use `*args` when:
- Number of inputs unknown ho  
- Order important ho  

### Use `**kwargs` when:
- Named parameters chahiye  
- Configuration type data ho  

---

## 🔹 One-Line Summary (Interview)

> Functions code reuse aur structure ke liye hote hain, aur `*args` aur `**kwargs` variable number of inputs ko handle karne ke liye use hote hain.

```python
# ============================================================
# 📘 PYTHON FUNCTIONS – ALL DEMO CODES (ONE FILE)
# ============================================================
# This file contains simple demo codes for:
# 1) Basic functions
# 2) Functions with arguments
# 3) Types of arguments
# 4) *args
# 5) **kwargs
# ------------------------------------------------------------


# ============================================================
# 1️⃣ BASIC FUNCTION (WHY & WHEN TO USE)
# ============================================================

def greet():
    # This function prints a welcome message
    # It shows how functions help in code reuse
    print("Hello, welcome to Python")

# Function calls
greet()
greet()



# ============================================================
# 2️⃣ FUNCTION WITH ARGUMENTS
# ============================================================

def greet_user(name):
    # 'name' is an argument passed to the function
    print("Hello", name)

# Passing different values
greet_user("Rahul")
greet_user("Aman")



# ============================================================
# 3️⃣ TYPES OF ARGUMENTS
# ============================================================

# ----------------------------
# Positional Arguments
# ----------------------------

def add(a, b):
    # Values are assigned based on position
    print("Sum:", a + b)

add(2, 3)



# ----------------------------
# Keyword Arguments
# ----------------------------

def student(name, age):
    # Values are passed using parameter names
    print("Name:", name)
    print("Age:", age)

student(age=20, name="Riya")



# ----------------------------
# Default Arguments
# ----------------------------

def country(name, country="India"):
    # 'India' is the default value
    print("Name:", name)
    print("Country:", country)

country("Amit")
country("John", "USA")



# ============================================================
# 4️⃣ *args (VARIABLE LENGTH NON-KEYWORD ARGUMENTS)
# ============================================================

def show_numbers(*args):
    # *args stores multiple values in a tuple
    print("Numbers:", args)

show_numbers(1, 2, 3)
show_numbers(10, 20, 30, 40)



def sum_all(*args):
    # This function calculates the sum of all values
    print("Total Sum:", sum(args))

sum_all(1, 2, 3)
sum_all(5, 10, 15, 20)



# ============================================================
# 5️⃣ **kwargs (VARIABLE LENGTH KEYWORD ARGUMENTS)
# ============================================================

def show_details(**kwargs):
    # **kwargs stores data as key-value pairs in a dictionary
    print("Details:", kwargs)

show_details(name="Aman", age=21, city="Delhi")



def profile(**kwargs):
    # Iterating over dictionary items
    for key, value in kwargs.items():
        print(key, ":", value)

profile(name="Riya", course="Python", year=2025)



# ============================================================
# 6️⃣ USING *args AND **kwargs TOGETHER
# ============================================================

def demo(*args, **kwargs):
    # *args captures positional arguments
    # **kwargs captures keyword arguments
    print("Args:", args)
    print("Kwargs:", kwargs)

demo(1, 2, 3, name="Aman", role="Student")
```

# 📘 Memory Level Working of Functions in Python

---

## 🔹 1️⃣ Function `def` milte hi kya hota hai? (Definition Time)

Jab Python interpreter ko `def` keyword milta hai:

- Python **function ko execute nahi karta**
- Sirf function ka **structure memory me store** karta hai
- Interpreter **jump maar kar aage chala jaata hai** (end ya next statement par)

👉 Hinglish:
> `def` dekh kar Python bolta hai:  
> “Theek hai, function yaad rakh liya… abhi nahi chalaunga.”

📌 Important:
- Function body tab tak run **nahi hoti**
- Jab tak function **call** na ho

---

## 🔹 2️⃣ Function Call hone par kya hota hai?

Jab function call hota hai:

- Python **call stack** me ek **naya stack frame** banata hai
- Function ke liye **alag memory allocate** hoti hai
- Control program se **function ke andar chala jaata hai**

👉 Hinglish:
> Function call = Python bolta hai  
> “Ab is function ka kaam shuru karo”

---

## 🔹 3️⃣ Call Stack kya hota hai?

**Call Stack** ek memory structure hai jisme:

- Active functions ka record hota hai
- Har function ka apna:
  - Local variables
  - Arguments
  - Execution state

👉 Hinglish:
> Call stack ek list jaisa hota hai jo yaad rakhta hai kaunsa function chal raha hai.

📌 Rule:
- Last aaya function → pehle complete hota hai  
  (LIFO: Last In First Out)

---

## 🔹 4️⃣ Function apni memory me kaise behave karta hai?

Function ke andar:

- Wo ek **chhota independent program** ki tarah behave karta hai
- Uske variables:
  - Sirf us function ke liye valid hote hain
  - Bahar access nahi ho sakte (local scope)

👉 Hinglish:
> Function ka apna ghar hota hai  
> Bahar wale variables directly andar nahi aate

---

## 🔹 5️⃣ Function complete hone par kya hota hai?

Jab function:

- Apna kaam complete kar leta hai
- Ya `return` statement mil jaata hai

Toh:
- Function ka **stack frame destroy** ho jaata hai
- Memory **free** ho jaati hai
- Control **wapas caller** ke paas aa jaata hai

👉 Hinglish:
> Function ka kaam khatam → memory clean → wapas main program

---

## 🔹 6️⃣ Agar function me `return` statement nahi ho?

### 📌 Important Concept ✅

Agar function ke andar:
- `return` statement nahi likha hai

Toh bhi Python:
- **Kuch na kuch return karta hai**
- Wo hota hai 👉 `None`

👉 Hinglish:
> Python bolta hai:  
> “Tumne kuch return nahi bola, main `None` de deta hoon.”

📌 Example (Conceptual):
- Return likha nahi → output = `None`
- Ye automatic behavior hai

---

## 🔹 7️⃣ `None` kya hota hai?

- `None` ek **special value** hai
- Ye batata hai:
  - “Yaha koi meaningful value nahi hai”

👉 Hinglish:
> `None` matlab khaali value, zero nahi, empty nahi

📌 Interview line:
> Every Python function returns something.  
> If nothing is specified, it returns `None`.

---

## 🔹 8️⃣ Complete Flow (Step-by-Step)

1. Python file read karta hai
2. `def` milta hai → function register hota hai
3. Execution aage badhta hai
4. Function call hota hai
5. New memory stack frame banta hai
6. Function body execute hoti hai
7. Return value milti hai (ya `None`)
8. Stack frame destroy hota hai
9. Control wapas caller ko milta hai

# 📘 Variable Scope in Python (Function Scope Notes)

---

## 🔹 What is Variable Scope?

**Scope** defines **where a variable can be accessed or modified** in a program.

👉 Hinglish:
> Scope matlab variable ki “reach” — kaha use ho sakta hai aur kaha nahi.

---

## 🔹 Types of Variable Scope in Python

Python me mainly **2 important scopes** hoti hain:
1. **Local Scope**
2. **Global Scope**

---

## 🔹 Local Variable

### 📌 Definition
- Jo variable **function ke andar** define hota hai
- Sirf **usi function ke andar accessible** hota hai

👉 Hinglish:
> Local variable function ka personal variable hota hai.

### 📌 Important Rule
- Local variable ko **function ke bahar use nahi** kar sakte

👉 Hinglish:
> Function ke andar ka variable bahar leak nahi hota.

---

## 🔹 Global Variable

### 📌 Definition
- Jo variable **function ke bahar** define hota hai
- Program ke **har part** me accessible hota hai

👉 Hinglish:
> Global variable sabke liye common hota hai.

---

## 🔹 Local vs Global Variable (Key Differences)

| Feature | Local Variable | Global Variable |
|------|---------------|----------------|
| Defined where | Function ke andar | Function ke bahar |
| Accessible | Sirf function ke andar | Poore program me |
| Lifetime | Function call tak | Program end tak |
| Memory | Function call stack | Global memory |

---

## 🔹 Function ke andar Global Variable ka use

### 📌 Important Concept ✅

Agar function ke andar:
- Koi variable **local nahi milta**

Toh Python:
- Pehle **local scope** check karta hai
- Phir **global scope** me jaata hai

👉 Hinglish:
> Function ko local variable nahi mila, toh wo global variable utha lega.

---

## 🔹 Can function change global variable directly? ❌

### ❌ Default behavior
- Function **global variable ko modify nahi kar sakta**
- Agar same naam ka variable assign kiya:
  - Python usko **local maan lega**

👉 Hinglish:
> Function ke andar assignment = naya local variable

---

## 🔹 Why local variable cannot be used globally?

- Local variable function ke **memory stack** me hota hai
- Function end hote hi:
  - Memory destroy ho jaati hai

👉 Hinglish:
> Function khatam = local variable khatam

---

## 🔹 What if I want to change a Global Variable from Local?

### ✅ Solution: `global` keyword

`global` keyword Python ko batata hai:
- “Ye variable local nahi, global hai”
- Isme change allowed hai

👉 Hinglish:
> `global` bolta hai Python ko:  
> “Is variable ko global samajh, naya mat bana”

---

## 🔹 How `global` works (Conceptually)

- Python ko signal milta hai
- Variable global memory se access hota hai
- Value update hoti hai

👉 Hinglish:
> Global keyword se function ko permission milti hai global value change karne ki.

---

## 🔹 Important Caution ⚠️

- Global variables zyada use karna **bad practice** hai
- Debugging hard ho jaati hai
- Unexpected bugs aa sakte hain

👉 Hinglish:
> Global variable kam use karo, function ko clean rakho.

---

## 🔹 Best Practice (Interview Tip)

- Global variables sirf:
  - Constants
  - Configuration values
- Baaki data:
  - Function arguments se pass karo
  - Return value se update karo

# 📘 Lambda Function & Higher-Order Functions — Python Notes

---

## 🔹 Lambda Function

### 📌 What is a Lambda Function?

A **lambda function** is a **small, anonymous (no-name) function** written in a **single line**.

👉 Hinglish:
> Lambda function ek chhota function hota hai jiska naam nahi hota aur jo ek hi line me likha jata hai.

---

### 📌 Key Characteristics

- Naam nahi hota (anonymous)
- Sirf **ek expression** allowed
- Automatically value **return** karta hai
- Short-term use ke liye hota hai

👉 Hinglish:
> Jab function simple ho aur baar-baar reuse na karna ho, tab lambda use karte hain.

---

### 📌 Why Lambda Functions are Used?

- Code short aur clean banane ke liye  
- Temporary logic ke liye  
- Higher-order functions ke saath use karne ke liye  

👉 Hinglish:
> Lambda tab use hota hai jab function ek hi line ka kaam kare.

---

### 📌 Limitations of Lambda

- Complex logic allowed nahi
- Multiple statements nahi likh sakte
- Readability kam ho sakti hai

👉 Hinglish:
> Lambda sirf simple kaam ke liye best hai.

---

## 🔹 Higher-Order Function (HOF)

### 📌 What is a Higher-Order Function?

A **Higher-Order Function** wo function hota hai jo:

- Ya toh **function ko argument** ke roop me leta hai  
- Ya **function ko return** karta hai  

👉 Hinglish:
> Jo function function ke saath kaam kare, wahi higher-order function hota hai.

---

### 📌 Why Higher-Order Functions are Used?

- Code ko **flexible** banane ke liye  
- Repetition avoid karne ke liye  
- Different behaviors ko easily switch karne ke liye  

👉 Hinglish:
> Same function, different kaam — bas logic pass kar do.

---

### 📌 Relation with First-Class Functions

Higher-order functions possible hote hain kyunki:
- Python me functions **first-class citizens** hote hain

👉 Hinglish:
> Agar function value ki tarah behave na kare, toh HOF possible hi nahi hoga.

---

## 🔹 Common Higher-Order Functions in Python

### 1️⃣ `map()`
- Har element par function apply karta hai

👉 Hinglish:
> Ek kaam sab par apply karna.

---

### 2️⃣ `filter()`
- Condition ke basis par elements select karta hai

👉 Hinglish:
> Sirf wahi values jo condition satisfy karein.

---

### 3️⃣ `reduce()`
- Sab values ko combine karke ek result banata hai

> Sab ko jod ke ek final answer.

---

### 4️⃣ Custom Higher-Order Functions
- Jo user-defined functions ko accept/return karein

👉 Hinglish:
> Apna logic, apna control.

```python
# ============================================================
# 📘 LAMBDA FUNCTION EXAMPLES (BEGINNER → PRACTICAL)
# ============================================================


# ============================================================
# 1️⃣ SIMPLE LAMBDA (ADD TWO NUMBERS)
# ============================================================

add = lambda a, b: a + b
print(add(2, 3))

# Explanation:
# - lambda creates an anonymous function
# - a, b are inputs
# - a + b is the returned expression



# ============================================================
# 2️⃣ LAMBDA WITH ONE ARGUMENT
# ============================================================

square = lambda x: x * x
print(square(5))

# Explanation:
# - Single argument lambda
# - Automatically returns x * x



# ============================================================
# 3️⃣ LAMBDA WITH CONDITIONAL (TERNARY STYLE)
# ============================================================

check_even = lambda x: "Even" if x % 2 == 0 else "Odd"
print(check_even(10))
print(check_even(7))

# Explanation:
# - Lambda supports conditional expressions
# - Only expressions allowed, not statements



# ============================================================
# 4️⃣ LAMBDA INSIDE MAP()
# ============================================================

numbers = [1, 2, 3, 4, 5]

squared_numbers = list(map(lambda x: x * x, numbers))
print(squared_numbers)

# Explanation:
# - map applies lambda to each element
# - Output is a new list



# ============================================================
# 5️⃣ LAMBDA INSIDE FILTER()
# ============================================================

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)

# Explanation:
# - filter keeps values where condition is True



# ============================================================
# 6️⃣ LAMBDA INSIDE REDUCE()
# ============================================================

from functools import reduce

total_sum = reduce(lambda a, b: a + b, numbers)
print(total_sum)

# Explanation:
# - reduce combines all values into one
# - a = accumulated value
# - b = next element



# ============================================================
# 7️⃣ LAMBDA WITH MULTIPLE CONDITIONS
# ============================================================

grade = lambda marks: "Pass" if marks >= 40 else "Fail"
print(grade(75))
print(grade(30))

# Explanation:
# - Lambda used as decision logic



# ============================================================
# 8️⃣ LAMBDA AS FUNCTION ARGUMENT
# ============================================================

def operate(a, b, func):
    return func(a, b)

result = operate(10, 5, lambda x, y: x - y)
print(result)

# Explanation:
# - Lambda passed as an argument
# - operate is a higher-order function



# ============================================================
# 9️⃣ LAMBDA FOR SORTING
# ============================================================

students = [("Aman", 21), ("Riya", 19), ("Karan", 23)]

sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)

# Explanation:
# - Lambda extracts age for sorting
# - key decides sorting logic



# ============================================================
# 🔟 LAMBDA VS NORMAL FUNCTION (COMPARISON)
# ============================================================

def multiply(a, b):
    return a * b

multiply_lambda = lambda a, b: a * b

print(multiply(3, 4))
print(multiply_lambda(3, 4))

# Explanation:
# - Both do the same work
# - Lambda is shorter but less readable for complex logic



# ============================================================
# ⚠️ IMPORTANT LIMITATIONS OF LAMBDA
# ============================================================

# ❌ No multiple statements
# ❌ No loops
# ❌ No try-except
# ❌ Only one expression allowed


# ============================================================
# ✅ WHEN TO USE LAMBDA
# ============================================================

# ✔ Short, one-time logic
# ✔ With map, filter, reduce
# ✔ Sorting & small transformations

# ❌ Avoid lambda for complex logic


# ============================================================
# 🧠 INTERVIEW ONE-LINER
# ============================================================

# "Lambda functions are anonymous, single-expression functions
# commonly used with higher-order functions like map and filter."
```

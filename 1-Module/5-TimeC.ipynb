# -----------------------------------------------------
# TIME COMPLEXITY IN PYTHON (and in Programming)
# -----------------------------------------------------
# Time Complexity ek measure hai ki ek program ya algorithm ko 
# apna kaam complete karne mein kitna time lagta hai input size ke hisaab se.

# -----------------------------------------------------
# 🔹 Definition:
# -----------------------------------------------------
# Time Complexity batata hai — jaise jaise input size (n) badhta hai,
# program ke run hone ka time kaise badhta hai.
# Isse hum kisi bhi algorithm ki efficiency compare kar sakte hain.

# -----------------------------------------------------
# 🔹 Why Important?
# -----------------------------------------------------
# Jab input chhota hota hai, koi bhi algorithm fast lagta hai.
# Lekin jab data bada ho jaata hai (jaise 10^6 elements),
# tab sirf efficient algorithms hi fast kaam karte hain.
# Isi liye Time Complexity ka analysis zaroori hota hai.

# -----------------------------------------------------
# 🔹 Big-O Notation:
# -----------------------------------------------------
# Big-O notation ek mathematical representation hai
# jo worst-case time complexity ko represent karta hai.
# Common notations:
#   O(1)       → Constant time
#   O(log n)   → Logarithmic time
#   O(n)       → Linear time
#   O(n log n) → Log-linear time
#   O(n^2)     → Quadratic time
#   O(2^n)     → Exponential time
#   O(n!)      → Factorial time

# -----------------------------------------------------
# 🔹 Example 1: O(1) → Constant Time
# -----------------------------------------------------
# Ye algorithm input ke size pe depend nahi karta.
# Chahe input 10 ho ya 10 lakh, time same lagega.

def constant_example(arr):
    return arr[0]                # Sirf ek operation

# Example:
print(constant_example([10, 20, 30]))    # Output: 10
# ✅ Time Complexity: O(1)
# ✅ Space Complexity: O(1)


# -----------------------------------------------------
# 🔹 Example 2: O(n) → Linear Time
# -----------------------------------------------------
# Jab algorithm har element ko ek baar visit karta hai.

def linear_example(arr):
    total = 0
    for x in arr:
        total += x
    return total

# Example:
print(linear_example([1, 2, 3, 4, 5]))   # Output: 15
# ✅ Time Complexity: O(n)
# ✅ Space Complexity: O(1)


# -----------------------------------------------------
# 🔹 Example 3: O(n^2) → Quadratic Time
# -----------------------------------------------------
# Nested loops wale programs (2-level loop)
# har element ke liye pura loop chala dete hain.

def quadratic_example(arr):
    for i in arr:
        for j in arr:
            print(i, j)

# Example:
quadratic_example([1, 2, 3])
# Output:
# (1,1) (1,2) (1,3)
# (2,1) (2,2) (2,3)
# (3,1) (3,2) (3,3)
# ✅ Time Complexity: O(n^2)
# ✅ Space Complexity: O(1)


# -----------------------------------------------------
# 🔹 Example 4: O(log n) → Logarithmic Time
# -----------------------------------------------------
# Jaise binary search jisme input har step pe half hota jaata hai.

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# Example:
arr = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print(binary_search(arr, 6))     # Output: 5
# ✅ Time Complexity: O(log n)
# ✅ Space Complexity: O(1)


# -----------------------------------------------------
# 🔹 Example 5: O(n log n) → Log-linear Time
# -----------------------------------------------------
# Jaise merge sort, quick sort, heap sort — divide and conquer algorithms.

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Example:
print(merge_sort([5, 2, 8, 3, 1]))       # Output: [1, 2, 3, 5, 8]
# ✅ Time Complexity: O(n log n)
# ✅ Space Complexity: O(n)


# -----------------------------------------------------
# 🔹 Example 6: O(2^n) → Exponential Time
# -----------------------------------------------------
# Recursive problems jisme har call do aur calls create karti hai (like Fibonacci)

def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

print(fib(5))                            # Output: 5
# ✅ Time Complexity: O(2^n)
# ✅ Space Complexity: O(n) (recursive stack)


# -----------------------------------------------------
# 🔹 Example 7: O(n!) → Factorial Time
# -----------------------------------------------------
# Ye sabse slow hota hai — har permutation ko generate karta hai (like travelling salesman)

import itertools
arr = [1, 2, 3]
for p in itertools.permutations(arr):
    print(p)
# ✅ Time Complexity: O(n!)
# ✅ Space Complexity: O(n)

# -----------------------------------------------------
# 🔹 Summary Table:
# -----------------------------------------------------
# O(1)      → Constant      (Fastest)
# O(log n)  → Logarithmic
# O(n)      → Linear
# O(n log n)→ Log-linear
# O(n^2)    → Quadratic
# O(2^n)    → Exponential
# O(n!)     → Factorial     (Slowest)


# -----------------------------------------------------
# Techniques to measure Time Efficiency
# -----------------------------------------------------

# 1. Measuring Time
# -----------------------------------------------------
# Jab hum kisi program ka performance dekhna chahte hain 
# (yaani ki kitna time usko execute hone mein lagta hai),
# toh hum "time module" ka use karte hain.
# Ye ek built-in Python module hai jo humko start aur end time
# ke beech ka difference batata hai (execution duration).

# Syntax:
# import time
# start = time.time()     --> program start hone se pehle time note kro
# .... (program code) ....
# end = time.time()       --> program khatam hone ke baad time note kro
# print(end - start)      --> dono ke beech ka difference nikal lo


# Example 1: Measuring time taken by a simple loop
import time

start = time.time()             # ⏱️ program start hone ka time note kiya
for i in range(1000000):        # ek simple loop run kiya
    x = i * i                   # loop mein koi simple computation
end = time.time()               # ⏱️ program end hone ka time note kiya

print("Time taken:", end - start, "seconds")      # difference print kiya
# Yahan output approx 0.1 - 0.2 seconds ke beech mein aayega 
# (system speed pe depend karta hai)


# -----------------------------------------------------
# 2. Using timeit module
# -----------------------------------------------------
# timeit module accurate measurement deta hai, 
# kyunki ye code ko multiple times run karke average time deta hai.
# Isse hum micro-benchmarks (bohot chhoti operations) ka bhi time nikalte hain.

import timeit

# Example 2: Measuring small operation ka accurate time
code_to_test = """
a = []
for i in range(1000):
    a.append(i)
"""

execution_time = timeit.timeit(stmt=code_to_test, number=1000)
print("Average time using timeit:", execution_time, "seconds")

# Yahan "number=1000" matlab ye code 1000 baar run hoga 
# aur total execution time return karega.
# Ye small scale performance comparison ke liye best hota hai.


# -----------------------------------------------------
# 3. Using process_time()
# -----------------------------------------------------
# Ye method "CPU time" measure karta hai (system wait time nahi).
# Matlab ye sirf wahi time count karega jab CPU actual computation kar raha ho.

start = time.process_time()
for i in range(1000000):
    x = i ** 2
end = time.process_time()

print("CPU Process Time:", end - start, "seconds")
# Ye sirf CPU cycles ko count karta hai, system idle ya wait time nahi.


# -----------------------------------------------------
# 4. Using perf_counter()
# -----------------------------------------------------
# perf_counter() sabse accurate clock deta hai (high precision timer).
# Ye background mein sleep, wait, or pause sab include karta hai.

start = time.perf_counter()
for i in range(1000000):
    x = i + i
end = time.perf_counter()

print("High precision time (perf_counter):", end - start, "seconds")


# -----------------------------------------------------
# ⚠️ Problem with this Strategy
# -----------------------------------------------------
# Ye saare time-based methods (time.time(), timeit, perf_counter, etc.)
# sirf "execution environment" ke hisaab se time batate hain.
# Par agar hum ek hi code ko alag systems par run karein, 
# toh timing alag-alag ho sakti hai.

# Example Problems:
# 1️⃣ Different Hardware:
#     - Faster CPU → kam time
#     - Slower CPU → zyada time
#
# 2️⃣ Background Processes:
#     - Agar system busy hai (Chrome, VS Code open hai),
#       toh execution time badh sakta hai.
#
# 3️⃣ Internet ya Disk Delay:
#     - Agar program file read/write kar raha hai ya network call le raha hai,
#       toh delay environment dependent hoga.
#
# 4️⃣ Input Size Dependency:
#     - Time har baar input size par depend karega,
#       isliye ek single run se algorithm ka true efficiency measure nahi hota.

# 💡 Isiliye — Hum sirf “Measured Time” pe depend nahi karte.
# Hum theoretical analysis bhi karte hain (like counting operations)
# jisse system-independent efficiency samajh aati hai.


# -----------------------------------------------------
# SUMMARY
# -----------------------------------------------------
# ✅ time.time()         → Wall-clock time (basic use)
# ✅ timeit.timeit()     → Accurate average time (for comparisons)
# ✅ time.process_time() → CPU-only time (no sleep counted)
# ✅ time.perf_counter() → Most precise timer (recommended)
#
# ⚠️ Limitation:
#     - Result machine aur environment pe depend karta hai.
#     - Isliye real analysis ke liye "Time Complexity" aur 
#       "Counting Operations" method use karte hain.


# -----------------------------------------------------
# 2. Counting Operations Involved
# -----------------------------------------------------

# ⏳ Ye technique tab use hoti hai jab hume algorithm ka 
# theoretical performance samajhna ho — bina system ke 
# actual execution time ko measure kiye.
#
# Is method mein hum code ke andar hone wale “operations” ko count karte hain.
# Har basic operation (like +, -, =, <, >, loops etc.) ek “unit step” maana jata hai.
#
# Phir hum total number of steps ko 'n' ke terms mein likhte hain 
# (jahan 'n' input size hota hai). Isi se time complexity nikalti hai.


# -----------------------------------------------------
# Example 1: Simple loop ka operation count
# -----------------------------------------------------

def count_operations_example_1(n):
    count = 0                           # 1 operation (assignment)
    for i in range(n):                  # runs n times
        count += 1                      # each iteration = 1 operation
    print("Total operations:", count)   # 1 print operation

# Agar n = 5 diya toh loop 5 baar chalega
count_operations_example_1(5)

# Total operations ≈ n + 2
# (1 for initialization + n times loop body + 1 for print)
# => O(n) time complexity


# -----------------------------------------------------
# Example 2: Nested Loop ka operation count
# -----------------------------------------------------

def count_operations_example_2(n):
    count = 0
    for i in range(n):                  # Outer loop → runs n times
        for j in range(n):              # Inner loop → runs n times for each i
            count += 1                  # Ek inner operation
    print("Total operations:", count)

count_operations_example_2(4)

# Yahan total operations = n * n = n²
# => O(n²) time complexity


# -----------------------------------------------------
# Example 3: Logarithmic Loop
# -----------------------------------------------------
import math

def count_operations_example_3(n):
    count = 0
    while n > 1:
        n = n // 2                     # input size half hota jata hai
        count += 1
    print("Total operations:", count)

count_operations_example_3(16)

# Yahan har step mein n ka size half ho raha hai
# Isliye total steps ≈ log₂(n)
# => O(log n) time complexity



# -----------------------------------------------------
# SUMMARY OF OPERATION COUNTS
# -----------------------------------------------------
# ✅ Sequential statements (fixed steps) → O(1)
# ✅ Single loop running n times → O(n)
# ✅ Nested loop (loop inside loop) → O(n²)
# ✅ Loop dividing input size (n/2 each time) → O(log n)
# ✅ Combination (like n * log n) → O(n log n)

# -----------------------------------------------------
# Example 4: Mix of Operations
# -----------------------------------------------------

def count_operations_example_4(n):
    count = 0
    for i in range(n):                  # O(n)
        for j in range(n):              # O(n)
            count += (i + j)            # constant operation
    for k in range(n):                  # another O(n)
        count += k
    print("Total operations:", count)

count_operations_example_4(3)

# Total operations = n² (first nested loops) + n (second loop)
# Dominant term = n²
# => Overall Time Complexity → O(n²)

# 3. Abstract notion of order of growth


# -----------------------------------------------------
# A function that searches for an element in a list
# -----------------------------------------------------

def search_for_elmt(L, e):
    for i in L:                   # har element ko check karte hain
        if i == e:                # agar element mil gaya
            return True           # toh True return kar do
    return False                  # agar loop pura chala aur element nahi mila

# -----------------------------------------------------
# BEST CASE:
# -----------------------------------------------------
# Jab 'e' list ke sabse pehle element mein mil jaaye
# → Sirf 1 comparison hoga
# → Time Complexity = O(1)

# -----------------------------------------------------
# WORST CASE:
# -----------------------------------------------------
# Jab 'e' list mein bilkul nahi ho
# → Poora loop n times chalega
# → Time Complexity = O(n)

# -----------------------------------------------------
# AVERAGE CASE:
# -----------------------------------------------------
# Jab 'e' kahin beech mein ho (approx half list traverse karni pade)
# → Roughly n/2 comparisons honge
# → Time Complexity = O(n)


# -----------------------------------------------------
# Law of Addition
# -----------------------------------------------------
# 🔹 Jab humare paas do independent operations (A aur B) hain 
# aur dono ko perform karne ka choice hai (yaani A ya B),
# toh total time = time(A) + time(B)
#
# Yani agar ek algorithm mein do different parts hain 
# jo sequentially execute ho rahe hain (one after another),
# toh total time dono ke sum ke barabar hota hai.
#
# 👉 Formula:
#     T(n) = T1(n) + T2(n)
#     Overall Time Complexity = O(max(T1, T2))
#
# Example: 
# Suppose ek algorithm ke do parts hain —
# 1. First loop runs n times
# 2. Second loop runs m times


def law_of_addition_example(n, m):
    # Part 1
    for i in range(n):                 # Loop 1 → runs n times
        x = i * 2

    # Part 2
    for j in range(m):                 # Loop 2 → runs m times
        y = j + 5


# Total operations = n + m
# Time Complexity = O(n + m)
# Lekin agar n aur m approx same size ke ho (n ≈ m)
# Toh overall = O(n)
# Kyunki O(n + n) → O(2n) → O(n)


# -----------------------------------------------------
# Example with Comments and Output
# -----------------------------------------------------

law_of_addition_example(5, 3)

# Explanation:
# Part 1 executes 5 times
# Part 2 executes 3 times
# Total 8 iterations → Total time proportional to (5 + 3)
# So T(n, m) = O(n + m)


# -----------------------------------------------------
# Law of Multiplication
# -----------------------------------------------------
# 🔹 Jab humare paas ek operation ke andar doosra operation run ho raha hai 
# (nested structure), toh total time = time(A) × time(B)
#
# Matlab agar ek loop ke andar doosra loop run ho raha hai,
# toh total iterations dono ke product ke barabar hote hain.
#
# 👉 Formula:
#     T(n) = T1(n) × T2(n)
#     Overall Time Complexity = O(n × m) (for nested loops)
#
# Example:
# Outer loop runs n times
# Inner loop runs m times


def law_of_multiplication_example(n, m):
    for i in range(n):                 # Outer loop → n times
        for j in range(m):             # Inner loop → m times per outer iteration
            z = i + j                  # Constant operation

# Total operations = n * m
# Time Complexity = O(n * m)
# Agar n = m ho, toh O(n²)


# -----------------------------------------------------
# Example with Comments and Output
# -----------------------------------------------------

law_of_multiplication_example(3, 4)

# Explanation:
# Outer loop (3 times)
# Inner loop (4 times each)
# Total iterations = 3 * 4 = 12
# T(n, m) = O(n * m)
# Agar dono same size ke hain (n = m),
# toh T(n) = O(n²)


# 📈 Complexity Growth

| **CLASS**     | **n = 10** | **n = 100** | **n = 1000** | **n = 1,000,000** |
|----------------|-------------|--------------|---------------|--------------------|
| **O(1)**       | 1           | 1            | 1             | 1                  |
| **O(log n)**   | 1           | 2            | 3             | 6                  |
| **O(n)**       | 10          | 100          | 1000          | 1,000,000          |
| **O(n log n)** | 10          | 200          | 3000          | 6,000,000          |
| **O(n²)**      | 100         | 10,000       | 1,000,000     | 1,000,000,000,000  |
| **O(2ⁿ)**      | 1024        | 1.26×10³⁰    | 1.07×10³⁰⁰⁰   | **Good luck!! 😅** |

---

### Notes:
- As `n` increases, **time complexity** grows dramatically for higher-order functions.
- `O(1)` and `O(log n)` scale very well even for large inputs.
- `O(2ⁿ)` becomes **practically impossible** for even moderate `n`.

---

### Example Interpretation:
| Complexity Type | Growth Description |
|-----------------|--------------------|
| **O(1)**        | Constant time – independent of input size. |
| **O(log n)**    | Grows slowly (like binary search). |
| **O(n)**        | Linear growth (simple loops). |
| **O(n log n)**  | Slightly worse than linear (efficient sorts like Merge Sort). |
| **O(n²)**       | Quadratic (nested loops). |
| **O(2ⁿ)**       | Exponential – infeasible beyond small `n`. |

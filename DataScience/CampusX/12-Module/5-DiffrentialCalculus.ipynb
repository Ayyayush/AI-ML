/**
 * =========================================================
 * ! DIFFERENTIATION
 * =========================================================
 * 
 * ! WHAT IS DIFFERENTIATION?
 * =========================================================
 * 
 * Definition:
 * Differentiation is the mathematical process
 * of finding derivative of a function.
 * 
 * Derivative tells:
 * 
 * → how fast output changes
 * → with respect to input
 * 
 * 
 * =========================================================
 * ! DERIVATIVE
 * =========================================================
 * 
 * Definition:
 * Derivative represents instantaneous rate of change
 * of a function with respect to its variable.
 * 
 * Represented as:
 * 
 * dy/dx
 * 
 * OR
 * 
 * f'(x)
 * 
 * =========================================================
 * ! FUNCTION
 * =========================================================
 * Definition:
 * A mathematical relationship where one value
 * depends on another value.
 * 
 * Example:
 * 
 * y = x²
 * 
 * Here:
 * 
 * x -> input
 * y -> output
 * 
 * =========================================================
 * ! RATE OF CHANGE
 * =========================================================
 * 
 * Definition:
 * Rate of change tells how quickly one quantity changes
 * compared to another quantity.
 * 
 * Example:
 * 
 * speed =
 * distance/time
 * 
 * 
 * =========================================================
 * ! WHY INSTANTANEOUS RATE OF CHANGE?
 * =========================================================
 * Average rate of change: 
 * total change / total interval
 * 
 * But in real life:
 * 
 * change har point pe different hota hai
 * 
 * Example:
 * 
 * car speed at exact moment
 * 
 * Isliye:
 * instantaneous rate of change important hota hai
 * 
 * =========================================================
 * ! EXAMPLE OF RATE OF CHANGE
 * =========================================================
 * Population growth:
 * 
 * 1990 -> 1000
 * 2000 -> 1800
 * 
 * Growth rate:
 * (1800 - 1000) / (2000 - 1990) 
 * = 800 / 10 
 * = 80
 * 
 * =========================================================
 * ! DIFFERENTIATION FORMULA
 * =========================================================
 * 
 * f'(x) =
 * 
 * lim(Δx → 0)
 * 
 * [ f(x + Δx) - f(x) ] / Δx
 * 
 * 
 * 
 * =========================================================
 * ! LIMIT
 * =========================================================
 * Definition:
 * Limit tells what value a function approaches
 * when input approaches a particular point.
 * 
 * Represented by:  lim
 * 
 * 
 * =========================================================
 * ! DELTA X (Δx)
 * =========================================================
 * Definition:
 * Δx represents very small change in x.
 * 
 * When:  
 * Δx → 0
 * 
 * it means:
 * change becomes extremely small
 * 
 * 
 * =========================================================
 * ! SLOPE
 * =========================================================
 *  Definition:
 * Slope represents steepness or inclination
 * of a curve or line.
 * 
 * Formula:
 * 
 * slope =
 * change in y / change in x
 * 
 * 
 * =========================================================
 * ! RELATION BETWEEN DERIVATIVE AND SLOPE
 * =========================================================
 * Derivative at a point
 * gives slope of curve at that point.
 * 
 * Example:
 * 
 * if:  
 * dy/dx = 5
 * 
 * then:   
 * slope at that point is 5
 * 
 * =========================================================
 * ! TANGENT
 * =========================================================
 * 
 * Definition:
 * A line touching curve at exactly one point
 * is called tangent.
 * 
 * Slope of tangent:
 * 
 * derivative
 * 
 * =========================================================
 * ! DIFFERENTIATION IN MACHINE LEARNING
 * =========================================================
 * 
 * In ML:
 * differentiation is used for:
 * 
 * → optimization
 * → minimizing loss
 * → gradient descent
 * → updating weights
 * 
 * 
 * =========================================================
 * ! OPTIMIZATION
 * =========================================================
 * Definition:
 * Process of finding best values
 * that minimize or maximize function.
 * 
 * 
 * 
 * =========================================================
 * ! MAXIMA
 * =========================================================
 * Definition:
 * Point where function reaches maximum value.
 * 
 * At maxima:
 * 
 * slope = 0
 * 
 * derivative = 0
 * 
 * =========================================================
 * ! MINIMA
 * =========================================================
 * Definition:
 * Point where function reaches minimum value.
 * 
 * At minima:
 * 
 * slope = 0
 * 
 * derivative = 0
 * 
 * 
 * =========================================================
 * ! CRITICAL POINT
 * =========================================================
 * 
 * Definition:
 * Point where derivative becomes:
 * 
 * → 0
 * OR
 * → undefined
 * 
 * =========================================================
 * ! EXAMPLE
 * =========================================================
 * 
 * y = x² + 2x
 * 
 * Differentiate:
 * 
 * dy/dx = 2x + 2
 * 
 * For minima/maxima:
 * 
 * 2x + 2 = 0
 * 
 * x = -1
 * 
 * =========================================================
 * ! INTUITION OF DERIVATIVE
 * =========================================================
 * 
 * Suppose:
 * 
 * y = x²
 * 
 * Then:
 * 
 * if x changes slightly,
 * y also changes.
 * 
 * Derivative measures:
 * 
 * how much y changes
 * due to small change in x.
 * 
 * =========================================================
 * ! DERIVATIVE OF x²
 * =========================================================
 * 
 * y = x²
 * 
 * Using definition:
 * 
 * dy/dx =  [ (x + Δx)² - x² ] / Δx
 * 
 * Expanding:
 *  = [ x² + 2xΔx + (Δx)² - x² ] / Δx
 *  = [ 2xΔx + (Δx)² ] / Δx
 *  = 2x + Δx
 * 
 *  Taking limit: Δx → 0 
 *  therefore: 
 *  dy/dx = 2x
 * 
 * =========================================================
 * ! DERIVATIVE OF CONSTANT
 * =========================================================
 * y = 5 
 * derivative:  dy/dx = 0
 * 
 * =========================================================
 * ! WHY DERIVATIVE OF CONSTANT IS 0?
 * =========================================================
 * Constant value never changes.
 * Therefore:  rate of change = 0
 * 
 * 
 * =========================================================
 * ! POWER RULE
 * =========================================================
 * 
 * Formula:
 * 
 * d/dx (xⁿ) = n*xⁿ⁻¹
 * 
 * =========================================================
 * ! EXAMPLES OF POWER RULE
 * =========================================================
 * 
 * d/dx (x²) = 2x
 * 
 * d/dx (x³) = 3x²
 * 
 * d/dx (x⁴) = 4x³
 * 
 * d/dx (x⁵) = 5x⁴
 * 
 * =========================================================
 * ! COMMON DERIVATIVES
 * =========================================================
 * 
 * d/dx (x) = 1
 * 
 * d/dx (sinx) = cosx
 * 
 * d/dx (cosx) = -sinx
 * 
 * d/dx (tanx) = sec²x
 * 
 * d/dx (secx) = secxtanx
 * 
 * d/dx (cosecx) = -cosecxcotx
 * 
 * d/dx (cotx) = -cosec²x
 * 
 * d/dx (e^x) = e^x
 * 
 * d/dx (lnx) = 1/x
 * 
 * =========================================================
 * ! TRIGONOMETRIC FUNCTIONS
 * =========================================================
 * 
 * Definition:
 * Functions related to angles and triangles.
 * 
 * Examples:
 * 
 * sinx
 * cosx
 * tanx
 * 
 * =========================================================
 * ! EXPONENTIAL FUNCTION
 * =========================================================
 * 
 * Definition:
 * Function where variable appears in exponent.
 * 
 * Example:
 * 
 * e^x
 * 
 * =========================================================
 * ! LOGARITHMIC FUNCTION
 * =========================================================
 * 
 * Definition:
 * Function inverse to exponential function.
 * 
 * Example:
 * 
 * lnx
 * 
 * =========================================================
 * ! SUM RULE
 * =========================================================
 * 
 * Formula:
 * 
 * d/dx [f(x) + g(x)]
 * 
 * =
 * 
 * f'(x) + g'(x)
 * 
 * =========================================================
 * ! EXAMPLE OF SUM RULE
 * =========================================================
 * 
 * d/dx (x² + logx)
 * 
 * =
 * 
 * d/dx(x²) + d/dx(logx)
 * 
 * =
 * 
 * 2x + 1/x
 * 
 * =========================================================
 * ! PRODUCT RULE
 * =========================================================
 * 
 * Formula:
 * 
 * d/dx [f(x)g(x)]
 * 
 * =
 * 
 * f'(x)g(x) + f(x)g'(x)
 * 
 * =========================================================
 * ! EXAMPLE OF PRODUCT RULE
 * =========================================================
 * 
 * y = x² sinx
 * 
 * dy/dx =
 * 
 * (2x)(sinx) + (x²)(cosx)
 * 
 * =========================================================
 * ! QUOTIENT RULE
 * =========================================================
 * 
 * Formula:
 * 
 * d/dx [f(x)/g(x)]
 * 
 * =
 * 
 * [f'(x)g(x) - f(x)g'(x)] / [g(x)]²
 * 
 * =========================================================
 * ! EXAMPLE OF QUOTIENT RULE
 * =========================================================
 * 
 * y = x²/sinx
 * 
 * dy/dx =
 * 
 * [2xsinx - x²cosx] / (sinx)²
 * 
 * =========================================================
 * ! CHAIN RULE
 * =========================================================
 * 
 * Formula:
 * 
 * d/dx [f(g(x))]
 * 
 * =
 * 
 * f'(g(x)) * g'(x)
 * 
 * =========================================================
 * ! COMPOSITE FUNCTION
 * =========================================================
 * 
 * Definition:
 * Function inside another function
 * is called composite function.
 * 
 * Example:
 * 
 * sin(x²)
 * 
 * Here:
 * 
 * outer function:
 * sin()
 * 
 * inner function:
 * x²
 * 
 * =========================================================
 * ! EXAMPLE OF CHAIN RULE
 * =========================================================
 * 
 * y = sin(x²)
 * 
 * dy/dx =
 * 
 * cos(x²) * 2x
 * 
 * =========================================================
 * ! ANOTHER EXAMPLE
 * =========================================================
 * 
 * y = sin(logx²)
 * 
 * dy/dx =
 * 
 * cos(logx²) * d/dx(logx²)
 * 
 * =
 * 
 * cos(logx²) * (2x/x²)
 * 
 * =
 * 
 * cos(logx²) * (2/x)
 * 
 * =========================================================
 * ! PARTIAL DIFFERENTIATION
 * =========================================================
 * 
 * Definition:
 * Differentiation with respect to one variable
 * while keeping all other variables constant.
 * 
 * =========================================================
 * ! PARTIAL DERIVATIVE SYMBOL
 * =========================================================
 * 
 * Represented by:
 * 
 * ∂
 * 
 * =========================================================
 * ! EXAMPLE OF PARTIAL DIFFERENTIATION
 * =========================================================
 * 
 * z = x² + y²
 * 
 * Partial derivative wrt x:
 * 
 * ∂z/∂x = 2x
 * 
 * y treated as constant
 * 
 * Partial derivative wrt y:
 * 
 * ∂z/∂y = 2y
 * 
 * x treated as constant
 * 
 * =========================================================
 * ! SURFACE
 * =========================================================
 * 
 * Definition:
 * Graph of multivariable function
 * in 3D space is called surface.
 * 
 * =========================================================
 * ! 3D PARABOLOID
 * =========================================================
 * 
 * Definition:
 * Bowl-shaped 3D surface formed by equations
 * like:
 * 
 * z = x² + y²
 * 
 * =========================================================
 * ! HIGHER ORDER DERIVATIVES
 * =========================================================
 * 
 * Definition:
 * Derivative of derivative
 * is called higher order derivative.
 * 
 * =========================================================
 * ! SECOND DERIVATIVE
 * =========================================================
 * 
 * Definition:
 * Derivative of first derivative.
 * 
 * Represented by:
 * 
 * d²y/dx²
 * 
 * =========================================================
 * ! EXAMPLE
 * =========================================================
 * 
 * y = x³
 * 
 * First derivative:
 * 
 * dy/dx = 3x²
 * 
 * Second derivative:
 * 
 * d²y/dx² = 6x
 * 
 * =========================================================
 * ! ROLE OF SECOND DERIVATIVE
 * =========================================================
 * 
 * if:
 * 
 * d²y/dx² > 0
 * 
 * then:
 * minima
 * 
 * if:
 * 
 * d²y/dx² < 0
 * 
 * then:
 * maxima
 * 
 * =========================================================
 * ! MATRIX DIFFERENTIATION
 * =========================================================
 * 
 * Definition:
 * Differentiation involving vectors and matrices.
 * 
 * Used heavily in:
 * 
 * → machine learning
 * → deep learning
 * → optimization
 * 
 * =========================================================
 * ! MATRIX TRANSPOSE
 * =========================================================
 * 
 * Definition:
 * Converting rows into columns
 * and columns into rows.
 * 
 * Represented by:
 * 
 * Aᵀ
 * 
 * =========================================================
 * ! MATRIX DIFFERENTIATION EXAMPLE
 * =========================================================
 * 
 * y = xᵀAx
 * 
 * derivative:
 * 
 * ∂y/∂x = 2Ax
 * 
 * if:
 * 
 * Aᵀ = A
 * 
 * meaning:
 * A is symmetric matrix
 * 
 * =========================================================
 * ! SYMMETRIC MATRIX
 * =========================================================
 * 
 * Definition:
 * Matrix whose transpose equals itself.
 * 
 * Aᵀ = A
 * 
 * =========================================================
 * ! DIFFERENTIATION IN DEEP LEARNING
 * =========================================================
 * 
 * Neural networks use:
 * 
 * → chain rule
 * → partial derivatives
 * → gradient descent
 * 
 * for:
 * 
 * updating weights
 * minimizing loss
 * 
 * =========================================================
 * ! SUMMARY
 * =========================================================
 * 
 * Differentiation:
 * → measures rate of change
 * 
 * Derivative:
 * → slope at a point
 * 
 * Important rules:
 * 
 * → power rule
 * → sum rule
 * → product rule
 * → quotient rule
 * → chain rule
 * 
 * ML usage:
 * 
 * → optimization
 * → gradient descent
 * → backpropagation
 * 
 * =========================================================

/**
 * Session on Optimization - the big picture
 * 
 * 
 * ------------------------------------------------------------
 * ! Optimization
 * ------------------------------------------------------------
 * Optimization Machine Learning ka bahut important topic hai.
 * 
 * Optimization ka main goal:
 * 
 * Loss ko minimum karna.
 * 
 * Matlab:
 * Model ko itna improve karna
 * ki:
 * Predictions actual answers ke closest ho jayein.
 * 
 * 
 * ------------------------------------------------------------
 * ! Function
 * ------------------------------------------------------------
 * Function ek mathematical rule hota hai
 * jo:
 * 
 * Input leta hai
 * Process karta hai
 * Output deta hai
 * 
 * Simple words mein:
 * 
 * Input -> Function -> Output
 * 
 * Example:
 * 
 * y = f(x)
 * 
 * Yahan:
 * x -> input
 * y -> output
 * 
 * Example:
 * 
 * y = x² + 2x
 * 
 * Agar:
 * x = 2
 * 
 * Toh:
 * y = 2² + 2(2)
 * y = 4 + 4
 * y = 8
 * 
 * Isliye:
 * f(2) = 8
 * 
 * 
 * ------------------------------------------------------------
 * ! Single Variable Function
 * ------------------------------------------------------------
 * Agar:
 * Sirf ek input ho
 * 
 * Example:
 * 
 * y = f(x)
 * 
 * Toh:
 * Usko single variable function bolte hain.
 * 
 * 
 * ------------------------------------------------------------
 * ! Multivariable Function
 * ------------------------------------------------------------
 * Agar:
 * Multiple inputs ho
 * 
 * Example:
 * 
 * z = f(x,y)
 * 
 * Toh:
 * Usko multivariable function bolte hain.
 * 
 * Example:
 * 
 * z = x² + y²
 * 
 * Agar:
 * x = 1
 * y = 1
 * 
 * Toh:
 * z = 1² + 1²
 * z = 2
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Dimensions in Multivariable Function
 * ------------------------------------------------------------
 * Example:
 * 
 * f(x1,x2,x3...xn)
 * Yahan:
 * n inputs hain
 * Toh:
 * Function:
 * n dimensional input space represent karega.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Parameters in a Function
 * ------------------------------------------------------------
 * Parameters:
 * Function ke andar present constants/cofficients hote hain
 * jo:
 * Function ka behaviour control karte hain.
 * 
 * Example:
 * 
 * f(x) = ax² + bx + c
 * 
 * Yahan:
 * a,b,c
 * 
 * parameters hain.
 * 
 * In parameters ko change karne se:
 * 
 * - Graph ka shape
 * - Position
 * - Curvature
 * 
 * change ho jata hai.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! ML Models as Mathematical Functions
 * ------------------------------------------------------------
 * Machine Learning model ko:
 * Mathematical function ki tarah dekh sakte hain.
 * 
 * Example:
 * 
 * placement = f(cgpa,iq)
 * 
 * Yahan:
 * 
 * Inputs:
 * - cgpa
 * - iq
 * 
 * Output:
 * placement
 * 
 * Matlab:
 * ML model inputs ko process karke
 * output predict karta hai.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Parametric vs Non Parametric Models
 * ------------------------------------------------------------
 * ------------------------------------------------------------
 * ! Parametric ML Models
 * ------------------------------------------------------------
 * Parametric models:
 * Fixed number of parameters use karte hain.
 * 
 * Example:
 * 
 * y = β0 + β1x1 + β2x2
 * 
 * Yahan:
 * β0,β1,β2
 * 
 * parameters hain.
 * 
 * Example:
 * Linear Regression
 * 
 * ------------------------------------------------------------
 * ! Important Point
 * ------------------------------------------------------------
 * Parametric models:
 * Data distribution ke baare mein
 * assumptions banate hain.
 * 
 * ------------------------------------------------------------
 * ! Non Parametric ML Models
 * ------------------------------------------------------------
 * Non-parametric models:
 * Fixed number of parameters assume nahi karte.
 * 
 * Ye:
 * Data ke according complexity badha sakte hain.
 * 
 * Example:
 * - Decision Tree
 * - KNN
 * 
 * ------------------------------------------------------------
 *  
 * ------------------------------------------------------------
 * Linear Regression:
 * Parametric model hai.
 * 
 * Formula:
 * 
 * y = β0 + β1x1 + β2x2
 * 
 * Yahan:
 * 
 * β0:
 * Intercept
 * 
 * β1,β2:
 * Slopes / weights
 * 
 * Initially:
 * Parameters random values se start hote hain.
 * 
 * Training ke baad:
 * Best values learn karte hain.
 * 
 * ------------------------------------------------------------
 * ! Loss Function
 * ------------------------------------------------------------
 * Loss Function:
 * Ek mathematical function hota hai
 * jo:
 * 
 * Actual output aur predicted output ke beech
 * difference measure karta hai.
 * 
 * ------------------------------------------------------------
 * ! Goal of Training
 * ------------------------------------------------------------
 * Machine Learning ka goal:
 * 
 * Loss Function ko minimize karna.
 * 
 * Kyunki:
 * Kam loss matlab:
 * Better predictions.
 * 
 * ------------------------------------------------------------
 * ! Loss Function Formula
 * ------------------------------------------------------------
 *        n
 * L = Σ(yi - ŷi)²
 *       i=1
 * 
 * Yahan:
 * yi:
 * Actual value
 * 
 * ŷi:
 * Predicted value
 * 
 * ------------------------------------------------------------
 * ! Optimization
 * ------------------------------------------------------------
 * Optimization ka matlab:
 * 
 * Parameters ki aisi values dhundna
 * jo:
 * Loss minimum kar dein.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Choosing a Good Loss Function
 * ------------------------------------------------------------
 * ------------------------------------------------------------
 * ! 1. Problem Type
 * ------------------------------------------------------------
 * Different problems ke liye
 * different loss functions use hote hain.
 * 
 * Regression:
 * - MSE
 * - MAE
 * 
 * Binary Classification:
 * - Cross Entropy
 * - Hinge Loss
 * 
 * Multi Class Classification:
 * - Categorical Cross Entropy
 * 
 * ------------------------------------------------------------
 * ! 2. Robustness to Outliers
 * ------------------------------------------------------------
 * Outliers:
 * Extreme values hote hain.
 * 
 * Example:
 * Sab salaries 5-10 lakh
 * aur ek salary 5 crore.
 * 
 * MSE:
 * Outliers se bahut affect hota hai.
 * 
 * MAE:
 * More robust hota hai.
 * 
 * ------------------------------------------------------------
 * ! 3. Interpretability
 * ------------------------------------------------------------
 * Loss function:
 * Simple aur understandable hona chahiye.
 * 
 * Example:
 * MSE
 * 
 * Easy to compute and differentiate.
 * 
 * ------------------------------------------------------------
 * ! 4. Differentiability
 * ------------------------------------------------------------
 * Most optimization algorithms:
 * Differentiation use karte hain.
 * 
 * Isliye:
 * Loss function differentiable hona chahiye.
 * 
 * ------------------------------------------------------------
 * ! 5. Compatibility with Model
 * ------------------------------------------------------------
 * Har model ke saath
 * har loss function compatible nahi hota.
 * 
 * Example:
 * Linear Regression:
 * MSE use karta hai.
 * 
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Calculating Parameters from Loss Function
 * ------------------------------------------------------------
 * Goal:
 * 
 * β0,β1...
 * ki best values find karna.
 * 
 * Mathematical approach:
 * 
 * Loss function ko differentiate karte hain.
 * 
 * Fir:
 * Derivative = 0
 * 
 * rakhte hain.
 * 
 * Kyunki:
 * Minimum point par slope zero hoti hai.
 * 
 * ------------------------------------------------------------
 * ! Convex Function
 * ------------------------------------------------------------
 * Convex function:
 * Bowl shaped graph hota hai.
 * 
 * Isme:
 * Ek global minimum hota hai.
 * 
 * Optimization easy hoti hai.
 * 
 * Example:
 * Quadratic graph.
 * 
 * ------------------------------------------------------------
 * ! Non Convex Function
 * ------------------------------------------------------------
 * Non-convex function:
 * Multiple local minima/maxima hote hain.
 * 
 * Problem:
 * Algorithm galat minimum par fas sakta hai.
 * 
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Problems with Analytical Solution
 * ------------------------------------------------------------
 * ------------------------------------------------------------
 * ! 1. Non Convexity
 * ------------------------------------------------------------
 * Multiple local minima ki wajah se
 * global minimum find karna difficult hota hai.
 * 
 * ------------------------------------------------------------
 * ! 2. Complexity
 * ------------------------------------------------------------
 * Deep Learning models mein:
 * Bahut zyada parameters hote hain.
 * 
 * Exact analytical solution difficult hota hai.
 * 
 * ------------------------------------------------------------
 * ! 3. Scalability
 * ------------------------------------------------------------
 * Large datasets mein:
 * Computation bahut expensive ho jata hai.
 * 
 * ------------------------------------------------------------
 * ! 4. Online Learning
 * ------------------------------------------------------------
 * Kabhi:
 * Data stream ki form mein aata hai.
 * 
 * Yahan:
 * Incremental learning chahiye hoti hai.
 * 
 * Gradient Descent useful hota hai.
 * 
 * 
 * 
 *  
 * ------------------------------------------------------------
 * ! Gradient Descent
 * ------------------------------------------------------------
 * Gradient Descent:
 * Optimization algorithm hai
 * jo:
 * Loss function ko minimum karta hai.
 * 
 * ------------------------------------------------------------
 * ! Basic Idea
 * ------------------------------------------------------------
 * Gradient:
 * Slope batata hai.
 * 
 * Descent:
 * Neeche jana.
 * 
 * Matlab:
 * 
 * Gradient Descent:
 * Loss kam karne wali direction mein move karta hai.
 * 
 * ------------------------------------------------------------
 * ! Gradient Descent Formula
 * ------------------------------------------------------------
 * βnew = βold - η(dL/dβ)
 * 
 * Yahan:
 * 
 * η:
 * Learning Rate
 * 
 * dL/dβ:
 * Gradient / slope
 * 
 * ------------------------------------------------------------
 * ! Learning Rate
 * ------------------------------------------------------------
 * Learning rate:
 * Step size hota hai.
 * 
 * Small learning rate:
 * Slow training
 * 
 * Large learning rate:
 * Overshooting
 * 
 * ------------------------------------------------------------
 * ! Epoch
 * ------------------------------------------------------------
 * Epoch:
 * Ek complete pass over dataset.
 * 
 * ------------------------------------------------------------
 * ! Local Minima
 * ------------------------------------------------------------
 * Kabhi:
 * Gradient Descent
 * local minima mein fas sakta hai.
 * 
 * Especially:
 * Deep Learning mein.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Gradient Descent with Multiple Parameters
 * ------------------------------------------------------------
 * Agar:
 * Multiple parameters hain
 * 
 *  Example:
 * β0,β1,β2...
 * 
 * Toh:
 * Har parameter separately update hota hai.
 * 
 * Formulas:
 * 
 * β0 = β0 - η(∂L/∂β0)
 * 
 * β1 = β1 - η(∂L/∂β1)
 * 
 * β2 = β2 - η(∂L/∂β2)
 * 
 * ------------------------------------------------------------
 * ! Gradient Vector
 * ------------------------------------------------------------
 * ∇L(β0...βn)
 * 
 * Isko:
 * Gradient Vector bolte hain.
 * 
 * Ye:
 * Saare partial derivatives ka collection hota hai.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Problems Faced in Optimization
 * ------------------------------------------------------------
 * ------------------------------------------------------------
 * ! 1. Non Convexity
 * ------------------------------------------------------------
 * Multiple local minima/maxima ki wajah se
 * optimization difficult ho jata hai.
 * 
 * ------------------------------------------------------------
 * ! 2. Ill Conditioning
 * ------------------------------------------------------------
 * Kuch dimensions mein:
 * Gradients bahut bade
 * aur kuch mein bahut small hote hain.
 * 
 * Isse:
 * Optimization unstable ho sakti hai.
 * 
 * ------------------------------------------------------------
 * ! 3. Vanishing Gradient
 * ------------------------------------------------------------
 * Deep Neural Networks mein:
 * Gradient bahut small ho jata hai.
 * 
 * Result:
 * Training bahut slow ho jati hai.
 * 
 * ------------------------------------------------------------
 * ! 4. Exploding Gradient
 * ------------------------------------------------------------
 * Kabhi:
 * Gradient bahut large ho jata hai.
 * 
 * Result:
 * Training unstable ho jati hai.
 * 
 * ------------------------------------------------------------
 * ! 5. Overfitting
 * ------------------------------------------------------------
 * Model:
 * Training data ko rat leta hai.
 * 
 * New unseen data par:
 * Poor performance deta hai.
 * 
 * ------------------------------------------------------------
 * ! 6. Scalability
 * ------------------------------------------------------------
 * Large scale problems mein:
 * Optimization computationally expensive hoti hai.
 * 
 * ------------------------------------------------------------
 * ! Other Optimization Techniques
 * ------------------------------------------------------------
 * Gradient Descent ke alawa bhi:
 * 
 * - Momentum
 * - RMSProp
 * - Adam
 * - Adagrad
 * 
 * jaise optimization techniques use hoti hain.
 * 
 */

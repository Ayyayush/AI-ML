```python
# GroupBy Attributes and Methods

# find total number of groups -> len
# Returns the total number of unique groups formed after applying groupby

# find items in each group -> size
# Gives the number of rows (including NaN) present in each group

# first()/last() -> nth item
# Returns the first or last row from each group respectively

# get_group -> vs filtering
# Extracts a specific group directly without applying conditional filtering

# groups
# Returns a dictionary mapping group names to their row indices

# describe
# Generates descriptive statistics for each group

# sample
# Randomly selects rows from each group

# nunique
# Counts the number of unique values in each group
```

# let's see more complex things done with help of groupby

```python
# agg method
# passing dict

genres.agg(
    {
        'Runtime'      : 'mean',     # Average movie runtime for each genre
        'IMDB_Rating'  : 'mean',     # Average IMDB rating per genre
        'No_of_Votes'  : 'sum',      # Total number of votes across movies in each genre
        'Gross'        : 'sum',      # Total gross earning per genre
        'Metascore'    : 'min'       # Minimum metascore within each genre
    }
)
```

```python
# agg method
# passing list

genres.agg(['min', 'max', 'mean', 'sum'])     # Applies all listed aggregation functions
                                             # to every numeric column in each genre group
```

```python
# agg method
# passing list (numeric columns only)

genres[['Runtime', 'IMDB_Rating', 'No_of_Votes', 'Gross', 'Metascore']] \
    .agg(['min', 'max', 'mean', 'sum'])        # Apply aggregations safely on numeric data
```

```python
# Adding both the syntax
genres.agg(
    {
        'Runtime':['min','mean'],
        'IMDB_Rating':'mean',
        'No_of_Votes':['sum','max'],
        'Gross':'sum',
        'Metascore':'min'
    }
)
```

```python
# looping on groups

# Create an empty DataFrame with the same columns as movies
df = pd.DataFrame(columns=movies.columns)

# Loop through each group (Genre) and its corresponding data
for group, data in genres:
    
    # Print the type of group and data
    # print(type(group))     # group is a single Genre value (usually string)
    # print(type(data))      # data is a DataFrame for that genre
    
    # Select the movie(s) with the maximum IMDB rating within the current genre
    top_rated = data[data['IMDB_Rating'] == data['IMDB_Rating'].max()]
    
    # Append the result to the final DataFrame
    df = pd.concat([df, top_rated], ignore_index=True)

# Final DataFrame containing top-rated movie(s) from each genre
df
```

```python
# split (apply) combine
# apply -> builtin function

genres.apply(min)        # Applies the built-in min() function to each genre group
                         # Returns the minimum value from every column within each group
```

```python
# Custom function to be applied on each group
def foo(group):
    
    # Check which movie titles start with letter 'A'
    # str.startswith('A') returns a boolean Series
    # sum() counts how many True values are present
    return group['Series_Title'].str.startswith('A').sum()

# Apply the custom function to each genre group
# This counts how many movies (per genre) start with 'A'
genres.apply(foo)
```

```python
# find ranking of each movie in the group according to IMDB score

def rank_movie(group):
    
    # Create a copy of the group to avoid SettingWithCopy warnings
    group = group.copy()
    
    # Rank movies within each genre based on IMDB rating
    # ascending=False → higher IMDB rating gets better (smaller) rank
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
    
    # Return the modified group
    return group

# Apply ranking function to each genre group
genres.apply(rank_movie)
```

```python
# find normalized IMDB rating group wise

def normal(group):
    
    # Create a copy to avoid SettingWithCopyWarning
    group = group.copy()
    
    # Min-Max normalize IMDB rating within each genre group
    # Formula: (x - min) / (max - min)
    group['norm_rating'] = (
        (group['IMDB_Rating'] - group['IMDB_Rating'].min()) /
        (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    )
    
    # Return the modified group
    return group

# Apply normalization function to each genre group
genres.apply(normal)
```

```python
ipl = pd.read_csv('content/deliveries.csv')
ipl.head()
```

```python
ipl.shape
```

```python
# find the top 10 batsman in terms of runs
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
six = ipl[ipl['batsman_runs'] == 6]

six.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# find batsman with most number of 4's and 6's in last 5 overs

# Filter deliveries from the last 5 overs (i.e., overs greater than 15)
temp_df = ipl[ipl['over'] > 15]

# Keep only those deliveries where the batsman scored a boundary (4 or 6)
temp_df = temp_df[
    (temp_df['batsman_runs'] == 4) | 
    (temp_df['batsman_runs'] == 6)
]

# Group by batsman and count number of boundaries (4s + 6s)
# Sort in descending order to get the highest boundary hitter
# Extract the batsman name with the maximum count
temp_df.groupby('batsman')['batsman'] \
       .count() \
       .sort_values(ascending=False) 
    #    .head(1) \
    #    .index[0]
```

```python
# Filter the dataset to include only deliveries faced by V Kohli
temp_df = ipl[ipl['batsman'] == 'V Kohli']

# Group the data by bowling team
# Calculate total runs scored by V Kohli against each team
temp_df.groupby('bowling_team')['batsman_runs'] \
        .sum() \
       # .reset_index()
```

```python
# Create a function that can return the highest score of any batsman
temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
def highest(batsman):
  temp_df = ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

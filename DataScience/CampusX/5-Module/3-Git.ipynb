# 📘 What is Git?

**Git** is a **Version Control System (VCS)**.  
It is used to **track changes in source code** during software development.

Git is also known as a **Source Code Manager (SCM)**.

---

## 🔹 VCS / SCM (Version Control System)

A **Version Control System** is a tool that:
- Records changes in files over time
- Allows you to go back to any previous version
- Helps multiple developers work together efficiently

---

## 🔹 Why Do We Need VCS?

Jab hum software banate hain, development **step-by-step** hota hai:
- New features add hote hain
- Bugs fix hote hain
- Code continuously change hota rehta hai

Without VCS:
- Purana code overwrite ho jata hai
- Mistakes ke baad wapas jana mushkil hota hai

👉 **VCS in problems ko solve karta hai**

---

## 🔹 Types of Version Control Systems

### 1️⃣ Local VCS
- Sirf local system pe kaam karta hai
- Collaboration support nahi hota

### 2️⃣ Centralized VCS (CVCS)
- Ek central server hota hai
- Example: SVN

### 3️⃣ Distributed VCS (DVCS)
- Har developer ke paas complete repository hoti hai
- Example: **Git**

---

## 🔹 Advantages of Version Control System

- ✅ Version history maintain hoti hai
- ✅ Non-linear development (branches)
- ✅ Easy collaboration
- ✅ Safe experimentation
- ✅ Backup & recovery

---

## 🔹 Git Snapshot Concept (Easy Explanation)

Git **file-by-file changes track nahi karta**,  
Git **project ke snapshots leta hai** 📸

👉 Jab bhi hum **commit** karte hain:
- Git poore project ka ek snapshot save karta hai

### Example:
- 1 week pehle ka snapshot available hai
- Aaj code mein gadbad ho gayi 😬
- Git se hum uss exact state pe wapas ja sakte hain

---

## 🔹 Git vs Undo / Redo

Undo / Redo:
- Temporary hota hai
- Limited steps hote hain

Git:
- Permanent history hoti hai
- Har commit ek safe checkpoint hota hai
- Pure project ke versions maintain karta hai

👉 **Git = Advanced Undo / Redo for projects**

---

## 🔹 Simple Real-Life Example

Socho:
- Tum assignment likh rahe ho
- Har important stage pe file save karte ho:
  - v1
  - v2
  - final

Git bhi **exactly isi tarah kaam karta hai**,  
bas zyada powerful aur automated 🔥

---

## 🔹 Common Git Commands

- `git init` → Initialize repository
- `git status` → Current repository state
- `git add .` → Stage all changes
- `git commit -m "message"` → Save snapshot
- `git log` → View commit history

---

## 🔹 Git vs GitHub (Very Important)

| Git | GitHub |
|---|---|
| Tool | Platform |
| Runs locally | Runs online |
| Version control system | Code hosting service |
| CLI based | Web + collaboration |

👉 **GitHub internally Git ka use karta hai**

---

## 🔹 Summary

- Git is a **Version Control System**
- Tracks code changes using **snapshots**
- Helps in collaboration, backup, rollback
- Git and GitHub are **not the same**

# 🔗 Git and GitHub

## 🔹 What is Git?
Git is a **Version Control System (VCS)** used to track changes in source code.
It works **locally on your system** and helps manage different versions of a project.

## 🔹 What is GitHub?
GitHub is an **online platform** that uses Git.
It allows developers to:
- Store code online
- Collaborate with others
- Share projects
- Review code

👉 **Git = tool**, **GitHub = service/platform**

---

# 📁 Repository (Repo)

A **repository** is a place where your project lives.

It contains:
- Project files
- Git history (commits)
- Branches and metadata

Types of repos:
- **Local Repository** → on your computer
- **Remote Repository** → on GitHub / GitLab

```bash
index.html
app.js
style.css


# ⚙️ Git Commands – Complete Notes (Theoretical)

---

## 🔹 git init

**What it is:**  
`git init` is used to initialize a new Git repository.

**Working:**  
- Creates a hidden `.git` directory  
- Enables version control for the project  

**Uses:**  
- Start tracking a new project  
- Convert an existing folder into a Git repository  

**Example:**  
Command used to initialize Git → git init

---

## 🔹 main (default branch)

**Important:**  
`main` is **not a Git command**.  
It is the **default branch name**.

**Working:**  
- Represents the primary branch of a repository  
- Usually contains stable or production-ready code  

**Uses:**  
- Acts as the base branch for development  
- Used while pushing final code  

**Example:**  
Renaming current branch to main → git branch -M main

---

## 🔹 git switch main

**What it is:**  
Used to switch to the `main` branch.

**Working:**  
- Moves the `HEAD` pointer to the main branch  
- Updates the working directory to match main  

**Uses:**  
- Return to main branch after feature development  
- Before pulling or pushing final code  

**Example:**  
Switching to main branch → git switch main  

(Older alternative: git checkout main)

---

## 🔹 git status

**What it is:**  
Shows the current state of the repository.

**Working:**  
- Compares Working Directory with Staging Area  
- Displays modified, staged, and untracked files  

**Uses:**  
- Check changes before adding or committing files  

**Example:**  
Checking repository state → git status

---

## 🔹 git add

**What it is:**  
Adds files to the staging area.

**Working:**  
- Moves files from Working Directory to Staging Area  

**Uses:**  
- Prepare files for commit  

**Examples:**  
Add all files → git add .  
Add specific file → git add filename

---

## 🔹 git commit

**What it is:**  
Creates a snapshot of staged changes.

**Working:**  
- Saves changes in the local repository  
- Generates a unique commit hash  

**Uses:**  
- Save a logical unit of work  
- Maintain version history  

**Example:**  
Creating a commit → git commit -m "commit message"

---

## 🔹 git log

**What it is:**  
Displays the commit history.

**Working:**  
- Shows commit hash, author, date, and message  

**Uses:**  
- Track project history  
- Debug issues  

**Examples:**  
View full history → git log  
View short history → git log --oneline

---

## 🔹 git show

**What it is:**  
Shows details of a specific commit.

**Working:**  
- Displays changes introduced by that commit  

**Uses:**  
- Inspect a particular commit  

**Example:**  
View commit details → git show <commit-hash>

---

## 🔹 git clone

**What it is:**  
Creates a local copy of a remote repository.

**Working:**  
- Downloads project files and complete commit history  
- Automatically sets the remote as `origin`  

**Uses:**  
- Start working on an existing GitHub project  

**Example:**  
Cloning a repository → git clone <repository-url>

---

## 🔹 git remote

**What it is:**  
Manages remote repository connections.

**Working:**  
- Stores URLs of remote repositories  

**Uses:**  
- Connect local repository to GitHub  

**Examples:**  
View remotes → git remote -v  
Add remote → git remote add origin <repo-url>

---


## 🔹 git push

**What it is:**  
Uploads local commits to a remote repository.

**Working:**  
- Sends commits from local repository to GitHub  

**Uses:**  
- Share code with others  
- Backup work on remote repository  

**Example:**  
Pushing code to main → git push origin main

---

## 🔹 git pull

**What it is:**  
Updates the local repository with remote changes.

**Working:**  
- Fetches changes from remote  
- Merges them into the current branch  

**Uses:**  
- Sync local repository with remote repository  

**Example:**  
Pulling latest changes → git pull origin main

---

## 🔹 git fetch

**What it is:**  
Downloads changes from remote repository without merging.

**Working:**  
- Updates remote-tracking branches  
- Does not change working directory  

**Uses:**  
- Review remote changes safely before merging  

**Example:**  
Fetching changes → git fetch origin

---

## 🔹 git branch

**What it is:**  
Used to manage branches in Git.

**Working:**  
- A branch is a pointer to a commit  
- Allows parallel development  

**Uses:**  
- Create independent feature branches  

**Examples:**  
List branches → git branch  
Create new branch → git branch feature-login

---

## 🔹 git merge

**What it is:**  
Combines changes from one branch into another.

**Working:**  
- Applies commits from source branch to target branch  

**Uses:**  
- Merge feature branch into main branch  

**Example:**  
Merging branch → git merge feature-login

---

## 🔹 git reset (Rollback – Unsafe)

**What it is:**  
Moves `HEAD` backward in commit history.

**Working:**  
- Can unstage files or remove commits  
- Alters commit history  

**Uses:**  
- Fix recent local mistakes  

**Examples:**  
Soft reset → git reset --soft HEAD~1  
Hard reset → git reset --hard HEAD~1  

⚠️ Hard reset permanently deletes changes.

---

## 🔹 git revert (Rollback – Safe)

**What it is:**  
Safely undoes a commit.

**Working:**  
- Creates a new commit that reverses previous changes  
- Does not delete history  

**Uses:**  
- Undo commits in shared repositories  

**Example:**  
Reverting a commit → git revert <commit-hash>

---

## 🔹 git stash

**What it is:**  
Temporarily saves uncommitted changes.

**Working:**  
- Clears working directory  
- Stores changes in a stack  

**Uses:**  
- Switch tasks without committing  

**Examples:**  
Save changes → git stash  
Restore changes → git stash pop

---

## 🔹 git log

**What it is:**  
Displays commit history.

**Working:**  
- Shows commit hash, author, date, and message  

**Uses:**  
- Track project history  
- Debug issues  

**Example:**  
Viewing commit history → git log

---

## 🔹 git log --stat

**What it is:**  
Shows commit history with file-level statistics.

**Working:**  
- Displays files changed in each commit  
- Shows number of lines added and removed  

**Uses:**  
- Understand impact of each commit  

**Example:**  
Viewing commit stats → git log --stat

---

## 🔹 git log -p

**What it is:**  
Shows commit history with patch (code changes).

**Working:**  
- Displays exact line-by-line changes for each commit  

**Uses:**  
- Review detailed code changes  
- Debug bugs introduced in commits  

**Example:**  
Viewing code changes → git log -p

---

## 🔹 Git Workflow (Quick Recall)

Working Directory → Staging Area → Local Repository → Remote Repository (GitHub)

---

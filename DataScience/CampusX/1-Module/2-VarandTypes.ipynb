# 📘 Variables in Python

A **variable** is a symbolic name that is a reference or pointer to an object stored in memory.  
In Python, variables are used to **store data temporarily** which can be accessed or modified later.
It is used to hold data during the execution of a program 
In python we don't need to declare varibales 
Just write the variable name and start using it 


Example:

pi = 3.14

Case-sensitive (Name and name are different)

# 📘 MEMORY ASSIGNMENT IN PYTHON

---

## 🔹 Python vs C++ Memory Assignment

- Python variables direct memory blocks (like C/C++) nahi hote.  
- Python me variables **containers nahi**, balki **labels/references** hote hain.  
- Jab hum `x = 5` likhte hain (conceptually):  
  - `5` ek object ban kar memory ke heap me store hota hai  
  - `x` stack me ek **reference** hota hai jo us object ke address ko point karta hai  
- Python me variable ke andar data nahi hota — variable sirf data ka **address hold** karta hai.  
- Isi wajah se Python ka assignment model C++ se bilkul alag hai.

---

# 📘 REFERENCING CONCEPT

## 🔹 1. Every Variable is a Reference
- Python me **har variable ek pointer jaise behave karta hai**.  
- Variable ka naam = object ka address store karne wala label.  
- Object heap me, label stack me.

## 🔹 2. Multiple Variables Can Refer to the Same Object
- Python me ek hi object ko **multiple names** refer kar sakte hain.  
- Object ek hi rehta hai, names alag alag.  
- Is concept ko “aliasing” bhi bolte hain.

## 🔹 3. Immutable Objects (int, str, tuple)
- Immutable values modify nahi hoti.  
- Jab tum value "change" karte ho:  
  - Python **naya object create karta hai**  
  - Variable ko naye object se **rebind** kar deta hai  
- Purana object as-is rehta hai.

## 🔹 4. Mutable Objects (list, dict, set)
- Mutable objects ka **content modify** ho sakta hai bina naye object banaye.  
- Agar same object ko multiple variables refer kar rahe hon,  
  to modification **sab jagah reflect** hota hai.

## 🔹 5. Reference Counting
- Python har object ka **reference count track** karta hai.  
- Kitne variable us object ko refer kar rahe hain.  
- Reference count zero → object memory se delete (garbage collection).

## 🔹 6. Python Memory Layout
- **Stack Memory:** variable names (references)  
- **Heap Memory:** actual objects (numbers, lists, dicts, strings)  
- Variable name stack me hota hai  
- Object heap me hota hai  
- Reference = heap object ka address

## 🔹 7. Binding vs Rebinding
- **Binding:** variable ko pehli baar kisi object se jodna.  
- **Rebinding:** variable ko naye object se jod dena.  
- Immutable types rebinding ka istemaal karte hain.  
- Mutable types me mostly modification hota hai, rebinding nahi.

---

# 📘 PYTHON’S ASSIGNMENT MODEL — KEY SUMMARY

- Assignment = object ka address variable ko dena.  
- Python values copy nahi karta, **address copy** karta hai.  
- Variable = label  
- Object = real data  
- Variable update ≠ object modify  
- Variable update = variable gets bound to a **new** object (for immutables)

---

# 📘 ADDITIONAL IMPORTANT CONCEPTS

## 🔹 Identity vs Equality
- Equality: do objects ka content same hai?  
- Identity: kya dono variables **same memory object** ko refer kar rahe hain?

## 🔹 Interning
- Python frequently-used values (small integers, common strings) ko pre-cache karta hai.  
- Isliye kuch objects automatically **same memory share** karte hain.

## 🔹 Garbage Collection
- Python automatic garbage collector use karta hai.  
- Jo objects kisi bhi variable se refer nahi ho rahe → delete ho jaate hain.

# Identifiers in Python

## 🔹 What are Identifiers?

In Python, **identifiers** are the names used to identify variables, functions, classes, modules, and objects.  
Ye basically **names** hote hain jinke through hum kisi object ko refer karte hain.

---

## 🔹 Rules for Identifiers

1. Identifier ka naam **letter (A-Z or a-z)** ya **underscore `_`** se start hona chahiye.
2. Naam mein **letters, digits (0-9), underscores** allowed hote hain.
3. **Spaces allowed nahi hote** identifiers mein.
4. Identifiers **case-sensitive** hote hain.  
   Example: `name`, `Name`, aur `NAME` teen alag identifiers hain.
5. Python ke **reserved keywords** ko identifiers ke roop mein use **nahi** kiya ja sakta.  
   Jaise: `for`, `class`, `if`, `else`, `try`, etc.

---

## 🔹 Valid vs Invalid Identifiers

| Valid Identifiers | Invalid Identifiers |
|-------------------|---------------------|
| `name`            | `1name`             |
| `user_name`       | `user-name`         |
| `_temp`           | `for` (keyword)     |
| `Data123`         | `my var` (space)    |

---

## 🔹 Best Practices

- Variable/function names should be **meaningful and readable**.
- Follow **snake_case** for variable and function names: `total_marks`, `user_input`
- Use **PascalCase** for class names: `StudentData`, `EmployeeRecord`
- Avoid using single-character names unless in loops (`i`, `j`, etc.)

---

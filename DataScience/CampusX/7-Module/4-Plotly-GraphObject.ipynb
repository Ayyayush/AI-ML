# Plotly

Plotly ek **interactive data visualization library** hai jo Python, JavaScript, R jaise languages ke saath use hoti hai.  
Iska main advantage ye hai ki ye **static graphs ke saath-saath interactive graphs** banane ki facility deta hai (zoom, hover, tooltip, filter, etc.).

Simple words mein:  
👉 **Matplotlib / Seaborn = mostly static graphs**  
👉 **Plotly = interactive + modern looking graphs**

Plotly ka use hota hai:
- Data Analysis
- Dashboards
- Data Science Projects
- Web-based visualizations


# Graph Objects (plotly.graph_objects)

Graph Objects Plotly ka **low-level interface** hai.

👉 Matlab:
- Har cheez ko manually control kar sakte ho  
- Axis, colors, labels, layout – sab kuch detail mein define karna hota hai

### Characteristics:
- Zyada **flexibility**
- Zyada **code likhna padta hai**
- Complex & customized graphs ke liye best

### Kab use kare?
- Jab tumhe **pure control** chahiye visualization par  
- Jab default graphs kaafi nahi lag rahe  
- Professional dashboards ke complex charts ke liye  

Example types:
- go.Figure()
- go.Scatter()
- go.Bar()
- go.Pie()

⚠️ Beginner ke liye thoda heavy lag sakta hai.





# Plotly Express

Plotly Express Plotly ka **high-level interface** hai.

👉 Matlab:
- Kam code mein graph ready  
- Pandas DataFrame ke saath directly kaam karta hai  

### Characteristics:
- **Fast & simple**
- Beginner-friendly
- Most common charts ek line mein ban jaate hain


📌 Plotly Express internally **Graph Objects ka hi use karta hai**, bas complexity hide kar deta hai.


# Dash

Dash ek **web framework** hai jo Plotly ke upar bana hai.

👉 Matlab:
- Tum Plotly graphs ko use karke **complete web dashboard** bana sakte ho  
- HTML, CSS, JS likhne ki zarurat nahi hoti  







# Why we most commonly use Plotly Express

Plotly Express sabse zyada use hone ke reasons 👇

### 1️⃣ Kam Code, Zyada Output  
Ek hi line mein professional graph ready.

### 2️⃣ Pandas Friendly  
Direct DataFrame pass kar sakte ho – no extra processing.

### 3️⃣ Built-in Aesthetics  
Colors, legends, labels automatically handle ho jaate hain.

### 4️⃣ Fast EDA  
Data ko jaldi samajhne ke liye perfect.

### 5️⃣ Interview & Projects Friendly  
Readable code + clean visualization = strong impression.

### 6️⃣ Easy Transition  
Agar future mein zyada customization chahiye,  
to Express se Graph Objects pe easily shift ho sakta hai.

📌 Isliye:
- **Beginners → Plotly Express**
- **Advanced Users → Graph Objects**
- **Dashboards → Dash**

# Demo Codes

```python
# Demo Codes using Plotly Graph Objects 

import plotly.graph_objects as go

# Data
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 17, 20]

# Figure create
fig = go.Figure()

# Add line trace
fig.add_trace(
    go.Scatter(
        x=x,
        y=y,
        mode='lines+markers', 
        name='Sales'
    )
)

# Layout customization
fig.update_layout(
    title="Basic Line Chart using Graph Objects",
    xaxis_title="Days",
    yaxis_title="Sales"
)

fig.show()
```

```python
import plotly.graph_objects as go

# Data
products = ['A', 'B', 'C', 'D']
sales = [100, 150, 80, 200]

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=products,
        y=sales,
        name='Product Sales',
        marker_color='orange'
    )
)

fig.update_layout(
    title="Bar Chart using Graph Objects",
    xaxis_title="Products",
    yaxis_title="Sales"
)

fig.show()
```

```python
# -------------------------------
# Data definition
# -------------------------------
x = [5, 7, 9, 11, 13]                                # X-axis ke values (independent variable)
y = [20, 30, 25, 40, 35]                             # Y-axis ke values (dependent variable)

# -------------------------------
# Figure creation
# -------------------------------
fig = go.Figure(                                    # Ek empty Figure object create kar rahe hain
    data=[                                          # Figure ke andar data (traces) define hote hain
        go.Scatter(                                 # Scatter trace create kar rahe hain
            x=x,                                    # X-axis data assign kiya
            y=y,                                    # Y-axis data assign kiya
            mode='markers',                         # Sirf markers draw honge (no lines)
            marker=dict(                            # Marker ki styling define kar rahe hain
                size=12,                            # Har point ka size
                color='red'                         # Har point ka color
            ),
            name='Points'                           # Trace ka naam (legend mein dikhega)
        )
    ]
)


# -------------------------------
# Layout customization
# -------------------------------
fig.update_layout(                                  # Graph ka overall layout update kar rahe hain
    title="Scatter Plot using Graph Objects",       # Graph ka title
    xaxis_title="X Values",                         # X-axis ka label
    yaxis_title="Y Values"                          # Y-axis ka label
)

# -------------------------------
# Render / display the plot
# -------------------------------
fig.show()                                          # Final interactive graph display karta hai
```

```python
import plotly.graph_objects as go                     # Plotly ka low-level Graph Objects module import kar rahe hain

# -------------------------------
# Data definition
# -------------------------------
x = [1, 2, 3, 4]                                     # X-axis ke common values (dono lines ke liye same)

# -------------------------------
# Figure creation
# -------------------------------
fig = go.Figure()                                   # Empty Figure object create kiya

# -------------------------------
# First trace (Line 1)
# -------------------------------
fig.add_trace(                                      # Figure ke andar pehla trace add kar rahe hain
    go.Scatter(                                     # Scatter trace (line type)
        x=x,                                        # X-axis data
        y=[10, 20, 30, 40],                          # Line 1 ke Y-axis values
        mode='lines',                               # Sirf line draw hogi (no markers)
        name='Line 1'                                # Trace ka naam (legend mein show hoga)
    )
)

# -------------------------------
# Second trace (Line 2)
# -------------------------------
fig.add_trace(                                      # Figure ke andar doosra trace add kar rahe hain
    go.Scatter(                                     # Scatter trace (line type)
        x=x,                                        # Same X-axis values use ho rahe hain
        y=[40, 30, 20, 10],                          # Line 2 ke Y-axis values
        mode='lines',                               # Sirf line draw hogi
        name='Line 2'                                # Legend ka naam
    )
)

# -------------------------------
# Layout customization
# -------------------------------
fig.update_layout(                                  # Graph ka overall layout set kar rahe hain
    title="Multiple Lines using Graph Objects",     # Graph ka title
    xaxis_title="X Axis",                           # X-axis label
    yaxis_title="Y Axis"                            # Y-axis label
)

# -------------------------------
# Render / display the plot
# -------------------------------
fig.show()                                          # Interactive graph render karta hai
```

```python
import plotly.graph_objects as go

labels = ['Python', 'Java', 'C++', 'JavaScript']
values = [40, 25, 20, 15]

fig = go.Figure(
    data=[
        go.Pie(
            labels=labels,
            values=values
        )
    ]
)

fig.update_layout(
    title="Pie Chart using Graph Objects"
)

fig.show()
```

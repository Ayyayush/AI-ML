# Categorical Scatter Plot

```python
sns.scatterplot(data=tips, x='total_bill',y='tip')
```

```python
# --------------------------------------------------
# Strip Plot (Axes-level Function)
# --------------------------------------------------

# Default strip plot (jitter = True by default)
sns.stripplot(
    data=tips,                   # dataset
    x='day',                     # categorical variable
    y='total_bill'               # numerical variable
)

# --------------------------------------------------
# Strip Plot without jitter
# --------------------------------------------------

sns.stripplot(
    data=tips,                   # dataset
    x='day',                     # categorical variable
    y='total_bill',              # numerical variable
    jitter=False                 # disable random horizontal spread
)
```

```python
# --------------------------------------------------
# Strip Plot using catplot (Figure-level Function)
# --------------------------------------------------

sns.catplot(
    data=tips,                   # dataset
    x='day',                     # categorical variable
    y='total_bill',              # numerical variable
    kind='strip'                 # strip plot
)
```

```python
# --------------------------------------------------
# Strip Plot with Custom Jitter (Figure-level Function)
# --------------------------------------------------

sns.catplot(
    data=tips,                   # dataset
    x='day',                     # categorical variable
    y='total_bill',              # numerical variable
    kind='strip',                # strip plot
    jitter=0.2,                  # control amount of random spread
    hue='sex'                    # color separation by gender
)
```

```python
# swarmplot
sns.catplot(data=tips, x='day',y='total_bill',kind='swarm')
```

```python
sns.swarmplot(data=tips, x='day',y='total_bill')
```

```python
# hue
sns.swarmplot(data=tips, x='day',y='total_bill',hue='sex')
```


# Categorical Distribution Plot

### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

image.png

```python
# --------------------------------------------------
# Box Plot (Axes-level Function)
# --------------------------------------------------

sns.boxplot(
    data=tips,                   # dataset
    x='day',                     # categorical variable
    y='total_bill'               # numerical variable
)
```

```python
# Using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
# hue
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
# --------------------------------------------------
# Single Box Plot (Univariate Numerical Column)
# --------------------------------------------------

sns.boxplot(
    data=tips,                   # dataset
    y='total_bill'               # numerical variable
)
```

### Violin Plot = (Box Plot + KDE Plot)

- Combines **summary statistics** of a box plot with the **distribution shape** from KDE.
- Shows **median, quartiles**, and **full data distribution**.
- Useful for comparing distributions across categories.
- Wider sections indicate **higher data density**.


```python
# violinplot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin')
```

```python
# hue

sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex',split=True)
```

# Categorical estimation plots 

### Bar Plot

- Used to visualize **categorical data** with numerical values.
- Displays **aggregated statistics** (mean by default).
- Includes **confidence intervals** by default.
- Commonly used for **category-wise comparison**.


```python
# --------------------------------------------------
# Bar Plot with custom estimator (min) and NO error bars
# --------------------------------------------------
import numpy as np
sns.barplot(
    data=tips,
    x='sex',
    y='total_bill',
    hue='smoker',
    estimator=np.min,     # custom aggregation (minimum)
    errorbar=None         # disable error bars
)
```

```python
sns.barplot(data=tips, x='sex', y='total_bill',ci=None)
```

When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# point plot
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker',ci=None)
```

### Bar Plot (Count Case)

- Used when you want to show the **number of observations per category**.
- Does **not compute a statistic** on a second variable.
- Similar to a **histogram for categorical data**.
- Commonly implemented using a **count plot**.


# countplot

```python
# --------------------------------------------------
# Count Plot (Axes-level Function)
# --------------------------------------------------

sns.countplot(
    data=tips,                   # dataset
    x='sex',                     # categorical variable
    hue='day'                    # split counts by day
)
```

# pointplot

```python
# --------------------------------------------------
# Point Plot (Axes-level Function)
# --------------------------------------------------

sns.pointplot(
    data=tips,                   # dataset
    x='sex',                     # categorical variable
    y='total_bill',              # numerical variable (aggregated)
    hue='smoker',                # subgroup comparison
    errorbar=None                # disable confidence intervals (modern seaborn)
)
```

```python
# --------------------------------------------------
# Faceted Box Plot using catplot (Figure-level Function)
# --------------------------------------------------

sns.catplot(
    data=tips,                   # dataset
    x='sex',                     # categorical variable (X-axis)
    y='total_bill',              # numerical variable (Y-axis)
    kind='box',                  # box plot
    col='smoker',                # create separate columns based on smoker status
    row='time'                   # create separate rows based on time (Lunch/Dinner)
)
```

### Multigrid Plots

- Used to create **multiple plots arranged in a grid layout**.
- Helps compare data across **different subsets or categories**.
- Commonly created using **Faceting** (`row`, `col`).
- Improves clarity when analyzing **patterns across groups**.



### Types of Multigrid Plots in Seaborn

- **FacetGrid** – Base grid for creating multiple plots by faceting data.
- **PairGrid** – Used for plotting pairwise relationships between variables.
- **JointGrid** – Used to plot bivariate relationships with marginal distributions.
- **PairPlot** – High-level wrapper around PairGrid for quick EDA.
- **JointPlot** – High-level wrapper around JointGrid for fast visualization.
- **CatPlot** – Figure-level categorical multigrid plot.
- **RelPlot** – Figure-level relational multigrid plot.
- **DisPlot** – Figure-level distribution multigrid plot.


```python
# --------------------------------------------------
# Violin Plot using catplot (Figure-level Function)
# --------------------------------------------------

sns.catplot(
    data=tips,                   # dataset
    x='sex',                     # categorical variable (X-axis)
    y='total_bill',              # numerical variable (Y-axis)
    kind='violin',               # violin plot
    col='day',                   # facet columns by day
    row='time'                   # facet rows by time (Lunch/Dinner)
)
```

### A second way to plot Facet plots -> FacetGrid

```python
# --------------------------------------------------
# FacetGrid (Low-level Figure Control)
# --------------------------------------------------

# Create a FacetGrid with rows, columns, and hue
g = sns.FacetGrid(
    data=tips,                   # dataset
    col='day',                   # facet columns
    row='time',                  # facet rows
    hue='smoker'                 # color grouping
)



# Map an axes-level function onto each facet
g.map(
    sns.boxplot,                 # plotting function (axes-level)
    'sex',                       # X-axis
    'total_bill'                 # Y-axis
)

# Add legend for hue
g.add_legend()
```

### `height` and `aspect`

- **`height`** – Controls the **height (in inches)** of each subplot.
- **`aspect`** – Controls the **width-to-height ratio** of each subplot.

**Final width = height × aspect**

Used mainly in **figure-level plots** (`relplot`, `catplot`, `displot`) to control overall plot size.


### Plotting Pairwise Relationships (PairGrid vs PairPlot)

- Used to visualize **pairwise relationships** between multiple variables.
- Helps identify **correlations, trends, and distributions**.

**PairGrid**
- More **flexible and customizable**
- Allows different plots for **upper, lower, and diagonal**
- Requires more code

**PairPlot**
- High-level, **quick and easy** wrapper around PairGrid
- Less customizable
- Ideal for **fast EDA**


```python
sns.pairplot(iris,hue='species')
```

```python
# --------------------------------------------------
# PairGrid (Low-level Multivariate Plotting)
# --------------------------------------------------

# Create a PairGrid with hue for grouping
g = sns.PairGrid(
    data=iris,                  # dataset
    hue='species'               # color grouping
)

# Map the same axes-level plot to all pairs
g.map(
    sns.scatterplot              # scatter plot for every (x, y) pair
)

# Add legend for hue
g.add_legend()
```

```python
# --------------------------------------------------
# PairGrid with different plots on diagonal & off-diagonal
# --------------------------------------------------

# Create PairGrid with hue for grouping
g = sns.PairGrid(
    data=iris,                  # dataset
    hue='species'               # color grouping
)

# Plot on diagonal (feature vs itself)
g.map_diag(
    sns.boxplot                 # boxplot for univariate distributions
)

# Plot on off-diagonal (feature vs feature)
g.map_offdiag(
    sns.kdeplot                 # 2D KDE for joint density
)

# Add legend for hue
g.add_legend()
```

```python
# --------------------------------------------------
# PairGrid with different plots on diagonal, upper & lower
# --------------------------------------------------

# Create PairGrid with hue for grouping
g = sns.PairGrid(
    data=iris,                  # dataset
    hue='species'               # color grouping
)

# Diagonal: univariate distributions
g.map_diag(
    sns.histplot                # histogram for each feature
)

# Upper triangle: bivariate density
g.map_upper(
    sns.kdeplot                 # 2D KDE contours
)

# Lower triangle: bivariate relationship
g.map_lower(
    sns.scatterplot             # scatter plot
)

# Add legend for hue
g.add_legend()
```

```python
# --------------------------------------------------
# PairGrid with selected variables using `vars`
# --------------------------------------------------

g = sns.PairGrid(
    data=iris,                                   # dataset
    hue='species',                               # color grouping
    vars=['sepal_width', 'petal_width']          # select only these variables
)

g=sns.PairGrid(
    data=iris,
    hue='species',
    vars=['sepal_width','sepal_length']

)
# Diagonal plots (univariate distribution)
g.map_diag(
    sns.histplot                                # histogram for each variable
)



# Upper triangle (bivariate relationship)
g.map_upper(
    sns.kdeplot                                 # 2D KDE density
)

# Lower triangle (bivariate relationship)
g.map_lower(
    sns.scatterplot                             # scatter plot
)

# Add legend for hue
g.add_legend()
```

### JointGrid vs JointPlot

- Used to visualize the **relationship between two variables** along with their **marginal distributions**.

**JointGrid**
- Low-level, **highly customizable**
- Separate control over **joint and marginal plots**
- Requires more code

**JointPlot**
- High-level, **easy-to-use** wrapper around JointGrid
- Limited customization
- Best for **quick exploratory analysis**


```python
sns.jointplot(data=tips,x='total_bill',y='tip',kind='hist',hue='sex')
```

```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.kdeplot,sns.violinplot)
```

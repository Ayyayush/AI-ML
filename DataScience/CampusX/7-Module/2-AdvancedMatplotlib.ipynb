# Advanced Matplotlib


```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris = pd.read_csv('content/iris.csv')
iris.sample(5)
```

```python
iris['Species'] = iris['Species'].replace({
    'Iris-setosa': 0,
    'Iris-versicolor': 1,
    'Iris-virginica': 2
})
iris.sample(5)

```

```python
# ------------------------------------------------------------
# Scatter plot:
# X-axis : Sepal Length
# Y-axis : Petal Length
# Color  : Species (0, 1, 2)
# ------------------------------------------------------------
plt.scatter(
    iris['SepalLengthCm'],      # X values
    iris['PetalLengthCm'],      # Y values
    c=iris['Species'],          # Color based on species
    cmap='summer',                 # Color map
    alpha=0.7                   # Transparency for better visibility
)

# Label the axes
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

# Show color scale for species
plt.colorbar()

# Display the plot
plt.show()
```

## 🎨 cmap and 🌫️ alpha (Matplotlib) — Hinglish Explanation

`cmap` aur `alpha` dono **data visualization ko readable aur meaningful** banane ke kaam aate hain.

---

## 🎨 cmap (Colormap)

**`cmap`** ka matlab hota hai **color map**.  
Ye batata hai ki **values ko colors kaise assign** honge.

👉 Mostly use hota hai:
- `plt.scatter()`
- `plt.imshow()`
- Heatmaps

### Simple Idea
- Chhoti value → ek color  
- Badi value → dusra color  

---

## 🔍 Common Colormaps


::contentReference[oaicite:0]{index=0}


| cmap name | Use case |
|---------|---------|
| `viridis` | Default, best readability |
| `plasma` | High contrast |
| `inferno` | Dark background |
| `magma` | Heat-style plots |
| `jet` | Old-style (avoid in serious analysis) |

---

## 🧠 When to Use `cmap`?

- Jab third variable ko color se represent karna ho
- Scatter plots with categories / intensity
- Heatmaps & image data

---

## 🌫️ alpha (Transparency)

**`alpha`** transparency control karta hai.  
Iski value hoti hai **0 se 1 ke beech**.

- `alpha = 0` → fully transparent (invisible)
- `alpha = 1` → fully opaque (solid)

---

## 🤔 alpha Kyun Zaroori Hai?

- Overlapping points ko clearly dikhane ke liye
- Dense data mein crowding kam karne ke liye

---

## 🧪 Typical Usage (Conceptual)

- Scatter plot mein:
  - `cmap` → color variation
  - `alpha` → overlap control

Example idea:
- Zyada dense area → dark shade
- Kam dense area → light shade

---

## ⚠️ Common Mistakes

- Bahut bright `cmap` use karna → misleading
- `alpha` bahut low rakhna → points gayab
- `jet` blindly use karna ❌

---

```python
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Create a figure with 2 rows and 1 column of subplots
# sharex=True means both plots will share the same X-axis
# ------------------------------------------------------------
fig, ax = plt.subplots(
    nrows=2,
    ncols=1,
    sharex=True,
    figsize=(10, 6)
)


# ------------------------------------------------------------
# First subplot: Average vs Strike Rate
# ------------------------------------------------------------
ax[0].scatter(
    batters['avg'],                 # X-axis values
    batters['strike_rate'],         # Y-axis values
    color='red'                     # Point color
)

# ------------------------------------------------------------
# Second subplot: Average vs Runs
# ------------------------------------------------------------
ax[1].scatter(
    batters['avg'],                 # X-axis values
    batters['runs']                 # Y-axis values
)

# ------------------------------------------------------------
# Titles and labels for first subplot
# ------------------------------------------------------------
ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')
ax[0].set_xlabel('Avg')


# ------------------------------------------------------------
# Titles and labels for second subplot
# ------------------------------------------------------------
ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')             # Shared X-axis label

# ------------------------------------------------------------
# Display the figure
# ------------------------------------------------------------
plt.show()
```

# 3. Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
# meshgrid converts 1D X and Y arrays into 2D grids
# so that we can compute values for every (x, y) pair.
# ------------------------------------------------------------
# Create a 2D grid (mesh grid) from x and y
# xx contains repeated rows of x values
# yy contains repeated columns of y values
# ------------------------------------------------------------
xx, yy = np.meshgrid(x, y)
xx.shape
yy.shape
```

```python

# ------------------------------------------------------------
# Define Z as a mathematical function of X and Y
# z = x² + y²  (paraboloid surface)
# ------------------------------------------------------------
z = xx**2 + yy**2

# ------------------------------------------------------------
# Check the shape of Z (optional but useful for understanding)
# ------------------------------------------------------------
z.shape    # Same shape as xx and yy
```

```python

# ------------------------------------------------------------
# Create a figure with custom size
# ------------------------------------------------------------
fig = plt.figure(figsize=(12, 8))

# ------------------------------------------------------------
# Add a 3D subplot
# ------------------------------------------------------------
ax = plt.subplot(projection='3d')

# ------------------------------------------------------------
# Create a 3D surface plot
# cmap controls the color mapping of the surface
# ------------------------------------------------------------
p = ax.plot_surface(
    xx, yy, z,
    cmap='viridis'
)

# ------------------------------------------------------------
# Add a color bar to show Z-value mapping
# ------------------------------------------------------------
fig.colorbar(p)

# ------------------------------------------------------------
# Display the plot
# ------------------------------------------------------------
plt.show()
```

```python
# ------------------------------------------------------------
# Define Z as a function of X and Y using existing meshgrid
# z = sin(x) + cos(y) evaluated at every (x, y) point
# ------------------------------------------------------------
z = np.sin(xx) + np.cos(yy)

# ------------------------------------------------------------
# Create a figure with specified size
# ------------------------------------------------------------
fig = plt.figure(figsize=(12, 8))

# ------------------------------------------------------------
# Add a 3D subplot to the figure
# ------------------------------------------------------------
ax = plt.subplot(projection='3d')

# ------------------------------------------------------------
# Plot a 3D surface using X, Y, and Z values
# cmap maps height (Z) values to colors
# ------------------------------------------------------------
p = ax.plot_surface(xx, yy, z, cmap='jet')

# ------------------------------------------------------------
# Add a color bar to show Z-value scale
# ------------------------------------------------------------
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(xx)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
delivery = pd.read_csv('content/IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
# ------------------------------------------------------------
# Filter the delivery DataFrame
# 1. Keep only valid ball numbers (1 to 6)
# 2. Keep only deliveries where batsman scored a six (6 runs)
# ------------------------------------------------------------
temp_df = delivery[
    (delivery['ballnumber'].isin([1, 2, 3, 4, 5, 6])) &   # Valid balls in an over
    (delivery['batsman_run'] == 6)                        # Six-hit deliveries
]
```

```python
# ------------------------------------------------------------
# Create a pivot table (matrix format) from filtered data
# Rows    -> Overs
# Columns -> Ball numbers within an over (1 to 6)
# Values  -> Count of sixes (batsman_run == 6)
# aggfunc -> Count how many times a six occurred
# ------------------------------------------------------------
grid = temp_df.pivot_table(
    index='overs',            # Rows represent overs
    columns='ballnumber',     # Columns represent ball numbers
    values='batsman_run',     # Values used for aggregation
    aggfunc='count'           # Count number of sixes
)
```

```python
# ------------------------------------------------------------
# Create a figure with large size for better visibility
# ------------------------------------------------------------
plt.figure(figsize=(20, 10))

# ------------------------------------------------------------
# Display the 2D data array as an image (heatmap)
# grid contains intensity values for each cell
# ------------------------------------------------------------
plt.imshow(grid)

# ------------------------------------------------------------
# Set Y-axis ticks
# delivery['overs'].unique() -> actual tick positions
# range(1, 21)               -> labels for overs (1 to 20)
# ------------------------------------------------------------
plt.yticks(
    delivery['overs'].unique(),
    list(range(1, 21))
)

# ------------------------------------------------------------
# Set X-axis ticks
# np.arange(0, 6) -> positions for balls in an over
# range(1, 7)     -> labels for balls (1 to 6)
# ------------------------------------------------------------
plt.xticks(
    np.arange(0, 6),
    list(range(1, 7))
)

# ------------------------------------------------------------
# Add a color bar to show value-to-color mapping
# ------------------------------------------------------------
plt.colorbar()

# ------------------------------------------------------------
# Display the plot
# ------------------------------------------------------------
plt.show()
```

```python
# ------------------------------------------------------------
# Create a DataFrame for pie chart analysis
# Rows    -> Batsmen
# Columns -> Runs scored in different matches
# ------------------------------------------------------------
df = pd.DataFrame(
    {
        'batsman': ['Dhawan', 'Rohit', 'Kohli', 'SKY', 'Pandya', 'Pant'],
        'match1':  [120, 90, 35, 45, 12, 10],
        'match2':  [0, 1, 123, 130, 34, 45],
        'match3':  [50, 24, 145, 45, 10, 90]
    }
)

# ------------------------------------------------------------
# Display the first few rows of the DataFrame
# ------------------------------------------------------------
df.head()
```

```python
# ------------------------------------------------------------
# Pie chart for Match 1 runs distribution
# Each slice represents a batsman's contribution
# ------------------------------------------------------------
df['match1'].plot(
    kind='pie',                           # Pie chart
    labels=df['batsman'].values,          # Labels for each slice
    autopct='%0.1f%%'                     # Show percentage with 1 decimal place
)
```

```python
# ------------------------------------------------------------
# Multiple pie charts for runs scored in each match
# Each column is plotted as a separate pie chart
# ------------------------------------------------------------
df[['match1', 'match2', 'match3']].plot(
    kind='pie',                 # Pie chart type
    subplots=True,              # Create a separate pie for each column
    figsize=(15, 8)             # Overall figure size
)
```

```python
# multiple separate graphs together
# using stocks

stocks.plot(kind='line',subplots=True)
```

```python
# ------------------------------------------------------------
# Create a pivot table and visualize it using multiple pie charts
# ------------------------------------------------------------

tips.pivot_table(
    index=['day', 'time'],        # Rows grouped by day and time (Lunch/Dinner)
    columns=['sex', 'smoker'],    # Columns grouped by gender and smoker status
    values='total_bill',          # Values to aggregate
    aggfunc='mean'                # Calculate average total bill
).plot(
    kind='pie',                   # Pie chart visualization
    subplots=True,                # One pie chart per column group
    figsize=(20, 10)              # Overall figure size
)
```

## 📊 Bivariate Analysis

**Bivariate Analysis** ka matlab hota hai:
👉 **do variables ke beech relationship** ko analyze karna.

Yaani hum dekhte hain:
- Ek variable dusre variable ko **kaise affect** karta hai
- Dono ke beech **koi pattern / trend / relation** hai ya nahi

---

## 🔁 Types of Bivariate Analysis

Variables ke type ke basis pe mainly 2 important cases hote hain:

1. **Categorical → Numerical**
2. **Numerical → Numerical**

---

## 1️⃣ Categorical → Numerical

### Matlab
- Ek variable **category** hai  
- Dusra variable **number** (numeric value) hai  

### Example
- Gender → Salary  
- City → Average Temperature  
- Product Category → Sales  

Yahan hum check karte hain:
👉 **Different categories ke liye numerical value kaise change ho rahi hai**

### Common Analysis
- Category-wise average / mean
- Comparison between groups

### Typical Questions
- Male vs Female → average salary ka difference?
- Product A vs Product B → kaun zyada bik raha hai?

---

## 2️⃣ Numerical → Numerical

### Matlab
- Dono variables **numbers** hain

### Example
- Height → Weight  
- Advertising Spend → Sales  
- Time → Stock Price  

Yahan hum dekhte hain:
👉 Kya ek number badhne se dusra bhi badh raha hai ya ghat raha hai?

### Common Analysis
- Correlation
- Trend detection
- Relationship strength

### Typical Questions
- Kya ads spend badhne se sales badhti hain?
- Kya temperature aur ice-cream sales related hain?

---

## ⏳ Use Case: Time Series Data

### What is Time Series Data?
Jab data **time ke order mein collect** hota hai, usse **time series data** kehte hain.

### Examples
- Daily stock prices  
- Monthly sales  
- Hourly website traffic  
- Yearly population growth  

Yahan:
- **Time** → Numerical (or ordered)
- **Value** → Numerical

👉 Ye ek **Numerical → Numerical** bivariate analysis ka special case hai

---

## Time Series Mein Kya Analyze Karte Hain?

- 📈 **Trend** → long-term increase / decrease  
- 🔁 **Seasonality** → fixed interval pe repeat pattern  
- 📉 **Fluctuations** → ups & downs  
- 🚨 **Anomalies** → sudden unusual change  

---

## 🧠 Summary

- **Bivariate Analysis** = relationship between 2 variables  
- Types:
  - Categorical → Numerical (group comparison)
  - Numerical → Numerical (trend / correlation)
- **Time Series Data** = numerical vs numerical over time  
- Real-world ML & Data Science mein bahut common concept

---

## 🎯 Interview One-Liner

> Bivariate analysis studies the relationship between two variables, such as categorical vs numerical or numerical vs numerical, including special cases like time series data.

## 📊 plt.plot vs plt.scatter (Matplotlib) – Hinglish Explanation

Dono functions **data visualization** ke liye use hote hain,  
lekin **use case aur meaning alag** hota hai.

---

## 🔹 plt.plot()

**`plt.plot()`** mainly use hota hai jab:
- Data **continuous** ho
- Order / sequence important ho
- Trend dikhana ho

👉 By default ye **points ko line se connect** kar deta hai.


::contentReference[oaicite:0]{index=0}


### Kab Use Karein?
- Time series data (date vs price)
- Trend analysis
- Continuous relationship

### Real-life Example
- Time → Temperature  
- Day → Stock price  

Matlab: **flow / continuity** important hai.

---

## 🔹 plt.scatter()

**`plt.scatter()`** use hota hai jab:
- Data **independent points** ho
- Sirf relationship dekhni ho
- Points ko connect nahi karna ho

👉 Har point **alag marker** hota hai.


::contentReference[oaicite:1]{index=1}


### Kab Use Karein?
- Correlation dekhna
- Outliers identify karna
- Numerical → Numerical bivariate analysis

### Real-life Example
- Height vs Weight  
- Avg vs Strike Rate  

Matlab: **pattern / relation** important hai, order nahi.

---

## 🧠 Core Difference (Side-by-Side)

| Feature | plt.plot() | plt.scatter() |
|------|-----------|---------------|
| Connection | Points connected | No connection |
| Data type | Continuous | Discrete |
| Order matters | ✅ Yes | ❌ No |
| Best for | Trends | Correlation |
| Time series | ✅ Perfect | ❌ Not ideal |

---

## 🎯 Easy Way to Remember

- **plt.plot** → “line mein flow”
- **plt.scatter** → “sirf points, no relation line”

---

## 🧪 Interview One-Liner

> `plt.plot()` is used for continuous data and trend visualization, while `plt.scatter()` is used to show relationships between independent data points.

```python
import numpy as np
import matplotlib.pyplot as plt

# Create bar positions for each batsman
x = np.arange(df.shape[0])     # Number of batsmen (rows in DataFrame)

# Bar plot for year 2015 (shifted left)
plt.bar(
    x - 0.2,                   # Shift bars left for grouping
    df['2015'],                # Runs in 2015
    width=0.2,                 # Width of each bar
    color='yellow'             # Bar color
)

# Bar plot for year 2016 (center)
plt.bar(
    x,                         # Center position
    df['2016'],                # Runs in 2016
    width=0.2,
    color='red'
)

# Bar plot for year 2017 (shifted right)
plt.bar(
    x + 0.2,                   # Shift bars right for grouping
    df['2017'],                # Runs in 2017
    width=0.2,
    color='blue'
)

# Set x-axis labels as batsman names
plt.xticks(x, df['batsman'])

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Number of children
children = [10, 20, 40, 10, 30]

# Correct color names (one per bar)
colors = ['red', 'blue', 'green', 'yellow', 'pink']

# Bar plot
plt.bar(colors, children, color=colors)

# Labels and title
plt.xlabel('Color Category')
plt.ylabel('Number of Children')
plt.title('Children Count by Color')

# Show plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Stacked Bar Chart

# Bottom layer: Runs scored in 2017
plt.bar(
    df['batsman'],          # X-axis: batsman names
    df['2017'],             # Height of bars for 2017
    label='2017'             # Legend label
)

# Middle layer: Runs scored in 2016 stacked on top of 2017
plt.bar(
    df['batsman'],
    df['2016'],
    bottom=df['2017'],      # Start 2016 bars from top of 2017 bars
    label='2016'
)

# Top layer: Runs scored in 2015 stacked on top of (2016 + 2017)
plt.bar(
    df['batsman'],
    df['2015'],
    bottom=(df['2016'] + df['2017']),   # Cumulative bottom
    label='2015'
)

# Show legend
plt.legend()

# Display the plot
plt.show()
```

## 📊 Histogram (Hinglish)

**Histogram** ek graph hota hai jo **numerical data ka distribution** dikhata hai.  
Ye batata hai ki data **kis range (bin)** mein **kitni frequency** ke saath aata hai.

---

## 🤔 Histogram Kyun Use Karte Hain?

- **X-axis** → Value ranges (bins)
- **Y-axis** → Frequency (count)

## 📦 Handling Bins in Histogram (Hinglish)

**Bins** histogram ka sabse important part hote hain.  
Bins decide karte hain:
👉 Data ko **kaunse ranges** mein divide kiya jaayega  
👉 Graph **clear dikhega ya confusing**

---

## 🤔 Bin Kya Hota Hai?

- Bin = **value range**
- Har bin count karta hai:
  > us range mein **kitne data points** aaye

Example:
- Bin: `0–10`
- Bin: `10–20`
- Bin: `20–30`

---

## 👀 Visual Understanding


::contentReference[oaicite:0]{index=0}


---

## 🎛️ Bin Size Ka Impact

### 🔹 Too Few Bins
- Data **over-simplified**
- Details chhup jaati hain

Example:
- Sirf 3 bins → poora pattern gayab

### 🔹 Too Many Bins
- Graph **noisy**
- Random spikes dikhte hain

Example:
- Har 1 value ka alag bin

👉 Goal: **Balance**

---

## 🧠 Bin Decide Karne Ke Tarike

### 1️⃣ Fixed Number of Bins
- Simple approach
- Small datasets ke liye ok

Example idea:
- `bins = 5`
- `bins = 10`

---

### 2️⃣ Bin Width (Range Based)
- Har bin ki width fixed hoti hai

Example:
- Width = 10  
- Bins: 0–10, 10–20, 20–30

✔ Useful jab data ka scale known ho

---

### 3️⃣ Auto Bin Selection
NumPy / Matplotlib smart rules use karta hai:

- `auto` → best default
- `sturges` → small datasets
- `fd` (Freedman–Diaconis) → skewed data
- `sqrt` → quick estimate

👉 Real projects mein **auto / fd** popular hain

---

## 📊 How to Think While Handling Bins

Ask yourself:
- Data **continuous** hai?
- Dataset **small ya large**?
- Distribution **normal ya skewed**?

| Data Type | Suggested Bins |
|---------|----------------|
| Small data | 5–10 |
| Large data | 20–50 |
| Skewed data | fd / auto |
| Unknown | auto |

---

## ⚠️ Common Mistakes

- Default bins blindly use karna
- Bin size ka logic ignore karna
- Histogram ko bar chart samajhna
- Extreme bins (too many / too few)

---

## 🎯 Interview One-Liner

> Bins define the value ranges in a histogram, and choosing the right bin size is crucial to correctly represent data distribution.


```python
import numpy as np
import matplotlib.pyplot as plt

# Load a large numerical array from .npy file
arr = np.load('content/big-array.npy')

# Plot histogram with logarithmic scale on Y-axis
plt.hist(
    arr,                                # Input data
    bins=[10, 20, 30, 40, 50, 60, 70],   # Custom bin ranges
    log=True                            # Use logarithmic scale for frequency
)

# Labels and title
plt.xlabel('Value Range')
plt.ylabel('Frequency (log scale)')
plt.title('Histogram with Logarithmic Scale')

# Display the plot
plt.show()
```

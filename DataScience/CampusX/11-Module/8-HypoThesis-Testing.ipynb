# Hypothesis Testing

## What is Hypothesis Testing?

👉 Hypothesis Testing ek statistical method hai jisme hum **data ke basis pe decision lete hain**  

Simple Hinglish:  
**Hypothesis Testing = guess ko test karna using data**

---

## Basic Idea

👉 Pehle hum ek assumption (hypothesis) banate hain  
👉 phir data use karke check karte hain ki wo sahi hai ya galat  

---

## Types of Hypothesis

### 1. Null Hypothesis (H₀)
👉 Default statement (generally "no effect" ya "no change")  

Example:  
H₀: average marks = 50  

---

### 2. Alternative Hypothesis (H₁ / Ha)
👉 Jo hum actually prove karna chahte hain  

Example:  
H₁: average marks ≠ 50  

---

## Steps in Hypothesis Testing

1. Hypothesis define karo (H₀ & H₁)  
2. Significance level choose karo (α, usually 0.05)  
3. Test statistic calculate karo  
4. p-value find karo  
5. Decision lo (reject ya accept H₀)  

---

## Important Terms

### Significance Level (α)
👉 Probability of rejecting a true H₀  
👉 Common value = 0.05  

---

### p-value
👉 Data kitna strong evidence deta hai H₀ ke against  

---

## Decision Rule

- Agar p-value ≤ α → Reject H₀  
- Agar p-value > α → Fail to reject H₀  

---

## Simple Example

Claim: average height = 170 cm  

- H₀: mean = 170  
- H₁: mean ≠ 170  

Data check karke decide karenge ki claim valid hai ya nahi  

---

## Conclusion

👉 Hypothesis Testing help karta hai:
- data-driven decisions lene me  
- assumptions verify karne me  

Simple line:  
**"Hypothesis Testing = data se decision lena"**

# Hypothesis Testing (Part 2)

# steps involved in hypothesis testing (rejection region approach)

1. formulate null and alternate hypothesis  
H0 (null) → jo default assumption hota hai  
H1 (alternate) → jo hum prove karna chahte hain  

---

2. select a significance level (alpha)  
ye probability hoti hai null hypothesis ko reject karne ki jab wo actually true ho  
usually alpha = 0.05 ya 0.01 lete hain  

👉 yaad rakh: alpha = type 1 error  

---

3. check assumptions  
example → data normal distributed hai ya nahi  
agar sample size >= 30 hai toh CLT se approx normal maan sakte hain  

---

4. decide which test is appropriate  
z-test → jab population std (sigma) given ho  
t-test → jab sigma given nahi ho  

maan lo data normal hai  
aur std bhi given hai → toh easily z-test use kar sakte hain  
agar std nahi hota → toh t-test use karte  

aage chalke ek generic flow bhi padhenge  

---

5. state the relevant test statistic  
z-test → z score nikalte hain  
t-test → t value nikalte hain  

---

6. conduct the test  
formula laga ke value nikalte hain  

---

7. reject or not reject the null hypothesis  
👉 rejection region ya p-value ke basis pe decide karte hain  

---

8. interpret the result  
final answer ko real life language me explain karte hain  

---

# performing a z test example 1

suppose karo company check kar rahi hai training program ka impact  

pehle:
per day 50 car ban rahi thi  
mean = 50  
std = 5  

phir company ne training di employees ko  

ab humne 30 employees ka sample liya  
sample mean = 53  

---

step 1 → hypothesis  
H0: mean = 50  
H1: mean > 50  

---

step 2 → alpha  
alpha = 0.05  

---

step 3 → assumption  
sample size = 30 → CLT se approx normal maan sakte hain  

---

step 4 → test select  
sigma given hai → z-test use karenge  

---

step 5 → formula  
z = (x̄ - mean) / (sigma / √n)  

---

step 6 → calculation  
z = (53 - 50) / (5 / √30)  
≈ 3 / (5 / 5.47)  
≈ 3 / 0.91  
≈ 3.29  

---

step 7 → decision  
critical value (one-tailed, alpha = 0.05) ≈ 1.645  
👉 since 3.29 > 1.645 → reject H0  

---

step 8 → interpretation  
👉 training program effective hai (productivity increase hui hai)  

---

# example 2

maan lo lays company check kar rahi hai packet ka weight  

claim → 50 gram  

sample liya = 40 packets  
sample mean = 49  
std = 4  

---

step 1 → hypothesis  
H0: mean = 50  
H1: mean ≠ 50  

---

baaki same steps follow karenge  

---

step 3 → assumption  
sample size = 40 → approx normal  

---

step 4 → test select  
sigma given nahi hai → t-test apply karenge  

---

step 5 → formula  
t = (x̄ - mean) / (s / √n)  

---

step 6 → calculation  
t = (49 - 50) / (4 / √40)  
≈ -1 / (4 / 6.32)  
≈ -1 / 0.63

---

step 8 → interpretation  
👉 enough evidence nahi hai prove karne ke liye ki weight 50g se different hai  

---

final goal → prove karna ki packet ka weight 50g hai ya nahi  

# part 3 lecture notes  

let's have some more discussion about rejection region  

significance level represent karta hai probability of rejecting null hypothesis when it is actually true  
maan lo significance level = 5%  

👉 ye approach type 1 error se related hai  
👉 significance level (alpha) = probability of type 1 error  

---

## rejection region approach  

the rejection region is region of values that corresponds to rejection of null hypothesis at some chosen probability level  

👉 matlab ek boundary (critical value) hoti hai  
👉 agar test statistic us region me aa jaye → reject H0  

---

## problem with rejection region  

rejection region me ek problem hai ki jo value aa rahi hai  
uski strength pata nahi chalti  

👉 bas itna pata chalta hai:
- reject hua ya nahi  
- but kitna strong evidence hai → nahi pata  

---

## isliye p-value approach use hota hai  

p-value hume value ke according strength bata deta hai  

👉 choti p-value → strong evidence against H0  
👉 badi p-value → weak evidence  

example samajh:
- p-value = 0.02 → strong evidence  
- p-value = 0.15 → weak evidence  

---

# very important topic  

# type 1 vs type 2 error  

hypothesis testing me 2 types ke errors hote hain  
aur high chance hai interview me ye poocha jaye  

---

## type 1 error (false positive)

type 1 error tab hota hai when sample results lead to rejection of null hypothesis  
which is actually true  

👉 in other words:
finding a significant relationship when there is none  

👉 simple:
"galti se bol diya ki effect hai, jab actually nahi tha"  

---

### probability

probability of committing type 1 error = alpha (α)  

---

## type 2 error (false negative)

type 2 error occurs when based on sample results  
null hypothesis reject nahi hota  
jabki wo actually false hota hai  

👉 simple:
"effect tha, par humne bola nahi hai"  

---

### probability

probability of committing type 2 error = beta (β)  

---

## power of the test  

power of the test = 1 - beta  

👉 matlab:
correctly reject karne ki probability jab H0 false ho  

👉 zyada power → better test  

---

## quick recap flow

H0 true + reject → type 1 error  
H0 false + not reject → type 2 error  

---

## final line

👉 rejection region basic approach hai  
👉 p-value better hai (strength batata hai)  
👉 alpha = type 1 error  
👉 beta = type 2 error  

simple line:  
**"hypothesis testing me decision ke saath risk bhi hota hai"**

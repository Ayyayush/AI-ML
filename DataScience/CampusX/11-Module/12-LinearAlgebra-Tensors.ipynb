/**
 * ! Linear Algebra Part - 0 : Tensors
 * 
 * Tensors are one of the most important concepts in Machine Learning
 * and Deep Learning.
 * 
 * TensorFlow ka naam bhi "Tensor" word se hi aaya hai.
 * 
 * Tensor is basically a data structure.
 * 
 * Simple words mein:
 * Tensor ek container hota hai numbers ko store karne ke liye.
 * 
 * Mostly tensors numbers store karte hain.
 * Kabhi kabhi characters aur strings bhi store ho sakte hain,
 * but woh rare cases hote hain.
 * 
 * Jo vectors aur matrices hum Maths mein padhte hain,
 * woh bhi actually tensors hi hote hain.
 * 
 * ------------------------------------------------------------
 * ! Types of Tensors
 * 
 * 1. 0D Tensor
 * 2. 1D Tensor
 * 3. 2D Tensor
 * 4. ND Tensor
 * 
 * ------------------------------------------------------------
 * ! Important Point
 * 
 * Jitne dimensions ka tensor hoga,
 * utni hi uski axes hongi.
 * 
 * ------------------------------------------------------------
 * ! 0D Tensor (Scalar)
 * 
 * Scalar means single value.
 * 
 * Example:
 * 
 * 5
 * 10
 * -3
 * 
 * Isme sirf ek value hoti hai.
 * 
 * Isliye:
 * Rank = 0
 * Axes = 0
 * 
 * ------------------------------------------------------------
 * ! 1D Tensor (Vector)
 * 
 * 1D tensor ko vector bhi bolte hain.
 * 
 * Example:
 * 
 * [1, 2, 3, 4]
 * 
 * Yeh ek 1D tensor hai.
 * 
 * Lekin confusion wali baat:
 * 
 * Tensor 1D hai
 * But iske andar 4 values hain.
 * 
 * Isliye iska:
 * Shape = (4)
 * Rank = 1
 * 
 * Matlab:
 * 
 * Axis ek hi hai
 * But us axis par 4 elements present hain.
 * 
 * Jo bhi vector hum banate hain,
 * woh tensor hi hota hai.
 * 
 * Agar scalar values ko continuously combine karte jao,
 * toh vector ban jata hai.
 * 
 * ------------------------------------------------------------
 * ! 2D Tensor (Matrix)
 * 
 * Multiple vectors ko combine kar do,
 * toh matrix ban jata hai.
 * 
 * Matrices are nothing but collection of vectors.
 * 
 * Example:
 * 
 * [
 *   [1, 2, 3],
 *   [4, 5, 6]
 * ]
 * 
 * Yeh ek 2D tensor hai.
 * 
 * Shape = (2, 3)
 * Rank = 2
 * 
 * NumPy mein matrix:
 * 
 * np.array(...)
 * 
 * se banaya ja sakta hai.
 * 
 * ------------------------------------------------------------
 * ! ND Tensor
 * 
 * ND tensor means:
 * 
 * Tensor having more than 2 dimensions.
 * 
 * Example:
 * 
 * 3D tensor ko cuboid ki tarah visualize kar sakte hain.
 * 
 * ------------------------------------------------------------
 * ! Rank, Axes and Shape
 * 
 * Number of Axes = Rank = Number of Dimensions
 * 
 * Example:
 * 
 * Matrix:
 * 
 * [
 *   [1, 2, 3],
 *   [4, 5, 6]
 * ]
 * 
 * Shape = (2, 3)
 * Rank = 2
 * 
 * ------------------------------------------------------------
 * ! Size of Tensor
 * 
 * Kisi bhi tensor ka total size nikalne ke liye:
 * 
 * Size = Product of Shape
 * 
 * Example:
 * 
 * Shape = (2, 3)
 * 
 * Size = 2 * 3 = 6
 * 
 * Scalar yani 0D tensor ka size hamesha 1 hota hai.
 * 
 * ------------------------------------------------------------
 * ! Practical Examples of Tensors in Machine Learning
 * 
 * Ye tensors Machine Learning aur Deep Learning mein
 * almost har jagah use honge.
 * 
 * ------------------------------------------------------------
 * ! Example of 1D Tensor
 * 
 * Maan lo students ka dataset hai:
 * 
 * Columns:
 * 
 * CGPA
 * IQ
 * State Rank
 * Placement Status
 * 
 * Ab agar sirf ek student ka data le:
 * 
 * [6.5, 120, 1]
 * 
 * Yeh ek vector hai.
 * 
 * Isliye:
 * 
 * Yeh 1D tensor hai.
 * 
 * Shape = (3)
 * Rank = 1
 * 
 * ------------------------------------------------------------
 * ! Example of 2D Tensor
 * 
 * Agar multiple students ka data store kare:
 * 
 * [
 *   [6.5, 120, 1],
 *   [8.2, 130, 0],
 *   [7.1, 110, 1]
 * ]
 * 
 * Toh yeh matrix ban jayega.
 * 
 * Isliye:
 * 
 * Yeh 2D tensor hai.
 * 
 * ------------------------------------------------------------
 * ! Example of 3D Tensor
 * 
 * 3D tensors thode rare lagte hain initially,
 * but NLP aur textual data mein bahut use hote hain.
 * 
 * Example:
 * 
 * Multiple sentences
 * Multiple words
 * Multiple embeddings
 * 
 * Time Series Data bhi 3D tensor ka
 * bahut achha example hai.
 * 
 * ------------------------------------------------------------
 * ! Example of 4D Tensor
 * 
 * Images are one of the best examples of 4D tensors.
 * 
 * Images mein bahut saare pixels hote hain.
 * 
 * Deep Learning mein image datasets ko mostly
 * 4D tensors ki form mein store kiya jata hai.
 * 
 * Example Dimensions:
 * 
 * (batch_size, height, width, channels)
 * 
 * ------------------------------------------------------------
 * ! Example of 5D Tensor
 * 
 * Videos are one of the best examples of 5D tensors.
 * 
 * Video basically images (frames) ka collection hota hai.
 * 
 * Frames itni speed se move karte hain ki
 * hamara brain unhe continuous motion samajhta hai.
 * 
 * Human persistence of vision approximately
 * 12 FPS ke around hota hai.
 * 
 * Usse upar frames chalenge toh
 * hume video jaisa feel hota hai.
 * 
 * Video Tensor Example Dimensions:
 * 
 * (
 *   batch_size,
 *   frames,
 *   height,
 *   width,
 *   channels
 * )
 * 
 */

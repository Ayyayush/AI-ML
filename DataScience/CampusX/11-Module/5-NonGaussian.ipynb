# Non-Gaussian Distribution

Non-Gaussian Distribution is also known as **Non-Normal Distribution**.

It refers to any distribution that **does NOT follow the bell-shaped (normal) curve**.

---

## 📌 Key Idea

* Data is **not symmetric**
* Mean, Median, Mode are **not equal**
* Shape can be:

  * Skewed
  * Multi-modal
  * Uniform
  * Exponential, etc.

---

## 📚 Types of Non-Gaussian Distributions

We will study in two parts:

1. **Discrete Distributions**
2. **Continuous Distributions**

---

# 1️⃣ Discrete Distributions

Discrete distributions deal with **countable values**.

👉 Example:

* Number of students
* Number of calls
* Number of defective items

---

## 🔹 Common Discrete Distributions

### 1. Bernoulli Distribution

* Only **2 outcomes**

  * Success (1)
  * Failure (0)

👉 Example:

* Tossing a coin (Head/Tail)

---

### 2. Binomial Distribution

* Repeated Bernoulli trials
* Fixed number of trials (n)

👉 Example:

* Number of heads in 10 coin tosses

---

### 3. Poisson Distribution

* Used for **rare events**
* Counts occurrences in a fixed interval

👉 Example:

* Number of calls in 1 hour
* Number of accidents per day

---

## 📌 Key Characteristics

* Values are **integers**
* Probability is assigned to each value
* Sum of probabilities = 1

---

# 2️⃣ Continuous Distributions

Continuous distributions deal with **real-valued data**.

👉 Example:

* Height
* Weight
* Time
* Temperature

---

## 🔹 Common Non-Gaussian Continuous Distributions

### 1. Uniform Distribution

* All values are **equally likely**

👉 Example:

* Random number between 0 and 1

---

### 2. Exponential Distribution

* Models **time between events**

👉 Example:

* Time between customer arrivals

---

### 3. Log-Normal Distribution

* Data is skewed to the right
* Log of data is normally distributed

👉 Example:

* Income distribution

---

### 4. Gamma Distribution

* Generalization of exponential distribution
* Used in waiting time problems

---

## 📌 Key Characteristics

* Values are **continuous (infinite possibilities)**
* Defined using **Probability Density Function (PDF)**
* Area under curve = 1

---

# ⚡ Difference: Normal vs Non-Normal

| Feature          | Normal Distribution | Non-Normal Distribution |
| ---------------- | ------------------- | ----------------------- |
| Shape            | Bell-shaped         | Any shape               |
| Symmetry         | Symmetric           | May be skewed           |
| Mean/Median/Mode | Equal               | Not equal               |
| Examples         | Height, marks       | Income, waiting time    |

---

# 🧠 ML Perspective (Very Important)

* Real-world data is **rarely perfectly normal**
* Need to:

  * Detect skewness
  * Apply transformations (log, sqrt)
* Some models assume normality:

  * Linear Regression
  * Naive Bayes

---

# ✅ Summary

* Non-Gaussian = Non-Normal distribution
* Can be **discrete or continuous**
* Shapes vary (not bell-shaped)
* Important for real-world ML data


# Kurtosis

## What is Kurtosis

Kurtosis is a measure of how **peaked or flat** a distribution is compared to a normal distribution.

It tells us about the **shape of the distribution**, especially:

* How sharp the peak is
* How heavy the tails are

---

## Types of Kurtosis

### 1. Mesokurtic

* Same shape as normal distribution
* Moderate peak and moderate tails

Example: Normal distribution

---

### 2. Leptokurtic

* More peaked (sharp peak)
* Heavy tails (more outliers)

Example: Data with extreme values

---

### 3. Platykurtic

* Flatter than normal distribution
* Light tails (fewer outliers)

Example: Uniform-like data

---

# Properties of Kurtosis

1. Measures peak and tail heaviness

2. Helps detect outliers

* High kurtosis → more outliers
* Low kurtosis → fewer outliers

3. Compared with normal distribution

4. Normal distribution kurtosis ≈ 3
   (Excess kurtosis = 0)

---

# Formula

Kurtosis = E[(X - μ)⁴] / σ⁴

Where:

* μ = mean
* σ = standard deviation

---

## Excess Kurtosis

Excess Kurtosis = Kurtosis - 3

* = 0 → Mesokurtic
* > 0 → Leptokurtic
* < 0 → Platykurtic

---

# Example

* Income data → Leptokurtic (few very high values)
* Uniform data → Platykurtic

---

# Summary

* Kurtosis tells shape of distribution
* Focus on peak + tails
* Helps identify outliers
* Important in data analysis and ML


# QQ Plots

QQ Plot (Quantile-Quantile Plot) ek graphical method hai jisse hum check karte hain ki **data kisi theoretical distribution (mostly normal distribution)** ko follow karta hai ya nahi.

---

## 📌 What is a QQ Plot?

* QQ Plot compare karta hai:

  * **Actual data ke quantiles**
  * vs
  * **Theoretical distribution ke quantiles**

👉 Agar data normal hai → points ek **straight line** pe aayenge

---

## 🎯 Purpose

* Check karna ki data **normally distributed hai ya nahi**
* Outliers detect karna
* Data ka shape samajhna

---

## 🧠 Intuition (Hinglish)

Socho:

* X-axis → theoretical values (normal distribution)
* Y-axis → actual data values

👉 Agar dono match karte hain → straight line
👉 Agar nahi → curve / deviation dikhega

---

## 📊 Interpretation

### 1. Straight Line

* Data ≈ Normal distribution
* Good for ML models

---

### 2. S-shaped Curve

* Data skewed hai

👉 Right skew → upward curve
👉 Left skew → downward curve

---

### 3. Points far from line

* Outliers present

---

### 4. Heavy tails

* Kurtosis high (Leptokurtic)

---

## ⚡ Key Observations

* Line ke paas points → good fit
* Deviations → non-normal behavior
* Ends (tails) pe deviation → outliers ya skewness

---

## 🧪 Python Code (DSMP Style)

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = sns.load_dataset('titanic')

# Drop missing values
data = df['age'].dropna()

# QQ Plot
sm.qqplot(data, line='s')
plt.title("QQ Plot of Age")
plt.show()
```

---

## 🧠 ML Use

* Normality check before:

  * Linear Regression
  * Statistical tests

* Helps in:

  * Feature transformation
  * Outlier detection

---

## ⚠️ Important Note

* Perfect straight line → rarely hota hai real data me
* Slight deviation → acceptable

---

## ✅ Summary

* QQ Plot = Quantile vs Quantile comparison
* Straight line → normal distribution
* Curve → skewness
* Extreme deviation → outliers

# Uniform Distribution

Uniform Distribution ek aisi probability distribution hai jisme **sabhi values equally likely hoti hain**.

👉 Matlab:
Har value ke aane ka chance **same hota hai**

---

## 📌 Types of Uniform Distribution

### 1. Discrete Uniform Distribution

* Finite values hoti hain
* Har value ka equal probability

👉 Example:

* Dice roll (1, 2, 3, 4, 5, 6)
* Har number ka probability = 1/6

---

### 2. Continuous Uniform Distribution

* Range ke beech infinite values hoti hain
* Har interval ka equal probability

👉 Example:

* Random number between 0 and 1

---

## ⚡ Probability Density Function (PDF)

For continuous uniform distribution:

[
f(x) =
\begin{cases}
\frac{1}{b - a} & \text{if } a \le x \le b \
0 & \text{otherwise}
\end{cases}
]

Where:

* **a** = minimum value
* **b** = maximum value

---

## 📊 Key Characteristics

* Shape = **Flat (Rectangle)**
* No peak (unlike normal distribution)
* Mean = (a + b) / 2
* Variance = (b - a)² / 12

---

## 🎯 Intuition (Hinglish)

Socho tum random number generate kar rahe ho between 0 and 10:

👉 2 aane ka chance = 8 aane ka chance = 5 aane ka chance

Sabka probability same hai → **Uniform Distribution**

---

## 🧠 Use in Machine Learning

### 1. Random Initialization

* Neural Networks me weights initialize karne ke liye
* Uniform distribution use hota hai (e.g., -0.5 to 0.5)

---

### 2. Random Sampling

* Train-test split
* Data sampling

---

### 3. Monte Carlo Simulation

* Random sampling based problems solve karne me

---

### 4. Exploration (Reinforcement Learning)

* Random actions choose karne ke liye

---

### 5. Baseline Assumption

* Jab koi prior knowledge nahi hoti → assume uniform

---

## ⚠️ Important Note

* Real-world data rarely uniform hota hai
* Mostly synthetic ya random processes me use hota hai

---

## 🔍 Visualization Idea

* Histogram → flat rectangle dikhega
* KDE → flat line

---

## ✅ Summary

* Uniform distribution = equal probability for all values
* Shape = flat
* Discrete + Continuous dono types
* ML me mainly randomization aur initialization me use hota hai


````python
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Generate a set of random data
x = np.random.uniform(low=0, high=1, size=1000)



````

````python
plt.hist(x)
````

````python
# Fit a uniform distribution to the data
params = stats.uniform.fit(x)
dist = stats.uniform(loc=params[0], scale=params[1])
````

````python
# Create a QQ plot of the data using the uniform distribution
fig = sm.qqplot(x, dist=dist, line='45')


# Add a title and labels to the plot
plt.title('QQ Plot of Uniform Distribution with Uniform Fit')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('Sample Quantiles')

# Show the plot
plt.show()
````

# Log-Normal Distribution

Log-Normal Distribution ek aisi distribution hai jisme **data ka logarithm (log) normal distribution follow karta hai**.

👉 Simple:
Agar ( \log(X) ) normal hai → ( X ) log-normal hai

---

## 📌 Key Idea

* Data **right-skewed (positive skew)** hota hai
* Values **0 se start hoti hain (negative nahi hoti)**
* Large values (outliers) ho sakte hain

---

## 🎯 Intuition (Hinglish)

Socho income data:

* Most log low–medium income me
* Few log bahut high earn karte hain

👉 Isliye graph right side me stretch hota hai

---

## 📊 Shape of Log-Normal Distribution

* Left side → steep
* Right side → long tail
* Not symmetric

---

## ⚡ Mathematical Definition

Agar:
[
Y = \log(X)
]
aur (Y) follows normal distribution

👉 Tab:
[
X \sim \text{Log-Normal Distribution}
]

---

## 📌 Key Characteristics

* Always **positive values (X > 0)**
* **Right skewed distribution**
* Mean > Median > Mode
* Heavy tail on right side

---

## 🔹 Real-Life Examples

* Income distribution
* Stock prices
* House prices
* Population growth
* Time to complete tasks

---

## 🧠 Use in Machine Learning

### 1. Real-World Data Modeling

* Financial data (prices, returns)
* Growth-based data

---

### 2. Feature Transformation

👉 Jab data skewed ho:

* Apply log transform
  [
  X' = \log(X)
  ]

* Data approx normal ban jata hai

---

### 3. Improves Model Performance

* Linear Regression better perform karta hai
* Variance stabilize hoti hai

---

### 4. Outlier Handling

* Large values compress ho jaate hain
* Extreme values ka effect reduce hota hai

---

## ⚠️ Important Note

* Direct data normal nahi hota
* Log lene ke baad normal hota hai

👉 Isliye ML me log transformation bahut important hai

---

## 🔍 Visualization Idea

* Histogram → right skewed
* Log transform ke baad → bell-shaped

---

## ✅ Summary

* Log-Normal = log(data) is normal
* Right skewed distribution
* Used for financial & real-world data
* ML me log transformation ke liye important

# Pareto Distribution

Pareto Distribution ek aisi probability distribution hai jo **80-20 rule (Pareto Principle)** ko represent karti hai.

👉 Simple:
**80% results → 20% causes se aate hain**

---

## 📌 Key Idea

* Data **highly skewed (right-skewed)** hota hai
* Few values bahut large hoti hain
* Majority values small hoti hain

---

## 🎯 Intuition (Hinglish)

Socho:

* 20% log → 80% paisa kamaate hain
* 20% products → 80% sales generate karte hain

👉 Yehi Pareto Distribution ka concept hai

---

## 📊 Shape of Distribution

* Left side → zyada data (small values)
* Right side → long tail (few large values)
* Highly skewed

---

## ⚡ Mathematical Formula (PDF)

[
f(x) = \frac{\alpha x_m^\alpha}{x^{\alpha+1}}, \quad x \ge x_m
]

Where:

* ( x_m ) = minimum value
* ( \alpha ) = shape parameter

---

## 📌 Key Characteristics

* Values **≥ minimum value (xₘ)**
* Heavy right tail
* Mean and variance depend on ( \alpha )
* Extreme values possible

---

## 🔹 Real-Life Examples

* Wealth distribution
* City population sizes
* Website traffic
* File sizes on internet
* Sales distribution

---

## 🧠 Use in Machine Learning

### 1. Business Analytics

* Identify top customers / products
* Revenue optimization

---

### 2. Feature Importance

* Few features dominate model output

---

### 3. Anomaly Detection

* Extreme values (tail) identify karne me useful

---

### 4. Resource Allocation

* Focus on high-impact areas (top 20%)

---

## ⚠️ Important Note

* Real-world data often Pareto-like hota hai
* Heavy skewness handle karne ke liye:

  * Log transformation use hota hai

---

## 🔍 Visualization Idea

* Histogram → right skewed
* Few large spikes in tail

---

## ✅ Summary

* Pareto = 80/20 rule
* Highly right-skewed distribution
* Few values dominate
* ML + business analytics me very important


````python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters of the Pareto distribution
alpha = 3
xm = 1

# Create an array of x values
x = np.linspace(0.1, 10, 1000)

# Calculate the y values of the Pareto distribution
y = alpha * (xm**alpha) / (x**(alpha+1))


plt.plot(x,y)
````

````python
# Create the log-log plot
plt.plot(np.log(x),np.log(y))

# Add labels and a title
plt.xlabel('x')
plt.ylabel('P(x)')
plt.title('Pareto distribution')

# Show the plot
plt.show()
````

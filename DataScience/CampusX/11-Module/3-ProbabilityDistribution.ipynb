# Random Variables

## What are Algebraic Variables

An algebraic variable is a symbol (like x, y) that represents an unknown but fixed value.

Example:
x + 5 = 10 → x = 5

Here, x is unknown initially but once solved, it has a fixed value.

---

## What are Random Variables in Statistics and Probability

A random variable is a variable whose value comes from a random experiment.
It represents all possible numerical outcomes of that experiment.

Example 1: Coin Toss
If we define:
Heads = 1, Tails = 0
Then the random variable X can take values → {0, 1}

Example 2: Rolling a Dice
Random variable X → {1, 2, 3, 4, 5, 6}

---

# Sample Space

Sample space is the set of all possible values that a random variable can take.

Example 1: Coin Toss
Sample Space → {H, T}
If mapped to numbers → {1, 0}

Example 2: Dice Roll
Sample Space → {1, 2, 3, 4, 5, 6}

---

# Types of Random Variables

## 1. Discrete Random Variable

A discrete random variable takes countable values.

Examples:

* Coin toss → {0, 1}
* Dice roll → {1, 2, 3, 4, 5, 6}

These values are finite or countable.

---

## 2. Continuous Random Variable

A continuous random variable takes infinite values within a range.

Examples:

* Height of a person → can be 170.1, 170.12, 170.123...
* Weight → can take any value in a range

These values are not countable and lie in an interval.


# Probability Distribution

A probability distribution tells us all the possible values a random variable can take and the probability of each value.

Example:
Dice → values = {1,2,3,4,5,6}
Each value has probability = 1/6

---

## Number of Outcomes May Be Infinite → Solution = Functions

In many real-life cases, outcomes are too many or even infinite.

Example:

* Height of people (can take infinite values)
* Measuring time

In such cases, writing a table is not possible.

👉 Solution: Use a mathematical function
We represent probability using a function:

y = f(x)

It is represented as:

y = f(x)

* x = value of random variable
* y = probability of that value

---

## Probability Distribution Function Types

### 1. Probability Mass Function (PMF)

Used for discrete random variables.

Properties:

* Probability ≥ 0
* Sum of all probabilities = 1

Example:
Dice → P(X=1) = 1/6

---

### 2. Cumulative Distribution Function (CDF)

Gives cumulative probability.

F(x) = P(X ≤ x)

Example:
Dice →
F(3) = P(X ≤ 3) = 1/6 + 1/6 + 1/6 = 3/6

---

### 3. Probability Density Function (PDF)

Used for continuous random variables.

* Probability at a single point = 0
* Probability is calculated over a range

Example:
Probability that height is between 170 and 180

---

# Probability Mass Function (PMF)

PMF is used for **discrete random variables**.
It gives the probability of each exact value.

### Definition

PMF assigns a probability to every possible value of a random variable.

### Properties

* Probability is always ≥ 0
* Sum of all probabilities = 1

### Example

Dice roll:
X → {1,2,3,4,5,6}

P(X = 1) = 1/6
P(X = 2) = 1/6
...

So, PMF gives probability for each value separately.

---

# Cumulative Distribution Function (CDF)

CDF gives the **cumulative probability** up to a value.

### Definition

F(x) = P(X ≤ x)

It means probability that the random variable is less than or equal to x.

---

### Example

Dice roll:

F(1) = 1/6
F(2) = 2/6
F(3) = 3/6
F(6) = 6/6 = 1

👉 CDF keeps adding probabilities.

---

# Probability Density Function (PDF)

PDF is used for **continuous random variables**.

### Definition

It is a function that describes how probability is distributed over a range of values.

---

### Important Points

* Probability at a single point = 0
* Probability is calculated over an interval
* Area under the curve = probability

---

### Example

Height of people:

We cannot say probability of height = exactly 170
Instead we say:
Probability that height is between 170 and 180

👉 This is calculated using area under the curve.

---

### Summary

* PMF → exact probability (discrete)
* CDF → cumulative probability
* PDF → probability over range (continuous)


# Density Estimation

Density estimation is a method used to estimate the probability distribution (PDF) of data based on given observations.

In simple words:
We try to understand how data is spread by creating a smooth curve from the data points.

---

# Real World Examples

1. Height of People
   We collect heights of many people and try to see how they are distributed
   → Most people fall around average height

2. Exam Scores
   We analyze marks of students
   → Helps understand how many students scored low, average, or high

3. Income Distribution
   We study how income is spread among people
   → Few people earn very high, most earn average

4. Website Traffic
   We analyze how users visit a website over time
   → Helps understand peak usage times

---

# Parametric Density Estimation

In parametric methods, we assume that the data follows a known distribution.

Example distributions:

* Normal distribution
* Poisson distribution
* Exponential distribution

### Key Idea

We assume a shape and then estimate parameters.

Example:
Height of people → assume Normal distribution
We estimate:

* Mean (average height)
* Standard deviation

---

# Non-Parametric Density Estimation

In non-parametric methods, we do NOT assume any fixed distribution.

We directly estimate the distribution from data.

---

### Example: Kernel Density Estimation (KDE)

* We place small curves (kernels) on each data point
* Combine them to form a smooth curve

---

### Advantages

* No assumption about data
* More flexible

### Disadvantages

* Needs more data
* More computation

---

# Summary

* Density estimation helps us understand data distribution
* Parametric → assume distribution
* Non-parametric → no assumption
* KDE is a popular non-parametric method


```python
import pandas as pd
import random
```

```python
L = []
for i in range(10000):
  a = random.randint(1,6)
  b = random.randint(1,6)

  L.append(a + b)
```

```python
len(L)
```

```python
L[:5]
```

```python
# --------------------------------------------------
# Calculate normalized frequency (probability distribution)
# Probability = frequency / total elements
# P(x) = (number of times x appears) / total values
# --------------------------------------------------

s = (
    pd.Series(L)                                  # convert list L into a Pandas Series
      .value_counts()                             # count frequency of each unique value
    / pd.Series(L).value_counts().sum()           # divide by total count → normalize
).sort_index()                                    # sort result by the actual values (index)
```

```python
import numpy as np
np.cumsum(s)
```

### Parametric Density Estimation

- Assumes the data follows a **known distribution** (e.g., Normal/Gaussian).
- Density is defined using a **fixed set of parameters** (mean, variance, etc.).
- Simple, fast, and works well when the assumption is correct.
- Less flexible if the real data distribution is complex or unknown.

**Example:**  
Normal distribution defined by μ (mean) and σ (standard deviation)


```python
import matplotlib.pyplot as plt
import numpy as np
from numpy.random import normal
# --------------------------------------------------
# Generate a random sample from a normal distribution
# nomrla function wil generate random values which follow normal distribution
# --------------------------------------------------

sample = normal(
    loc=50,        # mean (center) of the distribution
    scale=5,       # standard deviation (spread)
    size=1000      # number of random values to generate
)
```

```python
sample.mean()
```

```python
# plot histogram to understand the distribution of data
plt.hist(sample,bins=10)
```

```python
# calculate sample mean and sample std dev
sample_mean = sample.mean()
sample_std = sample.std()
```

```python
# fit the distribution with the above parameters

from scipy.stats import norm
dist = norm(60, 12)

"""
_summary_
    👉 norm() kya karta hai?
Yeh Normal Distribution (Gaussian) ka object banata hai
Matlab: ab tumhare paas ek mathematical model hai distribution ka
"""
```

```python
values = np.linspace(sample.min(),sample.max(),100)
```

```python
sample.max()
```

```python
# --------------------------------------------------
# ! Calculate Probability Density for given values
# --------------------------------------------------

probabilities = [
    dist.pdf(value)   # Har value ke liye probability density calculate kar rahe hain
    for value in values
]
```

```python
# --------------------------------------------------
# ! Plot Histogram + PDF Curve
# --------------------------------------------------

plt.hist(sample, bins=10, density=True)   # Histogram of data (normalized → probability density)
plt.plot(values, probabilities)           # PDF curve (smooth line)
```

### Kernel Density Estimation (KDE)

- A **non-parametric density estimation** technique.
- Does **not assume any predefined data distribution**.
- Creates a smooth curve to represent the data distribution.
- More flexible than parametric methods.
- Shape depends on the **kernel function** and **bandwidth**.

**Used in Seaborn:**  
`kdeplot`


```python
# generate a sample
sample1 = normal(loc=20, scale=5, size=300)
sample2 = normal(loc=40, scale=5, size=700)
sample = np.hstack((sample1, sample2))

#  normal is a function from NumPy that generates random numbers
#  from a Normal (Gaussian) distribution.
```

```python
# plot histogram bins=50
plt.hist(sample,bins=50)
```

```python
from sklearn.neighbors import KernelDensity

# --------------------------------------------------
# ! Create KDE (Kernel Density Estimation) Model
# --------------------------------------------------

model = KernelDensity(
    bandwidth=5,         # smoothing parameter (curve kitni smooth hogi)
    kernel='gaussian'    # Gaussian shape ka kernel use kar rahe hain
)

# --------------------------------------------------
# ! Convert data to 2D array (required by sklearn)
# --------------------------------------------------

# sklearn ko input shape chahiye: (n_samples, n_features)
sample = sample.reshape((len(sample), 1))

# --------------------------------------------------
# ! Fit KDE model to the data
# --------------------------------------------------

model.fit(sample)        # model ko data se train kar rahe hain
```

```python
values = np.linspace(sample.min(),sample.max(),100)
values = values.reshape((len(values), 1))
```

```python
# --------------------------------------------------
# ! Compute Probability Density using KDE Model
# --------------------------------------------------

# Step 1: Get log-density values for each point
probabilities = model.score_samples(values)   # returns log(probability density)

# Step 2: Convert log-density → actual density
probabilities = np.exp(probabilities)         # exponential se actual PDF milta hai
```

`score_samples(values)`
 it returns the log-density estimate of the input samples values. This is because the `score_samples()` method of the KernelDensity class returns the logarithm of the probability density estimate rather than the actual probability density estimate.

```python
# --------------------------------------------------
# Plot Histogram + KDE Curve
# --------------------------------------------------

# Plot histogram of the sample
plt.hist(
    sample,                      # raw data
    bins=50,                     # number of bins
    density=True                 # normalize histogram to probability density
)

# Plot KDE curve on top of histogram
plt.plot(
    values,                      # evenly spaced x-values
    probabilities                # KDE probability density values
)

# Display the plot
plt.show()
```

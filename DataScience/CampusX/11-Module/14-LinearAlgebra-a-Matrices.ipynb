 * ------------------------------------------------------------
 * ! What are Matrices ?
 * ------------------------------------------------------------
 * A Matrix is a rectangular array of:
 * - Numbers
 * - Symbols
 * - Expressions
 * 
 * arranged in:
 * - Rows
 * - Columns
 * 
 * Matrix ke andar jo values hoti hain unhe:
 * Elements of Matrix bolte hain.
 * 
 * Example:
 * [
 *   [1,2,3],
 *   [4,5,6]
 * ]
 * 
 * Rows = 2
 * Columns = 3
 * 
 * Shape = (Rows,Columns)
 * Example:
 * Shape = (2,3)
 * 
 * Total Elements = Rows * Columns
 * Example:
 * Total Elements = 2 * 3 = 6
 * 
 * ------------------------------------------------------------
 * ! Why are Matrices Useful ?
 * ------------------------------------------------------------
 * Matrix ek powerful mathematical tool hai.
 * Ye multiple domains mein use hota hai.
 * 
 * Real world examples:
 * - Images
 * - Videos
 * - Graphics
 * - Machine Learning Data
 * 
 * Ye sab internally matrices ki form mein represent kiye ja sakte hain.
 * 
 * ------------------------------------------------------------
 * ! Use Cases of Matrices
 * ------------------------------------------------------------
 * 1. Linear Systems
 * Multiple equations ko efficiently solve karne ke liye matrices use hote hain.
 * 
 * 2. Linear Transformations
 * Rotation, Scaling, Translation jaise operations mein matrices use hote hain.
 * 
 * 3. Eigen Values and Eigen Vectors
 * Deep Learning, PCA, Computer Vision mein heavily use hote hain.
 * 
 * 4. Graph Theory
 * Graph ko adjacency matrix ki form mein represent kar sakte hain.
 * 
 * 5. Markov Chains
 * Probability transitions ko represent karne ke liye matrices use hote hain.
 * 
 * 6. Computer Graphics
 * 2D aur 3D graphics transformations matrices ki help se perform hote hain.
 * 
 * 7. Optimization
 * Machine Learning optimization algorithms matrices ke bina possible nahi hote.
 * 
 * 8. Control Theory
 * Robotics aur automation systems mein matrices bahut use hote hain.
 * 
 * ------------------------------------------------------------
 * ! Rules of Operations on Matrices
 * ------------------------------------------------------------
 * ! 1. Matrix Addition Commutative Property
 * A + B = B + A
 * Matlab:
 * Addition ka order change karne par answer same rehta hai.
 * 
 * ------------------------------------------------------------
 * ! 2. Matrix Addition Associative Property
 * ------------------------------------------------------------
 * (A + B) + C = A + (B + C)
 * Matlab:
 * Addition kisi bhi grouping mein karo,
 * final answer same rahega.
 * 
 * ------------------------------------------------------------
 * ! 3. Scalar Multiplication Associative Property
 * ------------------------------------------------------------
 * a(bA) = (ab)A
 * Yahan:
 * a,b -> scalars
 * A -> matrix
 * 
 * ------------------------------------------------------------
 * ! 4. Distributive Property over Matrix Addition
 * ------------------------------------------------------------
 * A(B + C) = AB + AC
 * Aur:
 * (A + B)C = AC + BC
 * 
 * ------------------------------------------------------------
 * ! 5. Matrix Multiplication NOT Commutative
 * ------------------------------------------------------------
 * AB != BA
 * Ye matrices ka bahut important rule hai.
 * Mostly:
 * AB and BA different hote hain.
 * Kabhi kabhi:
 * BA exist bhi nahi karta.
 * 
 * ------------------------------------------------------------
 * ! 6. Matrix Multiplication Associative Property
 * ------------------------------------------------------------
 * (AB)C = A(BC)
 * 
 * ------------------------------------------------------------
 * ! 7. Identity Matrix Property
 * ------------------------------------------------------------
 * AI = IA = A
 * Yahan:
 * I -> Identity Matrix
 * 
 * Identity matrix matrices ke liye
 * waise hi kaam karti hai
 * jaise number system mein 1.
 * 
 * ------------------------------------------------------------
 * ! 8. Zero Matrix Property
 * ------------------------------------------------------------
 * A + 0 = A
 * A0 = 0A = 0
 * 
 * ------------------------------------------------------------
 * ! 9. Matrix Subtraction NOT Commutative
 * ------------------------------------------------------------
 * A - B != B - A
 * 
 * ------------------------------------------------------------
 * ! 10. Transpose Property
 * ------------------------------------------------------------
 * (AB)^T = B^T A^T
 * Yahan:
 * T -> transpose
 * 
 * Order reverse ho jata hai transpose mein.
 * 
 * (A+B)^T = A^T + B^T
 * (A^T)^T = A
 * 
 * ------------------------------------------------------------
 * ! 11. Inverse Property
 * ------------------------------------------------------------
 * AA^-1 = A^-1A = I
 * Yahan:
 * A^-1 -> inverse matrix
 * I -> identity matrix
 * 
 * ------------------------------------------------------------
 * ! 12. Condition for Matrix Multiplication
 * ------------------------------------------------------------
 * Agar:
 * A ka shape = (m,n)
 * B ka shape = (p,q)
 * 
 * Toh:
 * Multiplication tabhi possible hai jab:
 * n = p
 * 
 * Resultant matrix ka shape hoga:
 * (m,q)
 * 
 * ------------------------------------------------------------
 * ! Types of Matrices
 * ------------------------------------------------------------
 * Different types ki matrices hoti hain:
 * - Square Matrix
 * - Rectangular Matrix
 * - Identity Matrix
 * - Diagonal Matrix
 * - Symmetric Matrix
 * - Sparse Matrix
 * - Zero Matrix
 * - Null Matrix
 * - Upper Triangular Matrix
 * - Lower Triangular Matrix
 * - Skew Symmetric Matrix
 * 
 * Inhe hum aage detail mein padhenge.
 * 
 * ------------------------------------------------------------
 * ! Matrix Equality
 * ------------------------------------------------------------
 * Do matrices tab equal hoti hain jab:
 * 1. Dono ka shape same ho
 * 2. Corresponding elements equal ho
 * 
 * Example:
 * 
 * A = [
 *      [1,2],
 *      [3,4]
 *     ]
 * 
 * B = [
 *      [1,2],
 *      [3,4]
 *     ]
 * 
 * Toh:
 * A == B
 * 
 * ------------------------------------------------------------
 * ! Scalar Operations on Matrices
 * ------------------------------------------------------------
 * Scalar operations ka matlab:
 * Matrix ke har element par scalar value apply karna.
 * 
 * Scalar operations mainly 2 types ke hote hain:
 * 1. Scalar Addition
 * 2. Scalar Multiplication
 * 
 * ------------------------------------------------------------
 * ! Scalar Addition
 * ------------------------------------------------------------
 * Matrix:
 * [
 *   [1,2],
 *   [3,4]
 * ]
 * 
 * Scalar = 2
 * 
 * Matrix + 2
 * 
 * Result:
 * [
 *   [3,4],
 *   [5,6]
 * ]
 * 
 * ------------------------------------------------------------
 * ! Scalar Multiplication
 * ------------------------------------------------------------
 * Matrix:
 * [
 *   [1,2],
 *   [3,4]
 * ]
 * 
 * Scalar = 2
 * 
 * 2 * Matrix
 * 
 * Result:
 * [
 *   [2,4],
 *   [6,8]
 * ]
 * 
 * Har element ko 2 se multiply kiya gaya.
 * 
 * ------------------------------------------------------------
 * ! Matrix Operations
 * ------------------------------------------------------------
 * ! Addition
 * Matrix addition tabhi perform hoga
 * jab dono matrices ka shape same hoga.
 * 
 * Additive Identity:
 * Matrix mein kuch aisa add kiya
 * ki same matrix hi mil jaye.
 * 
 * A + 0 = A
 * 
 * Additive Inverse:
 * Matrix mein aisa kya subtract kare
 * ki zero matrix mil jaye.
 * 
 * A + (-A) = 0
 * 
 * ------------------------------------------------------------
 * ! Multiplication
 * ------------------------------------------------------------
 * Matrix multiplication mein
 * dot product use hota hai.
 * 
 * Formula for multiplication:
 * Multiplication ke liye:
 * Pehle wale matrix ka columns
 * = dusre wale matrix ka rows
 * 
 * Resultant shape:
 * Pehle wale ka row
 * aur dusre wale ka column
 * 
 * Example:
 * A -> (2,3)
 * B -> (3,4)
 * 
 * Result:
 * AB -> (2,4)
 * 
 * Rules for multiplication:
 * A.B != B.A
 * It is NOT commutative.
 * 
 * Example:
 * A = (2,3)
 * B = (3,3)
 * 
 * A*B possible hai
 * par B*A possible nahi.
 * 
 * (AB)C = A(BC)
 * 
 * A(B+C) = AB + AC
 * 
 * ------------------------------------------------------------
 * ! Multiplicative Identity
 * ------------------------------------------------------------
 * AI = IA = A
 * 
 * ------------------------------------------------------------
 * ! Transpose of a Matrix
 * ------------------------------------------------------------
 * Isme rows columns ban jate hain.
 * 
 * Symmetric matrix ka transpose
 * karne par bhi same matrix milta hai.
 * 
 * A^T = A
 * 
 * Skew Symmetric Matrix:
 * A^T = -A
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Determinant
 * ------------------------------------------------------------
 * Determinant ek scalar value hoti hai
 * jo kisi square matrix se compute ki jati hai.
 * 
 * Determinant matrix ke baare mein important information deta hai.
 * 
 * Uses:
 * - Linear Algebra
 * - Invertibility check
 * - Solving linear equations
 * - Eigen values
 * 
 * Determinant sirf:
 * Square Matrix ke liye hi nikalta hai.
 * 
 * ------------------------------------------------------------
 * ! Important Point
 * ------------------------------------------------------------
 * Agar matrix invertible hai
 * toh uska determinant zero nahi hoga.
 * 
 * Sirf square matrices ka inverse hota hai.
 * 
 * Non-square matrix ka inverse nahi hota.
 * 
 * Haan:
 * Kuch techniques use karke
 * pseudo inverse nikal sakte hain.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Types of Matrices based on Determinant
 * ------------------------------------------------------------
 * 1. Singular Matrix
 * Jiska determinant:
 * det(A) = 0
 * 
 * Is matrix ka inverse exist nahi karta.
 * 
 * 2. Non-Singular Matrix
 * Jiska determinant:
 * det(A) != 0
 * 
 * Is matrix ka inverse exist karta hai.
 * 
 * 
 * 
 * ------------------------------------------------------------
 * ! Minor
 * ------------------------------------------------------------
 * Minor of an element:
 * Kisi element ka row aur column remove karne ke baad
 * jo determinant bachta hai
 * use us element ka minor bolte hain.
 * 
 * Example:
 * A = [
 *      [1,2,3],
 *      [4,5,6],
 *      [7,8,9]
 *     ]
 * 
 * Agar:
 * M11 nikalna hai
 * 
 * Toh:
 * Row1 aur Col1 remove karenge.
 * 
 * Bachega:
 * [
 *   [5,6],
 *   [8,9]
 * ]
 * 
 * Iska determinant hi:
 * Minor hoga.
 * 
 * 
 * ------------------------------------------------------------
 * ! Cofactor
 * ------------------------------------------------------------
 * Cofactor of element Aij:
 * Aij = (-1)^(i+j) * Mij
 * 
 * Yahan:
 * Mij -> Minor
 * 
 * Simply:
 * Minor ke saath sign apply kar dete hain.
 * 
 * Sign pattern:
 * 
 * [
 *   [+,-,+],
 *   [-,+,-],
 *   [+,-,+]
 * ]
 * 
 * Example:
 * Agar:
 * M12 = 5
 * Toh:
 * C12 = (-1)^(1+2) * 5
 *      = -5
 * 
 * 
 * ------------------------------------------------------------
 * ! Adjoint / Adjugate Matrix
 * ------------------------------------------------------------
 * Adjoint matrix ko:
 * Adjugate Matrix
 * ya
 * Classical Adjoint
 * bhi bolte hain.
 * 
 * Steps:
 * 1. Har element ka cofactor nikalo
 * 2. Cofactor matrix banao
 * 3. Us matrix ka transpose le lo
 * 
 * Result:
 * Adjoint Matrix
 * 
 * 
 * ------------------------------------------------------------
 * ! Inverse of a Matrix
 * ------------------------------------------------------------
 * Inverse tabhi exist karega jab:
 * det(A) != 0
 * 
 * Formula:
 * A^-1 = (1 / det(A)) * adj(A)
 * Important Property:
 * 
 * A^-1 * A = I
 * 
 * Yahan:
 * I -> Identity Matrix
 * 
 * ------------------------------------------------------------
 * ! Methods to Find Inverse
 * ------------------------------------------------------------
 * 1. Adjugate Method
 * 2. Gaussian Elimination
 * 3. LU Decomposition
 * 
 * ------------------------------------------------------------
 * ! Solving System of Linear Equations
 * ------------------------------------------------------------
 * Example:
 * 
 * x + y = 5
 * 4x + 3y = 15
 * In equations ko matrices ki help se solve kar sakte hain.
 * 
 * General Form:
 * AX = B
 * Dono side A^-1 se multiply karenge:
 * A^-1 * AX = A^-1 * B
 * Identity property use karenge:
 * IX = A^-1 * B
 * 
 * Since:
 * IX = X
 * 
 * Final:
 * X = A^-1 * B
 * 
 * ------------------------------------------------------------
 * ! Important Point
 * ------------------------------------------------------------
 * Inverse nikalne se pehle
 * determinant check karte hain.
 * 
 * Agar:
 * det(A) = 0
 * 
 * Toh inverse exist nahi karega.

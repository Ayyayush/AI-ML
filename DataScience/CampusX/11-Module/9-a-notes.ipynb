# lecture 1 part notes

p-value is the probability of getting a sample **as extreme or more extreme than our observed sample**, given that the null hypothesis is true  

---

maan lo ek sikka hai jisko 100 baar toss kara  
aur hum note kar rahe hain ki head kitni baar aata hai  

ye experiment **binomial distribution** follow karega  
kyunki ye **Bernoulli trials** hain (2 possible outcomes: head ya tail)  

---

maan lo hum hypothesis testing kar rahe hain  

humne null hypothesis liya:
👉 coin fair hai  

matlab:
P(H) = P(T) = 0.5  

---

maan lo 100 toss me 53 heads aaye  

toh kya hum bas iss 53 ke basis pe keh sakte hain ki coin biased hai?  

👉 nahi  

kyunki:
53 head aana **normal fluctuation** ho sakta hai  
ye random variation ki wajah se bhi ho sakta hai  

👉 yani ye result itna extreme nahi hai ki H0 reject kar dein  

---

ab maan lo agar 100 toss me 80 heads aaye  

👉 ye kaafi extreme case hai  

agar coin fair hota  
toh 80 heads aane ki probability **bahut hi chhoti (almost 0)** hoti  

👉 aisa result dekh ke hum bolenge:
ye sirf chance se nahi hua  

👉 toh hum null hypothesis reject kar denge  

---

## key idea

👉 p-value batata hai:
agar H0 true ho  
toh jo result mila hai ya usse bhi extreme result aane ki probability kya hai  

---

# interpreting p-value

jab bhi hypothesis testing karte hain  
hum ek alpha value leke chalte hain  

👉 usually alpha = 0.05  

---

## decision rule

👉 p-value ≤ alpha  
→ reject null hypothesis  

👉 p-value > alpha  
→ fail to reject null hypothesis  

---

## simple samajh

- chhoti p-value → strong evidence against H0  
- badi p-value → weak evidence  

---

## final feel

👉 p-value = "ye result kitna rare hai agar H0 true ho"  

simple line:  
**"rare event mila → H0 galat ho sakta hai"**

# part 2 notes

# P value in context of z-test

pichle lecture wala hi problem hai, bas approach alag hai  

👉 yaha hum **p-value approach** use kar rahe hain  

---

## example 1 (same training wala)

null hypothesis test karni hai ki training ke baad bhi mean 50 hi hai  

H0: mean = 50  
H1: mean > 50  

alpha = 0.05  

given:
x̄ = 53  
mean = 50  
sigma = 5  
n = 30  

---

## z-statistic formula

z = (x̄ - mean) / (sigma / √n)  

= (53 - 50) / (5 / √30)  
≈ 3 / 0.91  
≈ 3.29  

---

## p-value approach

ab hume p-value nikalni hai  

👉 since ye one-tailed test hai (mean > 50)  

z = 3.29  

z-table se:
P(Z > 3.29) ≈ 0.0005  

---

## decision

p-value = 0.0005  
alpha = 0.05  

👉 p-value < alpha  
→ reject H0  

---

## interpretation

👉 training ka positive effect pada hai  

---

# advantages of p-value approach over rejection region

👉 exact strength pata chalti hai  
👉 bas reject ya not reject nahi, balki kitna strong evidence hai wo bhi pata chalta hai  
👉 more informative approach hai  

---

# 2nd example

suppose lays company check kar rahi hai packet ka weight  

claim → 50 gram  

sample liya = 40 packets  
sample mean = 49  
std = 4  

---

step 1 → hypothesis  

H0: mean = 50  
H1: mean ≠ 50  

---

given:

mean = 50  
x̄ = 49  
sigma = 4  
n = 40  
alpha = 0.05  

---

## z calculation

z = (49 - 50) / (4 / √40)  
≈ -1 / 0.63  
≈ -1.58  

---

## p-value calculation

👉 ye two-tailed test hai (≠)  

z = -1.58  

z-table se:
left area ≈ 0.057  

👉 since two-tailed:
p-value = 2 × 0.057  
≈ 0.114  

---

## decision

p-value = 0.114  
alpha = 0.05  

👉 p-value > alpha  
→ fail to reject H0  

---

## interpretation

👉 enough evidence nahi hai prove karne ke liye ki packet weight 50g se different hai  

---

## final feel

👉 p-value approach zyada clear hota hai  
👉 directly bata deta hai evidence kitna strong hai  

simple line:  
**"p-value chhota → H0 weak, p-value bada → H0 strong"**

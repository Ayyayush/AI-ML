/**
 * =========================================================
 * ! ASSUMPTIONS OF LINEAR REGRESSION - RECAP
 * =========================================================
 *
 * (Page 1-2)
 *
 * Till now hum Regression Summary kaafi had tak
 * samajh chuke hain:
 *
 * → R²
 * → Adjusted R²
 * → F-statistic
 * → Prob(F-statistic)
 * → t-statistic
 * → p-value
 * → Confidence Interval
 *
 * But ek important question:
 *
 * Kya hum in values par blindly trust kar sakte hain?
 *
 * Answer:
 *
 * NO.
 *
 * Ye saari statistics tabhi valid hoti hain
 * jab Linear Regression ki assumptions satisfy ho.
 *
 * Agar assumptions fail ho jaye:
 *
 * → Coefficients unreliable ho sakte hain
 * → p-values wrong ho sakti hain
 * → Confidence intervals invalid ho sakte hain
 * → Inference galat ho sakta hai
 *
 * =========================================================
 * ! 5 ASSUMPTIONS OF LINEAR REGRESSION
 * =========================================================
 *
 * 1. Linearity
 *
 * 2. Normality of Residuals
 *
 * 3. Homoscedasticity
 *
 * 4. No Autocorrelation
 *
 * 5. No or Little Multicollinearity
 *
 * Ye 5 assumptions Linear Regression ki backbone hain.
 *
 * =========================================================
 * ! WHAT IS A RESIDUAL?
 * =========================================================
 *
 * Definition:
 *
 * Residual is the difference between
 * actual value and predicted value.
 *
 * Formula:
 *
 * Residual = Actual - Predicted
 *
 * Residual = y - ŷ
 *
 * Example:
 *
 * Actual Salary = 10 LPA
 *
 * Predicted Salary = 8 LPA
 *
 * Residual:
 *
 * = 10 - 8
 *
 * = 2 LPA
 *
 * Residual ko Error Term bhi bol dete hain.
 *
 * =========================================================
 * ! 1. LINEARITY ASSUMPTION
 * =========================================================
 *
 * (Page 3-4)
 *
 * ---------------------------------------------------------
 * ! ASSUMPTION
 * ---------------------------------------------------------
 *
 * Independent variables aur dependent variable
 * ke beech linear relationship hona chahiye.
 *
 * Simple Meaning:
 *
 * X increase ho
 * toh Y proportional manner mein change ho.
 *
 * Example:
 *
 * Salary = β0 + β1(Experience)
 *
 * Experience increase ho
 * toh Salary linearly increase honi chahiye.
 *
 * Graph:
 *
 *          /
 *         /
 *        /
 *       /
 *      /
 *
 * Straight line relationship.
 *
 * =========================================================
 * ! WHAT IS LINEAR RELATIONSHIP?
 * =========================================================
 *
 * Definition:
 *
 * Relationship that can be represented
 * by a straight line.
 *
 * Example:
 *
 * y = β0 + β1x
 *
 * Here:
 *
 * β0
 * =
 * Intercept
 *
 * β1
 * =
 * Slope
 *
 * =========================================================
 * ! WHAT HAPPENS IF ASSUMPTION FAILS?
 * =========================================================
 *
 * 1. Biased Coefficients
 *
 * ---------------------------------------------------------
 *
 * Biased Estimate:
 *
 * Estimated coefficient
 * actual relationship ko properly represent
 * nahi karta.
 *
 * Example:
 *
 * Actual relation:
 *
 * y = x²
 *
 * But hum fit kar rahe:
 *
 * y = β0 + β1x
 *
 * Coefficients wrong milenge.
 *
 * ---------------------------------------------------------
 *
 * 2. Poor Prediction
 *
 * Model actual pattern learn nahi karega.
 *
 * Underfitting ho sakta hai.
 *
 * ---------------------------------------------------------
 *
 * ! UNDERFITTING
 * ---------------------------------------------------------
 *
 * Definition:
 *
 * Model data ka actual pattern
 * capture nahi kar pata.
 *
 * Training performance bhi poor.
 *
 * Test performance bhi poor.
 *
 * ---------------------------------------------------------
 *
 * 3. Invalid Hypothesis Testing
 *
 * t-test wrong
 *
 * F-test wrong
 *
 * Confidence intervals wrong
 *
 * =========================================================
 * ! HOW TO CHECK LINEARITY?
 * =========================================================
 *
 * ---------------------------------------------------------
 * 1. Scatter Plot
 * ---------------------------------------------------------
 *
 * X vs Y plot banao.
 *
 * Agar straight trend dikh raha hai
 * toh linearity likely satisfied.
 *
 * Example:
 *
 * Experience ---> Salary
 *
 * Scatter points line follow kare.
 *
 * ---------------------------------------------------------
 * 2. Residual Plot
 * ---------------------------------------------------------
 *
 * Plot:
 *
 * Residuals (y - ŷ)
 *
 * vs
 *
 * Predicted Values (ŷ)
 *
 * If assumption holds:
 *
 * Residuals randomly scatter hone chahiye.
 *
 * Example:
 *
 *      *     *      *
 *   *     *      *
 *      *      *
 * --------------------
 *          0
 *
 * Random pattern.
 *
 * ---------------------------------------------------------
 *
 * If curve dikhe:
 *
 *     *
 *   *
 * *
 *      *
 *         *
 *
 * Then:
 *
 * Non-linearity present.
 *
 * =========================================================
 * ! WHAT TO DO IF LINEARITY FAILS?
 * =========================================================
 *
 * ---------------------------------------------------------
 * 1. Transformation
 * ---------------------------------------------------------
 *
 * Variable transform karo.
 *
 * Common transformations:
 *
 * log(x)
 *
 * √x
 *
 * 1/x
 *
 * log(y)
 *
 * ---------------------------------------------------------
 * ! TRANSFORMATION
 * ---------------------------------------------------------
 *
 * Definition:
 *
 * Mathematical operation applied
 * on feature to change its distribution
 * or relationship.
 *
 * ---------------------------------------------------------
 * 2. Polynomial Regression
 * ---------------------------------------------------------
 *
 * Instead of:
 *
 * y = β0 + β1x
 *
 * Use:
 *
 * y = β0 + β1x + β2x²
 *
 * OR
 *
 * y = β0 + β1x + β2x² + β3x³
 *
 * PDF page 4 exactly introduces
 * polynomial feature idea.
 *
 * Example:
 *
 * x = 2
 *
 * Features become:
 *
 * x⁰ = 1
 * x¹ = 2
 * x² = 4
 *
 * New model:
 *
 * y =
 * β0
 * +
 * β1x⁰
 * +
 * β2x¹
 * +
 * β3x²
 *
 * ---------------------------------------------------------
 * ! POLYNOMIAL REGRESSION
 * ---------------------------------------------------------
 *
 * Definition:
 *
 * Regression technique which introduces
 * higher powers of features
 * to capture non-linear relationships.
 *
 * ---------------------------------------------------------
 * 3. Piecewise Regression
 * ---------------------------------------------------------
 *
 * Different regions ke liye
 * different linear models.
 *
 * ---------------------------------------------------------
 * ! PIECEWISE REGRESSION
 * ---------------------------------------------------------
 *
 * Definition:
 *
 * Data ko multiple segments mein divide karke
 * har segment par separate regression fit karna.
 *
 * ---------------------------------------------------------
 * 4. GAM
 * ---------------------------------------------------------
 *
 * Generalized Additive Model
 *
 * Non-linear relationship capture karta hai.
 *
 * ---------------------------------------------------------
 * ! GAM
 * ---------------------------------------------------------
 *
 * Full Form:
 *
 * Generalized Additive Model
 *
 * Definition:
 *
 * Flexible regression model
 * jo non-linear patterns ko learn kar sakta hai
 * without strict linear assumption.
 *
 * =========================================================
 * ! KEY TAKEAWAYS
 * =========================================================
 *
 * Linearity means:
 *
 * X aur Y ke beech linear relationship.
 *
 * Residual plot random hona chahiye.
 *
 * Curve pattern
 * =>
 * linearity violation.
 *
 * Fixes:
 *
 * → Transformations
 * → Polynomial Regression
 * → Piecewise Regression
 * → GAM
 *
 * =========================================================
 */

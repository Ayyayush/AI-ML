/**
 * ------------------------------------------------------------
 * ! Polynomial Regression
 * ------------------------------------------------------------
 * Polynomial Regression ek regression algorithm hai
 * jo:
 * Non-linear relationships ko model karne ke liye use hota hai.
 * 
 * Naam:
 * Polynomial Regression
 * 
 * Lekin:
 * Ye technically Linear Regression hi hai.
 * 
 * Kyunki:
 * Parameters (weights) linear hi rehte hain.
 * 
 * ------------------------------------------------------------
 * ! Why Polynomial Regression Needed ?
 * ------------------------------------------------------------
 * Simple Linear Regression:
 * 
 * y = mx + b
 * 
 * Sirf straight line fit kar sakta hai.
 * 
 * Problem:
 * Real world data bahut baar linear nahi hota.
 * 
 * Example:
 * - House Prices
 * - Population Growth
 * - Sales Forecasting
 * - Temperature Trends
 * 
 * In cases mein:
 * Data curved pattern follow karta hai.
 * 
 * ------------------------------------------------------------
 * ! Example
 * ------------------------------------------------------------
 * Maan lo:
 * 
 * Study Hours     Marks
 * 1               10
 * 2               25
 * 3               45
 * 4               70
 * 5               90
 * 
 * Is data ko dekh kar lag raha hai:
 * Marks constant rate se increase nahi ho rahe.
 * 
 * Yahan:
 * Straight line achha fit nahi hogi.
 * 
 * Hume:
 * Curve chahiye.
 * 
 * ------------------------------------------------------------
 * ! Polynomial Equation
 * ------------------------------------------------------------
 * Degree 1:
 * 
 * y = b0 + b1x
 * 
 * Ye:
 * Simple Linear Regression hai.
 * 
 * ------------------------------------------------------------
 * ! Degree 2
 * ------------------------------------------------------------
 * y = b0 + b1x + b2x²
 * 
 * Isme:
 * x² add ho gaya.
 * 
 * Graph:
 * Curve ban jayega.
 * 
 * ------------------------------------------------------------
 * ! Degree 3
 * ------------------------------------------------------------
 * y = b0 + b1x + b2x² + b3x³
 * 
 * Aur zyada flexible curve milega.
 * 
 * ------------------------------------------------------------
 * ! General Formula
 * ------------------------------------------------------------
 * y = b0 + b1x + b2x² + b3x³ + ... + bnxⁿ
 * 
 * Yahan:
 * n:
 * Degree of Polynomial
 * 
 * ------------------------------------------------------------
 * ! Important Interview Point
 * ------------------------------------------------------------
 * Polynomial Regression ko:
 * Non-linear Regression samajhte hain.
 * 
 * But:
 * Mathematically:
 * Ye Linear Regression hi hai.
 * 
 * Reason:
 * Parameters:
 * b0,b1,b2,b3...
 * 
 * linear form mein present hain.
 * 
 * Model:
 * Parameters ke respect mein linear hai.
 * 
 * ------------------------------------------------------------
 * ! Feature Transformation
 * ------------------------------------------------------------
 * Maan lo:
 * 
 * X = [2]
 * 
 * Degree = 3
 * 
 * Polynomial Features:
 * 
 * [1, x, x², x³]
 * 
 * [1, 2, 4, 8]
 * 
 * Ab:
 * Linear Regression isi transformed data par lagta hai.
 * 
 * ------------------------------------------------------------
 * ! Training Process
 * ------------------------------------------------------------
 * Step 1:
 * Polynomial Features create karo.
 * 
 * Step 2:
 * Linear Regression apply karo.
 * 
 * Step 3:
 * Best coefficients find karo.
 * 
 * Step 4:
 * Prediction karo.
 * 
 * ------------------------------------------------------------
 * ! Advantages
 * ------------------------------------------------------------
 * 1. Non-linear data ko model kar sakta hai.
 * 
 * 2. Curved relationships learn kar sakta hai.
 * 
 * 3. Simple Linear Regression se better fit de sakta hai.
 * 
 * ------------------------------------------------------------
 * ! Disadvantages
 * ------------------------------------------------------------
 * 1. High degree polynomial:
 * Overfitting kar sakta hai.
 * 
 * 2. Degree select karna difficult hota hai.
 * 
 * 3. Outliers se affect ho sakta hai.
 * 
 * 4. Degree bahut high ho jaye toh
 * model unstable ho sakta hai.
 * 
 * ------------------------------------------------------------
 * ! Underfitting vs Overfitting
 * ------------------------------------------------------------
 * Degree bahut low:
 * Underfitting
 * 
 * Model pattern nahi seekh paata.
 * 
 * Example:
 * Curved data par straight line fit kar dena.
 * 
 * ------------------------------------------------------------
 * ! Overfitting
 * ------------------------------------------------------------
 * Degree bahut high:
 * Overfitting
 * 
 * Model noise bhi learn kar leta hai.
 * 
 * Training performance:
 * Bahut achhi
 * 
 * Testing performance:
 * Poor
 * 
 * ------------------------------------------------------------
 * ! Choosing Degree
 * ------------------------------------------------------------
 * Commonly:
 * 
 * Degree:
 * 2
 * 3
 * 4
 * 
 * use ki jati hai.
 * 
 * Degree selection ke liye:
 * 
 * - Cross Validation
 * - Validation Set
 * - Adjusted R²
 * 
 * use kar sakte hain.
 * 
 * ------------------------------------------------------------
 * ! Evaluation Metrics
 * ------------------------------------------------------------
 * Polynomial Regression ko evaluate karne ke liye:
 * 
 * - MAE
 * - MSE
 * - RMSE
 * - R² Score
 * - Adjusted R² Score
 * 
 * use karte hain.
 * 
 * ------------------------------------------------------------
 * ! Real Life Examples
 * ------------------------------------------------------------
 * - Population Growth
 * - Stock Trends
 * - Sales Forecasting
 * - Temperature Prediction
 * - Growth Curves
 * - Medical Data Analysis
 * 
 * ------------------------------------------------------------
 * ! Interview Definition
 * ------------------------------------------------------------
 * Polynomial Regression ek regression technique hai
 * jo polynomial features use karke
 * non-linear relationships ko model karti hai,
 * while still using Linear Regression internally.
 * 
 */

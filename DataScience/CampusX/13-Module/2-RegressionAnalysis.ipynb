/**
 * =========================================================
 * ! F-STATISTIC
 * =========================================================
 * F-Statistic is one of the most important values
 * present in Regression Summary.
 *
 * Interviewers frequently ask:
 *
 * Q. Why do we need F-statistic?
 *
 * Q. What does F-statistic tell us?
 *
 * Q. Difference between F-test and t-test?
 *
 * =========================================================
 * ! PROBLEM BEFORE F-STATISTIC
 * =========================================================
 *
 * Suppose we have:
 *
 * Package = β0 + β1(CGPA) + β2(IQ)
 *
 * After training:
 *
 * β1 = 3.5
 * β2 = 1.2
 *
 * Now question:
 *
 * Are these coefficients actually useful?
 *
 * OR
 *
 * Are these values obtained just by chance?
 *
 * We need a statistical test
 * to verify this.
 *
 * Therefore:
 *
 * F-test is used.
 * 
 * 
 * 
 *  
 * =========================================================
 * ! ANOVA (ANALYSIS OF VARIANCE)
 * =========================================================
 *
 * ANOVA:
 * Analysis Of Variance
 *
 * One of the most important concepts
 * in Statistics and Machine Learning.
 *
 * =========================================================
 * ! WHY DO WE NEED ANOVA?
 * =========================================================
 *
 * Suppose:
 *
 * We have 3 teaching methods.
 *
 * Method A
 * Method B
 * Method C
 *
 * After exam:
 *
 * A -> Average Marks = 75
 * B -> Average Marks = 80
 * C -> Average Marks = 90
 *
 * Question:
 *
 * Is the difference real?
 *
 * OR
 *
 * Did it happen just by chance?
 *
 * ANOVA helps answer this question.
 *
 * 
 * 
 * =========================================================
 * ! DEFINITION
 * =========================================================
 *
 * ANOVA:
 * statistical technique used to compare
 * variances and determine whether
 * group means are significantly different.
 *
 * Simple Meaning:
 *
 * ANOVA checks whether different groups
 * actually differ from each other.
 *
 * =========================================================
 * ! INTUITION
 * =========================================================
 *
 * Suppose:
 *
 * Group A Mean = 80
 * Group B Mean = 81
 * Group C Mean = 79
 *
 * Means are very close.
 *
 * Probably:
 *
 * No significant difference.
 *
 * ---------------------------------------------------------
 *
 * Group A Mean = 50
 * Group B Mean = 80
 * Group C Mean = 95
 *
 * Means are far apart.
 *
 * Probably:
 *
 * Significant difference exists.
 *
 * 
 * 
 * =========================================================
 * ! F-STATISTIC COMES FROM ANOVA
 * =========================================================
 * F =
 * Between Group Variance
 * ----------------------
 * Within Group Variance
 *
 * This is the core idea of ANOVA.
 * ---------------------------------------------------------
 * Large F:
 * Group means are very different.
 *
 * ---------------------------------------------------------
 *
 * Small F:
 * Group means are almost same.
 *
 * 
 * 
 * =========================================================
 * ! ANOVA IN REGRESSION
 * =========================================================
 *
 * Regression also uses ANOVA concept.
 *
 * We already studied:
 *
 * TSS = ESS + RSS
 *
 * where:
 *
 * TSS:
 * Total Variation
 *
 * ESS:
 * Explained Variation
 *
 * RSS:
 * Unexplained Variation
 *
 * ---------------------------------------------------------
 *
 * F-statistic in Regression:
 *
 *          ESS/df_model
 * F =
 *          ------------
 *          RSS/df_residual
 *
 * 
 * =========================================================
 * ! CONNECTION BETWEEN ANOVA AND F-STATISTIC
 * =========================================================
 *
 * Interview Question:
 *
 * Q. Where does F-statistic come from?
 *
 * Answer:
 *
 * F-statistic comes from ANOVA.
 *
 * ANOVA compares explained variation
 * and unexplained variation.
 *
 * Regression F-test is actually
 * based on ANOVA.
 *
 * 
 * 
 * =========================================================
 * ! INTERVIEW QUESTION
 * =========================================================
 *
 * Q. Why not use multiple t-tests instead of ANOVA?
 *
 * Answer:
 *
 * Suppose:
 *
 * 5 groups exist.
 *
 * Multiple t-tests:
 *
 * A vs B
 * A vs C
 * A vs D
 * Too many tests.
 *
 * Error probability increases.
 * ANOVA performs one overall test.
 * Therefore:
 * ANOVA is preferred.
 *
 * 
 * =========================================================
 * ! VERY IMPORTANT INTERVIEW ANSWER
 * =========================================================
 *
 * Q. What is ANOVA?
 *
 * Answer:
 *
 * ANOVA (Analysis of Variance) is a statistical
 * technique used to determine whether there are
 * significant differences between the means of
 * multiple groups.
 *
 * It works by comparing:
 *
 * Between Group Variance
 * with
 * Within Group Variance
 *
 * and uses F-statistic to determine significance.
 *
 * 
 * =========================================================
 * ! ML INTERVIEW CONNECTION
 * =========================================================
 *
 * ANOVA is important because:
 *
 * → F-statistic comes from ANOVA
 * → Regression F-test uses ANOVA
 * → Feature Selection uses ANOVA
 * → Statistical Inference uses ANOVA
 *
 * Therefore:
 *
 * ANOVA is one of the foundations
 * of Regression Analysis.
 * 
 * 
 *
 * =========================================================
 * ! F-TEST
 * =========================================================
 * Definition:  
 * F-Test is a statistical test used to check
 * whether the regression model as a whole
 * is significant or not.
 *
 * Simple Meaning:
 * Does the model actually learn something?
 * OR
 * Is the model better than random guessing?
 *
 * =========================================================
 * ! NULL HYPOTHESIS (H0)
 * =========================================================
 *
 * Hypothesis:
 * statement we want to test.
 *
 * Null Hypothesis:
 *
 * H0:
 *
 * β1 = β2 = β3 = .... = βn = 0
 *
 * Meaning:
 *
 * None of the independent variables
 * have any effect on output.
 *
 * Model is useless.
 *
 * Example:
 *
 * Package prediction:
 *
 * CGPA effect = 0
 * IQ effect   = 0
 *
 * Then:
 *
 * Package does not depend on them.
 *
 * =========================================================
 * ! ALTERNATIVE HYPOTHESIS (H1)
 * =========================================================
 *
 * H1:
 *
 * At least one coefficient
 * is not equal to zero.
 *
 * Meaning:
 *
 * At least one feature contributes
 * towards prediction.
 *
 * Model is useful.
 *
 * =========================================================
 * ! WHY F-TEST?
 * =========================================================
 *
 * Suppose:
 *
 * 50 features present.
 *
 * Using t-test individually:
 *
 * feature1
 * feature2
 * feature3
 * ...
 * feature50
 *
 * Very expensive.
 *
 * F-test checks entire model
 * in one shot.
 *
 * =========================================================
 * ! IDEA BEHIND F-STATISTIC
 * =========================================================
 *
 * We already know:
 *
 * TSS = ESS + RSS
 *
 * where:
 *
 * TSS:
 * Total Sum of Squares
 *
 * ESS:
 * Explained Sum of Squares
 *
 * RSS:
 * Residual Sum of Squares
 *
 * ---------------------------------------------------------
 * GOOD MODEL
 * ---------------------------------------------------------
 *
 * ESS should be large
 *
 * RSS should be small
 *
 * ---------------------------------------------------------
 * BAD MODEL
 * ---------------------------------------------------------
 *
 * ESS small
 *
 * RSS large
 *
 * =========================================================
 * ! F-STATISTIC FORMULA
 * =========================================================
 *
 *                    ESS / df_model
 * F =
 *                  -----------------
 *                  RSS / df_residual
 *
 * where:
 *
 * df_model
 * =
 * model degree of freedom
 *
 * df_residual
 * =
 * residual degree of freedom
 *
 * =========================================================
 * ! INTERPRETATION
 * =========================================================
 * Numerator:
 * Explained variation per degree of freedom
 *
 * Denominator:
 * Unexplained variation per degree of freedom
 *
 * Therefore: *
 * F-statistic compares:
 * Explained Variation
 * -------------------
 * Unexplained Variation
 *
 * =========================================================
 * ! WHEN F-STATISTIC IS HIGH?
 * =========================================================
 *
 * ESS large
 *
 * RSS small
 *
 * Therefore:
 *
 * F-statistic becomes large.
 *
 * Meaning:
 *
 * Model explains data well.
 *
 * =========================================================
 * ! WHEN F-STATISTIC IS LOW?
 * =========================================================
 *
 * ESS small
 *
 * RSS large
 *
 * Therefore:
 *
 * F-statistic becomes small.
 *
 * Meaning:
 *
 * Model is poor.
 *
 * =========================================================
 * ! IMPORTANT INTERVIEW POINT
 * =========================================================
 *
 * Larger F-statistic
 *
 * generally indicates
 *
 * better regression model.
 *
 * But:
 *
 * Final decision is made
 * using p-value.
 *
 * =========================================================
 * ! P-VALUE
 * =========================================================
 *
 * Definition:
 *
 * Probability of obtaining observed result
 * assuming Null Hypothesis is true.
 *
 * Simple Meaning:
 *
 * Probability that results occurred
 * purely by chance.
 *
 * =========================================================
 * ! P-VALUE OF F-STATISTIC
 * =========================================================
 *
 * In statsmodels output:
 *
 * Prob(F-statistic)
 *
 * is shown.
 *
 * This is the p-value corresponding
 * to F-statistic.
 *
 * =========================================================
 * ! DECISION RULE
 * =========================================================
 *
 * If:
 *
 * p-value < 0.05
 *
 * Reject H0
 *
 * Meaning:
 *
 * Model is statistically significant.
 *
 * ---------------------------------------------------------
 *
 * If:
 *
 * p-value > 0.05
 *
 * Fail to Reject H0
 *
 * Meaning:
 *
 * Model is not statistically significant.
 *
 * =========================================================
 * ! WHY 0.05?
 * =========================================================
 *
 * 0.05
 *
 * =
 *
 * 5% significance level
 *
 * Means:
 *
 * We are willing to accept
 * only 5% chance of being wrong.
 *
 * =========================================================
 * ! SIGNIFICANCE LEVEL (α)
 * =========================================================
 *
 * Definition:
 *
 * Maximum acceptable probability
 * of making wrong decision.
 *
 * Represented by:
 *
 * α (alpha)
 *
 * Usually:
 *
 * α = 0.05
 *
 * =========================================================
 * ! EXAMPLE
 * =========================================================
 *
 * F-statistic = 120
 *
 * Prob(F-statistic)
 *
 * = 0.000001
 *
 * Since:
 *
 * 0.000001 < 0.05
 *
 * Reject H0
 *
 * Therefore:
 *
 * Model is statistically significant.
 *
 * =========================================================
 * ! INTERVIEW QUESTION
 * =========================================================
 *
 * Q. Can a model have high F-statistic
 * and still be bad?
 *
 * Answer:
 *
 * F-statistic alone is not enough.
 *
 * We must also check:
 *
 * → p-value
 * → R²
 * → Adjusted R²
 * → residual analysis
 *
 * =========================================================
 * ! DIFFERENCE BETWEEN F-TEST AND T-TEST
 * =========================================================
 *
 * F-Test:
 *
 * checks entire model.
 *
 * H0:
 *
 * β1 = β2 = β3 = ... = 0
 *
 * ---------------------------------------------------------
 *
 * T-Test:
 *
 * checks individual coefficient.
 *
 * Example:
 *
 * H0:
 *
 * β1 = 0
 *
 * Only CGPA feature tested.
 *
 * =========================================================
 * ! INTERVIEW ANSWER
 * =========================================================
 *
 * Q. What does F-statistic tell us?
 *
 * Answer:
 *
 * F-statistic tells whether the regression
 * model as a whole is statistically significant.
 *
 * It compares explained variation
 * with unexplained variation.
 *
 * Higher F-statistic and lower p-value
 * indicate a better model.
 *
 * =========================================================
 * ! KEY TAKEAWAYS
 * =========================================================
 *
 * F-test:
 * checks overall model significance
 *
 * H0:
 * all coefficients = 0
 *
 * H1:
 * at least one coefficient ≠ 0
 *
 * Formula:
 *
 * F =
 *
 * (ESS/df_model)
 * --------------
 * (RSS/df_residual)
 *
 * High F-statistic:
 * good sign
 *
 * Prob(F-statistic):
 * p-value of F-test
 *
 * p-value < 0.05
 * =>
 * statistically significant model
 *
 * F-test:
 * entire model
 *
 * t-test:
 * individual feature
 *
 * =========================================================
 */

/**
 * =========================================================
 * ! R-SQUARED (R² SCORE)
 * =========================================================
 *
 * (Page 13)
 *
 * After building a regression model,
 * next question is:
 *
 * "How good is my model?"
 *
 * R² helps answer this.
 *
 * ---------------------------------------------------------
 * ! R-SQUARED
 * ---------------------------------------------------------
 *
 * R²:
 * Coefficient of Determination
 *
 * Definition:
 *
 * R² measures the proportion of variation
 * in dependent variable that is explained
 * by independent variables.
 *
 * Simple Meaning:
 *
 * R² tells:
 *
 * "Kitna % data model explain kar paaya?"
 *
 * =========================================================
 * ! FORMULA
 * =========================================================
 *
 * We already know:
 *
 * TSS = ESS + RSS
 *
 * Therefore:
 *
 * R² = ESS/TSS
 *
 * OR
 *
 * R² = 1 - (RSS/TSS)
 *
 * =========================================================
 * ! INTUITION
 * =========================================================
 *
 * TSS:
 * Total Variation
 *
 * ESS:
 * Explained Variation
 *
 * Therefore:
 *
 * R² =
 *
 * Explained Variation
 * -------------------
 * Total Variation
 *
 * =========================================================
 * ! INTERPRETATION
 * =========================================================
 *
 * R² = 0
 *
 * Model explains nothing.
 *
 * ---------------------------------------------------------
 *
 * R² = 1
 *
 * Model explains everything.
 *
 * Perfect fit.
 *
 * ---------------------------------------------------------
 *
 * R² = 0.80
 *
 * Means:
 *
 * 80% variation explained.
 *
 * 20% variation unexplained.
 *
 * =========================================================
 * ! EXAMPLE
 * =========================================================
 *
 * Suppose:
 *
 * TSS = 100
 *
 * ESS = 80
 *
 * Then:
 *
 * R² = 80/100
 *
 * = 0.8
 *
 * = 80%
 *
 * Meaning:
 *
 * Model explains 80% variation.
 *
 * =========================================================
 * ! PROBLEM WITH R²
 * =========================================================
 *
 * Very important interview point.
 *
 * R² never decreases
 * when new features are added.
 *
 * Example:
 *
 * Feature1 -> useful
 * Feature2 -> useful
 * Feature3 -> useless
 * Feature4 -> useless
 *
 * Even after adding useless features:
 *
 * R² either:
 *
 * → increases
 * OR
 * → remains same
 *
 * It never decreases.
 *
 * Therefore:
 *
 * R² can be misleading.
 *
 * =========================================================
 * ! OVERFITTING CONNECTION
 * =========================================================
 *
 * Model keeps adding features.
 *
 * R² keeps increasing.
 *
 * But:
 *
 * Prediction quality may not improve.
 *
 * This can lead to:
 *
 * Overfitting.
 *
 * ---------------------------------------------------------
 * ! OVERFITTING
 * ---------------------------------------------------------
 *
 * Definition:
 *
 * Model starts learning noise
 * instead of actual pattern.
 *
 * Training accuracy high.
 *
 * Test accuracy poor.
 *
 * =========================================================
 * ! ADJUSTED R-SQUARED
 * =========================================================
 *
 * (Page 14)
 *
 * Problem:
 *
 * R² always increases.
 *
 * Solution:
 *
 * Adjusted R²
 *
 * =========================================================
 * ! ADJUSTED R²
 * =========================================================
 *
 * Definition:
 *
 * Modified version of R²
 * that penalizes unnecessary features.
 *
 * It considers:
 *
 * → number of observations
 * → number of predictors
 *
 * Therefore:
 *
 * Better metric for multiple regression.
 *
 * =========================================================
 * ! FORMULA
 * =========================================================
 *
 * Adjusted R² =
 *
 * 1 - [(1-R²)(n-1)]/(n-k-1)
 *
 * where:
 *
 * n:
 * number of observations
 *
 * k:
 * number of independent variables
 *
 * =========================================================
 * ! IMPORTANT PROPERTY
 * =========================================================
 *
 * If useful feature added:
 *
 * Adjusted R² increases.
 *
 * ---------------------------------------------------------
 *
 * If useless feature added:
 *
 * Adjusted R² decreases.
 *
 * =========================================================
 * ! WHY ADJUSTED R² IS BETTER?
 * =========================================================
 *
 * R²:
 *
 * rewards every new feature.
 *
 * ---------------------------------------------------------
 *
 * Adjusted R²:
 *
 * rewards only useful features.
 *
 * Penalizes model complexity.
 *
 * =========================================================
 * ! MODEL COMPLEXITY
 * =========================================================
 *
 * Definition:
 *
 * Complexity increases
 * as number of predictors increases.
 *
 * More features
 * =>
 * more complex model
 *
 * =========================================================
 * ! INTERVIEW QUESTION
 * =========================================================
 *
 * Q. Which is better?
 * R² or Adjusted R²?
 *
 * Answer:
 *
 * For Multiple Linear Regression:
 *
 * Adjusted R²
 *
 * Because it penalizes
 * unnecessary predictors.
 *
 * =========================================================
 * ! WHEN TO USE R²?
 * =========================================================
 *
 * Use R² when:
 *
 * understanding explained variance
 * is the primary goal.
 *
 * =========================================================
 * ! WHEN TO USE ADJUSTED R²?
 * =========================================================
 *
 * Use Adjusted R² when:
 *
 * comparing multiple models
 * having different number of features.
 *
 * =========================================================
 * ! PAGE 15
 * =========================================================
 *
 * WHICH ONE SHOULD BE USED?
 *
 * PDF Conclusion:
 *
 * For model comparison:
 *
 * Use Adjusted R²
 *
 * Because:
 *
 * it considers model complexity.
 *
 * ---------------------------------------------------------
 *
 * For understanding explanatory power:
 *
 * Use R²
 *
 * =========================================================
 * ! T-STATISTIC
 * =========================================================
 *
 * (Page 16)
 *
 * F-test:
 * checks entire model.
 *
 * T-test:
 * checks individual coefficient.
 *
 * =========================================================
 * ! WHY T-TEST?
 * =========================================================
 *
 * Suppose:
 *
 * Package =
 *
 * β0
 * +
 * β1(CGPA)
 * +
 * β2(IQ)
 *
 * We want to know:
 *
 * Is CGPA important?
 *
 * Is IQ important?
 *
 * For this:
 *
 * T-test is used.
 *
 * =========================================================
 * ! NULL HYPOTHESIS
 * =========================================================
 *
 * For coefficient β1:
 *
 * H0:
 *
 * β1 = 0
 *
 * Meaning:
 *
 * Feature has no effect.
 *
 * ---------------------------------------------------------
 *
 * H1:
 *
 * β1 ≠ 0
 *
 * Meaning:
 *
 * Feature affects output.
 *
 * =========================================================
 * ! T-STATISTIC
 * =========================================================
 *
 * Definition:
 *
 * Measures how many standard errors
 * coefficient is away from zero.
 *
 * Formula:
 *
 * t =
 *
 * coefficient
 * -----------
 * standard error
 *
 * =========================================================
 * ! STANDARD ERROR
 * =========================================================
 *
 * Definition:
 *
 * Measures uncertainty
 * in coefficient estimate.
 *
 * Small Standard Error:
 *
 * coefficient estimate reliable.
 *
 * Large Standard Error:
 *
 * coefficient estimate unstable.
 *
 * =========================================================
 * ! INTERPRETATION
 * =========================================================
 *
 * Large |t|
 *
 * =>
 *
 * strong evidence
 * against null hypothesis.
 *
 * ---------------------------------------------------------
 *
 * Small |t|
 *
 * =>
 *
 * weak evidence
 * against null hypothesis.
 *
 * =========================================================
 * ! P-VALUE USING T-TEST
 * =========================================================
 *
 * After calculating t-statistic,
 * corresponding p-value is obtained.
 *
 * Decision:
 *
 * p-value < 0.05
 *
 * Reject H0
 *
 * Feature is significant.
 *
 * ---------------------------------------------------------
 *
 * p-value > 0.05
 *
 * Fail to reject H0
 *
 * Feature is insignificant.
 *
 * =========================================================
 * ! INTERVIEW QUESTION
 * =========================================================
 *
 * Q. Difference between F-test and T-test?
 *
 * Answer:
 *
 * F-test:
 * entire model
 *
 * T-test:
 * individual coefficient
 *
 * =========================================================
 * ! CONFIDENCE INTERVALS
 * =========================================================
 *
 * (Pages 17-18)
 *
 * Instead of giving a single value,
 * statistics often provides a range.
 *
 * This range is called:
 *
 * Confidence Interval.
 *
 * =========================================================
 * ! CONFIDENCE INTERVAL
 * =========================================================
 *
 * Definition:
 *
 * Range within which true population
 * parameter is expected to lie
 * with certain confidence.
 *
 * Example:
 *
 * β1 = 5
 *
 * 95% Confidence Interval:
 *
 * [4.2 , 5.8]
 *
 * Meaning:
 *
 * We are 95% confident that
 * true coefficient lies
 * between 4.2 and 5.8.
 *
 * =========================================================
 * ! CONFIDENCE LEVEL
 * =========================================================
 *
 * Definition:
 *
 * Probability that interval estimation
 * procedure captures true parameter.
 *
 * Common values:
 *
 * 90%
 * 95%
 * 99%
 *
 * =========================================================
 * ! DEGREES OF FREEDOM
 * =========================================================
 *
 * For simple linear regression:
 *
 * df = n - 2
 *
 * Why 2?
 *
 * Because:
 *
 * β0 estimated
 * β1 estimated
 *
 * Two parameters.
 *
 * =========================================================
 * ! INTERVIEW INTERPRETATION
 * =========================================================
 *
 * If confidence interval contains 0:
 *
 * coefficient may be insignificant.
 *
 * Example:
 *
 * [-2 , 3]
 *
 * contains 0
 *
 * Not strong evidence.
 *
 * ---------------------------------------------------------
 *
 * If confidence interval does not contain 0:
 *
 * coefficient likely significant.
 *
 * Example:
 *
 * [2 , 7]
 *
 * does not contain 0.
 *
 * Strong evidence.
 *
 * =========================================================
 * ! KEY TAKEAWAYS
 * =========================================================
 *
 * R²:
 * explained variation percentage
 *
 * R² = ESS/TSS
 *
 * Problem:
 * never decreases
 *
 * ---------------------------------------------------------
 *
 * Adjusted R²:
 * penalizes unnecessary features
 *
 * Better for model comparison.
 *
 * ---------------------------------------------------------
 *
 * T-test:
 * significance of individual feature
 *
 * H0:
 * β = 0
 *
 * ---------------------------------------------------------
 *
 * Confidence Interval:
 * range containing true parameter
 *
 * Most common:
 * 95% confidence interval
 *
 * =========================================================
 */

## 📌 What is Data Gathering? (Definition)

**Data Gathering** ka matlab hota hai  
👉 **different sources se raw data collect karna** jisse hum us data par analysis kar sakein.

📌 Ye **Data Analysis Process ka first aur most critical step** hota hai  
kyunki agar data hi galat hoga → analysis bhi galat hoga.

---

## 🔄 Complete Data Analysis Process (High-Level Flow)

1. Asking Better Questions  
2. Understanding the Domain (SME)  
3. Data Gathering (Collection)  
4. Data Wrangling (Cleaning & Preparing)  
5. EDA – Exploratory Data Analysis  
6. Drawing Conclusions  
7. Communicating Results  

---

## 1️⃣ Asking Better Questions (Most Important Step)

Analysis start karne se pehle **right questions** poochna zaroori hota hai.

### ❓ What kind of questions?
- Problem kya hai?
- Business / real-life goal kya hai?
- Kya predict karna hai ya kya understand karna hai?
- Output kis format mein chahiye?

📌 **Rule:**  
> *Bad questions → Wrong data → Wrong analysis*

---

# 📂 Importing Data from CSV Files (Easy Hinglish)

CSV files se data import karna **data analysis ka first practical step** hota hai.  
Python mein sabse commonly hum **`read_csv()`** function use karte hain.

---

## 🧠 Basic Idea
CSV = **Comma Separated Values**  
Har row = ek record  
Har column = ek feature

---

## 📌 Common Parameters Used While Importing CSV Files

### 1️⃣ `filepath_or_buffer`
👉 CSV file ka **path ya filename**

**Example:**  
`sales.csv`  
`data/sales.csv`

📌 Ye **mandatory** parameter hai.

---

### 2️⃣ `sep` (separator)
👉 Columns ko separate karne wala character

**Common values:**
- `,` → comma (default)
- `;` → semicolon
- `\t` → tab

📌 Use jab file comma-separated na ho.

---

### 3️⃣ `header`
👉 Batata hai header row kahan hai

**Values:**
- `0` → first row header hai (default)
- `None` → header nahi hai

📌 Agar header na ho, to columns auto-numbered ho jaate hain.

---

### 4️⃣ `names`
👉 Custom column names dene ke liye

📌 Tab use hota hai jab:
- File mein header nahi ho
- Ya aap apna header chahte ho

---

### 5️⃣ `index_col`
👉 Kaun-sa column index banana hai

📌 Example:
- `id`
- `roll_no`

Useful jab ek unique column ho.

---

### 6️⃣ `usecols`
👉 Sirf **selected columns** read karne ke liye

📌 Benefit:
- Memory kam use hoti hai
- Faster loading

---

### 7️⃣ `dtype`
👉 Column ka **data type define** karna

📌 Example:
- `int`
- `float`
- `string`

Useful jab auto-detection galat ho.

---

### 8️⃣ `na_values`
👉 Missing values ko define karna

📌 Example:
- `"NA"`
- `"null"`
- `"-"`

In sabko missing value treat karega.

---

### 9️⃣ `skiprows`
👉 Starting ki rows skip karne ke liye

📌 Example:
- `skiprows=2` → pehli 2 rows skip

Useful jab upar metadata likha ho.

---

### 🔟 `nrows`
👉 Sirf limited rows read karni ho

📌 Use when:
- File bahut badi ho
- Quick testing karni ho

---

### 1️⃣1️⃣ `encoding`
👉 File ka text encoding batata hai

**Common values:**
- `utf-8` (most common)
- `latin1`

📌 Error aaye to encoding check karo.

---

### 1️⃣2️⃣ `parse_dates`
👉 Date columns ko date format mein read karna

📌 ML & time-series ke liye important.

---

## 🔑 When Things Go Wrong (Real-Life Tips)

- Column galat aa rahe → `sep` check karo  
- Header issue → `header` / `names` use karo  
- Data type galat → `dtype` define karo  
- Missing values odd lag rahe → `na_values` set karo  

---

sahi parameters use karke data ko correctly samajhna hai.**

# 4. Sep Parameter

```python
import pandas as pd

# ------------------------------------------------------------
# Reading a TSV (Tab Separated Values) file using pandas
# ------------------------------------------------------------

df = pd.read_csv(
    'movie_titles_metadata.tsv',                 # file name / file path of the TSV file
    sep='\t',                                    # separator is TAB (\t), not comma
    names=[                                     # custom column names for the dataset
        'sno',                                  # serial number / index-like column
        'name',                                 # movie name / title
        'release_year',                         # year in which movie was released
        'rating',                               # movie rating
        'votes',                                # number of votes received
        'genres'                                # movie genres
    ]
)

# ------------------------------------------------------------
# Quick checks after loading the data
# ------------------------------------------------------------

print(df.head())                                # shows first 5 rows to verify data
print(df.info())                                # shows column types & missing values
```

# 5. Index_col parameter

```python
import pandas as pd

# ------------------------------------------------------------
# Reading CSV file and setting a specific column as index
# ------------------------------------------------------------

df = pd.read_csv(
    'aug_train.csv',                            # CSV file containing training data
    index_col='enrollee_id'                     # 'enrollee_id' column is set as DataFrame index
                                                # useful when this column is unique
)

# ------------------------------------------------------------
# Verifying the result
# ------------------------------------------------------------

print(df.head())                                # displays first 5 rows
print(df.index)                                 # confirms 'enrollee_id' is now the index
```

# 6. Header parameter

```python
import pandas as pd

# ------------------------------------------------------------
# Reading CSV file with a custom header row
# ------------------------------------------------------------

df = pd.read_csv(
    'test.csv',                                # CSV file name / path
    header=1                                   # row index 1 (second row) is treated as column names
                                                # first row (index 0) is ignored as header
)

# ------------------------------------------------------------
# Verifying the imported data
# ------------------------------------------------------------

print(df.head())                               # shows first 5 rows
print(df.columns)                              # confirms correct column names
```

# 7. use_cols parameter

```python
# ------------------------------------------------------------
# Reading only selected columns from CSV using usecols
# ------------------------------------------------------------

df_selected = pd.read_csv(
    'aug_train.csv',                           # CSV file path
    usecols=['enrollee_id',                    # selecting only required columns
             'gender',
             'education_level']
)
```

# 8. Squeeze parameters

```python

# ------------------------------------------------------------
# Reading a single column as a Series using squeeze
# ------------------------------------------------------------

gender_series = pd.read_csv(
    'aug_train.csv',                           # same CSV file
    usecols=['gender'],                        # selecting only one column
    squeeze=True                               # converts DataFrame to Series (single column)
)
```

# 9. Skiprows/nrows Parameter

```python
# ------------------------------------------------------------
# Reading only first 100 rows from the CSV file
# ------------------------------------------------------------

df_limited = pd.read_csv(
    'aug_train.csv',                           # CSV file path
    nrows=100                                  # reads only first 100 rows (useful for large files)
)
```

# 10. Encoding parameter

```python
# ------------------------------------------------------------
# Reading CSV file with specific text encoding
# ------------------------------------------------------------

df = pd.read_csv(
    'zomato.csv',                              # CSV file path
    encoding='latin-1'                         # using latin-1 encoding to avoid Unicode errors
                                                # common for files with special characters
)
```

# 11. Skip bad lines

```python
# ------------------------------------------------------------
# Reading CSV file with custom separator, encoding,
# and handling bad / corrupted lines
# ------------------------------------------------------------

df = pd.read_csv(
    'BX-Books.csv',                            # CSV file path
    sep=';',                                  # semicolon is used as column separator
    encoding='latin-1',                       # latin-1 encoding for special characters
    error_bad_lines=False                     # skips rows with parsing errors (bad lines)
                                                # prevents crash due to corrupted rows
)
```

# 12. dtypes parameter

```python
# ------------------------------------------------------------
# Reading CSV file and explicitly defining data type for a column
# ------------------------------------------------------------

pd.read_csv(
    'aug_train.csv',                          # CSV file path
    dtype={                                   # enforcing data types manually
        'target': int                         # converting 'target' column to integer type
                                                # useful when pandas auto-detects wrong dtype
    }
).info()                                     # displays DataFrame structure, dtypes & memory usage
```

# 13. Handling Dates

```python
# ------------------------------------------------------------
# Reading CSV file and parsing date column into datetime format
# ------------------------------------------------------------

pd.read_csv(
    'IPL Matches 2008-2020.csv',               # CSV file containing IPL match data
    parse_dates=['date']                      # converts 'date' column to datetime dtype
                                                # useful for time-based analysis
).info()                                      # displays column info and data types
```

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

# 14. Convertors

```python
# ------------------------------------------------------------
# Reading CSV file and applying a custom function to a column
# using the converters parameter
# ------------------------------------------------------------

pd.read_csv(
    'IPL Matches 2008-2020.csv',               # CSV file containing IPL match data
    converters={
        'team1': rename                       # applying 'rename' function on each value of 'team1'
                                                # useful for standardizing / cleaning team names
    }
)

# NOTE:
# - 'rename' should be a user-defined function
# - It will be applied row-by-row on the 'team1' column
# - converters are used when we want custom transformation during import itself
```

# 15. na_values parameter

```python
# ------------------------------------------------------------
# Reading CSV file and treating specific values as missing (NaN)
# ------------------------------------------------------------

pd.read_csv(
    'aug_train.csv',                          # CSV file path
    na_values=['Male']                        # treating 'Male' as a missing value (NaN)
                                                # useful when certain values represent missing data
)

# NOTE:
# - Any occurrence of 'Male' in the dataset will be converted to NaN
# - Helpful for cleaning inconsistent or wrongly entered data during import itself
```

```python
# ============================================================
# Importing Excel Files using pandas
# Demonstrating commonly used parameters ONE BY ONE
# Dummy Excel file assumed: output.xlsx
# ============================================================

import pandas as pd


# ------------------------------------------------------------
# 1️⃣ Basic Excel Import
# ------------------------------------------------------------
df_basic = pd.read_excel(
    'output.xlsx'                             # Excel file name / path
)
# By default, reads the first sheet of the Excel file


# ------------------------------------------------------------
# 2️⃣ Reading a specific sheet (sheet_name)
# ------------------------------------------------------------
df_sheet = pd.read_excel(
    'output.xlsx',
    sheet_name='Sheet1'                       # reads data from Sheet1
)
# sheet_name can be sheet name or index (0, 1, 2...)


# ------------------------------------------------------------
# 3️⃣ Reading multiple sheets at once
# ------------------------------------------------------------
df_multiple = pd.read_excel(
    'output.xlsx',
    sheet_name=['Sheet1', 'Sheet2']           # returns a dictionary of DataFrames
)
# Useful when data is spread across sheets


# ------------------------------------------------------------
# 4️⃣ Skipping initial rows (skiprows)
# ------------------------------------------------------------
df_skip = pd.read_excel(
    'output.xlsx',
    skiprows=2                                # skips first 2 rows (metadata / notes)
)


# ------------------------------------------------------------
# 5️⃣ Setting a custom header row (header)
# ------------------------------------------------------------
df_header = pd.read_excel(
    'output.xlsx',
    header=1                                  # second row becomes column names
)

```

```python
# ------------------------------------------------------------
# Creating a connection to MySQL database using mysql.connector
# ------------------------------------------------------------

conn = mysql.connector.connect(
    host='localhost',                         # MySQL server address (localhost = same machine)
    user='root',                              # MySQL username
    password='',                              # MySQL password (empty here)
    database='world'                          # database name to connect with
)

# NOTE:
# - If connection is successful, 'conn' becomes a live DB connection object
# - This connection is later used to execute SQL queries using cursor()
# - Always close the connection after use: conn.close()
```

```python
# ------------------------------------------------------------
# Reading data from MySQL table into a pandas DataFrame
# ------------------------------------------------------------

df = pd.read_sql_query(
    "SELECT * FROM countrylanguage",           # SQL query to fetch all records
    conn                                       # active MySQL database connection object
)

# NOTE:
# - read_sql_query executes the SQL query on the database
# - Result is directly stored as a pandas DataFrame
# - Very useful for analysis without manually handling cursors
```

# Exporting diff kind of files in pandas


```python
# ============================================================
# Exporting Different Kinds of Files in Pandas
# All formats demonstrated in ONE place with comments
# ============================================================


# ------------------------------------------------------------
# 1️⃣ Exporting DataFrame to CSV file
# ------------------------------------------------------------
df.to_csv(
    'output.csv',                 # output file name
    index=False                   # do not save DataFrame index
)
# CSV is lightweight and widely supported


# ------------------------------------------------------------
# 2️⃣ Exporting DataFrame to Excel file
# ------------------------------------------------------------
df.to_excel(
    'output.xlsx',                # output Excel file
    index=False                   # excluding index column
)
# Excel is useful for business reports


# ------------------------------------------------------------
# 3️⃣ Exporting DataFrame to HTML file
# ------------------------------------------------------------
df.to_html(
    'output.html',                # output HTML file
    index=False                   # hides index in HTML table
)
# HTML is useful for web display


# ------------------------------------------------------------
# 4️⃣ Exporting DataFrame to JSON file
# ------------------------------------------------------------
df.to_json(
    'output.json',                # output JSON file
    orient='records',             # JSON as list of dictionaries
    indent=4                      # pretty-printed JSON
)
# JSON is commonly used in APIs


# ------------------------------------------------------------
# 5️⃣ Exporting DataFrame to SQL table
# ------------------------------------------------------------
df.to_sql(
    'exported_table',             # table name in database
    con=conn,                     # active database connection
    if_exists='replace',          # replace table if it already exists
    index=False                   # do not store index as column
)
# SQL export is useful for long-term storage
# ------------------------------------------------------------


# ✅ Best Practice:
# - Always verify exported files
# - For SQL, confirm table structure after export
```

# Web Scraping

## Definition
Web scraping is the process of **automatically extracting data from websites** when the data is not directly available through an API or downloadable format.  
It involves sending HTTP requests to a webpage, parsing the HTML content, and collecting the required information in a structured form such as tables, CSV files, or databases.

---

## When Web Scraping Is Used
Web scraping is used **when a website does not provide data via APIs** or official data export options.

Example scenarios:
- No public API is available
- Data is only visible on web pages
- Manual data collection is time-consuming
- Large-scale data extraction is required

---

## Why Web Scraping Is Needed
- To collect publicly available data
- To automate repetitive data collection tasks
- To build datasets for analysis or machine learning
- To monitor prices, news, or content changes

---

## Basic Steps in Web Scraping
1. Send a request to the website URL  
2. Receive the HTML response  
3. Parse the HTML using a parser (BeautifulSoup, lxml, etc.)  
4. Extract required data using tags, classes, or IDs  
5. Store the data in a structured format (CSV, DataFrame, Database)

---

## Common Tools Used for Web Scraping
- `requests` – for sending HTTP requests  
- `BeautifulSoup` – for parsing HTML  
- `Selenium` – for dynamic websites (JavaScript rendered)  
- `Scrapy` – for large-scale scraping projects  

---

## Important Notes (Interview Point)
- Web scraping should respect **robots.txt**
- Avoid scraping private or restricted data
- Follow website **terms and conditions**
- Use delays to prevent server overload

---

## One-Line Interview Answer
> Web scraping is used to extract data from websites when APIs are unavailable by programmatically parsing and collecting information from HTML pages.

# Handling 403 Forbidden Error in Web Scraping

## What is a 403 Error?
A **403 Forbidden** error occurs when a server **refuses to allow access** to a webpage.  
This usually happens when a website detects that the request is coming from a **bot or script** instead of a real browser.

---

## Why 403 Error Occurs in Web Scraping
Common reasons include:
- Missing or invalid `User-Agent`
- Website blocks automated requests
- Anti-bot protection enabled
- Too many requests in a short time
- Scraping restricted by website policies

---

## Solution: Send Browser-like Headers
To avoid a 403 error, we can send **HTTP headers** that mimic a real browser request.

---

## Example Code

```python
# ------------------------------------------------------------
# Handling 403 Forbidden Error using custom headers
# ------------------------------------------------------------

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 '
                  '(KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'
}

response = requests.get(                      # send GET request with headers
    'url',                                     # target webpage URL
    headers=headers                            # browser-like request headers
)

html = response.text                           # extract HTML content

# creating dataframe for all the pages

````python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np

# ------------------------------------------------------------
# ! Create empty DataFrame to store final result
# usage: store all pages data in one place
# ------------------------------------------------------------
final = pd.DataFrame()

# ------------------------------------------------------------
# ! Loop through multiple pages
# usage: pagination scraping
# ------------------------------------------------------------
for j in range(1, 1001):

    # --------------------------------------------------------
    # ! Send request and get HTML
    # usage: fetch webpage data
    # --------------------------------------------------------
    webpage = requests.get(
        'https://www.ambitionbox.com/list-of-companies?page={}'.format(j)
    ).text

    # --------------------------------------------------------
    # ! Parse HTML using BeautifulSoup
    # usage: convert HTML → searchable object
    # --------------------------------------------------------
    soup = BeautifulSoup(webpage, 'lxml')

    # --------------------------------------------------------
    # ! Get all company containers
    # usage: each block contains one company data
    # --------------------------------------------------------
    company = soup.find_all('div', class_='company-content-wrapper')

    # --------------------------------------------------------
    # ! Initialize lists to store data
    # usage: store extracted values column-wise
    # --------------------------------------------------------
    name = []
    rating = []
    reviews = []
    ctype = []
    hq = []
    how_old = []
    no_of_employee = []

    # --------------------------------------------------------
    # ! Loop through each company block
    # usage: extract data from each container
    # --------------------------------------------------------
    for i in company:

        # ------------------- NAME -------------------
        try:
            name.append(i.find('h2').text.strip())      # usage: company name
        except:
            name.append(np.nan)                         # usage: handle missing data

        # ------------------- RATING -------------------
        try:
            rating.append(i.find('p', class_='rating').text.strip())   # usage: rating
        except:
            rating.append(np.nan)

        # ------------------- REVIEWS -------------------
        try:
            reviews.append(i.find('a', class_='review-count').text.strip())  # usage: review count
        except:
            reviews.append(np.nan)

        # ------------------- COMPANY TYPE -------------------
        try:
            ctype.append(i.find_all('p', class_='infoEntity')[0].text.strip())  # usage: company type
        except:
            ctype.append(np.nan)

        # ------------------- HEADQUARTERS -------------------
        try:
            hq.append(i.find_all('p', class_='infoEntity')[1].text.strip())     # usage: location
        except:
            hq.append(np.nan)

        # ------------------- COMPANY AGE -------------------
        try:
            how_old.append(i.find_all('p', class_='infoEntity')[2].text.strip())  # usage: age
        except:
            how_old.append(np.nan)

        # ------------------- NUMBER OF EMPLOYEES -------------------
        try:
            no_of_employee.append(i.find_all('p', class_='infoEntity')[3].text.strip())  # usage: size
        except:
            no_of_employee.append(np.nan)

    # --------------------------------------------------------
    # ! Convert lists to DataFrame
    # usage: structured tabular data
    # --------------------------------------------------------
    df = pd.DataFrame({
        'name': name,
        'rating': rating,
        'reviews': reviews,
        'company_type': ctype,
        'Head_Quarters': hq,
        'Company_Age': how_old,
        'No_of_Employee': no_of_employee
    })

    # --------------------------------------------------------
    # ! Append page data to final DataFrame
    # usage: combine all pages
    # --------------------------------------------------------
    final = final.append(df, ignore_index=True)
````

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# ------------------------------------------------------------
# ! Step 1: Create empty DataFrame
# ------------------------------------------------------------

final = pd.DataFrame()

# ------------------------------------------------------------
# ! Step 2: Loop through pages (1 to 5 for practice)
# ------------------------------------------------------------

for page in range(1, 6):

    print(f"Scraping Page {page}...")

    # --------------------------------------------------------
    # ! Step 3: Send request
    # --------------------------------------------------------

    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url, timeout=5)

    # --------------------------------------------------------
    # ! Step 4: Parse HTML
    # --------------------------------------------------------

    soup = BeautifulSoup(response.text, 'lxml')

    # --------------------------------------------------------
    # ! Step 5: Find all quote containers
    # --------------------------------------------------------

    quotes = soup.find_all('div', class_='quote')

    # --------------------------------------------------------
    # ! Step 6: Create lists
    # --------------------------------------------------------

    text_list = []
    author_list = []
    tags_list = []

    # --------------------------------------------------------
    # ! Step 7: Extract data
    # --------------------------------------------------------

    for q in quotes:

        try:
            text_list.append(q.find('span', class_='text').text.strip())
        except:
            text_list.append(None)

        try:
            author_list.append(q.find('small', class_='author').text.strip())
        except:
            author_list.append(None)

        try:
            tags = q.find_all('a', class_='tag')
            tags_list.append([t.text for t in tags])   # list of tags
        except:
            tags_list.append(None)

    # --------------------------------------------------------
    # ! Step 8: Convert to DataFrame
    # --------------------------------------------------------

    df = pd.DataFrame({
        'quote': text_list,
        'author': author_list,
        'tags': tags_list
    })

    # --------------------------------------------------------
    # ! Step 9: Append to final DataFrame
    # --------------------------------------------------------

    final = final.append(df, ignore_index=True)

    time.sleep(1)   # ! avoid blocking

# ------------------------------------------------------------
# ! Step 10: Show result
# ------------------------------------------------------------

print(final.head())

# ------------------------------------------------------------
# ! Step 11: Save to CSV
# ------------------------------------------------------------

final.to_csv("quotes.csv", index=False)
````

```python
# -------------------------------------------------------------
# Series from Lists in Pandas
# -------------------------------------------------------------

import numpy as np
import pandas as pd

# Ek Python list create ki gayi jisme country names hain
country = ['India', 'Pakistan', 'USA', 'Nepal', 'SriLanka']

# Pandas Series banane ke liye pd.Series() ka use kiya jaata hai
# Agar hum directly list ko Series mein convert kar dete hain
# toh Pandas har element ke liye ek unique index assign kar deta hai
# Default index numeric hota hai: 0, 1, 2, 3, ...
# Index ko hum item ka address bhi bol sakte hain
# Output mein hume do cheezein milti hain:
# 1) Index values
# 2) Actual data (list items)
# Niche dtype: object bhi dikhega jo yeh batata hai ki data type string/object hai

series_country = pd.Series(country)

print(series_country)
```

```python
# -------------------------------------------------------------
# Series from Integers Data
# -------------------------------------------------------------

import pandas as pd

# Cricket runs ka ek simple Python list
# Iss list ke andar saare elements integers hain
runs = [13, 24, 56, 78, 100]

# List ko Pandas Series format mein convert kar rahe hain
# pd.Series() default indexing assign karta hai: 0, 1, 2, 3, 4
# Output mein hume do components milte hain:
# 1) Index: Har element ke liye ek positional label
# 2) Values: Actual integer data jo list mein tha
# Neeche dtype: int64 dikhega kyunki saare values integers hain
series_runs = pd.Series(runs)

print(series_runs)
```

```python
# -------------------------------------------------------------
# Custom Index in Pandas Series
# -------------------------------------------------------------

import pandas as pd

# Marks ka ek list create kiya gaya (integer values)
marks = [67, 57, 89, 100]

# Subjects ka ek list create kiya gaya (string values)
# Inhi subjects ko hum custom index ke roop mein assign karenge
subjects = ['maths', 'english', 'science', 'hindi']

# pd.Series() ke andar 'index' parameter use karke
# hum apna custom index provide kar sakte hain
# Iss case mein har marks value ko uske subject ke saath map kiya jayega
series_marks = pd.Series(marks, index=subjects)

print(series_marks)
```

```python
# -------------------------------------------------------------
# Setting a Name for a Pandas Series
# -------------------------------------------------------------

import pandas as pd

marks = [67, 57, 89, 100]
subjects = ['maths', 'english', 'science', 'hindi']

# Pandas Series create karte waqt 'name' parameter use karke
# Series ko ek meaningful label de sakte hain
series_marks = pd.Series(marks, index=subjects, name='Ayush ke Marks')

print(series_marks)
```

```python
# -------------------------------------------------------------
# Series from Dictionary in Pandas
# -------------------------------------------------------------

import pandas as pd

# Python dictionary create kiya gaya jisme:
# Key  Subject ka naam denote karta hai (index ban jaayega)
# Value  Marks denote karti hain (Series ke data ban jaayenge)
# Dictionary ka structure already index-value mapping jaisa hota hai
marks = {
    'maths': 67,
    'english': 57,
    'science': 89,
    'hindi': 100
}

# Dictionary ko directly Pandas Series mein convert kar rahe hain
# Is case mein keys automatically index ban jaati hain
# Aur unke respective values Series data ban jaate hain
marks_series = pd.Series(marks)

print(marks_series)
```

```python
# -------------------------------------------------------------
# Pandas Series Attributes with Explanation
# -------------------------------------------------------------

import pandas as pd

# Example Series
marks = {
    'maths': 67,
    'english': 57,
    'science': 89,
    'hindi': 100
}
marks_series = pd.Series(marks, name='Ayush ke Marks')

# -------------------------------------------------------------
# size
# -------------------------------------------------------------
# Total number of elements in the Series.
print("Size:", marks_series.size)

# -------------------------------------------------------------
# dtype
# -------------------------------------------------------------
# Data type of values stored inside the Series (here integers).
print("Data Type:", marks_series.dtype)

# -------------------------------------------------------------
# name
# -------------------------------------------------------------
# Name of the Series, works as a label or title for the data.
print("Series Name:", marks_series.name)

# -------------------------------------------------------------
# is_unique
# -------------------------------------------------------------
# Checks if all values in the Series are unique (True/False).
print("Is Unique:", marks_series.is_unique)

# -------------------------------------------------------------
# index
# -------------------------------------------------------------
# Shows the index labels assigned to each value.
print("Index:", marks_series.index)

# -------------------------------------------------------------
# values
# -------------------------------------------------------------
# Returns the actual data of the Series as a NumPy array.
print("Values:", marks_series.values)
```

# Pandas CSV: Reading CSV Files and Creating Series

## Introduction
CSV (Comma Separated Values) files are widely used for storing tabular data.  
They are similar to Excel files but much simpler:  

- Each row represents a record.  
- Each column is separated by a comma `,`.  
- CSV files can be opened in Excel, Notepad, or any text editor.  

Pandas provides the `read_csv()` function to read CSV files into **DataFrames** or **Series**.

---

## Characteristics of CSV

- **Simple text format**: Easy to read and write.  
- **Lightweight**: Can handle moderately large datasets.  
- **Row & Column structure**: Rows represent records, columns represent features.  
- **Comma-separated**: Each column value is separated by a comma (`,`).  
- **Portable**: Can be used across Excel, Python, R, SQL, etc.  
- **Flexible**: Can handle optional headers, indices, and selective columns.

---



## How CSV Extracts Data from Files

- Pandas reads the CSV file **line by line**.  
- Each line is split by the **delimiter** (default is comma `,`).  
- First row is treated as **header** by default (column names).  
- Remaining rows are stored as **data values**.  
- Pandas creates a **DataFrame** with rows as index and columns as header names.  
- Optional parameters allow:  
  - Selecting certain columns (`usecols`)  
  - Setting a custom index (`index_col`)  
  - Converting single-column DataFrame to Series (`squeeze=True`)

```python
# ------------------------------------------------------------
# 📘 Example Dataset
# ------------------------------------------------------------
# Maan lo humare paas ek "subs.csv" file hai
# jisme subscribers / categories ka data diya gaya hai.
# ------------------------------------------------------------

import pandas as pd

# CSV file read karna (make sure subs.csv file content folder mein ho)
subs_df = pd.read_csv("content/subs.csv")

# Single column ko Series me convert karna
subs = subs_df.iloc[:, 0]

# ------------------------------------------------------------
# 🔹 value_counts()
# ------------------------------------------------------------
"""
value_counts() ek Pandas Series ka function hai
jo har unique value ki frequency (kitni baar aayi hai) batata hai.

👉 Ye function data analysis ke liye kaafi useful hai.
Agar manually karna ho to loop chala kar count nikalna padta,
par Pandas me builtin function se ye kaam ek line me ho jata hai.

Use Case:
Agar hume pata lagana ho ki kaunsa subscriber/category
sabse zyada repeat hui hai.
"""

# ------------------------------------------------------------
# 🔸 Example:
# ------------------------------------------------------------

print("🔸 Frequency of each value:")
print(subs.value_counts())


# ------------------------------------------------------------
# 🔸 Agar hume frequency ke percentage chahiye hon to:
print("\n🔸 Frequency of each value (in percentage):")
print(subs.value_counts(normalize=True) * 100)


# ------------------------------------------------------------
# ✅ Summary:
# value_counts() → har unique value ka count return karta hai.
# value_counts(normalize=True) → percentage return karta hai.
# Data analysis me ye sabse common frequency check karne ka tool hai.
# ------------------------------------------------------------
```

```python
# ============================================================
# 🔢 Pandas Series Method → sort_values(inplace)
# ============================================================
# Ab hum dekh rahe hain sort_values() method ka use Series ke andar.
# Ye poori Series ko ascending ya descending order me sort karta hai.
# ============================================================

import pandas as pd

# ------------------------------------------------------------
# 📘 Example Dataset
# ------------------------------------------------------------
# Maan lo humare paas ek CSV file hai "subs.csv"
# jisme ek single column me subscribers ka data diya gaya hai.
# Is CSV ko hum Series me convert karke sort karenge.
# ------------------------------------------------------------

# CSV file read karna (default DataFrame)
subs_df = pd.read_csv("content/subs.csv")

# Single column ko Series me convert karna
subs = subs_df.iloc[:, 0]

# ------------------------------------------------------------
# 🔹 sort_values()
# ------------------------------------------------------------
"""
sort_values() Series ke elements ko sort karne ke liye use hota hai.

👉 By default ye ascending (chhote se bade) order me sort karta hai.
👉 Agar descending order chahiye ho to 'ascending=False' karna hota hai.
👉 Agar inplace=True likho to original Series modify ho jati hai.

⚙️ Syntax:
    series.sort_values(ascending=True/False, inplace=True/False)

📘 Parameters:
- ascending → True (ascending order) / False (descending order)
- inplace → True karne par original Series update ho jati hai
"""

# ------------------------------------------------------------
# 🔸 Example 1: Ascending order me sort karna
print("🔸 Ascending order me sorted Series:")
print(subs.sort_values())


# ------------------------------------------------------------
# 🔸 Example 2: Descending order me sort karna
print("\n🔸 Descending order me sorted Series:")
print(subs.sort_values(ascending=False))


# ------------------------------------------------------------
# 🔸 Example 3: Inplace sorting (original Series modify ho jati hai)
print("\n🔸 Inplace sorting (Original Series modified):")
subs = subs_df.iloc[:, 0].copy()
subs.sort_values(inplace=True)
print(subs)



# ------------------------------------------------------------
# 🔸 Example 4: Method Chaining
# ------------------------------------------------------------
"""
Pandas me method chaining ka matlab hai multiple operations ek line me karna.
Agar hume sorted Series ka sirf top 5 values chahiye,
to hum sort_values() ke saath head() ko chain kar sakte hain.
"""

# Method chaining example → sorted Series ke top 5 values
print("\n🔸 Smallest 5 values using method chaining:")
print(subs.sort_values().head(5))


# ------------------------------------------------------------
# ✅ Summary:
# sort_values() → Series ke elements ko sort karta hai.
# ascending=False → descending order.
# inplace=True → original Series modify hoti hai.
# Method chaining → ek hi line me multiple operations.
# ------------------------------------------------------------
```

```python
# Series Maths Methods
"""
Ab hum sabse jyada use hone wale Series ke mathematical methods padhnege.
Ye methods data ke analysis mein kaafi helpful hote hain.

🟢 Difference between count() and size:
    - count() missing values (NaN) ko ignore karta hai.
    - size total elements batata hai, chahe wo NaN ho ya valid value.

Example:
Agar Series = [10, 20, NaN, 40]
    count() = 3   (NaN ko ignore karega)
    size   = 4   (sabko count karega, including NaN)
"""

# Example Series:
import pandas as pd

numbers = pd.Series([10, 20, None, 40, 50])

# count()
print("Count:", numbers.count())                 # NaN values ko ignore karke count karega

# size
print("Size:", numbers.size)                     # Total number of elements including NaN

# sum()
print("Sum:", numbers.sum())                     # NaN ko ignore karke sum karega (10 + 20 + 40 + 50 = 120)

# mean()
print("Mean:", numbers.mean())                   # Average nikaalega (NaN ignore hoga)

# median()
print("Median:", numbers.median())               # Middle value return karega (after sorting valid numbers)

# mode()
print("Mode:", numbers.mode().values)            # Most frequently occurring value return karega

# min() and max()
print("Min:", numbers.min())                     # Smallest value
print("Max:", numbers.max())                     # Largest value

# std() → Standard Deviation
print("Standard Deviation:", numbers.std())      # Spread of data batata hai

# var() → Variance
print("Variance:", numbers.var())                # Kitna data mean se deviated hai

# describe() → Summary of all statistical info
print("\nSummary using describe():")
print(numbers.describe())                        # Count, Mean, Std, Min, Max sab ek hi bar mein mil jaata hai
```

### Indexing

```python
# Series Indexing

import pandas as pd

# Integer Indexing
"""
Pandas Series mein bhi hum list ki tarah index ke through elements access kar sakte hain.
Positive indexing yahan kaam karti hai, lekin negative indexing default Series ke liye kaam nahi karti.

👉 Important:
    - Positive index → valid
    - Negative index → by default error deta hai (kyunki pandas apne index label ke hisaab se kaam karta hai)
"""

# Example Series
x = pd.Series([10, 20, 30, 40, 50])
print("Original Series:")
print(x)

# Positive Indexing
print("\nPositive Indexing:")
print("x[0] =", x[0])                       # 0th index ka element (10)
print("x[2] =", x[2])                       # 2nd index ka element (30)

# Negative Indexing (Default behavior)
print("\nNegative Indexing (Default):")
# print(x[-1])                              # ❌ Ye error dega: "KeyError: -1"

"""
Agar hume negative indexing jaisa behavior chahiye 
toh hum iloc (integer location) ka use kar sakte hain.
iloc list-style integer indexing ko allow karta hai.
"""

# Correct way using iloc
print("\nUsing iloc for Negative Indexing:")
print("x.iloc[-1] =", x.iloc[-1])           # Last element (50)
print("x.iloc[-2] =", x.iloc[-2])           # Second last element (40)



# ----------------------------------------------------------------
# Custom Indexing (Overwriting Default Index)
"""
Agar humne pandas Series ke index ko overwrite kiya hai,
toh original positional index internally pandas ke paas rehta hai,
aur hum apne diye gaye index label ke through bhi access kar sakte hain.

🧠 Note:
    - Agar custom index string type hai → negative indexing iloc ke saath kaam karega.
    - Agar custom index number type hai → negative indexing error dega (kyunki -1 ko ek label samjhega).
"""

# Example 1 → Custom Index with Strings
y = pd.Series([100, 200, 300, 400], index=['a', 'b', 'c', 'd'])
print("\nCustom Indexed Series (String Index):")
print(y)

print("\nAccess using custom label:")
print("y['a'] =", y['a'])                   # label-based access
print("y['c'] =", y['c'])

print("\nAccess using iloc (position-based):")
print("y.iloc[-1] =", y.iloc[-1])           # works fine (last element)
print("y.iloc[-2] =", y.iloc[-2])           # second last element


# Example 2 → Custom Index with Numbers
z = pd.Series([11, 22, 33, 44], index=[10, 20, 30, 40])
print("\nCustom Indexed Series (Numeric Index):")
print(z)

print("\nAccess using custom label:")
print("z[10] =", z[10])                     # works using label 10

print("\nNegative Indexing on Numeric Index:")
# print(z[-1])                              # ❌ Error – kyunki -1 ko label samjhta hai
print("z.iloc[-1] =", z.iloc[-1])           # ✅ Correct way to get last element
```

```python
# Fancy Indexing

import pandas as pd

"""
Fancy Indexing ka matlab hota hai ek saath multiple indexes (ya labels) ko access karna.
Matlab agar hume Series ke multiple specific elements ek hi baar chahiye,
toh hum list ke form mein indexes pass karte hain.

👉 Important:
    - Fancy indexing Python list ki tarah hi kaam karta hai.
    - Hum index numbers ya custom index labels dono ka use kar sakte hain.
    - Output ek naya Series return karta hai.
"""

# Example 1 → Default Numeric Index
scores = pd.Series([45, 67, 89, 23, 56, 78, 91, 34, 66, 80])
print("Original Series:")
print(scores)

# Fancy Indexing using integer positions
print("\nFancy Indexing (using integer positions):")
print(scores[[1, 3, 5]])                 # Index 1, 3, 5 ke elements ek saath print honge

# Example 2 → Custom String Index
players = pd.Series(
    [55, 70, 45, 60, 90, 75],
    index=["Virat", "Rohit", "Gill", "Dhoni", "Hardik", "Surya"]
)
print("\nCustom Indexed Series (Players Scores):")
print(players)

# Fancy Indexing using labels
print("\nFancy Indexing (using labels):")
print(players[["Virat", "Dhoni", "Hardik"]])    # Multiple custom indexes ek saath access

# Repetition bhi allowed hai
print("\nRepetition Allowed in Fancy Indexing:")
print(players[["Rohit", "Rohit", "Gill"]])      # Same index multiple times print kar sakte hain

# Fancy Indexing with non-contiguous selection
print("\nNon-contiguous Index Selection:")
print(players[["Surya", "Virat", "Gill"]])      # Random order mein elements access kar sakte hain


"""
🧠 Notes:
    - Fancy indexing ek *list* ya *array* of indexes/labels leta hai.
    - Result ek naya Series hota hai (original Series modify nahi hoti).
    - Non-contiguous aur duplicate access dono possible hain.
"""
```

```python
# Boolean Indexing on Pandas Series

import pandas as pd

# Example 1 → Virat Kohli Scores
kohli_scores = pd.Series(
    [45, 100, 75, 50, 150, 60, 100, 20, 50, 120],
    index=[f"Match{i+1}" for i in range(10)]
)
print("Kohli Scores Series:")
print(kohli_scores)

# ------------------------------------------------------------
# 1️⃣ Find number of 50's and 100's scored by Kohli
# ------------------------------------------------------------
fifties = (kohli_scores == 50).sum()
hundreds = (kohli_scores == 100).sum()
print("\nNumber of 50's scored:", fifties)
print("Number of 100's scored:", hundreds)

# ------------------------------------------------------------
# 2️⃣ Find number of ducks (score = 0)
# ------------------------------------------------------------
ducks = (kohli_scores == 0).sum()
print("\nNumber of ducks scored:", ducks)

# ------------------------------------------------------------
# 3️⃣ Count number of days when he scored more than 100
# ------------------------------------------------------------
above_hundred = (kohli_scores > 100).sum()
print("\nNumber of days Kohli scored more than 100:", above_hundred)






# ------------------------------------------------------------
# Example 2 → Actors & Number of Movies as Lead Role
# ------------------------------------------------------------
actors_movies = pd.Series(
    [25, 60, 45, 70, 35, 55],
    index=["Actor1", "Actor2", "Actor3", "Actor4", "Actor5", "Actor6"]
)
print("\nActors Movies Series:")
print(actors_movies)



# 4️⃣ Find actors who have done more than 50 movies as lead
actors_above_50 = actors_movies[actors_movies > 50]
print("\nActors who have done more than 50 movies as lead role:")
print(actors_above_50)

"""
🧠 Notes:
1. Boolean Indexing → logical conditions ke basis pe Series ke subset select karna.
2. Example: series[series > 50] → elements > 50 select honge.
3. Counting → sum() method se True values count kar sakte hain.
4. Useful for filtering, analysis, and quick summary statistics.
"""
```

```python
# Slicing

import pandas as pd

"""
Pandas Series mein bhi slicing kar sakte hain —
iska matlab hota hai multiple indexes (ya range of elements) ko ek saath select karna.

🧠 Example:
Agar Virat Kohli ke match scores ki Series hai,
aur hume match 5 se 16 tak ke runs dekhne hain,
toh hum slicing ka use karenge.

👉 Important:
    - Slicing range-based hoti hai (start:end)
    - Negative slicing bhi kaam karti hai.
    - Agar hume last ke 5 elements dekhne hain, toh hum [-5:] likh sakte hain.
    - Agar alternate elements chahiye toh step value bhi de sakte hain (start:end:step)
"""

# Example Series
scores = pd.Series([45, 67, 89, 23, 56, 78, 91, 34, 66, 80, 55, 92, 49, 72, 60, 83, 77, 90])
print("Virat Kohli ke Match Scores:")
print(scores)

# Basic Slicing
print("\nSlicing Example (Match 5 to 10):")
print(scores[4:10])                       # Index 4 se lekar 9 tak ke elements print honge

# Slicing from start
print("\nSlicing from Start (First 5 Matches):")
print(scores[:5])                         # Pehle 5 matches ke scores

# Slicing till end
print("\nSlicing till End (From Match 10 onwards):")
print(scores[9:])                         # 10th match se aage ke saare scores

# Negative Slicing
print("\nNegative Slicing (Last 5 Matches):")
print(scores[-5:])                        # Last 5 elements print karega

# Slicing with Step
print("\nSlicing with Step (Alternate Matches):")
print(scores[::2])                        # Har dusre match ka score (alternate elements)

# Reverse Slicing
print("\nReverse Slicing:")
print(scores[::-1])                       # Pure Series ko reverse order mein print karega


"""
🧾 Notes:
    - scores[start:end] → start se end-1 tak elements return karta hai.
    - Negative slicing (-5:) → last ke 5 elements deta hai.
    - [::2] → step = 2, har alternate element deta hai.
    - [::-1] → reverse order mein pura Series print karta hai.
"""
```

### Editing Series

```python
# Editing Series

import pandas as pd

"""
Hum discuss karenge ki Pandas Series ke elements ko kaise edit kar sakte hain.

Example:
Agar English ke marks mein 50 hai aur usko 100 karna hai, toh hum Series ke index ke through update kar sakte hain.
"""

# Example Series
marks_series = pd.Series(
    [50, 60, 70, 80],
    index=['Math', 'English', 'Science', 'History']
)
print("Original Series:")
print(marks_series)

# ------------------------------------------------------------
# 1️⃣ Using Indexing
# ------------------------------------------------------------
# Update a value using its index
marks_series['English'] = 100
print("\nAfter updating English marks:")
print(marks_series)

# ------------------------------------------------------------
# What if an index does not exist?
# ------------------------------------------------------------
"""
Agar hum aise index ko edit karne ki koshish kare jo exist nahi karta, 
toh Pandas usko naya item add kar deta hai.
Example: 'Geography' index add karna
"""
marks_series['Geography'] = 90
print("\nAfter adding new index 'Geography':")
print(marks_series)

# Agar hum try kare sirf read karne ke liye
# print(marks_series['Physics'])  # ❌ KeyError: 'Physics' (exists nahi karta)

# ------------------------------------------------------------
# 2️⃣ Using Slicing
# ------------------------------------------------------------
"""
Slicing ke through bhi multiple values ko update kar sakte hain
ek hi line mein contiguous range ke elements ko modify karna easy hai
"""

# Update Math and English marks together
marks_series['Math':'English'] = [55, 105]
print("\nAfter slicing update (Math & English):")
print(marks_series)




# ------------------------------------------------------------
# 3️⃣ Using Fancy Indexing
# ------------------------------------------------------------
"""
Fancy indexing ke through non-contiguous elements ko update kar sakte hain
ek hi line mein multiple elements ko modify karna possible hai
"""

marks_series[['Science', 'History']] = [75, 85]
print("\nAfter fancy indexing update (Science & History):")
print(marks_series)




# ------------------------------------------------------------
# 4️⃣ Using Fancy Indexing with Labels
# ------------------------------------------------------------
"""
Fancy indexing mein hum custom labels ke through bhi update kar sakte hain.
Non-contiguous aur random order mein multiple elements ek saath update possible hai.
"""




# Example Series with custom string labels
players_scores = pd.Series(
    [50, 60, 70, 80, 90],
    index=['Virat', 'Rohit', 'Gill', 'Dhoni', 'Hardik']
)
print("\nOriginal Players Scores:")
print(players_scores)



# Fancy indexing using labels
players_scores[['Virat', 'Dhoni', 'Hardik']] = [55, 85, 95]
print("\nAfter Fancy Indexing update using labels:")
print(players_scores)

"""
🧠 Notes:
1. Indexing → single element update.
2. Non-existing index → automatically add new element.
3. Slicing → contiguous range of elements update.
4. Fancy Indexing → non-contiguous multiple elements update.
5. Fancy Indexing with labels → label-based non-contiguous updates possible.
6. All updates modify the original Series inplace.
"""
```

## 🔹 Core Python Functionalities


👉 Data store aur manipulate karne ke liye

```python
# Series With Python Functionalities

import pandas as pd

"""
Ab tak humne Pandas ke basics padhe hain.
Best part about Pandas aur Python libraries ye hai ki ye Python ki built-in
functionalities ko efficiently use kar pati hain.
"""

# Example Series
subs = pd.Series([50, 20, 70, 10, 90])
print("Original Series:")
print(subs)

# ------------------------------------------------------------
# 1️⃣ len() → Series ke elements ka count
# ------------------------------------------------------------
print("\nLength of Series:")
print(len(subs))                  # Total number of elements in the Series

# ------------------------------------------------------------
# 2️⃣ type() → Data type of the object
# ------------------------------------------------------------
print("\nType of Series object:")
print(type(subs))                 # <class 'pandas.core.series.Series'>

# ------------------------------------------------------------
# 3️⃣ dir() → All available attributes & methods of Series
# ------------------------------------------------------------
print("\nAvailable attributes and methods:")
print(dir(subs))                  # List of all methods and attributes

# ------------------------------------------------------------
# 4️⃣ sorted() → Sorts elements and returns as Python list
# ------------------------------------------------------------
print("\nSorted elements (as list):")
print(sorted(subs))               # Series ko sort karke Python list return karega

# ------------------------------------------------------------
# 5️⃣ max() and min() → Maximum and Minimum values in Series
# ------------------------------------------------------------
print("\nMaximum value in Series:", subs.max())
print("Minimum value in Series:", subs.min())

# ------------------------------------------------------------
# 6️⃣ sum() and mean() → Sum and average of elements
# ------------------------------------------------------------
print("\nSum of Series:", subs.sum())
print("Mean of Series:", subs.mean())

"""
🧠 Notes:
1. Pandas Series Python ki built-in functions ke saath seamlessly integrate hoti hai.
2. sorted() → Python list return karta hai, original Series ko modify nahi karta.
3. max(), min(), sum(), mean() → Series ke numerical operations ke liye useful.
"""
```

```python
# Type Conversion in Pandas Series

import pandas as pd

# Example Series
marks_series = pd.Series(
    [55, 105, 75, 85, 90],
    index=['Math', 'English', 'Science', 'History', 'Geography']
)
print("Original Series:")
print(marks_series)

# ------------------------------------------------------------
# 1️⃣ Convert Series to Python List
# ------------------------------------------------------------
marks_list = list(marks_series)
print("\nSeries converted to List:")
print(marks_list)   # [55, 105, 75, 85, 90]

# ------------------------------------------------------------
# 2️⃣ Convert Series to Python Dictionary
# ------------------------------------------------------------
marks_dict = dict(marks_series)
print("\nSeries converted to Dictionary:")
print(marks_dict)   # {'Math': 55, 'English': 105, 'Science': 75, 'History': 85, 'Geography': 90}

"""
🧠 Notes:
1. list(series) → Series ke values ko Python list mein convert karta hai.
2. dict(series) → Series ke index-value pairs ko Python dictionary mein convert karta hai.
3. Type conversion useful hai jab Pandas data ko Python native structures ke saath use karna ho.
"""
```

# Membership Operator

```python
# Membership Operator in Pandas Series

import pandas as pd

# Example Series
movies = pd.Series([
    "Avengers", "Inception", "Titanic", "Iron Man", "Dangal"
])
print("Original Movie Series:")
print(movies)

# ------------------------------------------------------------
# 1️⃣ Using 'in' to check membership
# ------------------------------------------------------------
"""
Ab maan lo hum pata karna chahte hain ki koi movie 
iss Series ke andar hai ya nahi, toh membership operator use karte hain.

⚠️ Important:
By default, 'in' operator **Series ke index** ke upar kaam karta hai, value ke upar nahi.
"""

# Check if index exists
print("\nCheck if index 2 exists in Series:")
print(2 in movies)     # True, because index 2 exists

# Check if value exists (correct way)
print("\nCheck if 'Titanic' exists in Series values:")
print("Titanic" in movies.values)   # True, use .values to check in data

print("\nCheck if 'Spiderman' exists in Series values:")
print("Spiderman" in movies.values) # False

"""
🧠 Notes:
1. 'value in series' → default mein index check karta hai.
2. 'value in series.values' → values ke andar membership check karega.
3. Useful for quick presence check of an item in Series.
"""
```

```python
# Looping in Pandas Series

import pandas as pd

# Example Series
movies = pd.Series([
    "Avengers", "Inception", "Titanic", "Iron Man", "Dangal"
])
print("Original Movie Series:")
print(movies)

# ------------------------------------------------------------
# 1️⃣ Looping over values
# ------------------------------------------------------------
"""
Loop values ke upar run karte hain.
Ye simplest way hai har element ko access karne ka.
"""
print("\nLooping over values:")
for movie in movies:
    print(movie)

# ------------------------------------------------------------
# 2️⃣ Looping over index
# ------------------------------------------------------------
"""
Agar hume index ke upar loop chalana hai, toh Series ke index attribute ka use karenge.
"""
print("\nLooping over indexes:")
for idx in movies.index:
    print(idx)

# ------------------------------------------------------------
# 3️⃣ Looping over index and value together
# ------------------------------------------------------------
"""
Index aur value dono saath me chahiye ho toh items() ka use kar sakte hain.
"""
print("\nLooping over index and value:")
for idx, movie in movies.items():
    print(f"Index: {idx}, Movie: {movie}")

"""
🧠 Notes:
1. for i in series → values ke upar loop chalta hai.
2. for i in series.index → indexes ke upar loop chalta hai.
3. for idx, val in series.items() → index aur value dono ke liye convenient.
"""
```

```python
# Operators in Pandas Series

import pandas as pd

# Example Series
marks_series = pd.Series([55, 105, 75, 85, 90])
print("Original Marks Series:")
print(marks_series)

# ------------------------------------------------------------
# 1️⃣ Arithmetic Operators
# ------------------------------------------------------------
"""
Arithmetic operators Series ke saath scalar ya dusre Series ke saath kaam karte hain.
Ye broadcasting ka example hai:
    - Ek scalar ko Series ke saath operate karne par woh automatically
      har element par apply ho jata hai.
"""

# Subtracting from 100
print("\n100 - marks_series (Broadcasting Example):")
print(100 - marks_series)   # 100 se har element minus hoga

# Adding scalar
print("\nmarks_series + 10:")
print(marks_series + 10)    # 10 add ho jayega har element me

# Multiplying by scalar
print("\nmarks_series * 2:")
print(marks_series * 2)

# Division by scalar
print("\nmarks_series / 5:")
print(marks_series / 5)

# Operations between two Series
bonus_series = pd.Series([5, 10, 5, 15, 10])
print("\nAdding two Series (marks_series + bonus_series):")
print(marks_series + bonus_series)  # element-wise addition

# ------------------------------------------------------------
# 2️⃣ Relational Operators
# ------------------------------------------------------------
"""
Relational operators true/false Series return karte hain.
Ye logical conditions ke liye useful hain.
"""

print("\nmarks_series > 80:")
print(marks_series > 80)    # element-wise comparison

print("\nmarks_series == 75:")
print(marks_series == 75)   # check equality element-wise

print("\nmarks_series != 90:")
print(marks_series != 90)   # check not equal element-wise

print("\nmarks_series <= 85:")
print(marks_series <= 85)   # less than or equal to comparison

"""
🧠 Notes:
1. Arithmetic operators → +, -, *, /, //, % applied element-wise.
2. Relational operators → >, <, >=, <=, ==, != applied element-wise.
3. Result of relational operator → Boolean Series (True/False per element).
4. Broadcasting → scalar automatically Series ke har element pe apply hota hai.
"""
```

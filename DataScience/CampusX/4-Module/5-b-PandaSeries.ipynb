```python
# Plotting Graph on Pandas Series

import pandas as pd
import matplotlib.pyplot as plt

"""
Pandas bohot powerful hai aur directly Series ke upar graphs plot kar sakta hai.
Example: Agar Kohli ke scores ka graph plot karna ho, toh Series ka plot() function use karenge.
"""

# Example Series → Kohli Scores
kohli_scores = pd.Series(
    [45, 100, 75, 50, 150, 60, 100, 20, 50, 120],
    index=[f"Match{i+1}" for i in range(10)]
)
print("Kohli Scores Series:")
print(kohli_scores)

# ------------------------------------------------------------
# 1️⃣ Simple Line Plot
# ------------------------------------------------------------
kohli_scores.plot(title="Virat Kohli Scores per Match", figsize=(10,5))
plt.xlabel("Matches")
plt.ylabel("Runs Scored")
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 2️⃣ Bar Plot
# ------------------------------------------------------------
kohli_scores.plot(kind='bar', title="Virat Kohli Scores (Bar Plot)", figsize=(10,5), color='skyblue')
plt.xlabel("Matches")
plt.ylabel("Runs Scored")
plt.show()

# ------------------------------------------------------------
# 3️⃣ Horizontal Bar Plot
# ------------------------------------------------------------
kohli_scores.plot(kind='barh', title="Virat Kohli Scores (Horizontal Bar)", figsize=(10,5), color='lightgreen')
plt.xlabel("Runs Scored")
plt.ylabel("Matches")
plt.show()

# ------------------------------------------------------------
# Notes:
# ------------------------------------------------------------
"""
1. Pandas ka plot() function matplotlib ka wrapper hai.
2. Default plot → line plot
3. kind='bar' → vertical bar plot
4. kind='barh' → horizontal bar plot
5. figsize → figure size
6. plt.grid(True) → grid add karne ke liye
7. Pandas Series ke index automatically x-axis labels ban jaate hain
"""
.
```

# IMPORTANT Series Method Notes and Short Description 

---

### 🧩 Overview
Pandas Series ke kuch **important built-in methods** yahan diye gaye hain jo data manipulation, cleaning, aur transformation mein frequently use hote hain.

---

### 📘 Methods with Short Descriptions

| **Method** | **Short Description** |
|-------------|------------------------|
| **astype()** | Convert karta hai Series ke data type ko (e.g. float → int, str → category) |
| **between()** | Check karta hai ki values kisi range ke andar hain ya nahi (True/False return karta hai) |
| **clip()** | Lower aur upper limit set karta hai — values ko given range ke andar bound kar deta hai |
| **drop_duplicates()** | Duplicate values ko Series se remove karta hai |
| **dropna()** | NaN (missing) values ko Series se remove karta hai |
| **fillna()** | Missing values ko kisi fixed ya calculated value (mean, median, etc.) se fill karta hai |
| **filter()** | Label, regex, ya condition ke base pe Series filter karta hai |
| **iloc[]** | Integer-based indexing (positions ke base pe data access) |
| **isnull()** | True/False return karta hai — batata hai kaunsi values NaN hain |
| **map()** | Har element pe ek function apply karta hai (element-wise operation) |
| **apply()** | Puri Series pe function lagata hai — complex logic ke liye use hota hai |
| **copy()** | Series ka independent copy create karta hai (original affect nahi hota) |



```python
# Some Important Series Methods

import numpy as np
import pandas as pd

# Sample Data ---------------------------------------------------
vk = pd.Series([45, 100, 75, np.nan, 150, 60, 100, 20, 50, np.nan])
print("Original Series:\n", vk)

# ---------------------------------------------------------------
# 1️⃣ astype() → data type convert karne ke liye
# ---------------------------------------------------------------
vk_int = vk.fillna(0).astype(int)        # NaN ko fill karke int mein convert kiya
print("\nastype() Example:\n", vk_int)
print("Type:", vk_int.dtype)

# ---------------------------------------------------------------
# 2️⃣ between() → check karta hai values ek range ke beech hain ya nahi
# ---------------------------------------------------------------
mask = vk.between(50, 120)               # 50 aur 120 ke beech wale scores True honge
print("\nbetween() Example:\n", mask)

# ---------------------------------------------------------------
# 3️⃣ clip() → lower aur upper limit set karta hai (values ko bound karta hai)
# ---------------------------------------------------------------
clipped = vk.clip(lower=50, upper=120)   # 50 se kam → 50, 120 se zyada → 120
print("\nclip() Example:\n", clipped)

# ---------------------------------------------------------------
# 4️⃣ drop_duplicates() → duplicate values remove karta hai
# ---------------------------------------------------------------
no_dupes = vk.drop_duplicates()
print("\ndrop_duplicates() Example:\n", no_dupes)

# ---------------------------------------------------------------
# 5️⃣ dropna() → missing values (NaN) hata deta hai
# ---------------------------------------------------------------
no_nans = vk.dropna()
print("\ndropna() Example:\n", no_nans)

# ---------------------------------------------------------------
# 6️⃣ fillna() → NaN values ko kisi value se fill karne ke liye
# ---------------------------------------------------------------
filled = vk.fillna(vk.mean())            # NaN ko mean se fill kiya
print("\nfillna() Example:\n", filled)

# ---------------------------------------------------------------
# 7️⃣ filter() → labels ya regex se filter karne ke liye
# ---------------------------------------------------------------
vk.index = [f"Match{i+1}" for i in range(10)]
filtered = vk.filter(like="1")           # index me “1” aane wale matches
print("\nfilter() Example:\n", filtered)

# ---------------------------------------------------------------
# 8️⃣ iloc[] → integer based indexing
# ---------------------------------------------------------------
first_five = vk.iloc[:5]                 # first 5 elements
print("\niloc Example:\n", first_five)

# ---------------------------------------------------------------
# 9️⃣ isnull() → NaN values ko check karta hai
# ---------------------------------------------------------------
null_check = vk.isnull()
print("\nisnull() Example:\n", null_check)

# ---------------------------------------------------------------
# 🔟 map() → har element par function apply karta hai
# ---------------------------------------------------------------
mapped = vk.fillna(0).map(lambda x: x * 2)
print("\nmap() Example:\n", mapped)

# ---------------------------------------------------------------
# 11️⃣ apply() → function poori Series par lagane ke liye
# ---------------------------------------------------------------
def performance(x):
    if x >= 100:
        return "Century"
    elif x >= 50:
        return "Half-Century"
    else:
        return "Low Score"

applied = vk.fillna(0).apply(performance)
print("\napply() Example:\n", applied)

# ---------------------------------------------------------------
# 12️⃣ copy() → independent copy banata hai (reference nahi)
# ---------------------------------------------------------------
copy_series = vk.copy()
copy_series[0] = 999                     # ye change original ko affect nahi karega
print("\ncopy() Example:\n", copy_series)
print("Original after copy modification:\n", vk)

# ---------------------------------------------------------------
# Bonus: sys.getsizeof() → size of object in memory
# ---------------------------------------------------------------
import sys
print("\nSeries Size in Bytes:", sys.getsizeof(vk))
```

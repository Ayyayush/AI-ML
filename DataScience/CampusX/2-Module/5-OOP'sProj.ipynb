# 💬 What is NLP Cloud Playground?

**NLP Cloud Playground** ek online tool hai jahan aap Natural Language Processing (NLP) models ko bina code likhe test kar sakte ho.  
Yeh ek interactive platform hota hai jahan aap text input dete ho aur model ka output instantly dekh sakte ho.  
Kuch common use-cases:

- Text Summarization  
- Sentiment Analysis  
- Question Answering  
- NER (Named Entity Recognition)  
- Translation, etc.

✅ Best for developers jo kisi NLP model ko integrate karne se pehle uska output dekhna chahte hain.

---

# 🧠 What is NER Extraction?

**NER** ka full form hai **Named Entity Recognition**.  
Yeh NLP ka ek task hota hai jisme kisi sentence ya paragraph ke andar se **important entities** ko identify kiya jaata hai.

Kuch example entities:

- 👤 Person: *Virat Kohli*  
- 🏙️ Location: *Delhi*  
- 🏢 Organization: *Google*  
- 💵 Money, 📅 Dates, ⏰ Times, etc.

**Example:**  
Text: `"Amazon opened a new office in Bengaluru in July 2025."`  
NER Output:  
- Organization: Amazon  
- Location: Bengaluru  
- Date: July 2025

# 🔌 What is API?

**API** ka matlab hota hai **Application Programming Interface**.  
Yeh ek set of rules aur endpoints hota hai jo allow karta hai do software systems ko aapas mein **communicate** karne ke liye.

Socho API ko ek **waiter** ki tarah — jo aap (client) ka order (request) leke kitchen (server) tak le jaata hai aur wapas result (response) lataa hai.

💡 Real Life Example:  
Zomato app pe jab aap search karte ho "Pizza in Delhi", tab app ek API call bhejta hai server ko. Server uske response mein list of pizza places bhejta hai.

````python
import nlpcloud                                     # Importing the NLP Cloud Python SDK

# Creating a client object to connect with NLP Cloud API
# "finetuned-llama-3-70b" → Model name being used
# Second argument → Your API token (used for authentication)
# gpu=True → Requests GPU-powered inference for faster processing
client = nlpcloud.Client(
    "finetuned-llama-3-70b",
    "4e73d98eed939d05d3e09f6d85f530f58be7aacc",
    gpu=False
)

# Calling the entities() method to perform Named Entity Recognition (NER)
# The text block is the input data on which entities will be extracted
# searched_entity specifies what type of entity we want to find
response = client.entities(
    """
    John Doe started learning Javascript when he was 15 years old.
    After a couple of years he switched to Python and started learning
    low level programming. He is now a Go expert at Google.
    """,
    searched_entity="programming languages"        # Filters entities to programming languages only
)

# Printing the response returned by the NLP Cloud API
# The response usually contains detected entities with confidence scores
print(response)
````

````python
import nlpcloud

class NLPApp:
    
    def __init__(self):                                      # Jab bhi class ka object banega, yeh constructor call hoga
        self.__database = {}                                 # Private dict jisme email: [name, password] store hoga
        self.__first_menu()                                  # App ka pehla menu dikhaya jaayega

    def __first_menu(self):                                  # User se input leke menu ka option process karega
        first_input = input("""
              Hi! How would you like to proceed?
              1. Not a member? Register
              2. Already a member? Login
              3. Galti se aa gye? Exit
        \n👉 Your choice: """)                               
        
        if first_input == '1':
            self.__register()
        elif first_input == '2':
            self.__login()
        else:
            exit()                                           # User ne exit choose kiya to program terminate

    def __register(self):                                    # User registration function
        name = input('Enter your name: ')
        email = input('Enter your email: ')
        password = input('Enter your password: ')

        if email in self.__database:
            print('⚠️ Email already exists. Please try logging in.')
            self.__first_menu()
        else:
            self.__database[email] = [name, password]
            print('✅ Registration successful! Now you can log in.')
            print(f'📋 Current Users: {self.__database}')
            self.__login()

    def __login(self):
        email = input('Enter your email: ')
        password = input('Enter your password: ')

        if email in self.__database:
            if self.__database[email][1] == password:
                print(f'✅ Welcome back, {self.__database[email][0]}!')
                self.__second_menu()
            else:
                print('❌ Incorrect password. Try again.')
                self.__first_menu()
        else:
            print('❌ This email is not registered. Please register first.')
            self.__first_menu()

    def __second_menu(self):
        second_input = input("""
        Hi! How would you like to proceed?
        1. Named Entity Recognition (NER)
        2. Language Detection
        3. Sentiment Analysis
        4. Logout 
        \n👉 Your choice: """)

        if second_input == '1':
            self.ner()
        elif second_input == '2':
            self.language_detection()
        elif second_input == '3':
            self.sentiment_analysis()
        elif second_input == '4':
            print("👋 Logged out successfully!")
            self.__first_menu()
        else:
            print("⚠️ Invalid input. Going back to main menu.")
            self.__second_menu()

    def ner(self):
        para = input("📝 Enter the paragraph for NER: ")
        search_term = input("🔍 What would you like to search (e.g., people, programming languages)? ")

        try:
            client = nlpcloud.Client(
                "finetuned-llama-3-70b",
                "4e73d98eed939d05d3e09f6d85f530f58be7aacc",
                gpu=True
            )
            response = client.entities(para, searched_entity=search_term)
            print("📌 Entities found:")
            print(response)

        except Exception as e:
            print(f"❌ Error occurred while fetching NER results: {e}")

        self.__second_menu()

    def language_detection(self):
        text = input("🌐 Enter text for language detection: ")
        try:
            client = nlpcloud.Client("gpt-neox-20b", "4e73d98eed939d05d3e09f6d85f530f58be7aacc")
            response = client.language_detection(text)
            print("🌍 Detected Language:")
            print(response)
        except Exception as e:
            print(f"❌ Error during language detection: {e}")
        self.__second_menu()

    def sentiment_analysis(self):                           # Sentiment Analysis
        text = input("💬 Enter text for sentiment analysis: ")
        try:
            client = nlpcloud.Client("bart-large-mnli", "4e73d98eed939d05d3e09f6d85f530f58be7aacc")
            response = client.classification(text, ["positive", "neutral", "negative"])
            
            # Extracting the label with highest score
            L = []
            for i in response['scored_labels']:
                L.append(i['score'])

            index = sorted(list(enumerate(L)), key=lambda x: x[1], reverse=True)[0][0]

            print("📊 Sentiment Result:")
            print(response['scored_labels'][index]['label'])

        except Exception as e:
            print(f"❌ Error during sentiment analysis: {e}")

        self.__second_menu()


# 🟢 Start the App
obj = NLPApp()


````

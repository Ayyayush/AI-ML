```python
# ==================== LIST ====================

l = [10, 20, 30, 20]  # List → Ordered, Mutable, Allows Duplicates

print(l[0])  # Access using index → 10
print(l[-1])  # Negative indexing → 20
print(l[1:3])  # Slicing → [20,30]

l.append(40)  # Add one element at end
l.extend([50, 60])  # Add multiple elements at end
l.insert(1, 15)  # Insert 15 at index 1

l.remove(20)  # Remove FIRST occurrence of 20
x = l.pop()  # Remove & return last element
x = l.pop(1)  # Remove & return element at index 1

print(l.index(30))  # Index of first occurrence of 30
print(l.count(20))  # Count occurrences of 20
print(30 in l)  # Membership check → True/False
print(len(l))  # Number of elements

l.sort()  # Sort ascending
l.sort(reverse=True)  # Sort descending
l.reverse()  # Reverse current list

l2 = l.copy()  # Create shallow copy

for x in l:  # Traverse list
    print(x)

l.clear()  # Remove all elements

# List supports indexing and slicing.
# List is MUTABLE → Elements can be modified.
# List → Ordered + Indexed + Mutable + Allows Duplicates


# ==================== TUPLE ====================

t = (10, 20, 30, 40)  # Tuple → Ordered, allows duplicates, IMMUTABLE

print(t[0])  # Access using index → 10
print(t[-1])  # Negative indexing → 40
print(t[1:3])  # Slicing → (20, 30)

print(len(t))  # Number of elements
print(t.count(20))  # Number of occurrences of 20
print(t.index(30))  # Index of first occurrence of 30

t2 = (50, 60)
t3 = t + t2  # Concatenate two tuples
print(t3)

a, b, c, d = t  # Tuple unpacking
print(a, b, c, d)

x = (10,)  # Single-element tuple MUST have comma

# t[0] = 100                      # ERROR → Tuple is immutable
# Tuple → Ordered + Indexed + Allows Duplicates + Immutable

# ==================== SET ====================

s = {10, 20, 30, 20}  # Set → Unordered, Unique elements
print(s)  # Duplicate 20 automatically removed

s.add(40)  # Add one element
s.update([50, 60])  # Add multiple elements

s.remove(20)  # Remove element → ERROR if not present
s.discard(100)  # Remove element → NO error if not present
x = s.pop()  # Remove and return an arbitrary element

print(30 in s)  # Membership check → True/False
print(len(s))  # Number of elements

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)  # Union → {1,2,3,4,5,6}
print(a & b)  # Intersection → {3,4}
print(a - b)  # Difference → {1,2}
print(a ^ b)  # Symmetric Difference → {1,2,5,6}

print(a.issubset(b))  # Check whether a is subset of b
print(a.issuperset(b))  # Check whether a is superset of b
print(a.isdisjoint(b))  # True if no common elements

for x in a:  # Traverse set
    print(x)

a.clear()  # Remove all elements

# a[0]                            # ERROR → Set does NOT support indexing
# Set → Unordered + Unique + Mutable + No Indexing


# ==================== DICTIONARY ====================

d = {
    "name": "Ayush",
    "age": 22,
    "cgpa": 7.5,
}  # Dictionary → Key-Value pairs, Keys unique

print(d["name"])  # Access value using key → Ayush
print(d.get("age"))  # Access safely using get() → 22
print(d.get("city", "Not Found"))  # Default value if key does not exist

d["age"] = 23  # Update existing value
d["city"] = "Kanpur"  # Add new Key-Value pair

print(d.keys())  # Get all keys
print(d.values())  # Get all values
print(d.items())  # Get all (key,value) pairs

print("name" in d)  # Check whether KEY exists → True
print(len(d))  # Number of Key-Value pairs

d.update({"age": 24, "course": "MCA"})  # Add/Update multiple items

x = d.pop("city")  # Remove key and return its value
item = d.popitem()  # Remove & return last inserted (key,value)

for key in d:  # Traverse keys
    print(key)

for key, value in d.items():  # Traverse Key-Value pairs
    print(key, value)

d2 = d.copy()  # Create shallow copy
d.clear()  # Remove all Key-Value pairs

# d["unknown"]                      # ERROR → KeyError if key doesn't exist
# d.get("unknown")                  # Returns None → No error
# Dictionary → Mutable + Unique Keys + Key-Value Pairs + Insertion Ordered


# ==================== COMPREHENSIONS ====================

# Comprehension → Short way to create List, Set or Dictionary from an iterable

# ==================== LIST COMPREHENSION ====================

nums = [1, 2, 3, 4, 5]

squares = [x * x for x in nums]  # Create list of squares
print(squares)  # [1,4,9,16,25]

even = [x for x in nums if x % 2 == 0]  # List comprehension with condition
print(even)  # [2,4]

result = ["Even" if x % 2 == 0 else "Odd" for x in nums]  # if-else
print(result)

# Normal Loop Equivalent
squares = []
for x in nums:
    squares.append(x * x)

# Syntax:
# [expression for item in iterable]
# [expression for item in iterable if condition]
# [value1 if condition else value2 for item in iterable]


# ==================== SET COMPREHENSION ====================

nums = [1, 2, 2, 3, 3, 4]

s = {x * x for x in nums}  # Creates Set → duplicates removed
print(s)  # {1,4,9,16}

even = {x for x in nums if x % 2 == 0}  # Set with condition
print(even)

# Syntax:
# {expression for item in iterable}


# ==================== DICTIONARY COMPREHENSION ====================

nums = [1, 2, 3, 4]

d = {x: x * x for x in nums}  # Key → x, Value → x²
print(d)  # {1:1,2:4,3:9,4:16}

even = {x: x * x for x in nums if x % 2 == 0}  # Dictionary with condition
print(even)  # {2:4,4:16}

# Syntax:
# {key:value for item in iterable}


# ==================== NESTED LIST COMPREHENSION ====================

matrix = [[1, 2], [3, 4]]

flat = [x for row in matrix for x in row]  # Convert 2D List → 1D List
print(flat)  # [1,2,3,4]


# ==================== GENERATOR EXPRESSION ====================

g = (x * x for x in range(5))  # () creates Generator, NOT Tuple
print(next(g))  # 0
print(next(g))  # 1

# Generator generates values one-by-one → Memory Efficient


# IMPORTANT:
# [] → List Comprehension
# {} → Set Comprehension
# {key:value} → Dictionary Comprehension
# () → Generator Expression
# There is NO direct Tuple


# ==================== CLASS & OBJECTS ====================

class Student:
    college = "NIT Jamshedpur"  # Class Variable → shared by ALL objects

    def __init__(self, name, age):  # __init__ automatically runs when object is created
        self.name = name  # self → reference to current object
        self.age = age  # Instance Variable → separate copy for each object

    def show(self):  # Instance Method → receives current object as self
        print(self.name, self.age)

    @classmethod
    def show_college(cls):  # Class Method → cls refers to current CLASS
        print(cls.college)  # Can directly access Class Variables using cls

    @classmethod
    def change_college(cls, name):  # Class Method can also modify Class Variables
        cls.college = name

    @staticmethod
    def add(a, b):  # Static Method → receives neither self nor cls
        return (
            a + b
        )  # Used when work is related to class but independent of object/class data


# ==================== OBJECT CREATION ====================

s1 = Student("Ayush", 21)  # Object created → __init__ called automatically
s2 = Student("Rahul", 22)  # Another object → gets separate name and age

print(s1.name)  # Access Instance Variable using object
print(Student.college)  # Access Class Variable using Class Name (preferred)


# ==================== INSTANCE METHOD ====================

s1.show()  # Normal way → Python automatically passes s1 as self
Student.show(s1)  # Calling Instance Method using Class Name → pass object manually
# s1.show() internally works like Student.show(s1)


# ==================== CLASS METHOD ====================

Student.show_college()  # Calling Class Method using Class Name (preferred)
s1.show_college()  # Can also call Class Method using object
# Python automatically passes Student class as cls

Student.change_college("NIT JS")  # Modify shared Class Variable
Student.show_college()


# ==================== STATIC METHOD ====================

print(Student.add(10, 20))  # Calling Static Method using Class Name (preferred)
print(s1.add(5, 6))  # Static Method can also be called using object
# No object/class is automatically passed → no self and no cls


# ==================== IMPORTANT ====================

# self → Current Object
# cls  → Current Class
# Instance Method → self → works mainly with object/instance data
# Class Method    → @classmethod + cls → works mainly with class data
# Static Method   → @staticmethod → neither self nor cls
# Instance Method using class → Student.show(s1) → object must be passed manually
# Class Method using class → Student.show_college() → class automatically passed as cls
# Static Method using class → Student.add(10,20) → nothing automatically passed


# ==================== INHERITANCE ====================
class Person:  #  ! Parent / Base / Super Class
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):  # ! Student inherits Person → Single Inheritance
    def __init__(self, name, age, course):
        super().__init__(name, age)  # Calls Parent Class constructor
        self.course = course  # Child-specific Instance Variable

    def show(self):  # ! Method Overriding → same method in Parent & Child
        super().show()  # Calls Parent Class show()
        print("Course:", self.course)

    def study(self):  # Child-specific method
        print(self.name, "is studying")


# ==================== OBJECT CREATION ====================
s1 = Student("Ayush", 21, "MCA")

s1.show()  # Calls Student's overridden show()
s1.study()  # Calls Child-specific method

print(s1.name)  # Inherited from Person
print(s1.course)  # Defined in Student


# ==================== PARENT METHOD ACCESS ====================

s1.show()  # Child version runs due to overriding
Person.show(s1)  # Explicitly call Parent version using Class Name
super(Student, s1).show()  # Another way to access Parent method


# ==================== isinstance() / issubclass() ====================

print(isinstance(s1, Student))  # True → s1 is Student object
print(isinstance(s1, Person))  # True → Student IS-A Person
print(issubclass(Student, Person))  # True → Student inherits Person


# ==================== IMPORTANT ====================

# Inheritance → Child Class acquires properties/methods of Parent Class
# Syntax → class Child(Parent):
# Person → Parent/Base/Super Class
# Student → Child/Derived/Sub Class
# super() → Used to access Parent Class methods/constructor
# Method Overriding → Child provides its own version of Parent's method
# Child can use inherited members + define its own members
# Student IS-A Person → inheritance represents an IS-A relationship


# ==================== ABSTRACTION + ENCAPSULATION ====================

from abc import ABC, abstractmethod  # ! ABC → Abstract Base Class module

class BankAccount(ABC):  # Inheriting ABC makes it an Abstract Base Class

    def __init__(self, name, balance, pin):
        self.name = name  # Public Member → accessible from anywhere
        self._balance = (
            balance  # Protected Member → convention: use inside class/child class
        )
        self.__pin = pin  # Private Member → name mangling prevents normal direct access

    @abstractmethod
    def account_type(self):  # ! Abstract Method → child class must implement it
        pass

    def show_balance(self):  # Public method providing controlled access to protected data
        print("Balance:", self._balance)

    def __show_pin(self):  # Private Method
        print("PIN:", self.__pin)

    def verify_pin(self, pin):  # Public method → controlled access to private member
        return self.__pin == pin


class SavingsAccount(BankAccount):

    def __init__(self, name, balance, pin):
        super().__init__(name, balance, pin)  # Calling Parent Class constructor

    def account_type(self):  # ! Implementing/Overriding Abstract Method
        print("Savings Account")

    def show_details(self):
        print("Name:", self.name)  # Public Member → directly accessible
        print("Balance:", self._balance)  # Protected Member → accessible in Child Class
        # print(self.__pin)                   # ERROR → Private Member not directly accessible in Child


# ==================== OBJECT CREATION ====================

# b1 = BankAccount("Ayush", 5000, 1234)      # ! ERROR → Cannot instantiate Abstract Class

s1 = SavingsAccount("Ayush", 5000, 1234)  # Object of concrete Child Class

s1.account_type()  # Child implementation of Abstract Method
s1.show_details()

print(s1.name)    # Public → accessible outside class
print(s1._balance)  # Protected → technically accessible, but should not be used outside
# print(s1.__pin)                            # ERROR → Private cannot be accessed normally from outside

print(s1.verify_pin(1234))  # ! Access private data through Public Method
s1.show_balance()


# ==================== IMPORTANT ====================

# Abstraction → Hide implementation details and expose only essential functionality
# ABC → Abstract Base Class; imported from abc module
# @abstractmethod → forces Child Class to implement that method
# Abstract Class → cannot be instantiated if it has unimplemented abstract methods

# Encapsulation → Bundle data + methods together and control access to data
# Public    → self.name      → accessible from anywhere
# Protected → self._balance  → single underscore; conventionally class + subclasses
# Private   → self.__pin     → double underscore; Python applies Name Mangling

# Python does NOT enforce Protected members strictly → _balance can still be accessed outside
# Private is implemented using Name Mangling, not absolute security


# ==================== DUNDER + ITERATOR + GENERATOR ====================
class Student:
    def __init__(self, name, marks):      #  ! Dunder → called automatically during object creation
        self.name = name
        self.marks = marks

    def __str__(self):  # Called by print(object) → user-friendly representation
        return f"Student: {self.name}, Marks: {self.marks}"

    def __repr__(self):  # Developer/debugging representation of object
        return f"Student('{self.name}', {self.marks})"
  
    def __len__(self):                    #  !  Called automatically by len(object)
        return len(self.marks)

    def __iter__(self):  # Makes Student object ITERABLE
        return iter(self.marks)  # Returns an Iterator for marks


s1 = Student("Ayush", [80, 90, 85])

print(s1)  # Automatically calls s1.__str__()
print(str(s1))  # Calls __str__()
print(repr(s1))  # Calls __repr__()
print(len(s1))  # Calls __len__()


# ==================== ITERABLE & ITERATOR ====================
# ! Iterable → Object that can produce an Iterator using iter()
# ! Iterator → Object that gives values one-by-one using next()

it = iter(s1)  # Calls s1.__iter__() → returns Iterator

print(next(it))  # 80
print(next(it))  # 90
print(next(it))  # 85
# print(next(it))                          # ERROR → StopIteration (no elements left)

for mark in s1:  # for loop internally uses iter() + next()
    print(mark)

# ==================== CUSTOM ITERATOR ====================
class Counter:
    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):  # Iterator must return itself from __iter__()
        return self

    def __next__(self):  # next(obj) automatically calls __next__()
        if self.current > self.limit:
            raise StopIteration  # Tells Python iteration is finished

        value = self.current
        self.current += 1
        return value

c = Counter(3)  # c is BOTH Iterable and Iterator
print(next(c))  # 1
print(next(c))  # 2
print(next(c))  # 3


# ==================== GENERATOR ====================

def numbers(n):
    for i in range(1, n + 1):
        yield i  # !  yield returns value & PAUSES function


g = numbers(3)  # Calling generator function returns Generator object

print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3

# Generator automatically implements __iter__() and __next__()
# Therefore → Every Generator is an Iterator
# Generator → values generated lazily/on-demand → memory efficient


# ==================== IMPORTANT ====================

# __init__() → Object initialization
# __str__()  → User-friendly representation → print(obj)
# __repr__() → Developer/debug representation → repr(obj)
# __len__()  → Defines behaviour of len(obj)
# __iter__() → Returns an Iterator
# __next__() → Returns next value; raises StopIteration when finished
# Iterable   → Can produce an Iterator using iter()
# Iterator   → Produces one value at a time using next()
# Generator  → Special Iterator created using yield
# yield      → Returns value + pauses function + remembers its state


# ==================== DECORATOR + CLOSURE + EXCEPTION HANDLING ====================

# ==================== CLOSURE ====================
def multiplier(x):
    def multiply(y):  # Inner function
        return x * y  # Remembers x even after outer function ends

    return multiply


double = multiplier(2)  # x = 2 is remembered by Closure
print(double(5))  # 10

#! Closure → Inner function remembers variables of Outer function
# even after the Outer function has finished execution.

# ==================== DECORATOR ====================
def login_required(func):  # !  Decorator receives another function
    def wrapper():
        print("Checking Login...")
        func()  # Calling original function
        print("Execution Completed")

    return wrapper


@login_required  # Same as → dashboard = login_required(dashboard)
def dashboard():
    print("Welcome to Dashboard")

dashboard()  # Actually calls wrapper()

# ! Decorator → Adds functionality to a function without modifying its original code
# ! Decorators internally commonly use Closures.


# ==================== EXCEPTION HANDLING ====================
def divide(a, b):
    try:  # ! Code which may generate Exception
        if b == 0:
            raise ValueError(
                "Cannot divide by zero"
            )  # raise → manually generate Exception

        result = a / b

    except ValueError as e:  # Handles ValueError
        print("Exception:", e)

    else:  # Runs ONLY when no Exception occurs
        print("Result:", result)

    finally:  # ALWAYS executes
        print("Division operation completed")


divide(10, 2)  # try → else → finally
divide(10, 0)  # try → except → finally


# ==================== CUSTOM EXCEPTION ====================

class AgeError(Exception):  #!  User-Defined Exception → inherit Exception
    pass


def check_age(age):
    try:
        if age < 18:
            raise AgeError("Age must be 18 or above")  #  Manually throw Custom Exception

        print("Eligible")

    except AgeError as e:
        print("Exception:", e)

    finally:
        print("Age verification completed")


check_age(16)


# ==================== IMPORTANT ====================

# Closure   → Inner function remembers Outer function's variables
# Decorator → Adds functionality to another function without changing its code
# @decorator → Syntactic sugar for applying a Decorator
# try       → Contains code that may generate an Exception
# except    → Handles the Exception
# else      → Executes when NO Exception occurs
# finally   → ALWAYS executes whether Exception occurs or not
# raise     → Manually generates an Exception
# Exception → Runtime error that interrupts normal program flow
# Custom Exception → Created by inheriting from Exception class


# ==================== FILE HANDLING ====================

# WRITE MODE → Creates file / overwrites existing content
with open("data.txt", "w") as f:
    f.write("Hello Python\n")
    f.write("Learning File Handling")

# READ MODE → Reads file content
with open("data.txt", "r") as f:
    data = f.read()
    print(data)

# APPEND MODE → Adds content at the end
with open("data.txt", "a") as f:
    f.write("\nNew Line Added")

# READ LINE BY LINE
with open("data.txt", "r") as f:
    for line in f:
        print(line.strip())

# with → Automatically closes the file
# r → Read | w → Write/Overwrite | a → Append | x → Create new file
# read() → Read entire file
# readline() → Read one line
# readlines() → Read all lines as a list
# write() → Write data into file


# ==================== NUMPY IMPORTANT CONCEPTS ====================

import numpy as np  # NumPy → Numerical Python

# ==================== ARRAY CREATION ====================
a = np.array([10, 20, 30, 40])  # Create 1D NumPy Array
b = np.array([[1, 2, 3], [4, 5, 6]])  # Create 2D Array

print(a)
print(b)

print(b.ndim)  # Number of dimensions → 2
print(b.shape)  # Shape → (2,3) = 2 rows, 3 columns
print(b.size)  # Total elements → 6
print(b.dtype)  # Data type of elements


# ==================== SPECIAL ARRAYS ====================
print(np.zeros((2, 3)))  # 2×3 array filled with 0
print(np.ones((2, 3)))  # 2×3 array filled with 1
print(np.arange(0, 10, 2))  # Like range() → [0,2,4,6,8]
print(np.linspace(0, 10, 5))  # 5 equally spaced values from 0 to 10


# ==================== INDEXING & SLICING ====================

print(a[0])  # First element → 10
print(a[-1])  # Last element → 40
print(a[1:3])  # Slicing → [20,30]

print(b[0, 1])  # Row 0, Column 1 → 2
print(b[0, :])  # Entire first row
print(b[:, 1])  # Entire second column


# ==================== RESHAPE / FLATTEN ====================

x = np.arange(1, 7)
x = x.reshape(2, 3)  # Change shape → 2×3
print(x)

print(x.flatten())  # Convert multidimensional → 1D
print(x.T)  # Transpose → Rows become Columns


# ==================== MATHEMATICAL OPERATIONS ====================

a = np.array([10, 20, 30])

print(a + 5)  # Element-wise addition
print(a * 2)  # Element-wise multiplication
print(a**2)  # Element-wise square

b = np.array([1, 2, 3])
print(a + b)  # Element-wise array addition
print(a * b)  # Element-wise multiplication


# ==================== AGGREGATE FUNCTIONS ====================

print(np.sum(a))  # Sum of elements
print(np.mean(a))  # Mean / Average
print(np.min(a))  # Minimum
print(np.max(a))  # Maximum
print(np.std(a))  # Standard Deviation
print(np.median(a))  # Median
print(np.argmin(a))  # Index of minimum element
print(np.argmax(a))  # Index of maximum element


# ==================== FILTERING ====================

a = np.array([10, 20, 30, 40, 50])

print(a[a > 25])  # Boolean Filtering → [30,40,50]
print(a[(a > 10) & (a < 50)])  # Multiple conditions

# ==================== SORT / UNIQUE ====================

a = np.array([30, 10, 20, 10])
print(np.sort(a))  # Sort array
print(np.unique(a))  # Return unique elements


# ==================== CONCATENATION ====================

a = np.array([1, 2])
b = np.array([3, 4])
print(np.concatenate((a, b)))  # Combine arrays → [1,2,3,4]


# ==================== COPY VS VIEW ====================

a = np.array([10, 20, 30])
b = a.copy()  # Independent copy
c = a.view()  # View → shares original array data

a[0] = 100

print(b)  # Copy unaffected
print(c)  # View reflects change


# ==================== RANDOM ====================

print(np.random.rand(3))  # 3 random values between 0 and 1
print(np.random.randint(1, 10, size=5))  # 5 random integers from 1 to 9


# ==================== IMPORTANT ====================

# NumPy Array → Faster & more memory-efficient than Python List for numerical data
# ndarray → Main NumPy array object
# ndim → Number of dimensions | shape → Dimensions | size → Total elements
# dtype → Data type of elements
# reshape() → Changes shape without changing data
# T → Transpose | flatten() → Convert to 1D
# NumPy supports Vectorization → operations performed on whole arrays without loops
# Boolean Indexing → Filter elements using conditions
# axis=0 → Operation column-wise | axis=1 → Operation row-wise


# ==================== PANDAS SERIES ====================

import pandas as pd
import numpy as np

# ==================== SERIES CREATION ====================
s = pd.Series([10, 20, 30, 40])  # ! Series → 1D labeled array
print(s)  # Default index → 0,1,2,3

s = pd.Series([90, 80, 85], index=["A", "B", "C"])  # ! Custom index
print(s)

d = {"Math": 90, "Python": 85, "SQL": 80}
s2 = pd.Series(d)  # Create Series from Dictionary


# ==================== IMPORTANT ATTRIBUTES ====================
print(s.values)  # Values as NumPy array
print(s.index)  # Index/labels
print(s.dtype)  # Data type
print(s.size)  # Number of elements
print(s.shape)  # Shape → (3,)


# ==================== ACCESSING DATA ====================

print(s["A"])  # Access using label → 90
print(s.iloc[0])  # Access using position → 90
print(s.loc["B"])  # Access using label → 80

print(s.iloc[0:2])  # Slice using positions
print(s.loc["A":"B"])  # Slice using labels


# ==================== MODIFY DATA ====================

s.loc["A"] = 95  # Modify value using label
s["D"] = 70  # Add new value with new index


# ==================== BASIC OPERATIONS ====================

print(s + 5)  # Add 5 to every element
print(s * 2)  # Multiply every element
print(s > 80)  # Boolean Series
print(s[s > 80])  # Filter values > 80


# ==================== AGGREGATE METHODS ====================

print(s.sum())  # Sum
print(s.mean())  # Mean
print(s.min())  # Minimum
print(s.max())  # Maximum
print(s.median())  # Median
print(s.std())  # Standard Deviation
print(s.count())  # Count non-null values
print(s.describe())  # Statistical summary


# ==================== SORTING ====================

print(s.sort_values())  # Sort according to values
print(s.sort_values(ascending=False))  # Sort descending
print(s.sort_index())  # Sort according to index


# ==================== NULL VALUES ====================

x = pd.Series([10, np.nan, 30, None])

print(x.isnull())  # True where value is missing
print(x.notnull())  # True where value is present
print(x.isnull().sum())  # Count missing values

print(x.dropna())  # Remove missing values
print(x.fillna(0))  # Replace missing values with 0


# ==================== USEFUL METHODS ====================
s = pd.Series([10, 20, 20, 30, 30, 30])

print(s.unique())  # Unique values
print(s.nunique())  # Number of unique values
print(s.value_counts())  # Frequency of each value

print(s.head(3))  # First 3 values
print(s.tail(3))  # Last 3 values


# ==================== APPLY ====================

s = pd.Series([1, 2, 3, 4])
print(s.apply(lambda x: x * x))  # Apply function to each value


# ==================== IMPORTANT ====================

# Series → One-dimensional labeled data structure in Pandas
# Series = Index + Values
# Series can store int, float, string, etc.
# loc[]  → Label-based access
# iloc[] → Position-based access
# isnull()/isna() → Detect missing values
# dropna() → Remove missing values
# fillna() → Replace missing values
# value_counts() → Frequency of each unique value
# unique() → Unique values | nunique() → Number of unique values
# apply() → Apply a function to every value


# ==================== PANDAS DATAFRAME ====================
import pandas as pd
import numpy as np

data = {
    "Name": ["Ayush", "Rahul", "Aman", "Riya"],
    "Age": [22, 23, 21, 22],
    "Marks": [85, 70, 90, 75],
    "City": ["Delhi", "Mumbai", "Delhi", "Lucknow"],
}
df = pd.DataFrame(data)  # 2D labeled tabular data

# ===== BASIC INFO =====
print(df.head())  # First 5 rows | tail() → Last 5
print(df.shape)  # (rows,columns)
print(df.columns)  # Column names
print(df.dtypes)  # Datatype of each column
df.info()  # Structure + datatype + null info
print(df.describe())  # Statistical summary

# ===== SELECTION =====
print(df["Name"])  # One column → Series
print(df[["Name", "Marks"]])  # Multiple columns → DataFrame
print(df.loc[0, "Name"])  # loc → Label based
print(df.iloc[0, 2])  # iloc → Position based
print(df.loc[0:2, ["Name", "Marks"]])  # Rows + columns

# ===== FILTERING =====
print(df[df["Marks"] > 80])  # Single condition
print(df[(df["Marks"] > 75) & (df["Age"] < 23)])  # Multiple conditions → & / |
print(df[df["City"].isin(["Delhi", "Mumbai"])])  # Match multiple values
print(df[df["Marks"].between(75, 90)])  # Values within range
print(df.query("Marks > 80"))  # SQL-like filtering


# ===== ADD / UPDATE =====
df["Passed"] = df["Marks"] >= 75  # Add new column
df.loc[0, "Marks"] = 95  # Update particular value
df["Grade"] = df["Marks"].apply(lambda x: "A" if x >= 80 else "B")  # Custom function
df["City"] = df["City"].map(str.upper)  # Transform each Series value

# ===== MISSING VALUES =====
df.loc[1, "Marks"] = np.nan
print(df.isna().sum())  # Count nulls column-wise
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())  # Fill using mean
# df.dropna(inplace=True)                   # Remove rows containing nulls

# ===== SORT / TOP VALUES =====
df.sort_values("Marks", ascending=False, inplace=True)  # Sort descending
print(df.nlargest(2, "Marks"))  # Top 2 Marks
print(df.nsmallest(2, "Marks"))  # Lowest 2 Marks

# ===== UNIQUE / DUPLICATES =====
print(df["City"].unique())  # Unique values
print(df["City"].nunique())  # Number of unique values
print(df["City"].value_counts())  # Frequency of each value
print(df.duplicated().sum())  # Count duplicate rows
df.drop_duplicates(inplace=True)  # Remove duplicates

# ===== GROUPBY / AGGREGATION =====
print(df.groupby("City")["Marks"].mean())  # Average Marks city-wise
print(df.groupby("City")["Marks"].agg(["count", "mean", "min", "max", "sum"]))

# ===== ROW / COLUMN OPERATIONS =====

df.rename(columns={"Marks": "Score"}, inplace=True)  #     ! Rename column
# df.drop("Passed",axis=1,inplace=True)      # Drop column → axis=1
# !   df.drop(0,axis=0,inplace=True)             # Drop row → axis=0
df.reset_index(drop=True, inplace=True)  # Reset index
# df.set_index("Name",inplace=True)          # ! Make column the index

# ===== MERGE / CONCAT =====

d2 = pd.DataFrame({"Name": ["Ayush", "Rahul"], "Salary": [50000, 40000]})
print(pd.merge(df, d2, on="Name", how="left"))  # SQL-like JOIN
print(pd.concat([df, df], ignore_index=True))  #   ! Combine DataFrames row-wise

# ===== PIVOT TABLE =====
print(pd.pivot_table(df, values="Score", index="City", aggfunc="mean"))

# ===== USEFUL ANALYSIS =====

print(df.select_dtypes(include="number"))  # Select numerical columns
print(df.select_dtypes(include="number").corr())  #  ! Correlation matrix
print(df.sample(2))  #    ! Random 2 rows
print(df["Score"].idxmax())  # Index of maximum Score
print(df["Score"].idxmin())  # Index of minimum Score

# ===== FILE HANDLING =====

# !   df = pd.read_csv("data.csv")               # CSV → DataFrame
# df = pd.read_excel("data.xlsx")            # Excel → DataFrame
#  !  df.to_csv("output.csv",index=False)         # DataFrame → CSV

# ===== REMEMBER =====
# DataFrame → 2D Rows + Columns | Series → 1D labeled data
# loc → Labels | iloc → Positions | axis=0 → Rows | axis=1 → Columns
# ! groupby → Split + Aggregate | merge → SQL-like Join | concat → Combine
# isna/dropna/fillna → Missing values | duplicated/drop_duplicates → Duplicates


# ==================== GROUPBY PRACTICE ====================

import pandas as pd

df = pd.DataFrame(
    {
        "Employee": ["A", "B", "C", "D", "E", "F"],
        "Department": ["IT", "HR", "IT", "HR", "Sales", "Sales"],
        "Gender": ["M", "F", "F", "M", "M", "F"],
        "Salary": [60000, 40000, 80000, 50000, 45000, 55000],
        "Experience": [2, 3, 5, 4, 2, 6],
    }
)

# Q1. Average salary of each department
print(df.groupby("Department")["Salary"].mean())

# Q2. Total salary of each department
print(df.groupby("Department")["Salary"].sum())

# Q3. Maximum salary in each department
print(df.groupby("Department")["Salary"].max())

# Q4. Number of employees in each department
print(df.groupby("Department")["Employee"].count())

# ! Q5. Min, Max and Average salary department-wise
print(df.groupby("Department")["Salary"].agg(["min", "max", "mean"]))

# ! Q6. Multiple aggregations on multiple columns
print(
    df.groupby("Department").agg(
        {"Salary": ["mean", "max"], "Experience": ["mean", "max"]}
    )
)

# Q7. Group using multiple columns
print(df.groupby(["Department", "Gender"])["Salary"].mean())

# Q8. Department-wise Salary + Experience average
print(df.groupby("Department")[["Salary", "Experience"]].mean())

# Q9. Department having highest average salary
avg = df.groupby("Department")["Salary"].mean()
print(avg.idxmax())  # Department name
print(avg.max())  # Highest average salary

#! Q10. Departments whose average salary > 50000
avg = df.groupby("Department")["Salary"].mean()
print(avg[avg > 50000])

# Q11. Sort departments according to average salary
print(df.groupby("Department")["Salary"].mean().sort_values(ascending=False))

# Q12. Get highest paid employee from each department
idx = df.groupby("Department")["Salary"].idxmax()
print(df.loc[idx])

# Q13. Add department average salary to every row
df["Dept_Avg"] = df.groupby("Department")["Salary"].transform("mean")
print(df)

# ! Q14. Employees earning more than their department average
print(df[df["Salary"] > df.groupby("Department")["Salary"].transform("mean")])

# Q15. Convert groupby result back into DataFrame
result = df.groupby("Department", as_index=False)["Salary"].mean()
print(result)

# REMEMBER:
# groupby("col") → Create groups
# ! groupby()["col"].mean() → Group + Aggregate
# agg() → Apply multiple aggregate functions
# groupby(["A","B"]) → Group using multiple columns
# transform() → Returns result matching original DataFrame size
# idxmax() → Index of maximum value in each group
# ! as_index=False → Group column remains a normal column



# ! ==================== PANDAS + SERIES TRICKY OUTPUT QUESTIONS ====================
import pandas as pd
import numpy as np

# Q1. Label vs Position
s = pd.Series([10, 20, 30], index=[1, 2, 3])
print(s[1])  # 10 → label 1
print(s.iloc[1])  # 20 → position 1

# Q2. Series Index Alignment
s1 = pd.Series([10, 20, 30], index=["a", "b", "c"])
s2 = pd.Series([1, 2, 3], index=["b", "c", "d"])
print(s1 + s2)  # a=NaN,b=21,c=32,d=NaN → aligns indexes

# Q3. Boolean Filtering
s = pd.Series([10, 20, 30, 40])
print(s[s > 20])  # 2:30, 3:40 → original indexes retained

# Q4. Single vs Multiple Columns
df = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
print(type(df["A"]))  # Series
print(type(df[["A"]]))  # DataFrame

# ! Q5. loc vs iloc Slicing
df = pd.DataFrame({"A": [10, 20, 30, 40]})
print(df.loc[1:2])  # Rows 1 AND 2 → end included
print(df.iloc[1:2])  # Only position 1 → end excluded

# Q6. NaN Comparison
s = pd.Series([10, np.nan, 30])
print(s == np.nan)  # False False False
print(s.isna())  # False True False

# Q7. count() vs size
s = pd.Series([10, np.nan, 30, np.nan])
print(s.count())  # 2 → ignores NaN
print(s.size)  # 4 → counts everything

# ! Q8. unique() / nunique()
s = pd.Series([10, 20, 20, 30, 30])
print(s.unique())  # [10 20 30]
print(s.nunique())  # 3

# Q9. value_counts()
s = pd.Series(["A", "B", "A", "C", "A"])
print(s.value_counts())  # A=3,B=1,C=1 → frequency

# Q10. duplicated()
s = pd.Series([10, 10, 20, 20, 30])
print(s.duplicated())  # F,T,F,T,F → first occurrence not duplicate

# Q11. Sorting Retains Index
s = pd.Series([30, 10, 20])
print(s.sort_values())  # 1:10,2:20,0:30 → index NOT reset

# Q12. drop() Without Assignment
df = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df.drop("A", axis=1)
print(df.columns)  # A,B → original df unchanged

# Q13. fillna() Without Assignment
s = pd.Series([10, np.nan, 30])
s.fillna(0)
print(s)  #  ! NaN remains → result wasn't assigned

# Q14. apply()
s = pd.Series([1, 2, 3, 4])
print(s.apply(lambda x: x * 2 if x % 2 == 0 else x))  # 1,4,3,8

# ! Q15. GroupBy Aggregation
df = pd.DataFrame({"Dept": ["IT", "IT", "HR", "HR"], "Salary": [10, 20, 30, 50]})
print(df.groupby("Dept")["Salary"].mean())  # HR=40, IT=15 → rows reduced

# Q16. GroupBy transform()
df["Avg"] = df.groupby("Dept")["Salary"].transform("mean")
print(df)  # Avg repeated for every original row

# Q17. Employee > Department Average
print(df[df["Salary"] > df.groupby("Dept")["Salary"].transform("mean")])
# IT:20 and HR:50

# ! Q18. DataFrame Column Alignment
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"B": [10, 20], "C": [30, 40]})
print(df1 + df2)  # B=13,24 | A,C=NaN → aligns columns

# Q19. any() vs all()
s = pd.Series([True, True, False])
print(s.any())  # True → at least one True
print(s.all())  # False → not all True

# ! Q20. isin() + between()
df = pd.DataFrame({"A": [10, 20, 30, 40]})
print(df[df["A"].isin([10, 30])])  # 10,30
print(df[df["A"].between(20, 30)])  # 20,30 → boundaries included

# Q21. idxmax()
s = pd.Series([50, 90, 70], index=["A", "B", "C"])
print(s.max())  # 90 → maximum VALUE
print(s.idxmax())  # B → INDEX of maximum value

# Q22. reset_index()
s = pd.Series([30, 10, 20]).sort_values()
print(s.index.tolist())  # [1,2,0]
print(s.reset_index(drop=True))  # New index → 0,1,2

# Q23. IMPORTANT VIEW/COPY RULE
df = pd.DataFrame({"A": [1, 2, 3]})
df.loc[0, "A"] = 100  # Recommended safe modification
print(df)  # 100,2,3

# ==================== REMEMBER ====================
# loc → Label | iloc → Position | loc slicing includes end
# Pandas arithmetic → aligns labels/indexes, NOT blindly by position
# count() ignores NaN | size counts NaN | NaN == NaN → False
# ! df["A"] → Series | df[["A"]] → DataFrame
# sort/filter → usually retain original index
# agg() → reduces rows | transform() → preserves original row count
#  ! drop()/fillna() → assign result back unless modifying appropriately
# max() → max value | idxmax() → index/label of max value
```

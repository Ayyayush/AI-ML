# Self in Python
Python me self ek instance ka reference hota hai jo har class ke method me pehla parameter hota hai.
Simple words me:
Jab aap kisi object ke method ko call karte ho, Python automatically us object ka reference method ko pass karta hai.
Us reference ko conventionally self kaha jaata hai.
self ka use karke aap instance variables (object ke data) aur dusre methods ko access karte ho.




### 1️⃣ Class Definition Rules

In Python, a class is defined using the keyword class.
It acts as a blueprint for creating objects (instances).

Syntax:

class ClassName:
    # class body
    pass

The class name should always follow the PascalCase convention (each word starts with a capital letter).
Example: Student, BankAccount, CarModel.
The body of the class must be properly indented since Python uses indentation to define scope.
You can include a docstring inside a class to describe its purpose.

 # The __init__() method is a constructor in Python.

It is automatically called when a new object is created.
It initializes the object’s attributes.
Inside the class, variables are of two types:
Instance variables → unique to each object.
Class variables → shared by all objects of the class.



### 2️⃣ Object Creation Rules

An object is an instance of a class.
It represents a specific implementation of the class blueprint.
To create an object, you simply call the class name like a function:
obj = ClassName()

When the object is created, Python automatically calls the __init__() method.
Each object gets its own copy of instance variables.
You can access the attributes and methods of an object using the dot (.) operator.
Example:

obj.attribute
obj.method()





### 3️⃣ The self Keyword

The self parameter refers to the current instance of the class.
It must be the first parameter in every instance method, including __init__().
You do not pass self explicitly while calling the method; Python passes it automatically.
It is used to access instance variables and methods from within the class.





### 4️⃣ Class and Instance Variables

Instance variables are defined inside the constructor (__init__() method).
Each object has its own copy of these variables.
Class variables are declared directly under the class and are shared among all objects.
They can be accessed using either the class name or the object name, but using the class name is preferred.





### 5️⃣ Types of Methods in a Class

Python provides three types of methods inside a class:

Instance Method — Works with instance variables; uses self.
Class Method — Works with class variables; uses cls and is defined using the decorator @classmethod.
Static Method — Independent of both instance and class; defined using @staticmethod.




### 6️⃣ Object Lifecycle

When an object is created, the __init__() method is automatically called to initialize it.
When the object is destroyed or goes out of scope, the __del__() method (destructor) is called automatically.





### 7️⃣ Best Practices for Creating Classes
Use meaningful class names that reflect the purpose.
Keep methods short and specific to one task.




# Use __str__() to provide a human-readable string representation of the object.
Follow OOP principles such as:

Encapsulation – hiding data inside the class.
Inheritance – reusing code through parent classes.
Polymorphism – same method behaving differently in different contexts.
Abstraction – showing only essential details.


````python
class ATM:
    def __init__(self):                       # Constructor to initialize PIN and balance
        self.pin = ""                         # Initially empty PIN
        self.balance = 0                      # Initially zero balance

    def menu(self):                           # Menu to interact with ATM
        print("\nHi, how can I help you?")
        print("1. Press 1 to create PIN")
        print("2. Press 2 to change PIN")
        print("3. Press 3 to check balance")
        print("4. Press 4 to withdraw")
        print("5. Anything else to exit")

        user_input = input("Choose your option: ")     # User selects option

        if user_input == '1':
            self.create_pin()              # Call create_pin method
        elif user_input == '2':
            self.change_pin()              # Call change_pin method
        elif user_input == '3':
            self.check_balance()           # Call check_balance method
        elif user_input == '4':
            self.withdraw()                # Call withdraw method
        else:
            print("Bye! 👋")               # Exit case

    def create_pin(self):                  # Method to create a PIN
        self.pin = input("Enter your new PIN: ")
        self.balance = int(input("Enter your starting balance: "))
        print("PIN created successfully ✅")

    def change_pin(self):                  # Method to change existing PIN
        old_pin = input("Enter your current PIN: ")
        if old_pin == self.pin:
            new_pin = input("Enter your new PIN: ")
            self.pin = new_pin
            print("PIN changed successfully ✅")
        else:
            print("Incorrect PIN ❌")

    def check_balance(self):              # Method to check balance
        entered_pin = input("Enter your PIN to view balance: ")
        if entered_pin == self.pin:
            print(f"Your current balance is ₹{self.balance}")
        else:
            print("Incorrect PIN ❌")

    def withdraw(self):                   # Method to withdraw money
        entered_pin = input("Enter your PIN to withdraw: ")
        if entered_pin == self.pin:
            amount = int(input("Enter amount to withdraw: "))
            if amount <= self.balance:
                self.balance -= amount
                print(f"₹{amount} withdrawn successfully ✅")
                print(f"Remaining balance: ₹{self.balance}")
            else:
                print("Insufficient balance ❌")
        else:
            print("Incorrect PIN ❌")


# Creating an object of ATM class and calling menu
obj = ATM()
obj.menu()
````

# Private vs Public in Python

Python me variables aur methods ke **access control** kaafi flexible hote hain.  
Unlike Java ya C++, Python me koi strict "private" ya "public" keyword nahi hota — lekin hum **naming conventions** ka use karke data ko public ya private banane ka signal dete hain.

---

## **1. Public Members**
- **Definition:** Ye wo variables ya methods hote hain jo class ke bahar se directly accessible hote hain.
- **Naming convention:** Normal naam use hota hai (koi underscore nahi).
- **Access:** Class ke andar aur bahar dono jagah se access kiya ja sakta hai.

**Example:**
```python
class Person:
    def __init__(self, name):
        self.name = name   # public attribute

    def greet(self):       # public method
        print(f"Hello, my name is {self.name}")

p = Person("Rahul")
print(p.name)   # Direct access allowed
p.greet()       # Method call allowed



## **2. Private Members



````python
class Example:

    def __init__(self):
        self.public_var = 10          # Public
        self._protected_var = 20      # Protected (convention)
        self.__private_var = 30       # Private (name mangling)

    def display(self):
        print(self.public_var)
        print(self._protected_var)
        print(self.__private_var)


obj = Example()
obj.display()

print(obj.public_var)         # ✅ Allowed
print(obj._protected_var)     # ⚠️ Allowed (but avoid)
# print(obj.__private_var)    # ❌ Error
````

# Methods vs Functions in Python

Python me **method** aur **function** dono ka kaam hota hai ek specific task perform karna, lekin inka scope aur use-case alag hota hai.

---

## 1. Function
- **Definition:** Function ek block of code hota hai jo **independent** hota hai, kisi class/object ka part nahi hota.
- **Use:** Jab ek kaam baar-baar karna ho bina kisi specific object ke.
- **Call:** Function ka naam likh ke call hota hai.
- **Syntax:**

def function_name(parameters):
    # code block
    return value



## 2. Methods 
- Definition: Method ek function hota hai jo class ke andar define hota hai aur kisi - object ka part hota hai.
- Use: Object ke data (attributes) ke saath kaam karne ke liye.
- Call: Object ke reference ke saath call hota hai (object.method()).
- First Parameter: Instance methods me pehla parameter hamesha self hota hai jo   object ka reference hota hai.

Syntax:

class ClassName:
    def method_name(self, parameters):
        # code block
        return value


# Magic Methods / Dunder Methods

Magic methods (jinhe **dunder methods** bhi kaha jata hai — “double underscore” ka short form) Python ke special methods hote hain jo **class ke behaviour ko define ya modify** karte hain.  
Ye methods **automatically call hote hain** specific situations me, bina explicitly unka naam likhe call kiye.

---

## 1. Why They Are Called Dunder Methods
- **Dunder** = **D**ouble **UNDER**score
- Har magic method ka naam double underscore (`__`) se start aur end hota hai.  
  Example: `__init__`, `__str__`, `__len__`

---

## 2. Key Characteristics
- Ye Python ke built-in features ka part hote hain.
- Inhe **directly call karne ki bajaye** Python internally call karta hai jab koi specific action perform hota hai.
- Ye **operator overloading**, **object creation**, **representation**, aur **comparison** jaisi functionalities ko control karte hain.

---

## 3. Common Categories of Magic Methods

### a) Object Initialization & Construction
- `__init__` → Object initialize karna
- `__new__` → Object create karna (memory allocation level pe)
- `__del__` → Object destroy karna

### b) String Representation
- `__str__` → Human-readable representation dena
- `__repr__` → Developer-friendly representation dena

### c) Comparison Operations
- `__eq__` → Equality check (`==`)
- `__ne__` → Not equal (`!=`)
- `__lt__` → Less than (`<`)
- `__gt__` → Greater than (`>`)
- `__le__` → Less than or equal (`<=`)
- `__ge__` → Greater than or equal (`>=`)

### d) Arithmetic Operations
- `__add__` → Addition (`+`)
- `__sub__` → Subtraction (`-`)
- `__mul__` → Multiplication (`*`)
- `__truediv__` → Division (`/`)
- `__floordiv__` → Floor division (`//`)
- `__mod__` → Modulus (`%`)
- `__pow__` → Power (`**`)

### e) Container Methods
- `__len__` → Length nikalna
- `__getitem__` → Index ke through element access
- `__setitem__` → Index pe element set karna
- `__delitem__` → Element delete karna
- `__contains__` → Membership check (`in`)

### f) Callable & Attribute Access
- `__call__` → Object ko function ki tarah call karna
- `__getattr__` → Jab attribute na mile tab ka behaviour
- `__setattr__` → Attribute set karna
- `__delattr__` → Attribute delete karna

---

## 4. Benefits of Using Magic Methods
- **Cleaner syntax** — direct operators ka use kar sakte ho.
- **Custom behaviour** — apne object ka default Python behaviour modify kar sakte ho.
- **Better readability** — objects ko human-friendly bana sakte ho.
- **Integration with built-ins** — apne class ko Python ke built-in functions jaise `len()`, `str()`, `sum()` ke saath compatible bana sakte ho.

---

## 5. Summary Table of Popular Magic Methods

| Category             | Method       | Purpose                               |
|----------------------|-------------|---------------------------------------|
| Initialization       | `__init__`  | Object initialization                |
| Representation       | `__str__`   | Human-readable string representation |
| Representation       | `__repr__`  | Developer/debug string representation|
| Arithmetic           | `__add__`   | Addition operator overloading        |
| Arithmetic           | `__sub__`   | Subtraction operator overloading     |
| Comparison           | `__eq__`    | Equality check (`==`)                |
| Comparison           | `__lt__`    | Less than (`<`)                      |
| Container            | `__len__`   | Length of object                     |
| Container            | `__getitem__`| Get element by index                 |
| Callable             | `__call__`  | Make object callable like function   |

---

## 6. Final Notes
- Magic methods ka use **carefully** karna chahiye, taki code readable aur maintainable rahe.
- Ye mostly **operator overloading** aur **Python built-in behaviour ke saath integration** ke liye use hote hain.
- Har magic method ka ek specific trigger hota hai jo Python automatically handle karta hai.


# 📘 What does `f` mean in a string? (f-String Explanation)

---

## 🔹 What does `f` mean?

The **`f`** before a string means **Formatted String Literal**, commonly called an **f-string**.

👉 Hinglish:
> `f` Python ko bolta hai:  
> “Is string ke andar jo `{}` hai, uske andar expression ko evaluate karo.”

---

## 🔹 Why do we need `f`?

Normally, Python strings **sirf text** hoti hain.  
Agar hum kisi variable ya expression ki value string ke andar dikhana chahte hain, to Python ko batana padta hai.

👉 `f` ka kaam:
- `{}` ke andar likhe code ko **run karna**
- Uski value ko **string me insert karna**

---

## 🔹 What happens WITHOUT `f`?

Agar `f` nahi likhenge:

- Python `{self.name}` ko **plain text** samjhega
- Expression evaluate **nahi** hoga

👉 Hinglish:
> Python bolega:  
> “Mujhe laga tum text likh rahe ho.”

---

## 🔹 What happens WITH `f`?

Agar `f` likha hai:

- Python `{}` ke andar ka code **execute karta hai**
- Result ko string me **replace** kar deta hai

👉 Hinglish:
> Python bolega:  
> “Achha, value chahiye — laata hoon.”

---

## 🔹 `{}` ke andar kya likh sakte hain?

- Variables  
- Expressions  
- Function calls  
- Calculations  

👉 Hinglish:
> `{}` ke andar jo valid Python expression hai, wo likh sakte ho.

---

## 🔹 Why f-strings are preferred?

- Readable hoti hain  
- Short syntax  
- Fast execution  
- Clean code  

👉 Hinglish:
> f-strings modern aur clean tareeka hai string formatting ka.

---

## 🔹 Memory-Level Understanding (Simple)

- `f` dekhte hi Python:
  1. String ko parse karta hai  
  2. `{}` ke expressions evaluate karta hai  
  3. Final string bana deta hai  

👉 Hinglish:
> Pehle value nikalo, phir string me chipkao.

---

## 🔹 Interview One-Liner 🧠

> The `f` in Python strings stands for formatted string literal and allows expressions inside `{}` to be evaluated at runtime.

---

## 🔹 One-Line Summary

> Without `f`, `{}` is just text.  
> With `f`, `{}` becomes executable expression.

````python
from math import gcd     # ✅ Fractions simplify karne ke liye gcd ka use karenge

class Fraction:
    
    # ✅ Parameterized constructor: initialize numerator and denominator
    def __init__(self, x, y):
        self.num = x                     # numerator set karo
        self.den = y                     # denominator set karo

        self.simplify()                 # har object banate hi fraction ko simplify karo
        
    # ✅ Simplify function to reduce the fraction to lowest terms
    def simplify(self):
        g = gcd(self.num, self.den)     # numerator aur denominator ka gcd nikalo
        self.num //= g                  # dono ko gcd se divide karo
        self.den //= g
        
    # ✅ String representation function (for print)
    def __str__(self):
        return '{}/{}'.format(self.num, self.den)     # fraction ko string mein return karo
    
    # ✅ Overloading + operator
    def __add__(self, other):
        # a/b + c/d = (ad + bc) / bd
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)             # result ko simplify karke return
        
    # ✅ Overloading - operator
    def __sub__(self, other):
        # a/b - c/d = (ad - bc) / bd
        new_num = self.num * other.den - other.num * self.den
        new_den = self.den * other.den
        return Fraction(new_num, new_den)             # result ko simplify karke return

    # ✅ Overloading * operator
    def __mul__(self, other):
        # a/b * c/d = (ac) / (bd)
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num, new_den)             # result ko simplify karke return

    # ✅ Overloading / operator
    def __truediv__(self, other):
        # a/b ÷ c/d = (a*d) / (b*c)
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num, new_den)             # result ko simplify karke return

# ------------------ Testing ------------------

# ✅ Create two fraction objects
fr1 = Fraction(1, 2)      # 1/2
fr2 = Fraction(1, 3)      # 1/3

# ✅ Perform operations
print("Fraction 1:", fr1)
print("Fraction 2:", fr2)

print("Addition:", fr1 + fr2)        # 1/2 + 1/3 = 5/6
print("Subtraction:", fr1 - fr2)     # 1/2 - 1/3 = 1/6
print("Multiplication:", fr1 * fr2)  # 1/2 * 1/3 = 1/6
print("Division:", fr1 / fr2)        # 1/2 ÷ 1/3 = 3/2
````

# Class Relationships in Python

Python mein classes ke beech commonly do tarah ke relationships hote hain:

1. Aggregation
2. Inheritance

---

## 🔸 Aggregation

**Definition:**  
Aggregation ka matlab hota hai **"has-a" relationship**. Iska matlab yeh hota hai ki ek class ke andar dusri class ka object use kiya jaata hai, lekin dono classes independent hoti hain.

**Key Points:**
- Aggregation ka use tab hota hai jab ek class ko dusri class ki functionality chahiye hoti hai.
- Lekin dono classes apne aap mein alag-alag exist kar sakti hain.
- Isme tight coupling nahi hoti — agar ek object destroy ho jaaye toh doosra phir bhi exist kar sakta hai.
- Ek object dusre object ke bina bhi jee sakta hai.

**Real-world Analogy:**  
- "A Customer has an Address" → Customer aur Address dono alag-alag exist kar sakte hain. Yeh **Aggregation** hai.

---



```python
# Aggregation example

# Customer class define kar rahe hain
class Customer:
    
    def __init__(self, name, gender, address):                          # Constructor jo name, gender aur address leta hai
        self.name = name                                                # name ko store kar rahe
        self.gender = gender                                            # gender ko store kar rahe
        self.address = address                                          # Address object ko store kar rahe (yeh hi aggregation hai)
        
    def print_address(self):                                           # Customer ka address print karne ke liye method
        print(self.address.city, self.address.pin, self.address.state) # Address object ke attributes ko access karke print kar rahe hain
        
    def edit_profile(self, new_name, new_city, new_pin, new_state):    # Customer ka profile update karne ke liye method
        self.name = new_name                                           # naya naam update kar rahe
        self.address.edit_address(new_city, new_pin, new_state)        # address object ke through nayi address details bhej rahe

# Address class define kar rahe hain
class Address:
    
    def __init__(self, city, pin, state):                              # Constructor for city, pin and state
        self.city = city                                               # city assign kar rahe
        self.pin = pin                                                 # pin assign kar rahe
        self.state = state                                             # state assign kar rahe
        
    def edit_address(self, new_city, new_pin, new_state):             # Address ko update karne ke liye method
        self.city = new_city                                           # nayi city assign
        self.pin = new_pin                                             # nayi pin assign
        self.state = new_state                                         # nayi state assign

# Pehle ek address object bana rahe
add1 = Address('gurgaon', 122011, 'haryana')                           # Address object create kiya

# Fir ek customer object bana rahe jisme address object pass kar rahe (Aggregation)
cust = Customer('nitish', 'male', add1)                                # Aggregation ho gaya - Address object ko Customer ke andar use kar rahe

cust.print_address()                                                  # Pehla address print hoga: gurgaon 122011 haryana

cust.edit_profile('ankit', 'mumbai', 111111, 'maharashtra')           # naam aur address ko update kar rahe

cust.print_address()                                                  # Updated address print hoga: mumbai 111111 maharashtra
```

## 🔹 Inheritance

**Definition:**  
Inheritance ka matlab hota hai **"is-a" relationship**. Yaani ek class (child class) dusri class (parent class) ke properties aur methods inherit karti hai.

**Key Points:**
- Child class parent class ke saare functions aur variables automatically le leti hai.
- Iska main purpose hai **code reusability** aur **hierarchical structure** banana.
- Inheritance se naye features existing structure par based banaaye ja sakte hain.
- Python mein single, multiple, multilevel, hierarchical inheritance possible hai.

**Real-world Analogy:**  
- "A Dog is an Animal" → Dog class Animal class se inherit karegi. Yeh **Inheritance** hai.

---


```python
# INHERITANCE IN PYTHON eg code ::

class user:
    
    def __init__(self):
        self.name='nitish'
    
    def login(self):
        print('login')
        
# bracket mein user likh ke hum bta rhe h student is child class of user
    class Student(user):
        
    # child class ka constructor hoga toh method overriding hoga 
    # child constrcutor mil gua toh parent constrcutor mein jayega hi nhi 
    # parent constrcutor chala hi nhi toh name create hi nhi hua  toh chal nhi paya  
    
        def __init__(self):
            self.rollno=100
            
        def enroll(self):
            print('enroll into the course')
            
    u=user()
    s=Student
    print(s.name)
    print(s.rollno)
    s.login()
    s.enroll()
```

```python
# Constructor example in Python using inheritance

class Phone:
    def __init__(self, price, brand, camera):                      # Constructor of parent class Phone
        print("Inside phone constructor")                          # Constructor call hone par print karega
        self.price = price                                         # Instance variable price set karna
        self.brand = brand                                         # Instance variable brand set karna
        self.camera = camera                                       # Instance variable camera set karna

    def buy(self):                                                 # Ek method jo kuch action perform karega
        print("Buying a phone")                                    # Print karega jab buy() method call hoga


# Child class SmartPhone jo Phone class ko inherit kar rahi hai
class SmartPhone(Phone):
    pass                                                           # Child class mein abhi koi extra functionality nahi di hai


# Object bana rahe hain SmartPhone class ka
s = SmartPhone(20000, "Apple", 13)                                 # Constructor automatically parent class ka call hoga
```

```python
# Accessing using private 

class Parent:
    def __init__(self, num):                         # Constructor: num ko __num variable mein store karte hain
        self.__num = num

    def get_num(self):                               # Getter method: __num return karta hai
        return self.__num

class Child(Parent):                                 # Child class, Parent class se inherit kar rahi hai
    def show(self):                                  # Child ka khud ka method
        print("This is in child class")

son = Child(100)                                     # Child class ka object banaya, constructor Parent se aaya
print(son.get_num())                                 # Parent class ka method call kiya
son.show()                                           # Child class ka method call kiya
```

# Method Overriding in Python

## 📌 Definition

Method Overriding tab hota hai jab **child class** (yaani derived class) ek method ko **dubara define karti hai** jo pehle se **parent class** (base class) mein defined hoti hai.

Iska main purpose hota hai:  
🔁 **Parent class ke behaviour ko modify karna** according to child class ki zarurat.

---

## 🔍 Key Points

- Parent class ke method ka **same name** child class mein dubara likha jaata hai.
- Child class ka method parent wale method ko **override** (yaani replace) kar deta hai.
- Run-time pe decide hota hai ki kaunsa method call hoga → isiliye ise **run-time polymorphism** bhi kehte hain.
- Agar child class override nahi karti, toh parent class ka method by default use hoga.

---

## ✅ Conditions for Overriding

- Method ka **naam same** hona chahiye.
- Parameters (function signature) bhi same hone chahiye (optional in Python due to dynamic typing).
- Inheritance ka hona zaroori hai — bina parent-child relationship ke overriding possible nahi hai.

---

## 🧠 Real-world Analogy

Agar ek `Animal` class ka `speak()` method "Animal sound" print karta hai, aur `Dog` class usi `speak()` method ko override karke "Dog barks" likhti hai — toh yeh method overriding hai.

---

## ⚙️ Use-Cases

- Jab child class ko parent class ke default behavior ko customize karna hota hai.
- Jab same function ka behavior har child class ke liye alag hona chahiye (like in polymorphism).
- Frameworks aur libraries mein method overriding bahut common hai, especially GUI ya Web apps mein.

---

## 🧩 Related Concepts

- **Method Overloading** (not natively supported in Python)
- **Polymorphism**
- **super() function** → Parent class ka overridden method call karne ke liye use hota hai.

---



```python

# -------------------------------
# EMPLOYEE → DEVELOPER (Inheritance)
# -------------------------------

class Employee:
    
    def __init__(self, name, salary):
        self.name = name                  # Employee ka naam store karta hai
        self.salary = salary              # Employee ki salary store karta hai
        
    def get_details(self):
        # Employee ki basic details return karta hai
        return f"Name: {self.name}, Salary: {self.salary}"


class Developer(Employee):
    
    def __init__(self, name, salary, language):
        super().__init__(name, salary)    # Parent class (Employee) ka constructor call
        self.language = language          # Developer ki programming language
        
    def get_details(self):
        # Parent method ko override kar rahe hain (Polymorphism)
        return (
            f"Name: {self.name}, "
            f"Salary: {self.salary}, "
            f"Language: {self.language}"
        )


# -------------------------------
# VEHICLE HIERARCHY
# -------------------------------

class Vehicle:
    
    def __init__(self, brand, speed):
        self.brand = brand                # Vehicle ka brand
        self.speed = speed                # Vehicle ki speed
        
    def show_info(self):
        return f"Brand: {self.brand}, Speed: {self.speed}"


class Car(Vehicle):
    
    def __init__(self, brand, speed, fuel_type):
        super().__init__(brand, speed)    # Vehicle constructor call
        self.fuel_type = fuel_type        # Car ka fuel type
        
    def show_info(self):
        # Method overriding
        return (
            f"Brand: {self.brand}, "
            f"Speed: {self.speed}, "
            f"Fuel Type: {self.fuel_type}"
        )


class Bike(Vehicle):
    
    def __init__(self, brand, speed, engine_cc):
        super().__init__(brand, speed)    # Vehicle constructor call
        self.engine_cc = engine_cc         # Bike ka engine capacity
        
    def show_info(self):
        # Method overriding
        return (
            f"Brand: {self.brand}, "
            f"Speed: {self.speed}, "
            f"Engine CC: {self.engine_cc}"
        )


# -------------------------------
# BANK ACCOUNT
# -------------------------------

class BankAccount:
    
    def __init__(self, holder_name, balance):
        self.holder_name = holder_name    # Account holder ka naam
        self.balance = balance            # Initial balance
        
    def deposit(self, amount):
        self.balance += amount            # Balance me paisa add kar rahe
        return f"Deposited: {amount}, New Balance: {self.balance}"
        
    def get_balance(self):
        return f"Current Balance: {self.balance}"


class SavingsAccount(BankAccount):
    
    def __init__(self, holder_name, balance, interest_rate):
        super().__init__(holder_name, balance)   # Parent constructor call
        self.interest_rate = interest_rate       # Interest rate store
        
    def add_interest(self):
        # Interest calculate karke balance me add kar rahe
        interest = self.balance * self.interest_rate / 100
        self.balance += interest
        return f"Interest Added: {interest}, New Balance: {self.balance}"


# -------------------------------
# OBJECT CREATION & TESTING
# -------------------------------

dev = Developer("Amit", 60000, "Python")
print(dev.get_details())

car = Car("Honda", 180, "Petrol")
bike = Bike("Yamaha", 140, 150)

print(car.show_info())
print(bike.show_info())

account = SavingsAccount("Rahul", 10000, 5)
print(account.deposit(2000))
print(account.add_interest())
print(account.get_balance())
```

# IMPORTANT Keywords in OOPS in Python

Object-Oriented Programming (OOPs) ke concepts ko implement karne ke liye Python kuch **special keywords** provide karta hai jo class aur object ke behavior ko control karte hain.

Neeche sabse important OOPS keywords aur unke uses diye gaye hain:

---

## 1️⃣ `class`

- Ye keyword ek **class define** karne ke liye use hota hai.
- Class ek **blueprint** hoti hai objects banane ke liye.
- Iske andar methods (functions) aur attributes (variables) define kiye jaate hain.

---

## 2️⃣ `self`

- `self` ek **reference** hota hai current object ka.
- Jab bhi class ka method banate hain, pehla parameter hamesha `self` hota hai.
- Iske through hum object ke andar ke variables ya methods ko access karte hain.

---

## 3️⃣ `__init__`

- Ye ek **constructor method** hai.
- Jab bhi object create hota hai, ye method **automatically call** hota hai.
- Iska kaam hota hai object ke initial values set karna.

---

## 4️⃣ `__str__`

- Ye ek **special method** hai jo object ko string form mein represent karta hai.
- Jab `print(object)` likhte ho, toh `__str__` method call hota hai.
- Custom output dene ke liye override kiya jaata hai.

---

## 5️⃣ `inheritance` related keywords

### 🔹 `super()`

- Parent class ke method ya constructor ko call karne ke liye use hota hai.
- Method overriding mein parent ka version access karne ke liye helpful hai.

---

## 6️⃣ `isinstance()` and `issubclass()`

- `isinstance(obj, Class)` → Check karta hai kya object us class ka instance hai.
- `issubclass(Derived, Base)` → Check karta hai kya ek class dusri class se inherit karti hai.

---

## 7️⃣ `@staticmethod` and `@classmethod`

### 🔸 `@staticmethod`
- Ye method class ke context mein hota hai, but `self` ya `cls` nahi leta.
- Object ya class dono ke through call ho sakta hai.

### 🔸 `@classmethod`
- Ye method class ko as argument leta hai (`cls`).
- Isse hum class-level data access ya modify kar sakte hain.

---

## 8️⃣ `pass`

- Jab class ya function ke andar abhi koi logic likhna nahi chahte ho, tab `pass` likh dete hain.
- Syntax error avoid karne ke liye use hota hai as a **placeholder**.

---

## 9️⃣ `del`

- Kisi object ya attribute ko **delete** karne ke liye use hota hai.
- Python garbage collector baad mein memory free karta hai.

---

## 🔟 `None`

- Ye ek **special constant** hai jo "koi value nahi hai" dikhata hai.
- Constructor ya method default return kare toh `None` hota hai.

---



```python
# Demonstration of 'super()' to call the constructor of the parent class

# Parent Class: Phone
class Phone:
    # Constructor of Phone class
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self._price = price         # _price is a protected variable
        self.brand = brand          # brand of the phone
        self.camera = camera        # camera megapixels
        

# Child Class: SmartPhone inherits from Phone
class SmartPhone(Phone):
    # Constructor of SmartPhone class
    def __init__(self, price, brand, camera, os, ram):
        # Calling the parent class constructor using super()
        super().__init__(price, brand, camera)
        
        # Additional attributes for SmartPhone
        self.os = os                # operating system
        self.ram = ram              # RAM size
        print("Inside smartphone constructor")
        
        

# Creating an object of SmartPhone
s = SmartPhone(20000, "Samsung", 12, "Android", 2)

# Accessing SmartPhone's and inherited Phone's attributes
print(s.os)        # Output: Android (from SmartPhone class)
print(s.brand)     # Output: Samsung (inherited from Phone class)
```

# TYPES OF INHERITANCE IN Python

Python mein inheritance ek powerful feature hai jo allow karta hai ek class ko dusri class ke properties aur methods **reuse** karne ke liye.

Isse code **modular** aur **maintainable** banta hai.

Python mein **multiple types of inheritance** supported hain:

---

## 1️⃣ Single Inheritance

- Jab ek **child class sirf ek parent class** se inherit karti hai, usse single inheritance kehte hain.
- Isme child class parent class ke saare attributes aur methods use kar sakti hai.

🧠 **Example sochne ka tarika:**  
"Dog inherits Animal" → Sirf ek base class involved hai.

---

## 2️⃣ Multiple Inheritance

- Jab ek **child class multiple parent classes** se inherit karti hai, use multiple inheritance kehte hain.
- Python is feature ko allow karta hai using comma-separated base classes.
- Isme method resolution order (MRO) kaafi important ho jaata hai.

🧠 **Example sochne ka tarika:**  
"Child inherits from Mother and Father"

---

## 3️⃣ Multilevel Inheritance

- Jab ek class ek parent se inherit karti hai, aur uska child bhi further kisi aur class ka parent ban jaata hai — to yeh multilevel inheritance hoti hai.
- Yaani **parent → child → grandchild** relationship.

🧠 **Example sochne ka tarika:**  
"Animal → Mammal → Dog"

---

## 4️⃣ Hierarchical Inheritance

- Jab **ek single parent class** se **multiple child classes** inherit karte hain, use hierarchical inheritance kehte hain.
- Sabhi child classes apne parent ke features inherit karte hain, lekin ek dusre se related nahi hote.

🧠 **Example sochne ka tarika:**  
"Animal → Dog, Cat, Horse"

---

## 5️⃣ Hybrid Inheritance

- Jab inheritance ka structure **mix hota hai multiple types ka** (e.g., multilevel + multiple), use hybrid inheritance kehte hain.
- Python isko handle karta hai **Method Resolution Order (MRO)** ke through to avoid ambiguity.

🧠 **Example sochne ka tarika:**  
Aisa structure jahan ek class multilevel mein bhi involved ho aur dusri multiple mein bhi — mixed structure.

---

## 💡 Method Resolution Order (MRO)

- Jab Python ko decide karna hota hai ki kaunsa method run karna hai jab multiple parents ho, tab wo **MRO** (C3 Linearization) use karta hai.
- `ClassName.__mro__` ya `help(ClassName)` se MRO dekh sakte ho.

---



```python
# Demo Codes for All Types of Inheritance in Python



## 🔹 Single Inheritance


# Parent class
class Animal:
    def speak(self):
        print("Animal speaks")

# Child class
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Object creation
d = Dog()
d.speak()   # Inherited from Animal
d.bark()    # Defined in Dog
```

```python
# MULTILEVEL
# Base class
class Animal:
    def breathe(self):
        print("Animal breathes")

# Intermediate class
class Mammal(Animal):
    def walk(self):
        print("Mammal walks")

# Derived class
class Dog(Mammal):
    def bark(self):
        print("Dog barks")

# Object creation
d = Dog()
d.breathe()   # From Animal
d.walk()      # From Mammal
d.bark()      # From Dog
```

```python
# MULTIPLE
# First parent class
class Father:
    def skills(self):
        print("Father knows gardening")

# Second parent class
class Mother:
    def skills(self):
        print("Mother knows cooking")

# Child class inheriting from both
class Child(Father, Mother):
    def show(self):
        print("Child inherits skills")

# Object creation
c = Child()
c.show()
c.skills()    # MRO: Father's skills() will be called first
```

```python
## 🔹 Hierarchical Inheritance


# Parent class
class Animal:
    def speak(self):
        print("Animal makes sound")

# Child class 1
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Child class 2
class Cat(Animal):
    def meow(self):
        print("Cat meows")

# Object creation
d = Dog()
d.speak()   # Inherited from Animal
d.bark()    # Specific to Dog

c = Cat()
c.speak()   # Inherited from Animal
c.meow()    # Specific to Cat
```

```python
## 🔹 Hybrid Inheritance

# Base class
class Animal:
    def breathe(self):
        print("Animal breathes")

# First-level derived class
class Mammal(Animal):
    def feed_milk(self):
        print("Mammal feeds milk")

# Another base class
class Aquatic:
    def swim(self):
        print("Aquatic animal swims")

# Final class inheriting from Mammal and Aquatic (Hybrid)
class Dolphin(Mammal, Aquatic):
    def sound(self):
        print("Dolphin clicks")

# Object creation
d = Dolphin()
d.breathe()      # From Animal (via Mammal)
d.feed_milk()    # From Mammal
d.swim()         # From Aquatic
d.sound()        # From Dolphin
```

```python
# BRAIN TESTER
class A:
    def m1(self):
        return 20                      # Class A ka method m1(), 20 return karta hai

class B(A):
    def m1(self):
        return 30                      # Class B m1() ko override karta hai, 30 return karta hai

    def m2(self):
        return 40                      # Class B ka method m2(), 40 return karta hai

class C(B):
    def m2(self):
        return 20                      # Class C m2() ko override karta hai, 20 return karta hai

# Object creation
obj1 = A()                             # obj1 is instance of class A
obj2 = B()                             # obj2 is instance of class B
obj3 = C()                             # obj3 is instance of class C

# Final print statement
print(obj1.m1() + obj3.m1() + obj3.m2())    
# Breakdown:
# obj1.m1() → Class A ka m1 → 20
# obj3.m1() → Class C inherits B, B overrides m1 → 30
# obj3.m2() → Class C overrides m2 → 20
# Final result = 20 + 30 + 20 = 70



```

```python
# BRAIN TESTER
# Base class A
class A:
    def m1(self):
        return 20    # Simply returns the value 20

# Class B inherits from A
class B(A):
    def m1(self):
        val = super().m1() + 30   # Calls m1() of class A, adds 30
        return val

# Class C inherits from B
class C(B):
    def m1(self):
        val = self.m1() + 20      # ❌ This line is problematic (infinite recursion)
        return val

# Creating object of class C
obj = C()
print(obj.m1())   # ❌ This will throw RecursionError
```

# POLYMORPHISM IN Python

## 📌 Definition

Polymorphism ka matlab hota hai — **"many forms"**.  
Object-oriented programming mein iska matlab hai ki ek hi method ya operation   **alag-alag objects ke liye alag tareeke se kaam kare**.

Python mein polymorphism ka use hota hai jab:
- Same method name ho sakta hai **different classes mein**, lekin behavior class ke according change hota hai.
- Code reuse hota hai aur flexibility milti hai.

---

## 🧠 Real-Life Example

Agar `Dog` aur `Cat` dono ke paas ek method `speak()` ho:  
- `Dog.speak()` karega "bark"  
- `Cat.speak()` karega "meow"  
Yeh dono **same method name** hone ke bawajood **alag behavior** show karte hain — that's polymorphism.

---

## 🔹 Types of Polymorphism in Python

1. **Duck Typing**  
   - "If it looks like a duck and quacks like a duck, it's a duck"  
   - Object ka type important nahi hota, uska behavior matter karta hai.

2. **Operator Overloading**  
   - `+` operator strings ke liye concatenate karta hai, integers ke liye add.  
   - Python internally methods override karta hai (`__add__`, `__mul__` etc.).

3. **Method Overriding (Runtime Polymorphism)**  
   - Jab child class parent class ka method **same naam se redefine** karti hai.  
   - Python run-time pe decide karta hai ki kaunsa version use hoga.

---

# 3 MAIN PRINCIPLES of OOP in Python

OOPs ke 4 pillars hote hain, lekin 3 ko **core/main** principles mana jaata hai:

---

## 1️⃣ Encapsulation  
- Data aur functions ko ek class ke andar **bundle** karna.  
- External access se bachane ke liye **data hiding** hoti hai using `_` or `__`.

---

## 2️⃣ Inheritance  
- Ek class doosri class se **properties aur methods inherit** karti hai.  
- Code reuse ke liye helpful hai. (e.g., `class Dog(Animal)`)

---

## 3️⃣ Polymorphism  
- Ek hi interface/method name ka use karke **multiple forms** support karna.  
- Code flexible, extendable, aur maintainable banta hai.

---

📝 **Note:**  
4th pillar hota hai **Abstraction**, jisme internal details hide ki jaati hain aur sirf essential features user ko dikhaye jaate hain.

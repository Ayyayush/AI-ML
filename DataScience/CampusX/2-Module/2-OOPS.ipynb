````python
import math     # math module import kiya distance aur abs() use karne ke liye

# ---------------------- Point Class ---------------------- #
class Point:
    def __init__(self, x, y):
        # Jab bhi koi Point object banega, uske x aur y coordinates assign karenge
        self.x_cod = x                                 # x coordinate store kiya
        self.y_cod = y                                 # y coordinate store kiya

    def __str__(self):
        # Jab bhi Point object ko print karenge, to ye method call hoga
        return '<{},{}>'.format(self.x_cod, self.y_cod)       # Point ko formatted string me return karega

    def euc_dis(self, other):
        # Yeh method do points ke beech ka Euclidean distance calculate karta hai
        return ((self.x_cod - other.x_cod) ** 2 + (self.y_cod - other.y_cod) ** 2) ** 0.5
        # √((x1 - x2)^2 + (y1 - y2)^2) formula use kiya gaya hai

    def dis_origin(self):
        # Origin (0,0) se current point ka distance nikalta hai
        return self.euc_dis(Point(0, 0))     # self hi first argument ban ke pass ho jata hai


# ---------------------- Line Class ---------------------- #
class Line:
    def __init__(self, A, B, C):
        # Line equation: Ax + By + C = 0 ke coefficients store kiye gaye
        self.A = A
        self.B = B
        self.C = C

    def __str__(self):
        # Line object ko print karte waqt uska string representation
        return '{}x + {}y + {} = 0'.format(self.A, self.B, self.C)

    def point_on_line(self, point):
        # Check karta hai kya point is line pe lie karta hai ya nahi
        val = self.A * point.x_cod + self.B * point.y_cod + self.C
        if val == 0:
            return "Point lies on the line"
        else:
            return "Point does NOT lie on the line"

    def distance_from_point(self, point):
        # Point aur line ke beech ka perpendicular distance return karta hai
        numerator = abs(self.A * point.x_cod + self.B * point.y_cod + self.C)
        denominator = math.sqrt(self.A ** 2 + self.B ** 2)
        return numerator / denominator
        # Formula: |Ax + By + C| / √(A² + B²)
        

# ---------------------- Sample Test ---------------------- #

p1 = Point(2, 4)
p2 = Point(1, 4)

print("Point 1:", p1)                                 # Output: <2,4>
print("Point 2:", p2)                                 # Output: <1,4>
print("Distance between p1 and p2:", p1.euc_dis(p2))  # Euclidean distance
print("Distance of p1 from origin:", p1.dis_origin()) # Origin se distance

l1 = Line(1, -1, -2)  # Line: x - y - 2 = 0

print("Line equation:", l1)
print("Check p1 on line:", l1.point_on_line(p1))              # Check if p1 lies on line
print("Distance of p1 from line:", l1.distance_from_point(p1))  # Distance of p1 from line
````

# 🔁 Reference Variable in Python


##  Reference Variable kya hota hai aaakhir?

Python mein variables directly value store nahi karte — balki **objects ka reference (yaani memory address)** store karte hain.  
Yeh ek **reference variable** hota hai — jo kisi object ko point karta hai.

---

## 🧠 Python mein reference kaise kaam karta hai?

Agar aap ek variable ko dusre variable ke equal kar dete ho, toh actual value copy nahi hoti, **sirf reference copy hota hai**.

### 🔍 Example:

```python
a = [1, 2, 3]     # 'a' ek list ko point karta hai
b = a             # 'b' bhi usi list ko point karega

b.append(4)       # b se list ko modify kiya

print(a)          # Output: [1, 2, 3, 4] — a bhi modify ho gaya


Jab aap object create karte ho bina kisi reference ke (jaise Person()), toh object temporarily memory mein exist karta hai, lekin use karne ka koi tareeka nahi hota. Yeh "Anonymous Object" bhi kaha jata hai.




````python
# 🔄 Object without a reference in Python

class Person:
    
    def __init__(self):
        self.name = 'nitish'        # Object ke andar 'name' property set ho rahi hai
        self.gender = 'male'        # Object ke andar 'gender' property set ho rahi hai

# ✅ Object create kiya but kisi variable mein store nahi kiya
Person()                            # Is line se ek Person object create hua
                                   # Lekin uska reference kisi variable ko assign nahi kiya gaya
                                   # Yani, yeh object memory mein bana lekin use access karne ka tareeka nahi hai
                                   # Python ka Garbage Collector ise kuch time baad memory se hata dega

# ❌ Agar hum aise likhte:
# p = Person()
# Toh ab 'p' variable ke through object ko access kar sakte the
# Jaise: print(p.name) ➝ "nitish"
````

````python
# creating attributes outside class

# Class Person banayi gayi jismein name aur gender attributes hain
class Person:
    
    def __init__(self, name, gender):               # Constructor to initialize name and gender
        self.name = name                            # Object ka name set karo
        self.gender = gender                        # Object ka gender set karo

# Class ke bahar ek function greet banaya gaya hai jo Person object ko accept karta hai
def greet(person):                                  # 'person' yahan ek reference hai Person class ke object ka
    print(id(person))
    print('Hi my name is', person.name, 'and I am a', person.gender)   # Object ke attributes ko access kiya jaa raha hai

    
# Ab ek Person class ka object banate hain
p = Person('nitish', 'male')                        # p naam ka object banaya aur values assign ki

# greet function ko call kiya aur object p ko pass kiya
greet(p)
# Output: Hi my name is nitish and I am a male
print(id(p))
````

# 🔐 Encapsulation in Python

Encapsulation is one of the four core concepts of Object-Oriented Programming (OOP), alongside **Abstraction**, **Inheritance**, and **Polymorphism**.

---

## ✅ What is Encapsulation?

> Encapsulation means **binding data (variables) and methods (functions)** that work on that data into a single unit — a class — and **restricting direct access** to some of the object's components.

In Python, encapsulation helps **hide the internal state** of objects and **only allow controlled access** via methods (getters/setters).

---

## 🧠 Why Use Encapsulation?

- ✅ To protect variables from accidental changes
- ✅ To enforce controlled access through methods
- ✅ To hide complexity and expose only necessary information
- ✅ To write cleaner, modular, and maintainable code

---

## 🔍 Access Modifiers in Python

| Modifier         | Syntax             | Meaning                                       |
|------------------|--------------------|-----------------------------------------------|
| Public           | `self.var`         | Accessible from anywhere                      |
| Protected        | `self._var`        | Accessible within class and subclasses only (by convention) |
| Private          | `self.__var`       | Name mangled → not accessible directly outside the class |

---

````python
#Why of Encapsulation 
class Atm:
    # Constructor (special function): Initializes the object when created
    def __init__(self):
        print(id(self))                                      # Shows the memory address of the object
        self.pin = ''                                        # User pin initialized as empty
        self.__balance = 0                                   # Private variable using encapsulation

    # Getter method to access balance (read-only)
    def get_balance(self):
        return self.__balance

    # Setter method to update balance (with type check)
    def set_balance(self, new_value):
        if type(new_value) == int:                           # Only allow integers
            self.__balance = new_value
        else:
            print("Beta, bahot maarenge")                    # Funny warning on wrong input

    # Menu method for user interaction
    def __menu(self):
        while True:
            print("""
            ===== ATM MENU =====
            1. Check Balance
            2. Deposit
            3. Withdraw
            4. Exit
            """)
            choice = input("Enter your choice: ")

            if choice == '1':
                print("Your balance is:", self.get_balance())

            elif choice == '2':
                amount = int(input("Enter amount to deposit: "))
                new_balance = self.get_balance() + amount
                self.set_balance(new_balance)
                print("Deposit successful!")

            elif choice == '3':
                amount = int(input("Enter amount to withdraw: "))
                if amount <= self.get_balance():
                    new_balance = self.get_balance() - amount
                    self.set_balance(new_balance)
                    print("Withdrawal successful!")
                else:
                    print("Insufficient Balance.")

            elif choice == '4':
                print("Thank you! Exiting...")
                break

            else:
                print("Invalid choice. Please try again.")

    # Public method to call the menu from outside the class
    def start(self):
        self.__menu()                                        # Call the private menu function
````

# 📦 Collection of Objects in Python

In Object-Oriented Programming, we often need to create **multiple objects** of a class and **store them together** — this is where the idea of a **Collection of Objects** comes into play.

---

## ✅ What is a Collection of Objects?

A **collection of objects** means storing multiple instances of a class (i.e., objects) in a **list**, **dictionary**, or any iterable data structure so that we can iterate, search, sort, and manage them efficiently.

---

## 📚 Example: Storing Multiple Students in a List

Let's say we have a `Student` class and want to store many student records:

```python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll

    def show(self):
        print(f"Name: {self.name}, Roll No: {self.roll}")


````python
# List of objects in Python

class Person:                                 # Person naam ki class define ki gayi hai
    
    def __init__(self, name, gender):         # Constructor define kiya gaya hai jismein name aur gender attributes hain
        self.name = name                      # name ko object ke attribute mein store kiya gaya
        self.gender = gender                  # gender ko object ke attribute mein store kiya gaya

# Ab 3 alag-alag Person ke objects banaye ja rahe hain:
p1 = Person('nitish', 'male')                 # p1 naam ka object bana jismein name = 'nitish', gender = 'male'
p2 = Person('ankit', 'male')                  # p2 object bana jismein name = 'ankit', gender = 'male'
p3 = Person('ankita', 'female')               # p3 object bana jismein name = 'ankita', gender = 'female'

# List banayi gayi jismein ye 3 Person objects add kiye gaye
L = [p1, p2, p3]                              # List L contains all 3 objects
````

````python
# Dictionary of Objects in Python
# Class define karte hain
class Student:
    def __init__(self, name, roll):
        self.name = name                        # Student ka naam
        self.roll = roll                        # Student ka roll number

    def show(self):
        print(f"Name: {self.name}, Roll: {self.roll}")    # Student info dikhata hai
        

# Dictionary jismein multiple Student objects honge
students = {
    101: Student("Aman", 101),                 # Key = 101, Value = Student object
    102: Student("Riya", 102),
    103: Student("Zaid", 103)
}

# Accessing and displaying each student
for key in students:
    print(f"Details of student with key {key}:")
    students[key].show()
````

````python
# Set of Objects in Python

# Student class define karte hain
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll

    def __eq__(self, other):
        return self.roll == other.roll              # Roll no se comparison

    def __hash__(self):
        return hash(self.roll)                      # Roll no ko hashable bana diya

    def __str__(self):
        return f"{self.name} ({self.roll})"

# Student objects
s1 = Student("Aman", 101)
s2 = Student("Riya", 102)
s3 = Student("Aman", 101)     # Same roll as s1 → Duplicate in terms of set logic

# Set of Student objects
student_set = {s1, s2, s3}

# Print all students in the set
print("Unique Students in the Set:")
for student in student_set:
    print(student)
````

# 🧲 Static Keyword in Python

In Python, we use the `@staticmethod` decorator to define a method that **does not depend on class instance or class variables**. It behaves just like a regular function but lives inside the class’s namespace.

---

## 🔍 What is a Static Method?

A **static method**:
- Doesn’t receive `self` (like instance methods).
- Doesn’t receive `cls` (like class methods).
- Can be called using the class name or an object.
- Can be used when the method **doesn’t need to access class/instance variables.**

---

## ✅ Syntax:

```python
class MyClass:
    @staticmethod
    def my_static_method():
        # logic here
        pass


class MathOperations:
    
    @staticmethod
    def add(a, b):
        return a + b

# Calling using class
print(MathOperations.add(5, 3))         # Output: 8

# Calling using object
obj = MathOperations()
print(obj.add(10, 7))                   # Output: 17


#🆚 Difference Between Static and Instance Variable in Python

In Python, **variables in a class** can be of two types:

- 🔹 **Instance Variable** — Belongs to an object (instance) of the class.
- 🔹 **Static (Class) Variable** — Belongs to the class itself and is shared among all objects.

---

## 📌 1. Instance Variable

- Defined using `self.variable` inside constructor or methods.
- Unique for each object.
- Changing it in one object doesn't affect others.

```python
class Person:
    def __init__(self, name):
        self.name = name                # instance variable

p1 = Person("Alice")
p2 = Person("Bob")

print(p1.name)       # Alice
print(p2.name)       # Bob


##  📌 2. Static Variables
class Person:
    species = "Human"                  # static variable

    def __init__(self, name):
        self.name = name              # instance variable

p1 = Person("Alice")
p2 = Person("Bob")

print(p1.species)      # Human
print(p2.species)      # Human

Person.species = "Robot"

print(p1.species)      # Robot
print(p2.species)      # Robot

````python
import numpy as np
import pandas as pd


# 1st
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
result = A * B
print("Element-wise Product:", result)

# 2nd 
C = np.array([2, -3, 1])
magnitude = np.linalg.norm(C)
print("Magnitude of Vector C:", magnitude)


# 3rd
# vectors
D = np.array([1, 2, 3])
E = np.array([4, 5, 6])

# dot product
dot_product = np.dot(D, E)

# magnitude of vectors
magnitude_D = np.linalg.norm(D)
magnitude_E = np.linalg.norm(E)

# cosine angle formula
cos_theta = dot_product / (magnitude_D * magnitude_E)

# output
print(cos_theta)
````

````python
#7th
import matplotlib.pyplot as plt
# points
a = np.array([1, 3])
b = np.array([5, 4])

# vector ab
ab = b - a

# vector sum
result = a + ab

# outputs
print("ab =", ab)
print("a + ab =", result)

# plotting arrows
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(a[0], a[1], ab[0], ab[1], angles='xy', scale_units='xy', scale=1)

# graph settings
plt.xlim(0, 6)
plt.ylim(0, 6)
plt.grid()
plt.show()
````

````python
# 8th 
# vectors
u = np.array([-1, 4])
v = np.array([3, -2])

# dot product
dot_product = np.dot(u, v)

# magnitudes
magnitude_u = np.linalg.norm(u)
magnitude_v = np.linalg.norm(v)

# cosine angle
cos_theta = dot_product / (magnitude_u * magnitude_v)

# angle in degrees
angle = np.degrees(np.arccos(cos_theta))

# projection of u onto v
projection = (dot_product / np.dot(v, v)) * v

# outputs
print("Dot Product:", dot_product)
print("Angle:", angle)
print("Projection:", projection)
````

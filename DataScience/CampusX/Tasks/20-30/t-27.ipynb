### SQLAlchemy

- **SQLAlchemy** is a Python library used to interact with relational databases.
- It acts as a bridge between **Python code and SQL databases**.
- Supports databases like MySQL, PostgreSQL, SQLite, etc.

---

### Why use SQLAlchemy?

- Write database operations in **Python instead of raw SQL**.
- Provides an **ORM (Object Relational Mapper)**.
- Helps in **database connection management**.
- Works smoothly with **Pandas (`to_sql`, `read_sql`)**.

---

### Two Main Components

1. **Core** – SQL Expression Language (write SQL in Python style).
2. **ORM** – Map database tables to Python classes.

---

### Example Use Case
```python
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:password@localhost/db_name")

Go to this url: https://www.flipkart.com/search?q=smartphones. This is the url to find phones in flipkart website. You have to extract the below things:

Extract all the phones which are available in this website. So you have to use the pagination concept. **Also after requesting every page through the url, please wait for a while (minimum 2-3 seconds), otherwise your IP address can be banned to access the flipkart website later.**

````python
# code here
import requests
from bs4 import BeautifulSoup
import time

import pandas as pd
import numpy as np
````

````python
webpage = requests.get('https://www.flipkart.com/search?q=smartphones&page=1').text
````

````python
soup = BeautifulSoup(webpage,'lxml')
````

````python
all_divs = soup.find_all('div', class_="_2kHMtA")
````

````python
import numpy as np
import pandas as pd

# Lists to store scraped phone details
phone_image_url = []          # Stores image URLs
phone_name = []               # Stores phone names
phone_rating = []             # Stores ratings
phone_num_ratings = []        # Stores number of ratings
phone_price = []              # Stores prices


# Loop through each phone container (div)
for phone in all_divs:

    # -----------------------------
    # Extract phone image URL
    # -----------------------------
    try:
        phone_image_url.append(
            phone.find("img", class_="_396cs4")['src']
        )
    except:
        phone_image_url.append(np.nan)    # If image not found


    # -----------------------------
    # Extract phone name
    # -----------------------------
    try:
        phone_name.append(
            phone.find("div", class_="_4rR01T").text.strip()
        )
    except:
        phone_name.append(np.nan)


    # -----------------------------
    # Extract phone rating
    # -----------------------------
    try:
        phone_rating.append(
            phone.find("div", class_="_3LWZlK").text.strip()
        )
    except:
        phone_rating.append(np.nan)


    # -----------------------------
    # Extract number of ratings
    # -----------------------------
    try:
        phone_num_ratings.append(
            phone.find_all('span', class_='_2_R_DZ')[0]
                 .find_all('span')[0]
                 .find_all('span')[0]
                 .text.strip()
        )
    except:
        phone_num_ratings.append(np.nan)


    # -----------------------------
    # Extract phone price
    # -----------------------------
    try:
        phone_price.append(
            phone.find("div", class_="_30jeq3 _1_WHN1").text.strip()
        )
    except:
        phone_price.append(np.nan)


# -----------------------------
# Create DataFrame from lists
# -----------------------------
df = pd.DataFrame({
    'phone_name': phone_name,
    'phone_price': phone_price,
    'phone_rating': phone_rating,
    'phone_num_rating': phone_num_ratings,
    'phone_img_url': phone_image_url
})

df    # Display final DataFrame
````

````python
import requests
import numpy as np
import pandas as pd
import time
from bs4 import BeautifulSoup

# ---------------------------------------
# Headers (VERY IMPORTANT for Flipkart)
# ---------------------------------------
headers = {
    "User-Agent": 
    "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
    "(KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36"
}

# List to store all page DataFrames
dfs = []

# ---------------------------------------
# Loop through pages
# ---------------------------------------
for i in range(1, 6):

    url = f"https://www.flipkart.com/search?q=smartphones&page={i}"
    response = requests.get(url, headers=headers)

    # If request fails, skip page
    if response.status_code != 200:
        continue

    soup = BeautifulSoup(response.text, "lxml")

    # Find all phone containers
    all_divs = soup.find_all("div", class_="_2kHMtA")

    # Temporary lists
    phone_image_url = []
    phone_name = []
    phone_rating = []
    phone_num_ratings = []
    phone_price = []

    # ---------------------------------------
    # Extract phone data
    # ---------------------------------------
    for phone in all_divs:

        try:
            phone_image_url.append(
                phone.find("img", class_="_396cs4")["src"]
            )
        except:
            phone_image_url.append(np.nan)

        try:
            phone_name.append(
                phone.find("div", class_="_4rR01T").text.strip()
            )
        except:
            phone_name.append(np.nan)

        try:
            phone_rating.append(
                phone.find("div", class_="_3LWZlK").text.strip()
            )
        except:
            phone_rating.append(np.nan)

        try:
            phone_num_ratings.append(
                phone.find("span", class_="_2_R_DZ")
                     .find("span")
                     .find("span")
                     .text.strip()
            )
        except:
            phone_num_ratings.append(np.nan)

        try:
            phone_price.append(
                phone.find("div", class_="_30jeq3 _1_WHN1").text.strip()
            )
        except:
            phone_price.append(np.nan)

    # ---------------------------------------
    # Create DataFrame for this page
    # ---------------------------------------
    df = pd.DataFrame({
        "phone_name": phone_name,
        "phone_price": phone_price,
        "phone_rating": phone_rating,
        "phone_num_rating": phone_num_ratings,
        "phone_img_url": phone_image_url
    })

    dfs.append(df)

    # Delay to avoid getting blocked
    time.sleep(2)

# ---------------------------------------
# Combine all pages into final DataFrame
# ---------------------------------------
final_df = pd.concat(dfs, ignore_index=True)

final_df
````

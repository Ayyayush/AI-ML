```python
# Basic practice questions on categorical plots

import seaborn as sns
import numpy as np

tips = sns.load_dataset('tips')

# level 1
# 1st

sns.countplot(
data=tips,
x='sex'                         # countplot only needs categorical axis
)

# 2nd

sns.countplot(
data=tips,
x='day',
hue='sex'                       # to compare gender across days
)

# 3rd

sns.barplot(
data=tips,
x='day',
y='total_bill',                 # must be numerical
errorbar=None,                  # spelling correction
estimator=np.mean               # default is mean (better than min)
)

# 4th

sns.barplot(
data=tips,
x='day',
y='total_bill',
hue='sex',
errorbar=None,
estimator=np.mean
)



# level 2 understanding distribution

# 5th

sns.boxplot(
data=tips,
x='day',
y='total_bill'
)

# 6th

sns.boxplot(
data=tips,
x='day',
y='total_bill',
hue='smoker'
)

# 7th

sns.violinplot(
data=tips,
x='day',
y='tip'
)

# 8th

sns.violinplot(
data=tips,
x='day',
y='tip',
hue='sex',
split=True                      # works only when hue has 2 categories
)

# data point visualization

# 9th

sns.stripplot(
data=tips,
x='day',
y='total_bill'
)

# 10th

sns.stripplot(
data=tips,
x='day',
y='total_bill',
hue='sex',
dodge=True
)

# 11th

sns.swarmplot(
data=tips,
x='day',
y='total_bill'
)

# level 4 (added missing ones)

# 12th

sns.barplot(
data=tips,
x='day',
y='tip',
hue='time',
errorbar=None
)

# 13th

sns.boxplot(
data=tips,
x='time',
y='total_bill',
hue='day'
)

# 14th

sns.violinplot(
data=tips,
x='sex',
y='tip',
hue='smoker',
split=True
)
```

```python
# part 1 regression plots

import seaborn as sns
tips = sns.load_dataset("tips")

# 1st

sns.regplot(
data=tips,
x='total_bill',
y='tip'
)


# 2nd

sns.regplot(
data=tips,
x='total_bill',
y='tip',
scatter=False                  # only regression line
)

# 3rd

sns.regplot(
data=tips,
x='total_bill',
y='tip',
color='red'
)

# 4th

sns.regplot(
data=tips,
x='total_bill',
y='tip',
marker='x'
)

# 5th

sns.regplot(
data=tips,
x='total_bill',
y='tip',
lowess=True                   # non-linear fit
)

# 6th

sns.lmplot(
data=tips,
x='total_bill',
y='tip'
)

# 7th

sns.lmplot(
data=tips,
x='total_bill',
y='tip',
hue='sex'
)

# 8th

sns.lmplot(
data=tips,
x='total_bill',
y='tip',
col='smoker'
)

# 9th

sns.lmplot(
data=tips,
x='total_bill',
y='tip',
row='time'
)

# 10th

sns.lmplot(
data=tips,
x='total_bill',
y='tip',
hue='sex',
col='time'
)

# part 2 multigrid plots

# --------------------------------------------------

# FacetGrid (Low-level Figure Control)

# --------------------------------------------------

g = sns.FacetGrid(
data=tips,
col='day',
row='time',
hue='smoker'
)

g.map(
sns.scatterplot,              # better for numeric vs numeric
'total_bill',
'tip'
)

g.add_legend()

# 11th

g = sns.FacetGrid(
data=tips,
col='time'
)

g.map(
sns.scatterplot,
'total_bill',
'tip'
)

# 12th

g = sns.FacetGrid(
data=tips,
col='day'
)

g.map(
sns.histplot,
'total_bill'
)

# 13th

g = sns.FacetGrid(
data=tips,
col='time',
row='smoker'
)

g.map(
sns.scatterplot,
'total_bill',
'tip'
)
```

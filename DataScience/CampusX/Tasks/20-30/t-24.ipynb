### **Problem-1:** Make a subplots which have 2 plots.

- For the first chart, draw a scatter plot "*Monitored Cap.(MW)*" vs "*Total Cap. Under Maintenance (MW)*" of top 5 most frequent power stations. Then draw the lines which indicate the average values of these two columns. Change the colors according to the names of the Power Stations.
- For the second chart, draw a scatter plot "*Monitored Cap. (MW)*" vs "*Actual(MU)*" of the top 5 most frequent power stations. Also draw the lines which indicates the average values of these two columns. Change the colors according to the names of the Power Stations.


````python
power= pd.read_csv('content/PowerGeneration.csv')
power.head()
````

````python
# --------------------------------------------------
# Problem-1
# Subplots with 2 scatter plots for top 5 power stations
# --------------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt


# --------------------------------------------------
# Step 1: Read the dataset
# --------------------------------------------------

power = pd.read_csv('content/PowerGeneration.csv')


# --------------------------------------------------
# Step 2: Find top 5 most frequent power stations
# --------------------------------------------------
# value_counts() gives frequency of each power station
# head(5) selects top 5
# index extracts station names

top5_stations = (
    power['Power Station']
    .value_counts()
    .head(5)
    .index
)


# --------------------------------------------------
# Step 3: Filter dataframe for only top 5 stations
# --------------------------------------------------

top5_df = power[power['Power Station'].isin(top5_stations)]


# --------------------------------------------------
# Step 4: Create subplots (1 row, 2 columns)
# --------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(24, 10))


# ==================================================
# Plot-1:
# Monitored Cap.(MW) vs Total Cap. Under Maintenance (MW)
# ==================================================

# Calculate average values
avg_monitored = top5_df['Monitored Cap.(MW)'].mean()
avg_maintenance = top5_df['Total Cap. Under Maintenace (MW)'].mean()

# Scatter plot for each power station (different colors)
for station in top5_stations:
    station_df = top5_df[top5_df['Power Station'] == station]

    axes[0].scatter(
        station_df['Monitored Cap.(MW)'],
        station_df['Total Cap. Under Maintenace (MW)'],
        label=station
    )

# Draw average reference lines
axes[0].axvline(avg_monitored)
axes[0].axhline(avg_maintenance)

# Labels and title
axes[0].set_xlabel("Monitored Cap.(MW)")
axes[0].set_ylabel("Total Cap. Under Maintenance (MW)")
axes[0].set_title("Monitored Capacity vs Maintenance Capacity")
axes[0].legend()


# ==================================================
# Plot-2:
# Monitored Cap.(MW) vs Actual(MU)
# ==================================================

# Calculate average actual generation
avg_actual = top5_df['Actual(MU)'].mean()

# Scatter plot for each power station
for station in top5_stations:
    station_df = top5_df[top5_df['Power Station'] == station]

    axes[1].scatter(
        station_df['Monitored Cap.(MW)'],
        station_df['Actual(MU)'],
        label=station
    )

# Draw average reference lines
axes[1].axvline(avg_monitored)
axes[1].axhline(avg_actual)

# Labels and title
axes[1].set_xlabel("Monitored Cap.(MW)")
axes[1].set_ylabel("Actual Generation (MU)")
axes[1].set_title("Monitored Capacity vs Actual Generation")
axes[1].legend()


# --------------------------------------------------
# Step 5: Adjust layout and show plots
# --------------------------------------------------

plt.tight_layout()
plt.show()
````

### **Problem-3:** Make a 3D *Surface* plot of this below mathematical equation.

$$z = |x| - |y|$$

````python
# --------------------------------------------------
# Problem-3
# 3D Surface Plot of: z = |x| - |y|
# --------------------------------------------------

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


# --------------------------------------------------
# Step 1: Create x and y values
# --------------------------------------------------

x = np.linspace(-10, 10, 100)           # x range
y = np.linspace(-10, 10, 100)           # y range

X, Y = np.meshgrid(x, y)                # create 2D grid


# --------------------------------------------------
# Step 2: Apply the equation z = |x| - |y|
# --------------------------------------------------

Z = np.abs(X) - np.abs(Y)


# --------------------------------------------------
# Step 3: Create 3D figure and axis
# --------------------------------------------------

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')


# --------------------------------------------------
# Step 4: Plot the surface
# --------------------------------------------------

ax.plot_surface(X, Y, Z)


# --------------------------------------------------
# Step 5: Label axes and title
# --------------------------------------------------

ax.set_xlabel("X axis")
ax.set_ylabel("Y axis")
ax.set_zlabel("Z axis")

ax.set_title("3D Surface Plot of z = |x| - |y|")


# --------------------------------------------------
# Step 6: Show the plot
# --------------------------------------------------

plt.show()
````

### **Problem-4:** Draw the 3D *Contour plot* of this below equation:

$$z = |x| - |y|$$

````python
# --------------------------------------------------
# Problem-4
# 3D Contour Plot of: z = |x| - |y|
# --------------------------------------------------

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


# --------------------------------------------------
# Step 1: Create x and y values
# --------------------------------------------------

x = np.linspace(-10, 10, 100)            # x range
y = np.linspace(-10, 10, 100)            # y range

X, Y = np.meshgrid(x, y)                 # create coordinate grid


# --------------------------------------------------
# Step 2: Apply the equation
# --------------------------------------------------

Z = np.abs(X) - np.abs(Y)


# --------------------------------------------------
# Step 3: Create 3D figure and axis
# --------------------------------------------------

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')


# --------------------------------------------------
# Step 4: Draw 3D contour plot
# --------------------------------------------------

ax.contour(
    X, Y, Z,
    levels=20                            # number of contour levels
)


# --------------------------------------------------
# Step 5: Label axes and title
# --------------------------------------------------

ax.set_xlabel("X axis")
ax.set_ylabel("Y axis")
ax.set_zlabel("Z axis")

ax.set_title("3D Contour Plot of z = |x| - |y|")


# --------------------------------------------------
# Step 6: Show plot
# --------------------------------------------------

plt.show()
````

### **Problem-5:** Draw a second type of Countour plot of the below equation:

$$z = |x| - |y|$$

````python
# --------------------------------------------------
# Problem-5
# 3D Filled Contour Plot of: z = |x| - |y|
# --------------------------------------------------

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


# --------------------------------------------------
# Step 1: Create x and y values
# --------------------------------------------------

x = np.linspace(-10, 10, 100)            # x range
y = np.linspace(-10, 10, 100)            # y range

X, Y = np.meshgrid(x, y)                 # create coordinate grid


# --------------------------------------------------
# Step 2: Apply the equation
# --------------------------------------------------

Z = np.abs(X) - np.abs(Y)


# --------------------------------------------------
# Step 3: Create 3D figure and axis
# --------------------------------------------------

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')


# --------------------------------------------------
# Step 4: Draw FILLED 3D contour plot
# --------------------------------------------------

ax.contourf(
    X, Y, Z,
    levels=20                             # number of filled levels
)


# --------------------------------------------------
# Step 5: Label axes and title
# --------------------------------------------------

ax.set_xlabel("X axis")
ax.set_ylabel("Y axis")
ax.set_zlabel("Z axis")

ax.set_title("3D Filled Contour Plot of z = |x| - |y|")


# --------------------------------------------------
# Step 6: Show plot
# --------------------------------------------------

plt.show()
````

## Freedom Ranking for Different Countries

You can get the dataset from [here](https://drive.google.com/drive/folders/1wr0J571rlfexEJ4-de6_tz1QvU-cJ0df?usp=share_link).

Some feature details of the dataset:

| Feature | Description |
| :--: | :-- |
| A | Electoral Process |
| B | Political Pluralism and Participation |
| C | Functioning of Government |
| D | Freedom of Expression and Belief |
| E | Associational and Organizational Rights |
| F | Rule of Law |
| G | Personal Autonomy and Individual Rights |
| CL | Civil Liberties Scores |
| Status | F=Free, PF=Partly Free, NF=Not Free |


###`Q-1` Find out top 10 countries' which have maximum A and D values.


###`Q-2` Find out highest CL value for 2020 for every region. Also sort the result in descending order. Also display the CL values in descending order.

SELECT 
    ab.country,
    ab.A,
    cd.D
FROM country_ab ab
JOIN country_cd cd
    ON ab.country = cd.country
ORDER BY 
    ab.A DESC,
    cd.D DESC
LIMIT 10;


SELECT 
    region,
    MAX(CL) AS highest_cl
FROM country_cl
WHERE year = 2020
GROUP BY region
ORDER BY highest_cl DESC;

SELECT 
    p.product_name,
    COUNT(s.sale_id) AS total_sales
FROM Sales s
JOIN Products p
    ON s.product_id = p.product_id
GROUP BY p.product_name
ORDER BY total_sales DESC
LIMIT 5;


SELECT 
    e.employee_name,
    COUNT(s.sale_id) AS total_products_sold
FROM Sales s
JOIN Employees e
    ON s.employee_id = e.employee_id
GROUP BY e.employee_name
ORDER BY total_products_sold DESC
LIMIT 1;


SELECT 
    e.employee_name,
    COUNT(DISTINCT s.customer_id) AS unique_customers
FROM Sales s
JOIN Employees e
    ON s.employee_id = e.employee_id
GROUP BY e.employee_name
ORDER BY unique_customers DESC
LIMIT 1;


SELECT 
    e.employee_name,
    SUM(s.quantity * p.price) AS total_revenue
FROM Sales s
JOIN Employees e
    ON s.employee_id = e.employee_id
JOIN Products p
    ON s.product_id = p.product_id
GROUP BY e.employee_name
ORDER BY total_revenue DESC
LIMIT 5;


SELECT 
    c.customer_name,
    COUNT(s.sale_id) AS total_purchases
FROM Sales s
JOIN Customers c
    ON s.customer_id = c.customer_id
GROUP BY c.customer_name
HAVING COUNT(s.sale_id) > 10;


SELECT 
    e.employee_name,
    COUNT(DISTINCT s.customer_id) AS total_customers
FROM Sales s
JOIN Employees e
    ON s.employee_id = e.employee_id
GROUP BY e.employee_name
HAVING COUNT(DISTINCT s.customer_id) > 5;



SELECT DISTINCT
    s1.customer_id AS customer_1,
    s2.customer_id AS customer_2,
    s1.employee_id

    ON s1.employee_id = s2.employee_id
   AND s1.customer_id < s2.customer_id;

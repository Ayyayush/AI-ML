###`Q8:` Write a decorator called `printer` which causes any decorated function to print their return values. If the return value of a given function is `None`, printer should do nothing.



````python
# ---------------------------------------------
# Q8: printer Decorator
# ---------------------------------------------
# This decorator prints the return value
# of a function ONLY if it is not None.
# ---------------------------------------------

def printer(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)     # Call original function and store return value

        # Print only if return value is NOT None
        if result is not None:
            print(result)

        return result                      # Return original result (good practice)

    return wrapper

@printer
def add(a, b):
    return a + b

add(3, 4)

````

### `Q10:` Write a decorator which doubles the return value of any function. And test that decoratos is working correctly or not using `asert`.

```
add(2,3) -> result in 10. Without decorator it should be 5.
```

````python
# ---------------------------------------------
# Q10: Double Return Value Decorator
# ---------------------------------------------
# This decorator doubles the return value
# of the decorated function.
# ---------------------------------------------

def double_return(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)   # Call original function
        return result * 2                # Double the return value
    return wrapper

def add_normal(a, b):
    return a + b

# Test without decorator
assert add_normal(2, 3) == 5

````

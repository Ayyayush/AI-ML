###`Q-1:` Create a random 3x4 matrix with value between 0-100. And perform below tasks

    i. Sort this matrix. np.sort()
    ii. Sort this matrix based on values in 2nd column.
    iii. Sort this matrix based on max value in each row.
    iv. Sort based on elements value.


See examples:

arr =
    [[92 90 74]
    [ 6 63 93]
    [15 93 96]
    [70 60 48]]
```
```
i. np.sort
[[74 90 92]
 [ 6 63 93]
 [15 93 96]
 [48 60 70]]

ii. based on 2nd column
[[70 60 48]
 [ 6 63 93]
 [92 90 74]
 [15 93 96]]

iii. based on row max- ascending
[[15 93 96]
 [ 6 63 93]
 [92 90 74]
 [70 60 48]]

iv. based on elements value
[[ 6 15 48]
 [60 63 70]
 [74 90 92]
 [93 93 96]]

```

````python
# ==============================
# NumPy Matrix Sorting Examples
# ==============================

import numpy as np

# --------------------------------------------------
# STEP 1: Create a random 3x4 matrix with values 0–100
# --------------------------------------------------
# np.random.randint(low, high, size)
# low = 0, high = 101 (100 include karne ke liye)
# size = (3, 4) => 3 rows, 4 columns

arr = np.random.randint(0, 101, size=(3, 4))

print("Original Matrix:")
print(arr)
print()

# --------------------------------------------------
# i) Sort the entire matrix (row-wise by default)
# --------------------------------------------------
# np.sort() by default axis = -1 (last axis)
# Matlab har row ke elements ko individually sort karega

sorted_matrix = np.sort(arr)

print("i) Matrix sorted row-wise (np.sort default):")
print(sorted_matrix)
print()

# --------------------------------------------------
# ii) Sort matrix based on values in 2nd column
# --------------------------------------------------
# 2nd column ka index = 1 (0-based indexing)
# arr[:, 1] => poora 2nd column
# np.argsort() indices return karta hai sorted order ke

sorted_by_2nd_col = arr[arr[:, 1].argsort()]

print("ii) Matrix sorted based on 2nd column values:")
print(sorted_by_2nd_col)
print()

# --------------------------------------------------
# iii) Sort matrix based on max value in each row
# --------------------------------------------------
# np.max(arr, axis=1) => har row ka max value
# argsort() => un max values ke sorted indices

row_max_indices = np.max(arr, axis=1).argsort()
sorted_by_row_max = arr[row_max_indices]

print("iii) Matrix sorted based on max value of each row:")
print(sorted_by_row_max)
print()

# --------------------------------------------------
# iv) Sort based on all element values (flatten sort)
# --------------------------------------------------
# flatten() => 2D matrix ko 1D array bana deta hai
# phir np.sort() se poore elements sort

sorted_by_values = np.sort(arr.flatten())

print("iv) All elements sorted (flattened):")
print(sorted_by_values)
````

###`Q-2:` There is an array of marks of 5 students in 4 subjects. Further you are asked to perform below task.
    i. Add marks every student of an extra subject in the same array.
    ii. Add two new students marks in respective 5 subjects.(one subject added in above task)
    iii. Add extra column with sum of all subjects(5-subjects) marks
    iv. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.

Note: Change dimension of arrays during concatenation or appending if required.

Given Array-
```
marks = [[13, 10,  9, 33],
       [63, 46, 90, 42],
       [39, 76, 13, 29],
       [82,  9, 29, 78],
       [67, 61, 59, 36]]

extra_subject = [41, 87, 72, 36, 92]
#Two extra students record-
rec1 = [77, 83, 98, 95, 89]
rec2 = [92, 71, 52, 61, 53]
```

````python
import numpy as np

# ---------------------------------------------------
# GIVEN: 5 students × 4 subjects
# ---------------------------------------------------
marks = np.array([
    [13, 10,  9, 33],
    [63, 46, 90, 42],
    [39, 76, 13, 29],
    [82,  9, 29, 78],
    [67, 61, 59, 36]
])

print("Initial Marks (5 students × 4 subjects)")
print(marks)
print("Shape:", marks.shape)
print()

# ===================================================
# PART i: Add extra subject marks for every student
# ===================================================

extra_subject = np.array([41, 87, 72, 36, 92])

# 1D → 2D column (required for column-wise append)
extra_subject = np.expand_dims(extra_subject, axis=1)

# axis=1 → column add
marks = np.append(marks, extra_subject, axis=1)

print("After adding extra subject (5 subjects)")
print(marks)
print("Shape:", marks.shape)
print()

# ===================================================
# PART ii: Add 2 new students with 5 subjects each
# ===================================================

new_students = np.array([
    [55, 67, 78, 88, 91],
    [72, 64, 59, 81, 77]
])

# axis=0 → row add
marks = np.append(marks, new_students, axis=0)

print("After adding 2 new students")
print(marks)
print("Shape:", marks.shape)
print()

# ===================================================
# PART iii: Add total marks column (sum of 5 subjects)
# ===================================================

# axis=1 → row-wise sum (per student)
total_marks = np.sum(marks, axis=1)

# Convert to column for append
total_marks = np.expand_dims(total_marks, axis=1)

# Append total column
marks = np.append(marks, total_marks, axis=1)

print("After adding total marks column")
print(marks)
print("Shape:", marks.shape)
print()

# ===================================================
# PART iv: Sort by total marks (descending)
#         and show top 2 students
# ===================================================

# Last column index = -1 (total marks)
sorted_marks = marks[marks[:, -1].argsort()[::-1]]

top_2 = sorted_marks[:2]

print("Top 2 students based on total marks (descending):")
print(top_2)
````

arr = np.array([[1,2,3,3,1,1],

                [1,2,3,3,1,1]])

### `Q-5:` Get row numbers of NumPy array having element larger than X.
```
arr = [[1,2,3,4,5],
      [10,-3,30,4,5],
      [3,2,5,-4,5],
      [9,7,3,6,5]]

X = 6
```

````python
import numpy as np

elt = 10   # threshold value

# ==================================================
# METHOD 1: Nested loops (basic logic)
# ==================================================

arr1 = np.arange(16).reshape(4,4)
result = []

for i in range(arr1.shape[0]):          # rows
    for j in range(arr1.shape[1]):      # columns
        if arr1[i][j] > elt:
            result.append(i)
            break                       # ek row me mil gaya, aage mat dekho

print("Method 1 - Row indices:", result)


# ==================================================
# METHOD 2: Boolean masking + loop
# ==================================================

arr2 = np.tile(np.arange(6), (5,1))     # 5x6 array
result = []

for i in range(arr2.shape[0]):
    if np.any(arr2[i] > elt):           # row me koi bhi element > elt
        result.append(i)

print("Method 2 - Row indices:", result)


# ==================================================
# METHOD 3: NumPy vectorized (BEST)
# ==================================================

arr3 = np.arange(30).reshape(5,6)

# arr3 > elt → boolean matrix
# np.any(..., axis=1) → row-wise check
# np.where(...) → row numbers

indices = np.where(np.any(arr3 > elt, axis=1))[0]

print("Method 3 - Row indices:", indices)
````

### `Q-6:` How to convert an array of arrays into a flat 1d array?


````python
# These arrays are given.
arr1 = np.arange(3)
arr2 = np.arange(3,7)
arr3 = np.arange(7,10)
````

````python
import numpy as np

# ==================================================
# METHOD 1: Using loops (manual flatten)
# ==================================================

arr1 = np.array([[1, 2, 3],
                 [4, 5, 6]])

flat1 = []

for row in arr1:
    for element in row:
        flat1.append(element)

flat1 = np.array(flat1)

print("Method 1 - Using loops:")
print(flat1)
print()


# ==================================================
# METHOD 2: Using reshape()
# ==================================================

arr2 = np.array([[7, 8],
                 [9, 10],
                 [11, 12]])

# -1 → NumPy automatically calculate correct size
flat2 = arr2.reshape(-1)

print("Method 2 - Using reshape:")
print(flat2)
print()


# ==================================================
# METHOD 3: Using flatten()
# ==================================================

arr3 = np.array([[13, 14, 15],
                 [16, 17, 18]])

flat3 = arr3.flatten()

print("Method 3 - Using flatten:")
print(flat3)
print()


# ==================================================
# METHOD 4: Using ravel()
# ==================================================

arr4 = np.array([[19, 20],
                 [21, 22]])

flat4 = arr4.ravel()

print("Method 4 - Using ravel:")
print(flat4)
````

### `Q-7:` You are given a array. You have to find the minimum and maximum array element and remove that from the array.

```python
import numpy as np

np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
```

````python
import numpy as np

# ==================================================
# METHOD 1: Using loops (manual way)
# ==================================================

np.random.seed(400)
arr1 = np.random.randint(100, 1000, 200)

print("Original Array (Method 1):")
print(arr1)

# find min and max manually
min_val = arr1[0]
max_val = arr1[0]

for x in arr1:
    if x < min_val:
        min_val = x
    if x > max_val:
        max_val = x

# remove min and max
arr1_removed = [x for x in arr1 if x != min_val and x != max_val]

print("\nMin:", min_val, "Max:", max_val)
print("After removing min & max:")
print(arr1_removed)
print()


# ==================================================
# METHOD 2: Using NumPy functions (recommended)
# ==================================================

np.random.seed(200)
arr2 = np.random.randint(10, 100, 100)

print("Original Array (Method 2):")
print(arr2)

min_val = np.min(arr2)
max_val = np.max(arr2)

# boolean masking
arr2_removed = arr2[(arr2 != min_val) & (arr2 != max_val)]

print("\nMin:", min_val, "Max:", max_val)
print("After removing min & max:")
print(arr2_removed)
print()


# ==================================================
# METHOD 3: Using argmin & argmax (index-based)
# ==================================================

arr3 = arr2.copy()

min_idx = np.argmin(arr3)
max_idx = np.argmax(arr3)

print("Min index:", min_idx, "Max index:", max_idx)

# delete using indices
arr3_removed = np.delete(arr3, [min_idx, max_idx])

print("After removing min & max (using indices):")
print(arr3_removed)
````

### `Q-8:` You are given an arrays. You have to limit this array's elements between 100 to 200. $arr ∈ [100, 700]$. So replace those values accordingly with the minimum and maximum value. Then sort the array and perform the cumulative sum of that array.

````python
import numpy as np

# ==================================================
# METHOD 1: Using loops (manual clipping)
# ==================================================

# Random values in range [0,700)
np.random.seed(10)
arr1 = (np.random.random((5,5)) * 700).astype(int)

print("Original Array (Method 1):")
print(arr1)
print()

# Manual clipping using loops
for i in range(arr1.shape[0]):
    for j in range(arr1.shape[1]):
        if arr1[i][j] < 100:
            arr1[i][j] = 100
        elif arr1[i][j] > 200:
            arr1[i][j] = 200

print("After limiting values [100, 200]:")
print(arr1)
print()

# Sorting + cumulative sum
sorted_arr1 = np.sort(arr1, axis=None)     # flatten sort
cumsum_arr1 = np.cumsum(sorted_arr1)

print("Sorted array:")
print(sorted_arr1)
print("Cumulative sum:")
print(cumsum_arr1)
print()


# ==================================================
# METHOD 2: Using NumPy clip() (BEST & RECOMMENDED)
# ==================================================

arr2 = np.linspace(100, 700, 15).reshape(3,5)

print("Original Array (Method 2):")
print(arr2)
print()

# np.clip → values ko min & max ke beech limit karta hai
clipped_arr2 = np.clip(arr2, 100, 200)

print("After limiting values [100, 200]:")
print(clipped_arr2)
print()

# Sorting + cumulative sum
sorted_arr2 = np.sort(clipped_arr2, axis=None)
cumsum_arr2 = np.cumsum(sorted_arr2)

print("Sorted array:")
print(sorted_arr2)
print("Cumulative sum:")
print(cumsum_arr2)
print()


# ==================================================
# METHOD 3: Boolean masking
# ==================================================

arr3 = np.array([50, 120, 250, 180, 90, 300])

print("Original Array (Method 3):")
print(arr3)
print()

# Boolean masking
arr3[arr3 < 100] = 100
arr3[arr3 > 200] = 200

print("After limiting values [100, 200]:")
print(arr3)
print()

sorted_arr3 = np.sort(arr3)
cumsum_arr3 = np.cumsum(sorted_arr3)

print("Sorted array:")
print(sorted_arr3)
print("Cumulative sum:")
print(cumsum_arr3)
````

### `Q-9:` You are given a array ($arr ∈ [0, 1]$). First you have round off the elements upto 3 decimal places and compare that
- 0th percentile == minimum value of the array
- 100th percentile == maximum value of the array
- also find the difference betwen 51th percenile and 50th percentile values

````python
import numpy as np

# --------------------------------------------------
# Step 1: Create array in range [0,1]
# --------------------------------------------------
np.random.seed(42)
arr = np.random.random(10)

print("Original Array:")
print(arr)
print()

# --------------------------------------------------
# Step 2: Round off elements to 3 decimal places
# --------------------------------------------------
arr_rounded = np.round(arr, 3)

print("Rounded Array (3 decimal places):")
print(arr_rounded)
print()

# --------------------------------------------------
# Step 3: Percentile calculations
# --------------------------------------------------

# 0th percentile → minimum value
p0 = np.percentile(arr_rounded, 0)

# 100th percentile → maximum value
p100 = np.percentile(arr_rounded, 100)

# 50th percentile → median
p50 = np.percentile(arr_rounded, 50)

# 51st percentile → just above median
p51 = np.percentile(arr_rounded, 51)

print("0th Percentile (Min):", p0)
print("Actual Min Value   :", np.min(arr_rounded))
print()

print("100th Percentile (Max):", p100)
print("Actual Max Value      :", np.max(arr_rounded))
print()

# --------------------------------------------------
# Step 4: Difference between 51st & 50th percentile
# --------------------------------------------------
diff = p51 - p50

print("50th Percentile:", p50)
print("51st Percentile:", p51)
print("Difference (51st - 50th):", diff)
````

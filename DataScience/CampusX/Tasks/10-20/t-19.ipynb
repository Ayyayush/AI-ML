### `Q-1:` Using `groupby` make groups using the `"Pclass"` column and find out the average age and total number of missing values in the `"Age"` column for every group.

````python
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/e/2PACX-1vQjh5HzZ1N0SU7ME9ZQRzeVTaXaGsV97rU8R7eAcg53k27GTstJp9cRUOfr55go1GRRvTz1NwvyOnuh/pub?gid=1562145139&single=true&output=csv"
titanic_df = pd.read_csv(url)
titanic_df
````

````python
# method -1 
result = titanic_df.groupby('Pclass')['Age'].agg(
    average_age='mean',
    missing_values=lambda x: x.isna().sum()
)

result

````

### `Q-2:` Using `groupby` make groups using the `"Pclass"` column and fill every group's `"Embarked"` column's missing values with the mode value of that group. After that, print every group's `"Embarked"` column's value counts in ascending order.

````python
# Step 1: Fill missing Embarked values using group-wise mode
titanic_df['Embarked'] = (
    titanic_df
    .groupby('Pclass')['Embarked']
    .transform(lambda x: x.fillna(x.mode()[0]))
)

# Step 2: Value counts per Pclass in ascending order
result = (
    titanic_df
    .groupby('Pclass')['Embarked']
    .value_counts()
    .sort_values(ascending=True)
)

result
````

### `Q-3:` Make groups based on `"Embarked"` column. And for each of this embarked group, make another group based on `"Pclass"` and find out the average fare (round off up to 2 decimal places) for each "Pclass" for each group of "Embarked".

**Sample Output:**

```bash
{'C': {1: 105, 2: 25, 3: 11},
 'Q': {1: 90, 2: 12, 3: 11},
 'S': {1: 70, 2: 20, 3: 15}}
```

````python
result = (
    titanic_df
    .groupby(['Embarked', 'Pclass'])['Fare']
    .mean()
    .round(2)
    .unstack()
)

result
````

### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $

After that find out the below values for each group:
- minimum "Passess"
- maximum "Passess"
- minimum "Yellow Cards"
- maximum "Yellow cards"
- average "Yellow Cards"
- maximum "Attempted Line Breaks"
- minimum "Attempted Line Breaks"
- standardard deviation of "Attempted Line Breaks"
- average Possession

````python
import pandas as pd

fifa_df = pd.read_csv(
    "https://docs.google.com/spreadsheets/d/e/2PACX-1vT3D_x_4DS6d51LKJ7ze1sxT5WpV5uiSVOFYHLwBiGru6vFyVv5h5-83AwFjxWYiWfCDjDAaarHAV-k/pub?gid=0&single=true&output=csv"
)

# Z-normalization function
def z_normalization(group, cols_to_perform):
    for col in cols_to_perform:
        group[col] = (group[col] - group[col].mean()) / group[col].std()
    return group

cols = [
    'Passes',
    'Passes Completed',
    'Attempted Line Breaks',
    'Completed Line Breaks'
]

# Apply Z-normalization team-wise
norm_df = fifa_df.groupby('Team', group_keys=False) \
                  .apply(z_normalization, cols_to_perform=cols)

# Required aggregations
result = norm_df.groupby('Team').agg(
    min_passes=('Passes', 'min'),
    max_passes=('Passes', 'max'),
    min_yellow=('Yellow Cards', 'min'),
    max_yellow=('Yellow Cards', 'max'),
    avg_yellow=('Yellow Cards', 'mean'),
    max_line_breaks=('Attempted Line Breaks', 'max'),
    min_line_breaks=('Attempted Line Breaks', 'min'),
    std_line_breaks=('Attempted Line Breaks', 'std'),
    avg_possession=('Possession (%)', 'mean')
)

result
````

###`Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 100+ balls


> Chasing mean team batting in second inning

````python
# Filter only chasing innings (2nd innings)
chasing_df = ipl[ipl['innings'] == 2]

# -------------------------------
# 1️⃣ Highest score while chasing
# -------------------------------
highest_score = (
    chasing_df
    .groupby('batter')['batsman_run']
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

highest_score


````

````python
strike_rate_df = (
    chasing_df
    .groupby('batter')
    .agg(
        runs=('batsman_run', 'sum'),
        balls=('ballnumber', 'count')
    )
)

# Filter batsmen who faced 100+ balls
strike_rate_df = strike_rate_df[strike_rate_df['balls'] >= 100]

# Calculate strike rate
strike_rate_df['strike_rate'] = (strike_rate_df['runs'] / strike_rate_df['balls']) * 100

# Best strike rate
best_strike_rate = strike_rate_df.sort_values('strike_rate', ascending=False).head(1)

best_strike_rate
````

###`Q-6` Most Successful bowler against any batsman. Find that pair of bowler and batsman.
> Most Successful in terms of dissmissal. A bowler who have dissmissed any batsman most no of times. If any two pairs have same no of dissmisal, consider runs conceded by bowler to that batsman. Those who have concede lesser runs is more successful.

````python
# Keep only rows where a dismissal happened
dismissals_df = ipl[ipl['isWicketDelivery'] == 1]

# Count dismissals bowler vs batter
pair_stats = (
    dismissals_df
    .groupby(['bowler', 'batter'])
    .agg(
        dismissals=('isWicketDelivery', 'count'),
        runs_conceded=('batsman_run', 'sum')
    )
    .reset_index()
)

# Sort based on problem statement
result = pair_stats.sort_values(
    by=['dismissals', 'runs_conceded'],
    ascending=[False, True]
).head(1)

result
````

###`Q-7`: Most successful batting pair in IPL. Batting pair who have scored most runs playing together.


````python

# Create batting pairs (order-independent)
ipl['pair'] = ipl.apply(
    lambda x: tuple(sorted([x['batter'], x['non-striker']])),
    axis=1
)

# Calculate total runs scored by each pair
pair_runs = (
    ipl
    .groupby('pair')['batsman_run']
    .sum()
    .sort_values(ascending=False)
)

# Most successful batting pair
most_successful_pair = pair_runs.head(1)

most_successful_pair
````

###`Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# -------------------------------
# Step 1: Create batting pairs (order independent)
# -------------------------------
ipl[['Batsman1', 'Batsman2']] = (
    ipl[['batter', 'non-striker']]
    .apply(lambda x: sorted(x), axis=1, result_type='expand')
)

# -------------------------------
# Step 2: Aggregate required values
# -------------------------------
pair_stats = (
    ipl
    .groupby(['Batsman1', 'Batsman2'])
    .agg(
        Runs=('batsman_run', 'sum'),
        Balls=('ballnumber', 'count'),
        Dismissals=('isWicketDelivery', 'sum')
    )
    .reset_index()
)

# -------------------------------
# Step 3: Calculate Avg & Strike Rate
# -------------------------------
pair_stats['Avg'] = pair_stats['Runs'] / pair_stats['Dismissals']
pair_stats['StrikeRate'] = (pair_stats['Runs'] / pair_stats['Balls']) * 100

# Replace inf Avg (pairs never dismissed)
pair_stats['Avg'] = pair_stats['Avg'].replace(float('inf'), pair_stats['Runs'])

pair_stats
````

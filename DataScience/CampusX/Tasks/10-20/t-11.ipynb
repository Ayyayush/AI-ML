###`Q-1`: You are given a function definition. There might be several issues on execution of this function. You are asked to do exception handling for diffrent errors that this function goes in to `without altering this function`. And print error text.



Function parameters `l -> list, s -> could be anything`

```
def function(l: list, s, **args):
    last_element = l[-1]
    
    l[int(s)]=10
    any_element = l[int(s)+10]
    l[s]=10
    
    res = sum(l)
    
    p = args['p']
    # print(p)
    return res/last_element * p + any_element

```
Check for different function calls:-

```
function([1,2,1], 12)
function([1,2,1]*9, '1-2')
function([1,'2',1]*9, 12)
function([1,'2',1]*9, 12)
function([1,2,0]*9, 12  )
function([1,2,1]*9, 12, p=None)
function([1,2,0]*9, 12, p=10)
```

````python
# -------------------------------------------------
# GIVEN FUNCTION (❌ DO NOT MODIFY)
# -------------------------------------------------
def function(l: list, s, **args):
    last_element = l[-1]                 # May cause IndexError if list empty
    
    l[int(s)] = 10                       # ValueError / IndexError
    any_element = l[int(s) + 10]         # IndexError
    l[s] = 10                            # TypeError (invalid index type)
    
    res = sum(l)                         # TypeError if non-numeric element
    
    p = args['p']                        # KeyError if 'p' missing
    return res / last_element * p + any_element   # ZeroDivisionError / TypeError


# -------------------------------------------------
# SAFE WRAPPER WITH EXCEPTION HANDLING
# -------------------------------------------------
def safe_function(l, s, **args):
    try:
        result = function(l, s, **args)
        print("Result:", result)

    except ValueError as e:
        print("ValueError:", e)           # int(s) conversion failed

    except IndexError as e:
        print("IndexError:", e)           # Index out of range

    except TypeError as e:
        print("TypeError:", e)            # Invalid type usage

    except ZeroDivisionError as e:
        print("ZeroDivisionError:", e)    # Division by zero

    except KeyError as e:
        print("KeyError:", e)             # Missing keyword argument 'p'

    except Exception as e:
        print("Unhandled Exception:", e)


# -------------------------------------------------
# FUNCTION CALLS WITH INLINE ERROR COMMENTS
# -------------------------------------------------

safe_function([1, 2, 1], 12)
# ❌ IndexError → l[int(s)] where int(12) is out of range

safe_function([1, 2, 1] * 9, '1-2')
# ❌ ValueError → int('1-2') cannot be converted to integer

safe_function([1, '2', 1] * 9, 12)
# ❌ TypeError → sum(l) fails because '2' is a string

safe_function([1, '2', 1] * 9, 12)
# ❌ TypeError → same as above (string inside list)

safe_function([1, 2, 0] * 9, 12)
# ❌ ZeroDivisionError → last_element = 0

safe_function([1, 2, 1] * 9, 12, p=None)
# ❌ TypeError → None used in arithmetic ( * p )

safe_function([1, 2, 0] * 9, 12, p=10)
# ❌ ZeroDivisionError → division by zero (last_element = 0)
````

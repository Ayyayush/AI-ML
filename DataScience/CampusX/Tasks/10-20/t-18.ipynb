###`Q-6:` Make a dataframe of each team in IPL with details like - Team Name, Matches Played, Win%, Home Win%, Away Win%.
Show sorted dataframe on Win%

Replace old team name as new name before performing any tasks.
```
Delhi Daredevils ->Delhi Capitals
Kings XI Punjab -> Punjab Kings
Rising Pune Supergiants -> Rising Pune Supergiant
```

Note: Team1 represents Home team. Exclude not result matches.


````python
# code here
ipl_matches = "https://docs.google.com/spreadsheets/d/e/2PACX-1vRy2DUdUbaKx_Co9F0FSnIlyS-8kp4aKv_I0-qzNeghiZHAI_hw94gKG22XTxNJHMFnFVKsO4xWOdIs/pub?gid=1655759976&single=true&output=csv"
matches = pd.read_csv(ipl_matches)
changed_name = {'Delhi Daredevils':'Delhi Capitals', 
                'Kings XI Punjab':'Punjab Kings',
                'Rising Pune Supergiants':'Rising Pune Supergiant'}
matches.replace(changed_name.keys(), changed_name.values(),inplace=True )
````

````python
# Remove matches where the winning team is NaN (i.e., match had no result)
df = matches[~matches.WinningTeam.isna()]

# Create an empty DataFrame to store final results
new_df = pd.DataFrame()

# Get all unique teams from Team1 column
teams = df.Team1.unique()

# List to collect computed data for each team
data = []

# Loop through each team
for team in teams:
    
    # Total matches played by the team (home + away)
    played = df[(df.Team1 == team) | (df.Team2 == team)].shape[0]
    
    # Total matches won by the team
    wins = df[df.WinningTeam == team].shape[0]
    
    # Home win percentage:
    # Matches where team is Team1 AND wins ÷ total home matches
    home_win = (
        df[(df.WinningTeam == team) & (df.Team1 == team)].shape[0]
        / df[df.Team1 == team].shape[0]
    ) * 100
    
    # Away win percentage:
    # Matches where team is Team2 AND wins ÷ total away matches
    away_win = (
        df[(df.WinningTeam == team) & (df.Team2 == team)].shape[0]
        / df[df.Team2 == team].shape[0]
    ) * 100
    
    # Append calculated metrics for the team
    data.append([
        team,                      # Team name
        played,                    # Matches played
        wins / played * 100,       # Overall win percentage
        home_win,                  # Home win percentage
        away_win                   # Away win percentage
    ])

# Assign collected data to DataFrame with column names
new_df[[
    "Team",
    "Matches Played",
    "Win Percentage",
    "Home Win Percentage",
    "Away Win Percentage"
]] = data

# Sort teams by overall win percentage (highest first)
new_df.sort_values("Win Percentage", inplace=True, ascending=False)

# Display the final table
new_df
````

###`Q-8:` Player with most appearance in final match.

`Team1Players` and `Team2Players` have all players name. It is not a list of players name instead it is str. So handle it as string.

Hint: split and strip will help; Make a series of all players in final and do value counts


````python
# Function to extract individual player names from the string representation
def getPlayers(player_string):
    
    # Remove leading "'[" and trailing "]'" from the string
    cleaned = player_string.lstrip("'[").rstrip("]'")
    
    # Split the string into a list of players using comma as separator
    players = cleaned.split(", ")
    
    # Remove single quotes from each player name
    players = [player.strip("'") for player in players]
    
    # Convert the list into a Pandas Series
    return pd.Series(players)


# Filter the dataframe to get only the Final match
df_final = matches[matches["MatchNumber"] == "Final"]

# Create an empty Series to store all players from the final
playerSeries = pd.Series(dtype=object)

# Extract players from Team1Players column
for player_list in df_final.Team1Players:
    
    # Get individual players as a Series
    x = getPlayers(player_list)
    
    # Append players to the main Series
    playerSeries = playerSeries.append(x)


# Extract players from Team2Players column
for player_list in df_final.Team2Players:
    
    # Get individual players as a Series
    x = getPlayers(player_list)
    
    # Append players to the main Series
    playerSeries = playerSeries.append(x)


# Count how many times each player appears and show top 5
playerSeries.value_counts().head()
````

###`Q-9:` IPL Point Table

Make a function `point_table` which take `season` as parameter and show points table in non-ascendng order of points and in ascending order of team name.

For winning - 2 Ponits;
For loosing - 0 Point
For not result both team gets 1 points.

Make dataframe which will have
`TeamName` `MatchesPlayed` `MatchesWon` `NoResult` `Points`
make `TeamName` as index.

```
season parametr should be one of these->
['2022', '2021', '2020/21', '2019', '2018', '2017', '2016', '2015', '2014', '2013', '2012', '2011', '2009/10', '2009', '2007/08']
```


Output of two Top 2 in season 2022
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points

Gujarat Titans	    16	   12	       0	     24
Rajasthan Royals	  17	   10	       0	     20

```

````python
# Function to calculate total matches played by a team
def matches_played(df, team):
    
    # Count matches where the team appears as Team1 or Team2
    return df[(df.Team1 == team) | (df.Team2 == team)].shape[0]


# Function to calculate total matches won by a team
def matches_won(df, team):
    
    # Count matches where the team is the winning team
    return df[df.WinningTeam == team].shape[0]


# Function to calculate matches with no result for a team
def matches_no_result(df, team):
    
    # Match played by team AND WinningTeam is null (no result)
    return df[
        ((df.Team1 == team) | (df.Team2 == team)) &
        (df.WinningTeam.isnull())
    ].shape[0]


# Function to generate points table for a given season
def point_table(season):
    
    # Filter matches of the given season
    df = matches[matches.Season == season]
    
    # Create an empty dataframe for points table
    new_df = pd.DataFrame()
    
    # Add unique team names
    new_df['Team Name'] = df.Team1.unique()
    
    # Calculate matches played for each team
    new_df["Matches Played"] = new_df["Team Name"].apply(
        lambda x: matches_played(df, x)
    )
    
    # Calculate matches won for each team
    new_df["Matches Won"] = new_df["Team Name"].apply(
        lambda x: matches_won(df, x)
    )
    
    # Calculate no-result matches for each team
    new_df["No Result"] = new_df["Team Name"].apply(
        lambda x: matches_no_result(df, x)
    )
    
    # Calculate total points
    # Win = 2 points, No Result = 1 point
    new_df["Points"] = new_df["Matches Won"] * 2 + new_df["No Result"]
    
    # Sort teams by points in descending order
    new_df.sort_values("Points", ascending=False, inplace=True)
    
    # Set Team Name as index
    new_df.set_index("Team Name", inplace=True)
    
    # Return final points table
    return new_df


# Generate points table for season 2022
point_table("2022")
````

###`Q-10:` IPL Point Table cont.
Extend the above IPL Point Table with an extra column as `SeasonPosition`

Team below top 4 after sorting on `Points` and then on `TeamName` Will have same `SeasonPosition` as there rank. use rank function.

Teams in Top four will have `SeasonPosition` as:
```
    'Winner' - Team won final
    'Runner' - Team lost Final
    3 - Losing Team in Qualifier2
    4 - Losing Team in Eliminator
```

For changing value of pariticular cell use `df.at[row_index, col_label] = value`

Output of two Top 2 in season 2022. Your result should have all teams.
```
TeamName    MatchesPlayed	MatchesWon	NoResult	Points   SeasonPosition

Gujarat Titans	    16	   12	       0	     24         Winner
Rajasthan Royals	  17	   10	       0	     20         Runner

```

Note: If you try to chnage value of view of any dataframe a warnig will be shown. To avoid it, make a copy of the dataframe you want to change in by `df.copy()`

````python
# Function to extend the points table with final season positions
def point_table_extension(season):
    
    # Get the base points table for the season
    df3 = point_table(season)
    
    # Filter match data for the given season
    df = matches[matches.Season == season].copy()
    
    # Rank teams based on points (highest points = rank 1)
    # 'first' ensures unique ranking in case of ties
    df3["SeasonPosition"] = (
        df3.Points.rank(ascending=False, method='first')
        .astype('object')
    )
    
    # Identify losing teams:
    # If Team1 wins → Team2 is losing team
    # If Team2 wins → Team1 is losing team
    df["LoosingTeam"] = (
        df[df.WinningTeam == df.Team1]["Team2"]
        .append(df[df.WinningTeam == df.Team2]["Team1"])
    )
    
    # Extract Final match
    final = df[df["MatchNumber"] == "Final"]
    
    # Get winner and runner-up from the Final
    wining_team = final.WinningTeam.values[0]
    runner = final.LoosingTeam.values[0]
    
    # Assign special season positions
    df3.at[wining_team, "SeasonPosition"] = "Winner"
    df3.at[runner, "SeasonPosition"] = "Runner"
    
    # Extract Qualifier 2 match (for third place)
    q = df[df["MatchNumber"] == "Qualifier 2"]
    
    # Extract Eliminator match (for fourth place)
    e = df[df["MatchNumber"] == "Eliminator"]
    
    # Losing team of Qualifier 2 → Third place
    third = q.LoosingTeam.values[0]
    
    # Losing team of Eliminator → Fourth place
    fourth = e.LoosingTeam.values[0]
    
    # Assign third and fourth positions
    df3.at[third, "SeasonPosition"] = "Third"
    df3.at[fourth, "SeasonPosition"] = "Fourth"
    
    # Return the extended points table
    return df3


# Generate extended points table for season 2022
point_table_extension("2022")
````

###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

````python
# Import numpy library
import numpy as np  

# Target value
a = 23  

# Array of numbers
arr = np.array([10, 17, 24, 31, 38, 45, 52, 59])  

# Calculate absolute difference between each element and target
result = abs(arr - a)  

# Find the index of the element with the minimum difference
# This gives the element closest to 'a'
index_of_closest = np.argmin(result)  

# Print the index
print(index_of_closest)  
````

###`Q-2:` Replace multiples of 3 or 5 as 0 in the given array.
```
arr=[1 2 3 4 5 6 7 9]

result-> [1 2 0 4 0 0 7 0]
```

````python
import numpy as np

# Original array
arr = np.array([10, 20, 25, 30, 40, 50])

# -----------------------------
# Method 1: Using a for loop
# -----------------------------
# Loop through the indices of the array
for i in range(len(arr)):
    if arr[i] % 3 == 0:      # Check if element is divisible by 3
        arr[i] = 0           # Replace element with 0

print("After Method 1:", arr)

# Reset array for next methods
arr = np.array([10, 20, 25, 30, 40, 50])

# -----------------------------
# Method 2: Using fancy indexing
# -----------------------------
indices = np.where(arr % 3 == 0)  # Get indices where elements are divisible by 3
arr[indices] = 0                  # Replace those elements with 0

print("After Method 2:", arr)

# Reset array for next method
arr = np.array([10, 20, 25, 30, 40, 50])

# -----------------------------
# Method 3: Using boolean indexing
# -----------------------------
arr[arr % 3 == 0] = 0             # Boolean mask directly replaces elements divisible by 3

print("After Method 3:", arr)
````

###`Q-3:` Use Fancy Indexing.
1. Double the array elements at  given indexes
```
arr = np.arrange(10)
indexes = [0,3,4,9]
```
Result -> `[ 0  1  2  6  8  5  6  7  8 18]`

2. Using a given array make a different array as in below example
```
array = [1,2,3]
result array -> [1 1 1 2 2 2 3 3 3]
```
* Internal-repetion should be as length of the array.

Hint:
```
if a is an array
a = [2,4]
a[[1,1,0,1]] will result in-> [4 4 2 4]
```

````python
import numpy as np


# Part -1 
# Original array
arr = np.arange(10)  # Corrected from np.arrange
print("Original array:", arr)

# -----------------------------
# Method 1: Using boolean/fancy indexing
# -----------------------------
indexes = [0, 3, 4, 9]      # Indices to double
arr[indexes] = arr[indexes] * 2  # Multiply selected elements by 2
print("After Method 1:", arr)

# Reset array for next methods
arr = np.arange(10)

# -----------------------------
# Method 2: Using a for loop
# -----------------------------
indices = [0, 3, 4, 9]
for i in range(len(indices)):
    arr[indices[i]] *= 2      # Multiply each selected element by 2
print("After Method 2:", arr)

# Reset array for next methods
arr = np.arange(10)

# -----------------------------
# Method 3: Using fancy indexing (NumPy style)
# -----------------------------
arr[np.array([0, 3, 4, 9])] *= 2  # Fancy indexing directly multiplies selected elements
print("After Method 3:", arr)



# =====================================================
# PART - 2 : Internal Repetition using Fancy Indexing
# =====================================================

array = np.array([1, 2, 3])
print("\nOriginal array:", array)

# Length of array
n = len(array)

# Create index pattern using fancy indexing
# np.arange(n) → [0,1,2]
# np.repeat(..., n) → [0,0,0, 1,1,1, 2,2,2]
indexes = np.repeat(np.arange(n), n)

# Fancy indexing to repeat elements
result = array[indexes]

print("Result array:", result)
# Output → [1 1 1 2 2 2 3 3 3]

````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
import numpy as np
from collections import Counter

# =====================================================
# Q-4: Fill NaN values with the most common element
# =====================================================

# Original array
arr = np.array([[1, 2, np.nan],
                [4, 2, 6],
                [np.nan, np.nan, 5]])

print("Original Array:\n", arr)


# =====================================================
# METHOD 1: Pure NumPy (Best & Recommended)
# =====================================================

# Step 1: Remove NaN values using boolean indexing
flat = arr[~np.isnan(arr)]                 # keeps only valid numbers

# Step 2: Find unique values and their counts
values, counts = np.unique(flat, return_counts=True)

# Step 3: Get the most frequent element
most_common = values[np.argmax(counts)]

# Step 4: Replace NaN with most common value
arr[np.isnan(arr)] = most_common

print("\nMethod 1 Result:\n", arr)


# =====================================================
# Reset array for next method
# =====================================================
arr = np.array([[1, 2, np.nan],
                [4, 2, 6],
                [np.nan, np.nan, 5]])


# =====================================================
# METHOD 2: Using Loops + Counter
# =====================================================

elements = []                              # to store non-NaN values

# Step 1: Collect all non-NaN elements
for i in range(len(arr)):
    for j in range(len(arr[i])):
        if not np.isnan(arr[i][j]):
            elements.append(arr[i][j])

# Step 2: Find most common element
most_common = Counter(elements).most_common(1)[0][0]

# Step 3: Replace NaN values
for i in range(len(arr)):
    for j in range(len(arr[i])):
        if np.isnan(arr[i][j]):
            arr[i][j] = most_common

print("\nMethod 2 Result:\n", arr)


# =====================================================
# Reset array for next method
# =====================================================
arr = np.array([[1, 2, np.nan],
                [4, 2, 6],
                [np.nan, np.nan, 5]])


# =====================================================
# METHOD 3: NumPy Flatten + Counter
# =====================================================

# Step 1: Flatten array
flat = arr.flatten()

# Step 2: Remove NaN values
clean = flat[~np.isnan(flat)]

# Step 3: Find most common element
most_common = Counter(clean).most_common(1)[0][0]

# Step 4: Replace NaN using boolean indexing
arr[np.isnan(arr)] = most_common

print("\nMethod 3 Result:\n", arr)


# =====================================================
# METHOD 4: One-liner NumPy Approach
# =====================================================

arr = np.array([[1, 2, np.nan],
                [4, 2, 6],
                [np.nan, np.nan, 5]])

arr[np.isnan(arr)] = np.bincount(
    arr[~np.isnan(arr)].astype(int)
).argmax()

print("\nMethod 4 Result:\n", arr)
````

###`Q-6:` Given two arrays, X and Y, construct the Cauchy matrix C.
`Cij =1/(xi - yj)`

http://en.wikipedia.org/wiki/Cauchy_matrix
```
x = numpy.array([1,2,3,4]).reshape((-1, 1)
y = numpy.array([5,6,7])
```

````python
import numpy as np

# =====================================================
# Q-6: Construct a Cauchy Matrix
# Formula: Cij = 1 / (xi - yj)
# =====================================================


# =====================================================
# METHOD 1: Using NumPy Broadcasting (Recommended)
# =====================================================

# Step 1: Create array x
# reshape((-1, 1)) converts x into a column vector
# Shape becomes (4, 1)
x = np.array([1, 2, 3, 4]).reshape((-1, 1))

# Step 2: Create array y
# Shape is (3,)
y = np.array([5, 6, 7])

# Step 3: Broadcasting happens automatically here
# x shape -> (4,1)
# y shape -> (3,) treated as (1,3)
# Result -> (4,3) matrix
C1 = 1 / (x - y)

print("Cauchy Matrix using Broadcasting (Method 1):")
print(C1)


# =====================================================
# METHOD 2: Using Nested Loops (Conceptual / Learning)
# =====================================================

# Step 1: Define x and y as 1D arrays
x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7])

# Step 2: Initialize empty matrix of size (len(x), len(y))
C2 = np.zeros((len(x), len(y)))

# Step 3: Fill the matrix using formula Cij = 1 / (xi - yj)
for i in range(len(x)):                 # loop over rows (x values)
    for j in range(len(y)):             # loop over columns (y values)
        C2[i][j] = 1 / (x[i] - y[j])    # apply Cauchy formula

print("\nCauchy Matrix using Loops (Method 2):")
print(C2)
````

###`Q-7:` Plot this below equation.

$$\large y = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

**Note: This equation is called tanh activation function. In deep learning, many times this function is used. If you find some difference between the sigmoid function and this tanh function, note that to your notebook.**

````python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate x values
# linspace creates evenly spaced values between -5 and 5
x = np.linspace(-5, 5, 400)

# Step 2: Apply the given tanh formula
# y = (e^x - e^-x) / (e^x + e^-x)
y = (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

# Step 3: Plot the graph
plt.figure()
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Tanh Activation Function")
plt.show()
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
def check_broadcast(shape_a, shape_b):
    """
    Checks whether two arrays with given shapes
    can be broadcasted together according to NumPy rules.
    """

    # Convert shapes to lists for easy handling
    a = list(shape_a)
    b = list(shape_b)

    # Reverse shapes to compare from the last dimension
    a.reverse()
    b.reverse()

    # Compare each dimension
    for i in range(min(len(a), len(b))):
        # If dimensions are not equal and neither is 1 → cannot broadcast
        if a[i] != b[i] and a[i] != 1 and b[i] != 1:
            return False

    # If all dimensions satisfy broadcasting rules
    return True


# ==========================
# Test Case
# ==========================

shape_a = (3, 2, 2)
shape_b = (2, 2)

result = check_broadcast(shape_a, shape_b)
print(result)    # Expected output: True

````

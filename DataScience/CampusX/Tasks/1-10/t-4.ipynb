###`Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
# Write code here
num1 = [23, 45, 67, 78, 89, 34]
num2 = [34, 89, 55, 56, 39, 67]
nums3 = []

for n in num1:
    if n  in num2:                 # ✅ corrected syntax
        nums3.append(n)

for n in num2:
    if n in num1 and n not in nums3:                 # ✅ corrected syntax
        nums3.append(n)

nums3.sort()
print(nums3)
````

# Problem 7:
# Sort a list of alphanumeric strings based on product value of numeric character in it.
# If in any string there is no numeric character, take its product value as 1.

# Input:
L = ['1ac21', '23fg', '456', '098d', '1', 'k1s']


# output 
['456', '23fg', '1ac21', '1', 'k1s', '098d']


````python
# Sample list of alphanumeric strings
L = ['1ac21', '23fg', '456', '098d', '1', 'kls']

# Empty list jisme hum har string ka product store karenge
prod_value = []

# 🔁 Step 1: Iterate over every string in the list
for item in L:
    product = 1                                  # Initial product value = 1 (multiplicative identity)
    
    # 🔁 Step 2: Iterate over every character in the string
    for char in item:
        if char.isdigit():                       # Agar character ek digit hai
            product *= int(char)                 # Multiply karte jao product mein
    
    # Step 3: Final product value ko prod_value list mein daal do
    prod_value.append(product)


# ⚙️ Step 4: Pair each product value with its corresponding string
# zip() se dono lists ko combine karte hain tuple ke form mein
paired = list(zip(prod_value, L))
# Example: [(2, '1ac21'), (6, '23fg'), (120, '456'), (0, '098d'), (1, '1'), (1, 'kls')]


# ⚙️ Step 5: Sort the paired list in descending order based on the product values
paired.sort(reverse=True)
# After sorting → [(120, '456'), (6, '23fg'), (2, '1ac21'), (1, '1'), (1, 'kls'), (0, '098d')]


# ⚙️ Step 6: Extract only the strings from the sorted paired list
result = [string for _, string in paired]
# Output list will contain strings in order of decreasing product value
# result = ['456', '23fg', '1ac21', '1', 'kls', '098d']


# Step 7: Print the final sorted list
print(result)                                   # 👉 Output: ['456', '23fg', '1ac21', '1', 'kls', '098d']


# 🔁 Alternative One-Liner (Same logic in short)
# print([i[1] for i in sorted(list(zip(prod_value, L)), reverse=True)])


# 🧾 Dry Run (for understanding)
# L = ['1ac21', '23fg', '456', '098d', '1', 'kls']
# Products → [2, 6, 120, 0, 1, 1]
# Paired → [(2,'1ac21'), (6,'23fg'), (120,'456'), (0,'098d'), (1,'1'), (1,'kls')]
# Sorted (desc) → [(120,'456'), (6,'23fg'), (2,'1ac21'), (1,'1'), (1,'kls'), (0,'098d')]
# Final result → ['456', '23fg', '1ac21', '1', 'kls', '098d']

````

`Problem 8:` Split String of list on K character.

**Example :**

Input:
```bash
['CampusX is a channel', 'for data-science', 'aspirants.']
```

Output:
```bash
['CampusX', 'is', 'a', 'channel', 'for', 'data-science', 'aspirants.']
```

````python

# -------------------------------------------------------------------
# 🧩 METHOD 1 — Using extend() and split()
# -------------------------------------------------------------------

lst = ['CampusX is a channel', 'for data-science', 'aspirants.']    # Original list of sentences
result = []                                                         # Empty list to store final words

# 🔁 Step 1: Iterate over each sentence in the list
for sentence in lst:
    # sentence.split() → splits sentence into words wherever space (' ') appears
       
    # result.extend() → adds all those words into the result list (not nested)
    result.extend(sentence.split())                                 # Add split words to final list

# Step 2: Print final list of words
print(result)



# -------------------------------------------------------------------
# 🧩 METHOD 2 — Using List Comprehension (Shorter and Pythonic way)
# -------------------------------------------------------------------

lst = ['CampusX is a channel', 'for data-science', 'aspirants.']    # Same list as before

# ✅ This is a nested list comprehension
result = [word for sentence in lst for word in sentence.split()]

# Step 2: Print final flattened list
print(result)
# 👉 Output: ['CampusX', 'is', 'a', 'channel', 'for', 'data-science', 'aspirants.']

````

### `Problem 9:` Convert Character Matrix to single String using string comprehension.

**Example 1:**

Input:
```bash
[['c', 'a', 'm', 'p', 'u', 'x'], ['i', 's'], ['b', 'e', 's', 't'], ['c', 'h', 'a', 'n', 'n', 'e', 'l']]
```

Output:
```bash
campux is best channel
```

````python
# 🎯 Objective:
# Humein ek nested list di gayi hai jisme har inner list ek word ke characters contain karti hai.
# Task → Un characters ko join karke poora sentence banana hai.
# Example:
# Input  : [['c','a','m','p','u','x'], ['i','s'], ['b','e','s','t'], ['c','h','a','n','n','e','l']]
# Output : "campux is best channel"


# -------------------------------------------------------------------
# 🧩 METHOD 1 — Using Nested Loops
# -------------------------------------------------------------------

list1 = [['c', 'a', 'm', 'p', 'u', 'x'], ['i', 's'], ['b', 'e', 's', 't'], ['c', 'h', 'a', 'n', 'n', 'e', 'l']]   # Given nested list
temp = ""                                              # Empty string to store final sentence

# 🔁 Outer loop → iterate over each sublist (word)
for i in range(len(list1)):
    # 🔁 Inner loop → iterate over each character of that word
    for j in range(len(list1[i])):                     # ✅ fixed: iterate over current sublist length
        temp += list1[i][j]                            # Add each character to temp (form word)
    
    temp += ' '                                        # Add space after each word

print(temp)                                            # 👉 Output: campux is best channel




# -------------------------------------------------------------------
# 🧩 METHOD 2 — Using Nested Join (Shortest & Pythonic Way)
# -------------------------------------------------------------------

# ✅ Inner join: "".join(i) → joins each sublist’s characters to form a word
# ✅ Outer join: " ".join([...]) → joins all words with a space between them
# Combined → creates final sentence directly
print(" ".join(["".join(i) for i in list1]))           # 👉 Output: campux is best channel



# -------------------------------------------------------------------
# 🧩 METHOD 3 — Step-by-step join using intermediate list
# -------------------------------------------------------------------

matrix = [['c', 'a', 'm', 'p', 'u', 'x'], ['i', 's'], ['b', 'e', 's', 't'], ['c', 'h', 'a', 'n', 'n', 'e', 'l']]   # Same input list

result = []                                            # Step 1: Empty list to store each complete word

for word in matrix:                                    # Step 2: Loop over each character list
    result.append(''.join(word))                       # Join all chars in sublist → single word → append to result
# result = ['campux', 'is', 'best', 'channel']

final_string = ' '.join(result)                        # Step 3: Join all words with space between them
print(final_string)                                    # 👉 Output: campux is best channel

````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

```bash
['campusxIs', 'bestFor', 'dataScientist']
```

Output:
```bash
['campusx Is', 'best For', 'data Scientist']
```

````python

# =====================================================================
# 🧩 METHOD 1 — Using Loops and isupper() Check
# =====================================================================

lst = ['campusxIs', 'bestFor', 'dataScientist']
result = []                                    # Final list to store modified strings

for word in lst:
    new_word = ""                              # Empty string to build updated version
    for char in word:
        if char.isupper():                     # Agar uppercase letter mile
            new_word += " " + char             # uske pehle ek space daal do
        else:
            new_word += char                   # warna normal add karo
    result.append(new_word.strip())            # Extra space hata ke final list mein daal do

print("Method 1 Output:", result)
# 👉 Output: ['campusx Is', 'best For', 'data Scientist']




# =====================================================================
# 🧩 METHOD 2 — Using Regular Expression (re.sub)
# =====================================================================

import re                                     # Regular expressions ke liye module

lst = ['campusxIs', 'bestFor', 'dataScientist']
result = [re.sub(r'([a-z])([A-Z])', r'\1 \2', word) for word in lst]
# Explanation:
# ([a-z])([A-Z]) → pattern find karta hai jahan lowercase ke baad uppercase letter ho
# r'\1 \2' → unke beech ek space daal deta hai (Group 1 + space + Group 2)

print("Method 2 Output:", result)
# 👉 Output: ['campusx Is', 'best For', 'data Scientist']


# =====================================================================
# 🧩 METHOD 3 — Using List Comprehension + Loop Logic Inline
# =====================================================================

lst = ['campusxIs', 'bestFor', 'dataScientist']
result = [''.join([' ' + ch if ch.isupper() else ch for ch in word]).strip() for word in lst]

print("Method 3 Output:", result)
# 👉 Output: ['campusx Is', 'best For', 'data Scientist']


# =====================================================================
# 🧩 METHOD 4 — Using map() with Helper Function
# =====================================================================

def add_space(word):                           # Helper function to process one word
    res = ""
    for ch in word:
        if ch.isupper():
            res += " " + ch
        else:
            res += ch
    return res.strip()

lst = ['campusxIs', 'bestFor', 'dataScientist']
result = list(map(add_space, lst))             # map() applies add_space on each element
print("Method 4 Output:", result)
# 👉 Output: ['campusx Is', 'best For', 'data Scientist']

````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

````python
# Write code here
# Input lists
list1 = [1, 2, 3, 4, 5, 1]
list2 = [2, 3, 5, 7, 8]

# Create an empty result list
union_list = []

# Step 1: Add all elements from list1 to the result (but only once)
for num in list1:
    if num not in union_list:
        union_list.append(num)

# Step 2: Add elements from list2 only if they are not already in the result
for num in list2:
    if num not in union_list:
        union_list.append(num)

# Print the result
print("Union of two lists:", union_list)





#2nd method 
# Input lists
list1 = [1, 2, 3, 4, 5, 1]
list2 = [2, 3, 5, 7, 8]

# Perform union using set to remove duplicates
union_result = list(set(list1 + list2))

# Optional: Sort the result if you want a sorted union
union_result.sort()

# Print the result
print(union_result)
````

### `Problem 12:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

````python
# Write code here
# 1st method
# Input matrix
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Empty list to store maximums
max_in_rows = []

# Loop through each row in the matrix
for row in matrix:
    max_value = row[0]  # assume first element is max
    for num in row:
        if num > max_value:
            max_value = num
    max_in_rows.append(max_value)

# Print the result
print("Maximum of each row:", max_in_rows)


#2nd method
result=[]
for i in matrix:
    result.append(max(i))

print(result)
````

### `Problem 13:` Write a list comprehension to print the following matrix

[[0, 1, 2], [3, 4, 5], [6, 7, 8]]

````python
# Write code here
#1st method 
matrix=[[0, 1, 2], [3, 4, 5], [6, 7, 8]]
flattened = []                        # Empty list to store all values

for row in matrix:                   # Loop through each row
    for val in row:                  # Loop through each value in row
        flattened.append(val)        # Add it to the flat list

print(flattened)

#2nd method 
flattened = [value for row in matrix for value in row]
print(flattened)

````

### `Problem 14:` Write a list comprehension that can transpose a given matrix

matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

[1, 4, 7]<br>
[2, 5, 8]<br>
[3, 6, 9]<br>

````python
# Dimensions of matrix
rows = len(matrix)
cols = len(matrix[0])

# Create an empty matrix of size cols x rows
transpose = []

# Outer loop for columns of original (will become rows of transpose)
for i in range(cols):
    new_row = []
    for j in range(rows):
        new_row.append(matrix[j][i])   # Pick i-th element from each row
    transpose.append(new_row)          # Add new row to transposed matrix

# Print the transposed matrix
for row in transpose:
    print(row)
````

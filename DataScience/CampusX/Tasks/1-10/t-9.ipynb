### `Problem-1:` Class inheritence

Create a **Bus** child class that inherits from the Vehicle class. The default fare charge of any vehicle is seating capacity * 100. If Vehicle is Bus instance, we need to add an extra 10% on full fare as a maintenance charge. So total fare for bus instance will become the final amount = total fare + 10% of the total fare.

Note: The bus seating capacity is 50. so the final fare amount should be 5500. You need to override the fare() method of a Vehicle class in Bus class.

````python
# Base class
class Vehicle:
    
    def __init__(self, type, capacity):   # Constructor method
        # Instance variables
        self.type = type           # Vehicle ka type (e.g., Car, Bus)
        self.capacity = capacity   # Kitne log baith sakte hain
    
    def fare(self):
        # Basic fare calculation (per person = ₹100)
        return 100 * self.capacity
     

# Child class (inherits from Vehicle)
class Bus(Vehicle):
    
    def fare(self):
        # Parent class ka fare() call karna using super()
        base_fare = super().fare()
        
        # Bus ka extra 10% maintenance charge add karna
        bus_fare = base_fare + 0.1 * base_fare
        
        return bus_fare


# Object creation for Bus
bus = Bus('School Bus', 50)  # Type = "School Bus", Capacity = 50

# Printing the total fare for the bus
print(bus.fare())  # Output: 5500.0
````

### `Problem-3:` Write a program that has a class Point. Define another class Location which has two objects (Location & Destination) of class Point. Also define a function in Location that prints the reflection of Destination on the x axis.

````python
# =========================
# Point CLASS
# =========================
# This class represents a single point in 2D space
# A point has two coordinates:
#   x → x-axis value
#   y → y-axis value
class Point:
    
    def __init__(self, x, y):
        # Constructor of Point class
        # 'self' refers to the current object being created
        
        self.x = x        # Stores x-coordinate of the point
        self.y = y        # Stores y-coordinate of the point
     
    def show_point(self):
        # This method returns the point in "x,y" string format
        # It does NOT print directly, it just returns the value
        return '{},{}'.format(self.x, self.y)


# =========================
# Location CLASS
# =========================
# This class represents a location having:
#   - a source point
#   - a destination point
#
# IMPORTANT:
# Location does NOT inherit from Point
# Instead, it USES Point objects inside it
# This is called COMPOSITION (HAS-A relationship)
class Location:
    
    def __init__(self, x1, y1, x2, y2):
        # Constructor of Location class
        
        # Creating a Point object for source
        # Location HAS-A Point (source)
        self.source = Point(x1, y1)
        
        # Creating a Point object for destination
        # Location HAS-A Point (destination)
        self.destination = Point(x2, y2)
        
    def show(self):
        # This method prints both source and destination points
        
        # Calling show_point() of Point object (source)
        print("Source is:", self.source.show_point())
        
        # Calling show_point() of Point object (destination)
        print("Destination is:", self.destination.show_point())
        
    def reflection(self):
        # Reflection means mirroring the destination point
        # across the Y-axis
        
        # To mirror across Y-axis:
        # x-coordinate becomes negative
        self.destination.x = -self.destination.x
        
        # Printing the reflected destination point
        print("Reflection is:", self.destination.show_point())


# =========================
# OBJECT CREATION & METHOD CALLS
# =========================

# Creating an object of Location class
# Source point  → (0, 0)
# Destination  → (1, 1)
L = Location(0, 0, 1, 1)

# Calling show() method
# This prints source and destination points
L.show()

# Calling reflection()
# This changes destination x-coordinate to negative
L.reflection()
````

### `Problem-4:` Write a program that has an abstract class Polygon. Derive two classes Rectangle and Triamgle from Polygon and write methods to get the details of their dimensions and hence calculate the area.

````python
from abc import ABC, abstractmethod   # ABC = Abstract Base Class

# Abstract Class
class Polygon(ABC):   # Har polygon ek abstract class se derive karega
    
    @abstractmethod   # jis bhi method pe abstract laga h mtlb usko implemetn karna  mandatory h subclasses mein 
    def get_data(self, *args):    # Har polygon apna apna data lega (length, breadth, height, etc.)
        pass
    
    @abstractmethod
    def area(self):               # Har polygon apna area calculate karega
        pass


# Rectangle class (Polygon se inherit karegi)
class Rectangle(Polygon):
    
    def get_data(self, l, b):   # Rectangle ke liye length aur breadth
        self.l = l
        self.b = b
          
    def area(self):             # Area = length * breadth
        return self.l * self.b
          

# Triangle class (Polygon se inherit karegi)
class Triangle(Polygon):
    
    def get_data(self, h, b):   # Triangle ke liye height aur base
        self.h = h
        self.b = b
          
    def area(self):             # Area = 1/2 * base * height
        return 0.5 * self.h * self.b


# Pentagon class (Polygon se inherit karegi)
class Pentagon(Polygon):
    
    def get_data(self, a, p):   # Pentagon ke liye apothem (a) aur perimeter (p)
        self.a = a
        self.p = p
          
    def area(self):             # Area = 1/2 * apothem * perimeter
        return 0.5 * self.a * self.p


# --------------------------
# Ab objects banate hain
# --------------------------

rect = Rectangle()
rect.get_data(4, 5)
print("Rectangle Area:", rect.area())     # Output: 20

tri = Triangle()
tri.get_data(4, 5)
print("Triangle Area:", tri.area())       # Output: 10.0

pent = Pentagon()
pent.get_data(6, 30)
print("Pentagon Area:", pent.area())      # Output: 90.0
````

### `Problem-5:` Write a program with class Bill. The users have the option to pay the bill either by cheque or by cash. Use the inheritance to model this situation.

````python
# Base class
class Bill:
    
    def __init__(self, amount):
        self.amount = amount

    def show_bill(self):
        print("Total Bill Amount:", self.amount)


# Child class for cash payment
class CashPayment(Bill):
    
    def pay_cash(self):
        self.show_bill()
        print("Payment done using CASH")


# Child class for cheque payment
class ChequePayment(Bill):
    
    def pay_cheque(self):
        self.show_bill()
        print("Payment done using CHEQUE")
        
amount = int(input("Enter bill amount: "))
choice = int(input("Pay by Cash(1) or Cheque(2): "))

if choice == 1:
    c = CashPayment(amount)
    c.pay_cash()
else:
    ch = ChequePayment(amount)
    ch.pay_cheque()

````

###`Q-6:` FlexibleDict
As of now we are accessing values from dictionary with exact keys. Now we want to amend accessing values functionality. if a dict have key `1` (int) the even if we try to access values by giving `'1'` (1 as str) as key, we should get the same result and vice versa.

Write a class `FlexibleDict` upon builtin `dict` class with above required functionality.

Hint- `dict[key] => dict.__getitem__(key)`


Ex.
```
fd = FlexibleDict()
fd['a'] = 100
print(fd['a']) # Like regular dict

fd[5] = 500
print(fd[5]) # Like regular dict

fd[1] = 100
print(fd['1']) # actual Key is int but still trying to access through str key.
fd['1'] = 100
print(fd[1])

```
`Output:`
```
100
500
100
100

```

````python
class FlexibleDict(dict):
    """
    FlexibleDict extends the built-in dict class.
    It allows accessing values using:
    - int key even if stored as str
    - str key even if stored as int
    """

    def __getitem__(self, key):
        """
        This method is called internally when we do:
        dict[key]

        Example:
        fd[1]   -> calls __getitem__(1)
        fd['1'] -> calls __getitem__('1')
        """

        # 1️⃣ First, try normal dictionary access
        # This works when exact key exists
        try:
            return super().__getitem__(key)
        except KeyError:
            # 2️⃣ If exact key is NOT found, try alternate form
            
            # Case 1: key is string and looks like a number
            if isinstance(key, str) and key.isdigit():
                alt_key = int(key)
            
            # Case 2: key is integer → convert to string
            elif isinstance(key, int):
                alt_key = str(key)
            
            else:
                # Any other key type → raise error
                raise KeyError(key)

            # 3️⃣ Try accessing dictionary with converted key
            try:
                return super().__getitem__(alt_key)
            except KeyError:
                # If still not found, raise KeyError
                raise KeyError(key)
            
            
            
fd = FlexibleDict()

fd['a'] = 100
print(fd['a'])     # 100 (normal dict behavior)

fd[5] = 500
print(fd[5])       # 500 (normal dict behavior)

fd[1] = 100
print(fd['1'])     # 100 (int → str access)

fd['1'] = 100
print(fd[1])       # 100 (str → int access)

````

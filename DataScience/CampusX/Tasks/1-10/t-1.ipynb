### Q2:- Write a program that will convert celsius value to fahrenheit.

````python
# Convert Celsius to Fahrenheit

# Method 1: Basic formula method
c = int(input('Enter Celsius temperature: '))                     # user se Celsius temperature input liya
print("Fahrenheit is:", ((9 / 5) * c) + 32)                        # formula lagaya: F = (9/5)*C + 32


# Method 2: Direct method using function
def celsius_to_fahrenheit(celsius):                               # function define kiya conversion ke liye
    return (celsius * 9 / 5) + 32                                  # formula lagaya
print("Using function:", celsius_to_fahrenheit(c))                # function ko call kiya


# Method 3: Using list comprehension (L.C.)
temps = [c]                                                       # ek list banayi input temperature ke saath
fahrenheit_list = [((9 / 5) * temp) + 32 for temp in temps]       # list comprehension se Fahrenheit convert kiya
print("Using list comprehension:", fahrenheit_list[0])            # result print kiya


# Method 4: Using lambda function
convert = lambda cel: (9 / 5) * cel + 32                           # lambda function banaya conversion ke liye
print("Using lambda:", convert(c))                                # lambda call karke result print kiya
````

### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python
# Swap two numbers without using third variable (Method 1: Arithmetic Method)
a = int(input('Enter a number (a): '))                             # pehla number input
b = int(input('Enter a number (b): '))                             # doosra number input

a = a * b                                                          # a ab dono ka product hai
b = a // b                                                         # b ab original a ban gaya
a = a // b                                                         # a ab original b ban gaya

print("After swap:")
print("a =", a)                                                    # swapped value of a
print("b =", b)                                                    # swapped value of b



# Method 2: Direct method using Python’s tuple unpacking
a, b = b, a                                                        # Pythonic way to swap
print("Direct method:")
print("a =", a)
print("b =", b)



# Method 3: Using List Comprehension (L.C.)
a, b = [b if i == 0 else a for i in range(2)]                      # i=0 pe b, i=1 pe a assign kar diya (swap)
print("Using list comprehension:")
print("a =", a)
print("b =", b)



# Method 4: Using Lambda function
swap = lambda x, y: (y, x)                                         # lambda se tuple swap return karega
a, b = swap(a, b)                                                  # lambda ko call karke result assign
print("Using lambda:")
print("a =", a)
print("b =", b)
````

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Squared Distance Between Two Points (x1, y1) & (x2, y2)

# Method 1: Basic Formula Method
a = int(input('Enter x1: '))                                       # pehla point ka x1
b = int(input('Enter y1: '))                                       # pehla point ka y1
c = int(input('Enter x2: '))                                       # doosra point ka x2
d = int(input('Enter y2: '))                                       # doosra point ka y2

dis = (c - a)**2 + (d - b)**2                                      # squared distance ka formula
print("Squared distance (Basic):", dis)                            # output print



# Method 2: Direct method using function
def squared_distance(x1, y1, x2, y2):                              # function define
    return (x2 - x1)**2 + (y2 - y1)**2                             # squared distance return
print("Squared distance (Function):", squared_distance(a, b, c, d))



# Method 3: Using List Comprehension (L.C.)
coords = [a, b, c, d]                                              # 4 coordinates list mein store kiye
dist_lc = sum([(coords[i+2] - coords[i])**2 for i in [0, 1]])      # (x2 - x1)^2 + (y2 - y1)^2 via L.C.
print("Squared distance (List Comprehension):", dist_lc)



# Method 4: Using Lambda Function
squared_dist_lambda = lambda x1, y1, x2, y2: (x2 - x1)**2 + (y2 - y1)**2  # lambda function
print("Squared distance (Lambda):", squared_dist_lambda(a, b, c, d))
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Simple Interest Calculation (All Methods Combined)

# Method 1: Basic Formula Method
p = int(input('Enter principal amount (P): '))                    # Principal amount
r = int(input('Enter rate of interest (R): '))                    # Interest rate
t = int(input('Enter time in years (T): '))                       # Time duration

si = (p * r * t) / 100                                            # SI = (P × R × T) / 100
print("Simple Interest (Basic):", si)



# Method 2: Direct Method using Function
def simple_interest(prin, rate, time):                            # Function to calculate SI
    return (prin * rate * time) / 100
print("Simple Interest (Function):", simple_interest(p, r, t))



# Method 3: Using List Comprehension (L.C.)
inputs = [p, r, t]                                                # All inputs in a list
si_lc = [ (inputs[0] * inputs[1] * inputs[2]) / 100 ][0]          # Single element list se SI nikala
print("Simple Interest (List Comprehension):", si_lc)



# Method 4: Using Lambda Function
si_lambda = lambda p, r, t: (p * r * t) / 100                     # Lambda to calculate SI
print("Simple Interest (Lambda):", si_lambda(p, r, t))
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Q6: Count Dogs and Chickens given total heads and legs (All Methods Combined)

# ===========================
# Method 1: Basic Formula Method
# ===========================
h = int(input('Enter number of heads: '))                         # total heads input (dogs + chickens)
l = int(input('Enter number of legs: '))                          # total legs input (dogs*4 + chickens*2)

# Formula derive kiya from maths equations:
#   dogs + chickens = heads
#   4*dogs + 2*chickens = legs
# Solve karne ke baad: chickens = (4h - l)/2
chickens = (4 * h - l) // 2                                       # integer division se exact count
dogs = h - chickens                                               # dogs = total heads - chickens
print("Basic Method:")
print("Dogs:", dogs)
print("Chickens:", chickens)


# ===========================
# Method 2: Direct Method using Function
# ===========================
def count_animals(heads, legs):                                   # function banaya to reuse logic
    chickens = (4 * heads - legs) // 2                           # chickens count formula
    dogs = heads - chickens                                      # dogs = heads - chickens
    return dogs, chickens

d1, c1 = count_animals(h, l)                                     # function call with h and l
print("\nFunction Method:")
print("Dogs:", d1)
print("Chickens:", c1)


# ===========================
# Method 3: Using List Comprehension
# ===========================
vals = [h, l]                                                     # heads & legs ko list mein rakha
res = [vals[0] - ((4 * vals[0] - vals[1]) // 2),                  # dogs count nikal kar list mein
       (4 * vals[0] - vals[1]) // 2]                              # chickens count nikal kar list mein
print("\nList Comprehension Method:")
print("Dogs:", res[0])                                           # res[0] = dogs
print("Chickens:", res[1])                                       # res[1] = chickens


# ===========================
# Method 4: Using Lambda Function
# ===========================
count_lambda = lambda heads, legs: (heads - ((4 * heads - legs) // 2),   # tuple return: (dogs, chickens)
                                    (4 * heads - legs) // 2)

d2, c2 = count_lambda(h, l)                                      # lambda call with h and l
print("\nLambda Method:")
print("Dogs:", d2)
print("Chickens:", c2)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Q7: Sum of squares of first n natural numbers (All Methods Combined)

# Method 1: Basic Formula Method
n = int(input('Enter a number: '))                                # user se n input liya
sum_squares = (n * (n + 1) * (2 * n + 1)) // 6                     # formula: n(n+1)(2n+1)/6
print("Sum of squares (Basic):", sum_squares)



# Method 2: Using Function (Direct Method)
def sum_of_squares(n):
    return (n * (n + 1) * (2 * n + 1)) // 6

print("Sum of squares (Function):", sum_of_squares(n))



# Method 3: Using List Comprehension
squares = [i**2 for i in range(1, n + 1)]                         # list of squares from 1 to n
sum_lc = sum(squares)                                            # total sum of squares
print("Sum of squares (List Comprehension):", sum_lc)



# Method 4: Using Lambda Function
sum_lambda = lambda x: (x * (x + 1) * (2 * x + 1)) // 6           # lambda for sum of squares
print("Sum of squares (Lambda):", sum_lambda(n))
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Q9: Add Two Fractions (All Methods Combined)

# Method 1: Basic Formula Method
a = int(input("Enter numerator of first fraction: "))             # numerator of 1st fraction
b = int(input("Enter denominator of first fraction: "))           # denominator of 1st fraction
c = int(input("Enter numerator of second fraction: "))            # numerator of 2nd fraction
d = int(input("Enter denominator of second fraction: "))          # denominator of 2nd fraction

numerator = a * d + b * c                                         # cross multiply & add
denominator = b * d                                               # multiply denominators
print("Sum of fractions (Basic):", f"{numerator}/{denominator}")



# Method 2: Using Function
def add_fractions(a, b, c, d):
    num = a * d + b * c
    den = b * d
    return num, den

n1, d1 = add_fractions(a, b, c, d)
print("Sum of fractions (Function):", f"{n1}/{d1}")



# Method 3: Using List Comprehension
vals = [a, b, c, d]
frac_sum = [vals[0]*vals[3] + vals[1]*vals[2], vals[1]*vals[3]]   # [numerator, denominator]
print("Sum of fractions (List Comprehension):", f"{frac_sum[0]}/{frac_sum[1]}")



# Method 4: Using Lambda Function
add_frac = lambda a, b, c, d: (a*d + b*c, b*d)
n2, d2 = add_frac(a, b, c, d)
print("Sum of fractions (Lambda):", f"{n2}/{d2}")
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Q10: Compute (d × l × b) / (π × r² × h) using all methods

# Method 1: Basic Formula Method
d = float(input('Enter value for d: '))                            # input values as float
l = float(input('Enter value for l: '))
b = float(input('Enter value for b: '))
h = float(input('Enter value for h: '))
r = float(input('Enter value for r: '))

result = (d * l * b) / (3.14 * r * r * h)                          # formula calculation
print("Result (Basic):", round(result, 2))                         # rounded to 2 decimals



# Method 2: Direct Method using Function
def compute_formula(d, l, b, r, h):
    return round((d * l * b) / (3.14 * r * r * h), 2)              # same formula inside function

print("Result (Function):", compute_formula(d, l, b, r, h))



# Method 3: Using List Comprehension
vals = [d, l, b, r, h]
result_lc = [ round((vals[0]*vals[1]*vals[2]) / (3.14*vals[3]**2*vals[4]), 2) ][0]
print("Result (List Comprehension):", result_lc)



# Method 4: Using Lambda Function
compute = lambda d, l, b, r, h: round((d * l * b) / (3.14 * r * r * h), 2)
print("Result (Lambda):", compute(d, l, b, r, h))
````

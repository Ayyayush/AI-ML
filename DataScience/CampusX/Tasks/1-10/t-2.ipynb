### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
salary = int(input('Enter the salary in lakhs: '))

# Convert lakhs to actual amount
annual_salary = salary * 100000

# Calculate deductions: HRA(10%), DA(5%), PF(3%)
hra = annual_salary * 0.10
da = annual_salary * 0.05
pf = annual_salary * 0.03

# Calculate salary after basic deductions
salary_after_deductions = annual_salary - hra - da - pf

# Apply tax based on salary slabs
if annual_salary <= 500000:  # Below 5 lakhs
    tax = 0
elif annual_salary <= 1000000:  # 5-10 lakhs
    tax = salary_after_deductions * 0.10
elif annual_salary <= 2000000:  # 10-20 lakhs
    tax = salary_after_deductions * 0.20
else:  # Above 20 lakhs
    tax = salary_after_deductions * 0.30

# Calculate final in-hand salary
final_salary = salary_after_deductions - tax

# Convert to monthly salary
monthly_salary = final_salary / 12

print(f"Annual CTC: ₹{annual_salary:,.2f}")
print(f"HRA Deduction: ₹{hra:,.2f}")
print(f"DA Deduction: ₹{da:,.2f}")
print(f"PF Deduction: ₹{pf:,.2f}")
print(f"Tax Deduction: ₹{tax:,.2f}")
print(f"Monthly In-hand Salary: ₹{monthly_salary:,.2f}")
````

### `Problem 2`: Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

````python
# ✅ METHOD 1: Direct Input Method
print("------ Method 1: Direct Input ------")
a = int(input("Enter first angle: "))                      # Pehla angle input le rahe hain
b = int(input("Enter second angle: "))                     # Dusra angle input
c = int(input("Enter third angle: "))                      # Teesra angle input

# ✅ Condition check: sum == 180 & all angles > 0
if (a + b + c == 180 and a > 0 and b > 0 and c > 0):
    print('Triangle can be formed')                        # Valid triangle
else:
    print('Triangle cannot be formed')                     # Invalid triangle

print("\n------ Method 2: Function Method ------")

# ✅ METHOD 2: Function-Based Approach
def is_valid_triangle(a, b, c):
    # Check kar rahe hain ki sum 180 ho aur koi angle 0 ya negative na ho
    return (a + b + c == 180 and a > 0 and b > 0 and c > 0)

# Input from user
x = int(input("Enter first angle: "))
y = int(input("Enter second angle: "))
z = int(input("Enter third angle: "))

# Function call se check karenge
if is_valid_triangle(x, y, z):
    print("Triangle can be formed")
else:
    print("Triangle cannot be formed")


print("\n------ Method 3: List Method ------")

# ✅ METHOD 3: List-Based Approach
angles = []                                                # Empty list to hold 3 angle inputs

# Loop to take 3 inputs
for i in range(3):
    val = int(input(f"Enter angle {i+1}: "))
    angles.append(val)                                     # Har input list mein add kar rahe hain

# ✅ Check: sum should be 180 and all values > 0
if sum(angles) == 180 and all(angle > 0 for angle in angles):
    print("Triangle can be formed")                        # Agar dono conditions pass hui
else:
    print("Triangle cannot be formed")                     # Agar koi condition fail hui
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
# ✅ 1st Method: Direct input and condition check (Basic Version)

a = int(input("Enter numerator of first fraction: "))             # Jaise 3 in 3/5
b = int(input("Enter denominator of first fraction: "))           # Jaise 5 in 3/5

if a < b:                                                         # Agar numerator chhota hai
    print('Profit')                                               # Toh profit
elif a > b:                                                       # Agar bada hai
    print('Loss')                                                 # Toh loss
else:
    print('Neither loss nor profit')                              # Equal hone par no profit no loss

print("\n----------\n")


# ✅ 2nd Method: Function-based approach

def check_fraction_status(numerator, denominator):
    if numerator < denominator:
        return 'Profit'
    elif numerator > denominator:
        return 'Loss'
    else:
        return 'Neither loss nor profit'

# Calling function with user inputs
num = int(input("Enter numerator again: "))
den = int(input("Enter denominator again: "))
result = check_fraction_status(num, den)
print(result)

print("\n----------\n")


# ✅ 3rd Method: List of fractions, loop through and evaluate

# Format: [(numerator1, denominator1), (numerator2, denominator2), ...]
fractions = [(3, 5), (10, 3), (4, 4), (2, 7)]

# Loop through each pair in list
for i, (num, den) in enumerate(fractions):
    print(f"Checking fraction #{i+1}: {num}/{den}")
    if num < den:
        print("→ Profit")
    elif num > den:
        print("→ Loss")
    else:
        print("→ Neither loss nor profit")
    print("----------------------------")
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Method 1: Using For Loop (Standard Approach)
# --------------------------------------------

num = int(input('Enter number: '))             # User se input lo
prod = 1                                       # Factorial ka initial value 1 hota hai

for i in range(1, num + 1):                    # 1 se leke num tak sabko multiply karo
    prod *= i                                  # prod = prod * i

print("Factorial using loop:", prod)           # Final factorial print karo



# Method 2: Using List and reduce()
# ---------------------------------

from functools import reduce                   # reduce() use karne ke liye import

nums = list(range(1, num + 1))                 # List banao from 1 to num
fact = reduce(lambda x, y: x * y, nums)        # List ke elements ko multiply kar do

print("Factorial using list:", fact)           # Final result print karo
````

### `Problem 7` - Reverse a given integer number.

Example:

`Input:`

```bash
76542
```

`Output:`

```bash
24567
```

````python
# Method 1: Using While Loop
# --------------------------

num = int(input('Enter number: '))             # User se number lo
rev = 0                                        # Reverse number ko store karne ke liye variable

while num > 0:                                 # Jab tak number khatam na ho
    rev = (rev * 10) + (num % 10)              # Last digit ko reverse mein daalo
    num = num // 10                            # Last digit hata do (integer division)

print("Reverse using loop:", rev)              # Reversed number print karo


# Method 2: Using List Method
# ---------------------------

num2 = input('Enter number again: ')           # Dusra input as string lete hain for list processing

digits = list(num2)                            # Har digit ko ek list element bana diya
digits.reverse()                               # List ko reverse kar diya

reversed_num = ''.join(digits)                 # List ko string mein wapas joda

print("Reverse using list:", reversed_num)     # Reversed number print karo
````

### `Problem 8`: Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash
30
```

`Output:`

```bash
276
```

````python
# Write code here
num = int(input("Enter number: "))
i = 1
total = 0

while i <= num:
    if i % 5 == 0:               # Agar number 5 se divisible hai, skip karo
        i += 1
        continue

    total += i                   # Add to total
    if total > 300:             # Agar total 300 se bada ho gaya toh break kar do
        break

    i += 1                      # Next number check karo

print(total)                    # Final sum print karo
````

### `Problem 9`: Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python
# Write code here
sum = 0
count = 0

while True:
    num = int(input("Enter number (0 to stop): "))   # Input lo

    if num == 0:                                     # 0 aaye toh break
        break

    sum += num                                       # Add karo sum mein
    count += 1                                       # Count increase karo

if count > 0:
    print("Sum:", sum)
    print("Average:", sum / count)
else:
    print("No numbers entered.")




# 2nd method 
# Write code here using walrus operator (Python 3.8+)
sum = 0
count = 0

while (num := int(input("Enter number (0 to stop): "))) != 0:
    sum += num                         # Number add karo sum mein
    count += 1                         # Count badhao

if count > 0:
    print("Sum:", sum)
    print("Average:", sum / count)
else:
    print("No numbers entered.")
````

### **`Problem-1:`** Write a Python function that takes a list and returns a new list with unique elements of the first list.

**Exercise 1:**

Input:

```bash
[1,2,3,3,3,3,4,5]
```

Output:

```bash
[1, 2, 3, 4, 5]
```

````python
# Method 1: Using manual loop (preserves order)
def unique1(l1):
    l2 = []
    for n in l1:
        if n not in l2:
            l2.append(n)
    return l2


# Testing Method 1
l1 = [1, 2, 3, 4, 5, 3, 2]
print("Method 1:", unique1(l1))


# Method 2: Using set (order not guaranteed)
def unique2(l1):
    l3 = list(set(l1))
    return l3


# Testing Method 2
print("Method 2:", unique2(l1))


# Method 3: Using dictionary keys (preserves order, fast)
def unique3(l1):
    return list(dict.fromkeys(l1))


# Testing Method 3
print("Method 3:", unique3(l1))


# Method 4: Using list comprehension (requires ref list)
l4 = []
[l4.append(x) for x in l1 if x not in l4]

# Testing Method 4
print("Method 4:", l4)
````

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
# Given string
s = "green-red-yellow-black-white"

# Method 1: Using split + sort + join
parts = s.split('-')       # Split into list of words
parts.sort()               # Sort alphabetically
new_s1 = "-".join(parts)   # Join back using '-'
print("Method 1:", new_s1)


# Method 2: Using sorted()
new_s2 = "-".join(sorted(s.split('-')))
print("Method 2:", new_s2)


# Method 3: Using list comprehension + sorted()
new_s3 = "-".join([word for word in sorted(s.split('-'))])
print("Method 3:", new_s3)
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Given string
s = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'


# Method 1: Manual loop and character comparison
small = 0
large = 0

for c in s:
    if 'a' <= c <= 'z':
        small += 1
    elif 'A' <= c <= 'Z':
        large += 1

print("Small =", small, ", Large =", large)


# Method 2: Using additional Data Structures (list)
lowercase = []
uppercase = []

for c in s:
    if c.islower():
        lowercase.append(c)
    elif c.isupper():
        uppercase.append(c)

print("Small =", len(lowercase), ", Large =", len(uppercase))


# Method 3: Using List Comprehension
count_small = [c for c in s if c.islower()]
count_large = [c for c in s if c.isupper()]

print("Small =", len(count_small), ", Large =", len(count_large))
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here

# method -1 (Loop and divisor sum)
num = int(input("Enter number: "))
sum = 0

for i in range(1, num):
    if num % i == 0:
        sum += i

if sum == num:
    print("Perfect Number")
else:
    print("Not a Perfect Number")


# method -2 (Using function + early return)
def is_perfect(n):
    if n <= 1:
        return False
    s = 1
    for i in range(2, n):
        if n % i == 0:
            s += i
    return s == n

print("Using Method 2:", "Perfect Number" if is_perfect(num) else "Not Perfect")


# method -3 (Optimized divisor sum up to sqrt(n))
def is_perfect_fast(n):
    if n <= 1:
        return False
    s = 1
    i = 2
    while i * i <= n:
        if n % i == 0:
            s += i
            if i != n // i:
                s += n // i
        i += 1
    return s == n

print("Using Method 3:", "Perfect Number" if is_perfect_fast(num) else "Not Perfect")
````

`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
s = "hello how are you i am fine thank you"
words = s.split()  # split string into list of words

# ------------------------
# Method 1: Looping and manual count
freq1 = {}
for word in words:
    if word in freq1:
        freq1[word] += 1
    else:
        freq1[word] = 1

most_freq_word1 = max(freq1, key=freq1.get)
print("Method 1:", most_freq_word1, "->", freq1[most_freq_word1])

# ------------------------
# Method 2: Using dictionary get()
freq2 = {}
for word in words:
    freq2[word] = freq2.get(word, 0) + 1

most_freq_word2 = max(freq2, key=freq2.get)
print("Method 2:", most_freq_word2, "->", freq2[most_freq_word2])

# ------------------------
# Method 3: Using collections.Counter
from collections import Counter
freq3 = Counter(words)
most_freq_word3, count3 = freq3.most_common(1)[0]
print("Method 3:", most_freq_word3, "->", count3)

# ------------------------
# Method 4: Using pandas value_counts
import pandas as pd
freq4 = pd.Series(words).value_counts()
most_freq_word4 = freq4.idxmax()
count4 = freq4.max()
print("Method 4:", most_freq_word4, "->", count4)
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here
# histogram particular range mein frequency count btata hai ek particular defined bin mein 
# pure data ko range mein divide karte h 
# bin size ke baad start karte h aur range ka size bin jitna hota h
import math 

def histogram(L):
    """
    Creates a histogram of frequencies in bins of size 10.
    Returns a dictionary where keys are ranges (as strings) and values are counts.
    """

    # example input list
    # l=[13,42,15,37,22,39,41,50]  # commented out because L is passed as argument

    # find minimum and maximum value of the list
    min_val = min(L)   # minimum value in the data
    max_val = max(L)   # maximum value in the data

    # we should use floor for min and ceil for max to cover full range 
    # floor ensures min value fits in first bin
    # ceil ensures max value fits in last bin
    min_bin = math.floor(min_val / 10) * 10
    max_bin = math.ceil(max_val / 10) * 10

    # dictionary to store frequency counts for each bin
    d = {}

    # iterate over bins from min_bin to max_bin in steps of 10
    for i in range(min_bin, max_bin, 10):
        count = 0
        # iterate through each number in the list
        for j in L:
            # check if element j falls into current bin
            # i < j <= i+10 ensures no double counting at bin edges
            if i < j <= i + 10:
                count += 1
        # store the count in dictionary
        # using str here to make keys readable, e.g., "11-20"
        d[str(i + 1) + '-' + str(i + 10)] = count

    return d

# example list
L = [13, 42, 15, 37, 22, 39, 41, 50]

# call the function and print histogram
hist = histogram(L)
print(hist)
````

````python
# Write code here
# series of 2-D points coordinates
# fir query point milega aur batana h ki iss query point ke sabse paas kaun sa point h

import math

points = [(1, -21), (2, -2), (3, 33), (-4, 4)]
query_point = (0, 0)

# -----------------------------------
# Method -1 : Using list comprehension
# distance = sqrt((x2-x1)^2 + (y2-y1)^2) for each point
distances = [math.sqrt((x - query_point[0])**2 + (y - query_point[1])**2) for x, y in points]

# find index of minimum distance
closest_index = distances.index(min(distances))  # min finds smallest distance, index finds its position
closest_point = points[closest_index]  # use index to get actual point

print("Method 1: Closest Point:", closest_point, "Distance:", distances[closest_index])

# -----------------------------------
# Method -2 : Using loop and enumerate (your original approach)
temp = []  # store distances

for i, point in enumerate(points):
    # calculate squared distance (faster than sqrt for comparison)
    distance = (point[0] - query_point[0])**2 + (point[1] - query_point[1])**2
    temp.append(distance)
    # note: if you use max, order may change; use enumerate to keep track of original index
    # enumerate gives (index, element) for each item in list
    # here i = index of point, point = actual coordinates

# sorted list of (index, distance) tuples
sorted_distances = sorted(list(enumerate(temp)), key=lambda x: x[1])
# the first element is the closest point
closest_index2 = sorted_distances[0][0]  # index of closest point
closest_point2 = points[closest_index2]

print("Method 2: Closest Point:", closest_point2, "Distance squared:", temp[closest_index2])

# -----------------------------------
# Method -3 : Using min with key function
closest_point3 = min(points, key=lambda point: (point[0]-query_point[0])**2 + (point[1]-query_point[1])**2)
closest_distance3 = math.sqrt((closest_point3[0]-query_point[0])**2 + (closest_point3[1]-query_point[1])**2)

print("Method 3: Closest Point:", closest_point3, "Distance:", closest_distance3)

# -----------------------------------
# Notes on enumerate:
# enumerate(iterable) returns a sequence of (index, element) pairs
# Example:
# lst = ['a','b','c']
# for idx, val in enumerate(lst):
#     print(idx, val)
# Output:
# 0 a
# 1 b
# 2 c
# Useful for keeping track of original indices when sorting or looping.
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
# Write code here
# Bag of Words (BoW)
# Bag of words is an algorithm to convert text into numbers
# Here we perform bag-of-words operation on a list of strings

"""
Example:
    'cat mat bat'
    'cat cat cat'
    'bat bat mat'

Steps:
1. Find all unique words across all sentences → this forms the vocabulary
2. For each sentence, count the occurrence of each word from the vocabulary
3. Return a list of lists (or vectors) representing each sentence numerically
"""

def bow(L):
    # Step 1: Build vocabulary (unique words across all sentences)
    vocab = set()
    for sentence in L:
        vocab.update(sentence.split())  # update adds all words from sentence to the set
    
    # Convert set to list so that order is fixed
    vocab = sorted(list(vocab))  # optional, makes output reproducible
    print("Vocabulary:", vocab)

    # Step 2: Create Bag-of-Words representation
    result = []
    
    for sentence in L:
        vector = []  # empty list for this sentence
        words_in_sentence = sentence.split()
        for word in vocab:
            # count how many times this word appears in the sentence
            vector.append(words_in_sentence.count(word))
        result.append(vector)
    
    return result

# Example list of strings
L = [
    'cat mat rat cat',
    'sat bat fat cat rat',
    'pat cat mat rat'
]

# Call Bag of Words function
bow_result = bow(L)
print("Bag of Words Vectors:")
for vec in bow_result:
    print(vec)
````

`Problem-14`: Use reduce to convert a 2D list to 1D

````python
# Write code here
# Sample 2D list
matrix = [[1, 2], [3, 4], [5, 6]]

# -----------------------------------
# Method -1 : Using nested loops
flat_list1 = []
for row in matrix:
    for item in row:
        flat_list1.append(item)
print("Method 1 - Nested loops:", flat_list1)

# -----------------------------------
# Method -2 : Using ready-made function (itertools.chain)
from itertools import chain
flat_list2 = list(chain.from_iterable(matrix))
print("Method 2 - itertools.chain:", flat_list2)

# -----------------------------------
# Method -3 : Using list comprehension
flat_list3 = [item for row in matrix for item in row]
print("Method 3 - List comprehension:", flat_list3)

# -----------------------------------
# Method -4 : Using lambda and map
# flatten each row and then sum using reduce
from functools import reduce
flat_list4 = reduce(lambda x, y: x + y, matrix)
print("Method 4 - Lambda with reduce:", flat_list4)
````

````python
# ======================================
# PRACTICE: *args (Variable Arguments)
# ======================================

# --------------------------------------
# Q1: Print all arguments
# --------------------------------------

def a1(*args):
    for item in args:
        print(item)

a1(1, 2, 3, 4)

# --------------------------------------
# Q2: Sum of all numbers using *args
# --------------------------------------

def a2(*args):
    print(sum(args))

# Alternative (manual):
# def a2(*args):
#     total = 0
#     for num in args:
#         total += num
#     return total

a2(10, 20, 30)

# --------------------------------------
# Q3: Count number of arguments
# --------------------------------------

def a3(*args):
    print(len(args))

# Alternative:
# def a3(*args):
#     count = 0
#     for _ in args:
#         count += 1
#     return count

a3(1, 2, 3, 4, 5)

# --------------------------------------
# Q4: Maximum number using *args
# --------------------------------------

def a4(*args):
    print(max(args))

# Alternative (manual):
# def a4(*args):
#     maximum = args[0]
#     for num in args:
#         if num > maximum:
#             maximum = num
#     return maximum

a4(3, 9, 1, 6)

# --------------------------------------
# Q5: Print only even numbers using *args
# --------------------------------------

def a5(*args):
    even = []
    for num in args:
        if num % 2 == 0:
            even.append(num)
    print(even)

# Alternative (list comprehension):
# def a5(*args):
#     print([num for num in args if num % 2 == 0])

a5(1, 2, 3, 4, 5, 6)

# ======================================
# PRACTICE: **kwargs (Keyword Arguments)
# ======================================

# --------------------------------------
# Q6: Print all keys and values
# --------------------------------------

def a6(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

a6(name="Ayush", age=21, city="Delhi")

# --------------------------------------
# Q7: Check if key "name" exists
# --------------------------------------

def a7(**kwargs):
    if "name" in kwargs:
        print(True)
    else:
        print(False)

a7(age=20, city="Ballia")
a7(name="Ayush", age=21)

# --------------------------------------
# Q8: Keep only numeric values from kwargs
# --------------------------------------

def a8(**kwargs):
    dct = {}
    for key, value in kwargs.items():
        if isinstance(value, int) or isinstance(value, float):
            dct[key] = value
    return dct

print(a8(a=10, b="hello", c=3.5, d=True))

# ======================================
# PRACTICE: *args + **kwargs (MIXED)
# ======================================

# --------------------------------------
# Q9: Accept *args and **kwargs together
# --------------------------------------

def a9(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

a9(10, 20, 30, name="Ayush", age=21)

# --------------------------------------
# Q10: Total marks using *args & info using **kwargs
# --------------------------------------

def a10(*marks, **info):
    total = sum(marks)
    print("Total Marks:", total)
    print("Student Info:")
    for key, value in info.items():
        print(key, ":", value)

a10(80, 90, 70, name="Ayush", roll=101)

# ======================================
# PRACTICE: NESTED FUNCTIONS
# ======================================

# --------------------------------------
# Q11: Simple nested function
# --------------------------------------

def n1():
    def inner():
        print("Inner function called")
    inner()

n1()

# --------------------------------------
# Q12: Nested function accessing outer variable
# --------------------------------------

def n2():
    msg = "Hello from outer function"

    def inner():
        print(msg)                       # Accessing outer variable

    inner()

n2()

# --------------------------------------
# Q13: Inner function returning value
# --------------------------------------

def n3():
    def inner():
        return 10
    return inner()

print(n3())

# --------------------------------------
# Q14: Function returning inner function (closure)
# --------------------------------------

def n4():
    def inner():
        print("Returned inner function")
    return inner

func = n4()
func()

# --------------------------------------
# Q15: Nested function to check even/odd
# --------------------------------------

def n5(num):
    def check():
        if num % 2 == 0:
            return "Even"
        return "Odd"
    return check()

print(n5(7))
print(n5(10))
````

````python
# ======================================
# PRACTICE: VARIABLE SCOPE & LAMBDA
# ======================================

# --------------------------------------
# Q1 & Q2: Local vs Global Scope
# --------------------------------------

a = 5                                  # Global variable

def a1():
    b = 6                              # Local variable
    print(a)                           # Can access global variable

a1()

# ❌ WRONG:
# print(b)                             # NameError (b is local to a1)

# --------------------------------------
# Q3 & Q4: Global variable modification
# --------------------------------------

count = 0                              # Global variable

def a3():
    global count                       # Needed to modify global variable
    count += 10
    print(count)

a3()
print(count)

# --------------------------------------
# Q5: Enclosing vs Local scope
# --------------------------------------

x = 10                                 # Global variable

def a4():
    print(x)                           # Accesses global x

    def inner():
        x = 5                          # Local to inner()
        print(x)

    inner()

a4()

# --------------------------------------
# Q6: Scope Order (LEGB Rule)
# --------------------------------------

y = 100                                # Global

def scope_demo():
    y = 50                             # Enclosing

    def inner():
        y = 10                         # Local
        print(y)

    inner()

scope_demo()

# --------------------------------------
# Q7: Closure (Function accessing outer variable)
# --------------------------------------

def a7():
    z = 5                              # Enclosing variable

    def inner():
        return z                       # Closure

    return inner()

print(a7())

# ======================================
# PRACTICE: LAMBDA FUNCTIONS
# ======================================

# --------------------------------------
# Q8: Lambda to add two numbers
# --------------------------------------

add = lambda a, b: a + b
print(add(3, 4))

# --------------------------------------
# Q9: Lambda to square a number
# --------------------------------------

square = lambda x: x * x
print(square(5))

# --------------------------------------
# Q10: Lambda with map()
# --------------------------------------

numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))
print(squares)

# --------------------------------------
# Q11: Lambda to find maximum of two numbers
# --------------------------------------

# ❌ WRONG:
# lambda a,b: a>b   -> returns True/False

# ✅ CORRECT:
maximum = lambda a, b: a if a > b else b
print(maximum(10, 20))

# --------------------------------------
# Q12: Lambda with filter() (Even numbers)
# --------------------------------------

even = list(filter(lambda x: x % 2 == 0, numbers))
print(even)

# --------------------------------------
# Q13: Lambda with sorted()
# --------------------------------------

students = [("Aman", 21), ("Riya", 19), ("Karan", 23)]

# Sort by age (index 1)
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)

# Sort by name (index 0)
sorted_names = sorted(students, key=lambda x: x[0])
print(sorted_names)

# ======================================
# PRACTICE: SCOPE + LAMBDA (CLOSURES)
# ======================================

# --------------------------------------
# Q14: Function returning lambda (closure)
# --------------------------------------

def a9():
    p = 5
    return lambda x: p * x              # p remembered (closure)

f = a9()
print(f(4))

# --------------------------------------
# Q15: Using nonlocal keyword
# --------------------------------------

def a10():
    p = 5

    def inner():
        nonlocal p                     # Modify enclosing variable
        p += 10
        return p

    return inner()

print(a10())

# --------------------------------------
# Q16: Normal function vs Lambda
# --------------------------------------

def normal_square(x):
    return x * x

lambda_square = lambda x: x * x

print(normal_square(6))
print(lambda_square(6))

# --------------------------------------
# Q17: Why lambda cannot have multiple statements
# --------------------------------------

# ❌ INVALID:
# lambda x:
#     x = x + 1
#     return x

# ✅ Reason:
# Lambda expressions allow ONLY ONE expression
# No assignment, no loops, no multiple statements
````

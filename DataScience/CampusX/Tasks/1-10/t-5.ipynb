###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
# Original list of tuples
# Method -1
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

unique = []                                 # List to store unique first elements

# Step 1: Extract all first elements from tuples
for i in test_list:
    unique.append(i[0])                     # First element i[0] ko unique list mein daalo

unique = set(unique)                        # Duplicates hatane ke liye set bana diya

result = []                                 # Final result list

# Step 2: Har unique element ke liye usse related values group karo
for i in unique:
    result.append([i])                      # Har unique first element se ek nayi sublist start karo
    for j in test_list:
        if j[0] == i:                       # Agar tuple ka first element current unique value ke barabar hai
             result[-1].append(j[1])         # Toh uska second element result ke last sublist mein daalo


#result = [[5, 6, 7, 8], [6, 10], [7, 13]]
#tuple([5, 6, 7, 8]) → (5, 6, 7, 8)
#tuple([6, 10])      → (6, 10)
#tuple([7, 13])      → (7, 13)
#map object → [(5, 6, 7, 8), (6, 10), (7, 13)]



# Step 3: List of lists ko list of tuples mein convert karo
print(list(map(tuple, result)))             # Final result ko print karo as list of tuples
````

###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
def multiply_adjacent(t):
    n = len(t)
    res = []
    for i in range(n):
        if i == 0:
            res.append(t[i] * t[i+1])
        elif i == n - 1:
            res.append(t[i] * t[i-1])
        else:
            res.append(t[i]*t[i-1] + t[i]*t[i+1])
    return tuple(res)

# Example
t = (1, 5, 7, 8, 10)
print(multiply_adjacent(t))   # Output: (5, 40, 91, 136, 80)
````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
# Comparing two tuples using different methods

t1 = (1, 2, 3, 0)
t2 = (0, 1, 2, 3)

# Method 1: Direct comparison
print("Method 1:", t1 == t2)         # False, because order matters in tuple

# Method 2: Using zip() and all() for element-wise comparison
print("Method 2:", all(x == y for x, y in zip(t1, t2)))     # False, checks each element one by one


# Method 3: List comprehension + all()
print("Method 3:", all([t1[i] == t2[i] for i in range(len(t1))]))   # False, same as above but using indexing

# Method 4: Lambda function
compare = lambda a, b: a == b
print("Method 4:", compare(t1, t2))     # False
````

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
# 🧠 Problem: Count number of Lists, Sets, and Tuples inside a mixed list
# -----------------------------------------------------------
# Example:
# Input  → [{'hi', 'bye'}, {'Geeks', 'forGeeks'}, ('a', 'b'), ['hi', 'bye'], ['a', 'b']]
# Output → List = 2, Set = 2, Tuple = 1
# -----------------------------------------------------------

list1 = [{'hi', 'bye'}, {'Geeks', 'forGeeks'}, ('a', 'b'), ['hi', 'bye'], ['a', 'b']]

# -----------------------------------------------------------
# Method - 1 : Using type() and simple loop
# T.C : O(n)
# S.C : O(1)
# -----------------------------------------------------------
list_count = 0
set_count = 0
tuple_count = 0

for item in list1:
    if type(item) == list:                   # check if item is of type list
        list_count += 1
    elif type(item) == set:                  # check if item is of type set
        set_count += 1
    elif type(item) == tuple:                # check if item is of type tuple
        tuple_count += 1

print("Method 1:")
print("List-", list_count)
print("Set-", set_count)
print("Tuples-", tuple_count)


# -----------------------------------------------------------
# Method - 2 : Using isinstance() (recommended in interviews)
# T.C : O(n)
# S.C : O(1)
# -----------------------------------------------------------
# isinstance() is safer and works with inheritance
list_count = sum(isinstance(i, list) for i in list1)
set_count = sum(isinstance(i, set) for i in list1)
tuple_count = sum(isinstance(i, tuple) for i in list1)

print("\nMethod 2 (Using isinstance):")
print("List-", list_count)
print("Set-", set_count)
print("Tuples-", tuple_count)


# -----------------------------------------------------------
# Method - 3 : Using dictionary + type() count mapping
# T.C : O(n)
# S.C : O(1)
# -----------------------------------------------------------
# More compact, pythonic approach using dictionary
count_dict = {'list': 0, 'set': 0, 'tuple': 0}

for item in list1:
    t = type(item)
    if t == list:
        count_dict['list'] += 1
    elif t == set:
        count_dict['set'] += 1
    elif t == tuple:
        count_dict['tuple'] += 1

print("\nMethod 3 (Using dict mapping):")
print(count_dict)
````

###`Q1:` Write a program to find set of common elements in three lists using sets.
```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
# 🧠 Problem: Find common elements in three sorted arrays
# -----------------------------------------------------------
# Example:
# ar1 = [1, 5, 10, 20, 40, 80]
# ar2 = [6, 7, 20, 80, 100]
# ar3 = [3, 4, 15, 20, 30, 70, 80, 120]
# Output → [20, 80]
# -----------------------------------------------------------


ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

# -----------------------------------------------------------
# Method - 1 : Brute Force (Triple Membership Check)
# T.C : O(n1 * (n2 + n3))     because we check membership multiple times
# S.C : O(n)                  to store result
# -----------------------------------------------------------

res = []

for n in ar1:
    if n in ar2 and n in ar3:               # agar element teeno arrays mein hai
        res.append(n)

# ✅ Optional: avoid duplicates (though sorted input generally won’t have)
res = list(set(res))
print("Method 1:", res)                     # Output → [80, 20] or [20, 80]



# -----------------------------------------------------------
# Method - 2 : Using List Comprehension + Set
# T.C : O(n1 + n2 + n3)
# S.C : O(n)
# -----------------------------------------------------------
# Compact pythonic approach with set comprehension
res = list({x for x in ar1 if x in ar2 and x in ar3})
print("Method 2:", res)                     # Output → [20, 80]



# -----------------------------------------------------------
# Method - 3 : Using Set Intersection (Most Efficient & Clean)
# T.C : O(n1 + n2 + n3)
# S.C : O(n)
# -----------------------------------------------------------
# Python’s set intersection gives common elements directly
res = list(set(ar1) & set(ar2) & set(ar3))
print("Method 3 (Set Intersection):", res)  # Output → [80, 20]


````

###`Q2:` Write a program to count unique number of vowels using sets in a given string. Lowercase and upercase vowels will be taken as different.

`Input:`
```
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
```
`Output:`
```
No of unique vowels-6
```

````python
# 🔤 Problem: Count number of unique vowels in a string

# --------------------------------------------------------------
# Method 1: Using simple loop + set (basic & clear way)
# --------------------------------------------------------------
def method1(s):
    vowels = {'a', 'e', 'i', 'o', 'u'}                       # Lowercase vowels ka set
    found = set()                                            # Unique vowels store karne ke liye
    for char in s.lower():                                   # Case-insensitive bana diya
        if char in vowels:                                   # Agar character vowel hai
            found.add(char)                                  # Set me add ho jayega (duplicates ignore)
    print("Method 1 → Unique vowel count:", len(found))       # Output count


# --------------------------------------------------------------
# Method 2: Using list comprehension + set (short & pythonic)
# --------------------------------------------------------------
def method2(s):
    vowels = {'a', 'e', 'i', 'o', 'u'}
    found = {ch for ch in s.lower() if ch in vowels}          # Direct set comprehension
    print("Method 2 → Unique vowel count:", len(found))


# --------------------------------------------------------------
# Method 3: Using filter() + lambda (functional style)
# --------------------------------------------------------------
def method3(s):
    vowels = {'a', 'e', 'i', 'o', 'u'}
    found = set(filter(lambda ch: ch.lower() in vowels, s))   # Filter vowels and convert to set
    print("Method 3 → Unique vowel count:", len(found))


# --------------------------------------------------------------
# Method 4: Using set intersection (very neat one-liner)
# --------------------------------------------------------------
def method4(s):
    vowels = set("aeiou")
    found = set(s.lower()) & vowels                           # Intersection of string chars and vowels
    print("Method 4 → Unique vowel count:", len(found))


# --------------------------------------------------------------
# Input String
# --------------------------------------------------------------
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"

# --------------------------------------------------------------
# Calling all methods
# --------------------------------------------------------------
method1(Str1)
method2(Str1)
method3(Str1)
method4(Str1)
````

### `Q4`: find union of n arrays.

**Example 1:**

Input:
```bash
[[1, 2, 2, 4, 3, 6],
 [5, 1, 3, 4],
 [9, 5, 7, 1],
 [2, 4, 1, 3]]
```

Output:

```bash
[1, 2, 3, 4, 5, 6, 7, 9]
```

````python
# 🔤 Problem: Find Union of N Arrays (Unique Elements from All Lists)


# --------------------------------------------------------------
# ✅ Method 1: Using Nested Loops + Set (Basic & Clear)
# --------------------------------------------------------------
def method1(arrays):
    result_set = set()                                       # Empty set for unique values
    for array in arrays:
        for num in array:
            result_set.add(num)                              # Add each element
    return sorted(result_set)                                # Sorted list as output


# --------------------------------------------------------------
# ✅ Method 2: Using Set Comprehension (Pythonic Way)
# --------------------------------------------------------------
def method2(arrays):
    result_set = {num for arr in arrays for num in arr}      # Flatten and make unique
    return sorted(result_set)


# --------------------------------------------------------------
# ✅ Method 3: Using set.union() (Clean & Readable)
# --------------------------------------------------------------
def method3(arrays):
    result_set = set()
    for arr in arrays:
        result_set = result_set.union(set(arr))              # Combine sets one by one
    return sorted(result_set)


# --------------------------------------------------------------
# Example Input
# --------------------------------------------------------------
input_arrays = [
    [1, 2, 2, 4, 3, 6],
    [5, 1, 3, 4],
    [9, 5, 7, 1],
    [2, 4, 1, 3]
]


# --------------------------------------------------------------
# Output for All Methods
# --------------------------------------------------------------
print("Method 1 →", method1(input_arrays))
print("Method 2 →", method2(input_arrays))
print("Method 3 →", method3(input_arrays))
````

### `Q1`: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

Input:

```bash
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
CampusX
```

**Example 2:**

Input:
```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
Best
```

````python
# 🔤 Q1: Find the key with maximum number of unique values in a dictionary


# --------------------------------------------------------------
# ✅ Method 1: Basic Loop + set()  (Most beginner-friendly)
# --------------------------------------------------------------
def method1(test_dict):
    max_count = 0
    res_key = None

    for key, val_list in test_dict.items():
        unique_count = len(set(val_list))               # Count unique values
        if unique_count > max_count:                    # Track maximum
            max_count = unique_count
            res_key = key
    return res_key


# --------------------------------------------------------------
# ✅ Method 2: Dictionary Comprehension + max() (Pythonic)
# --------------------------------------------------------------
def method2(test_dict):
    # Create dict: key -> number of unique values
    unique_counts = {key: len(set(val)) for key, val in test_dict.items()}
    return max(unique_counts, key=unique_counts.get)     # Get key with max value


# --------------------------------------------------------------
# ✅ Method 3: Using lambda inside max() (Compact + Interview favorite)
# --------------------------------------------------------------
def method3(test_dict):
    return max(test_dict, key=lambda k: len(set(test_dict[k])))   # Direct one-liner


# --------------------------------------------------------------
# Example Inputs
# --------------------------------------------------------------
test_dict1 = {"CampusX": [5, 7, 9, 4, 0], "is": [6, 7, 4, 3, 3], "Best": [9, 9, 6, 5, 5]}
test_dict2 = {"CampusX": [5, 7, 7, 7, 7], "is": [6, 7, 7, 7], "Best": [9, 9, 6, 5, 5]}


# --------------------------------------------------------------
# Outputs
# --------------------------------------------------------------
print("Example 1:")
print("Method 1 →", method1(test_dict1))
print("Method 2 →", method2(test_dict1))
print("Method 3 →", method3(test_dict1))

print("\nExample 2:")
print("Method 1 →", method1(test_dict2))
print("Method 2 →", method2(test_dict2))
print("Method 3 →", method3(test_dict2))
````

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:

```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.
```

````python
# 🔤 Q1: Replace words in a string using a replacement dictionary

# --------------------------------------------------------------
# ✅ Method 1: Basic Loop + split() (Beginner-friendly)
# --------------------------------------------------------------

def method1(test_str, repl_dict):
    words = test_str.split()                             # Split string into words
    for i in range(len(words)):
        word_clean = words[i].rstrip('.,!?')            # Remove punctuation for matching
        if word_clean in repl_dict:                     # Check if word is in replacement dict
            words[i] = repl_dict[word_clean] + words[i][len(word_clean):]  # Replace and keep punctuation
    return ' '.join(words)                              # Join back to string




# --------------------------------------------------------------
# ✅ Method 2: Using List Comprehension (Pythonic)
# --------------------------------------------------------------
def method2(test_str, repl_dict):
    words = test_str.split()
    # Replace words in a single line
    words = [repl_dict[word.rstrip('.,!?')] + word[len(word.rstrip('.,!?')):] 
             if word.rstrip('.,!?') in repl_dict else word 
             for word in words]
    return ' '.join(words)

# --------------------------------------------------------------
# ✅ Method 3: Using regex + re.sub (Compact + Interview favorite)
# --------------------------------------------------------------
import re
def method3(test_str, repl_dict):
    # Create regex pattern for all keys
    pattern = re.compile(r'\b(' + '|'.join(map(re.escape, repl_dict.keys())) + r')\b')
    # Replace using dictionary
    return pattern.sub(lambda x: repl_dict[x.group()], test_str)

# --------------------------------------------------------------
# Example Input
# --------------------------------------------------------------
test_str = 'CampusX best for DS students.'
repl_dict = {"best": "is the best channel", "DS": "Data-Science"}

# --------------------------------------------------------------
# Outputs
# --------------------------------------------------------------
print("Method 1 →", method1(test_str, repl_dict))
print("Method 2 →", method2(test_str, repl_dict))
print("Method 3 →", method3(test_str, repl_dict))
````

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
# 🔤 Q3: Convert List to List of dictionaries
# Given list of values and keys list, convert values to key-value pairs in the form of list of dictionaries

# --------------------------------------------------------------
# ✅ Method 1: Basic Loop + zip() (Beginner-friendly)
# --------------------------------------------------------------
def method1(test_list, key_list):
    res = []
    size = len(key_list)
    # iterate over test_list in chunks of size equal to key_list
    for i in range(0, len(test_list), size):
        chunk = test_list[i:i+size]               # get consecutive values
        res.append(dict(zip(key_list, chunk)))    # create dict and append
    return res

# --------------------------------------------------------------
# ✅ Method 2: List Comprehension + zip() (Pythonic)
# --------------------------------------------------------------
def method2(test_list, key_list):
    size = len(key_list)
    # list comprehension over chunks
    return [dict(zip(key_list, test_list[i:i+size])) for i in range(0, len(test_list), size)]

# --------------------------------------------------------------
# ✅ Method 3: Using iterator + zip (Compact + Efficient)
# --------------------------------------------------------------
def method3(test_list, key_list):
    it = iter(test_list)                          # create iterator
    return [dict(zip(key_list, [next(it) for _ in key_list])) for _ in range(len(test_list)//len(key_list))]

# --------------------------------------------------------------
# Example Inputs
# --------------------------------------------------------------
test_list1 = ["DataScience", 3, "is", 8]
test_list2 = ["CampusX", 10]
key_list = ["name", "id"]

# --------------------------------------------------------------
# Outputs
# --------------------------------------------------------------
print("Example 1:")
print("Method 1 →", method1(test_list1, key_list))
print("Method 2 →", method2(test_list1, key_list))
print("Method 3 →", method3(test_list1, key_list))

print("\nExample 2:")
print("Method 1 →", method1(test_list2, key_list))
print("Method 2 →", method2(test_list2, key_list))
print("Method 3 →", method3(test_list2, key_list))
````

### `Q4`: Convert a list of Tuples into Dictionary.

**Example 1:**

[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]

**Example 2:**

Input:
```bash
[('A', 1), ('B', 2), ('C', 3)]

{'A': [1], 'B': [2], 'C': [3]}
```

````python
# 🔤 Q4: Convert List of Tuples to Dictionary

# --------------------------------------------------------------
# ✅ Method 1: Using dict() constructor (Beginner-friendly)
# --------------------------------------------------------------
def method1(test_tup):
    return dict(test_tup)

# --------------------------------------------------------------
# ✅ Method 2: Using dictionary comprehension (Pythonic)
# --------------------------------------------------------------
def method2(test_tup):
    return {key: value for key, value in test_tup}

# --------------------------------------------------------------
# ✅ Method 3: Using loop + assignment (Manual / Beginner-friendly)
# --------------------------------------------------------------
def method3(test_tup):
    res = {}
    for t in test_tup:
        res[t[0]] = t[1]      # key = first element, value = second element
    return res

# --------------------------------------------------------------
# Example Input
# --------------------------------------------------------------
test_tup = [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]

# --------------------------------------------------------------
# Outputs
# --------------------------------------------------------------
print("Method 1 →", method1(test_tup))
print("Method 2 →", method2(test_tup))
print("Method 3 →", method3(test_tup))
````

### `Q5`: Sort Dictionary key and values List.

**Example 1:**

Input:

```bash
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

Output:

```bash
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

**Example 2:**

Input:

```bash
{'c': [10, 34, 3]}
```

Output:

```bash
{'c': [3, 10, 34]}
```

````python
# 🔤 Q5: Sort dictionary keys and sort their value lists

# --------------------------------------------------------------
# ✅ Method 1: Basic loop + sort() (Beginner-friendly)
# --------------------------------------------------------------
def method1(test_dict):
    res = {}
    for key in sorted(test_dict.keys()):             # sort keys
        res[key] = sorted(test_dict[key])           # sort each value list
    return res

# --------------------------------------------------------------
# ✅ Method 2: Dictionary comprehension + sorted() (Pythonic)
# --------------------------------------------------------------
def method2(test_dict):
    return {key: sorted(test_dict[key]) for key in sorted(test_dict)}

# --------------------------------------------------------------
# ✅ Method 3: Using collections.OrderedDict + sorted() (Interview-friendly)
# --------------------------------------------------------------
from collections import OrderedDict
def method3(test_dict):
    return OrderedDict((key, sorted(test_dict[key])) for key in sorted(test_dict))

# --------------------------------------------------------------
# Example Inputs
# --------------------------------------------------------------
test_dict1 = {'c': [3], 'b': [12, 10], 'a': [19, 4]}
test_dict2 = {'c': [10, 34, 3]}

# --------------------------------------------------------------
# Outputs
# --------------------------------------------------------------
print("Example 1:")
print("Method 1 →", method1(test_dict1))
print("Method 2 →", method2(test_dict1))
print("Method 3 →", dict(method3(test_dict1)))   # convert OrderedDict to dict for display

print("\nExample 2:")
print("Method 1 →", method1(test_dict2))
print("Method 2 →", method2(test_dict2))
print("Method 3 →", dict(method3(test_dict2)))
````

```python
# 🔰 What is Aliasing in Python?

"""
Aliasing ek technique hai jiske through hum **memory optimize** karte hain.

🔸 Jab Python program run hota hai, to saara data RAM me store hota hai.
🔸 RAM ke andar Registers hote hain jisme data **binary format** me store hota hai.
🔸 Variables us memory address ko point karte hain jahan actual data store hai.

🔹 Example:
    a = 4          # 'a' ek memory address ko point karega jahan 4 store hai
    b = a          # 'b' bhi wahi address point karega
    c = b          # 'c' bhi wahi address point karega

    👉 Is situation me a, b, c teeno **same memory location** ko refer kar rahe hote hain.

🔹 Isko hi kehte hain **Aliasing** — multiple names (variables) for the same memory location.
"""

# ✅ Benefits of Aliasing:
"""
🔸 Memory efficient hota hai — multiple variables ek hi jagah ko point karte hain
🔸 Large data structures ke saath kaam karte waqt memory usage kam hoti hai
"""

# 🔍 Deleting a variable
"""
Agar hum del a karte hain, to kya b aur c delete ho jayenge?
❌ Nahi!

Python me jab hum koi variable delete karte hain, to:
    - Us variable ka **reference hata diya jaata hai**
    - Lekin memory location tab tak delete nahi hoti jab tak **sabhi references** hat na jaayein
"""

# 🧠 Reference Count in Python
"""
Python me har object ke sath ek **reference count** maintain hota hai.
Hum `sys.getrefcount()` se yeh jaan sakte hain ki koi value ko kitne variables refer kar rahe hain.

⚠️ Note: `getrefcount()` me khud ka reference bhi include hota hai, so result me +1 aata hai.
"""

# 🔸 Demo Code for Aliasing:
import sys

a = 4
b = a
c = b

print("ID of a:", id(a))
print("ID of b:", id(b))
print("ID of c:", id(c))        # ✅ Teeno ka ID same hoga

del a                           # 'a' ka reference hat gaya
print("Value of b:", b)         # 'b' abhi bhi valid hai
print("Value of c:", c)         # 'c' bhi valid hai

# 🔍 Check reference count of 4 (will include b, c, and getrefcount’s internal call)
print("Reference count of 4:", sys.getrefcount(4))
```

```python
# 🔰 Garbage Collection in Python

"""
🔸 Jab hum kisi variable ko `del` karte hain, to uska **reference** delete hota hai,
    lekin actual object memory se turant delete nahi hota.

🔸 Ab agar us object ko koi aur variable bhi refer nahi kar raha,
    to wo object RAM me **useless** (garbage) ban jaata hai.

🗑️ Aise objects ko remove karne ke liye Python me hota hai **Garbage Collection**.
"""

# 🔎 Why Garbage Collection is needed?
"""
Socho agar Python unused (garbage) objects ko memory me rakhta rahe,
to memory slowly bhar jaayegi — jisko hum kehte hain **memory leak**.

Isliye Python me ek **automatic garbage collector** hota hai jo:
    ✔️ Detect karta hai unused objects
    ✔️ Unko memory se free karta hai
"""

# ⚙️ How it works in Python:
"""
Python ka internal module `gc` (garbage collector) is kam ko handle karta hai.
Isme ek mechanism hota hai called **reference counting**.

✅ Jab kisi object ka reference count 0 ho jaata hai:
    -> Python samajh jaata hai ki koi bhi us object ko use nahi kar raha
    -> Aur use memory se hata diya jaata hai
"""

# 🔹 Important Note:
"""
User ko manually garbage collector chalane ki zarurat nahi padti.
Lekin agar chahe to `gc` module ka use karke:
    - Garbage collector ko control kar sakte hain
    - Manually collect kar sakte hain
"""

# 🔸 Example Code (Using gc module):
import gc

a = [1, 2, 3]
b = a                     # b and a dono same list ko refer kar rahe
del a                    # a ka reference hat gaya
del b                    # b ka bhi reference hat gaya

# Ab list ka reference count 0 ho gaya — eligible for garbage collection

print("Collecting garbage...")
collected = gc.collect()   # Manually trigger garbage collector
print("Garbage Collected:", collected)
```

```python
# 🔰 Mutability in Python – All-in-One Notes

"""
📌 Definition:
Mutability ek property hai jisme ek object **bina apna memory address badle** modify ho sakta hai.

🔹 Mutable object: 
    ➤ Memory address same rehta hai 
    ➤ Content change ho jaata hai

🔹 Immutable object:
    ➤ Har update par **naya memory address** milta hai
    ➤ Old value untouched rehti hai
"""

# -----------------------------------------------
# 🔹 Mutable vs Immutable Examples
# -----------------------------------------------

# ✅ Mutable Data Types:
"""
- list
- set
- dict
"""

# ✅ Immutable Data Types:
"""
- int
- float
- bool
- str
- tuple
- complex
"""

# -----------------------------------------------
# 🔍 Address Check Example (Using id())
# -----------------------------------------------

L = [1, 2, 3]
print("🔸 ID of L before mutation:", id(L))

L.append(4)                    # Mutating the list
print("🔸 List after append:", L)
print("🔸 ID of L after mutation:", id(L))    # ✅ Same ID → proves list is mutable

# -----------------------------------------------
# 🔁 Aliasing: Why is it Dangerous?
# -----------------------------------------------

"""
Agar ek hi object ko multiple variables refer kar rahe hain (Aliasing),
aur wo object mutable hai, toh ek variable se change hone par **sab jagah impact padta hai**.
"""

M = L
M.append(5)
print("🔸 M mutated. Now L is:", L)     # ✅ L bhi change ho gaya

# -----------------------------------------------
# ⚠️ Function Mutating Argument Example
# -----------------------------------------------

def func(data):
    data.append(100)           # Direct mutation

a = [10, 20, 30]
func(a)
print("🔸 List after function call:", a)     # ✅ [10, 20, 30, 100]

# -----------------------------------------------
# ✅ Immutable Types are Safe
# -----------------------------------------------

tup = (1, 2, 3)
print("🔹 Tuple ID before:", id(tup))

tup = tup + (4,)                    # Creates a new tuple (no mutation)
print("🔹 Tuple after addition:", tup)
print("🔹 Tuple ID after:", id(tup))   # ✅ Different ID → proves tuple is immutable

# -----------------------------------------------
# 🛡️ Real-Life Danger of Mutability
# -----------------------------------------------

"""
Socho ek global list ko kisi function ne change kar diya:
    - Toh baaki sab jagah wo changed data hi propagate ho jaayega

❗ Isse bugs aate hain, especially in shared config or sensitive data.

✅ Isiliye, jab kabhi original data ko protect karna ho:
    - Use immutable types (like tuple or frozenset)
"""

# -----------------------------------------------
# 🧠 Summary Table
# -----------------------------------------------

"""
| Feature             | Mutable                          | Immutable                     |
|---------------------|----------------------------------|-------------------------------|
| Examples            | list, set, dict                  | int, float, str, tuple, bool  |
| Address (ID)        | Same after mutation              | New after update              |
| Aliasing impact     | Risky (shared mutation)          | Safe                          |
| Function behavior   | Can modify passed object         | Cannot change original        |
| Real use            | When in-place updates needed     | When data integrity matters   |
"""

# -----------------------------------------------
# ✅ Final Tip:
"""
💡 Jab data ko accidentally modify hone se bachana ho,
   ya functional-style programming follow karni ho → use **immutable types**.
"""
```

```python
# 🔁 CLONING IN PYTHON – All-in-One Notes

"""
📌 Cloning ka matlab hota hai: ek **naya object banana** jisme original object ka **same content** ho,
lekin memory address **alag ho**.

⚠️ Ye mutable types ke liye important hota hai — kyunki agar direct assign kar diya (aliasing), 
toh dono variable same memory ko point karenge → changes dono jagah reflect honge.
"""

# -------------------------------------------------
# ❌ Example of ALIASING (Not cloning)
# -------------------------------------------------

a = [1, 2, 3]
b = a                 # Yeh sirf alias hai, cloning nahi

b.append(4)           # b ko change kiya, par a bhi change ho gaya
print("🔸 a:", a)      # Output: [1, 2, 3, 4]
print("🔸 b:", b)

print("🆔 ID of a:", id(a))
print("🆔 ID of b:", id(b))    # ✅ Same ID → aliasing confirmed

# -------------------------------------------------
# ✅ SHALLOW CLONING – Slice [:] or .copy()
# -------------------------------------------------

a = [1, 2, 3]
b = a[:]              # Yeh ek nayi list banata hai (same values, new ID)
# b = a.copy()        # Alternate syntax for same thing

b.append(4)
print("🔸 a:", a)      # Output: [1, 2, 3] → original unchanged
print("🔸 b:", b)      # Output: [1, 2, 3, 4]

print("🆔 ID of a:", id(a))
print("🆔 ID of b:", id(b))    # ✅ Different IDs → cloning success
```

```python
# Shallow vs Deep Copy in Python

""" 
Cloning and Copy are the same thing 
b=a.copy()
ye shallow copy h 
isme problem ye h ki nested list type ya kuc haur nested hua 
toh andar wali chiz ka same memory address paas hoo jata h 
jisko hum mutate kr skte h 

agar hum chahte h ki pura hi naya memory location mile 
toh deep copy use krni padti h 
deep


"""

# -------------------------------------------------
# ⚠️ Nested List Case – Shallow Clone se dikkat
# -------------------------------------------------

original = [[1, 2], [3, 4]]
clone = original[:]       # Shallow copy – outer list cloned, inner list same

clone[0].append(99)
print("🔸 Original:", original)    # ❌ Changed!
print("🔸 Clone:", clone)

print("🆔 ID of outer list:", id(original), id(clone))        # Alag ID
print("🆔 ID of inner list:", id(original[0]), id(clone[0]))  # Same ID ❗

# -------------------------------------------------
# ✅ DEEP CLONING – full independent copy
# -------------------------------------------------

import copy
original = [[1, 2], [3, 4]]
clone = copy.deepcopy(original)

clone[0].append(99)
print("✅ Deep Clone ->")
print("🔸 Original:", original)     # ✅ Unchanged
print("🔸 Clone:", clone)

# -------------------------------------------------
# 📘 Summary Table – Aliasing vs Shallow vs Deep
# -------------------------------------------------

"""
| Type          | Syntax          | New Memory | Nested Safe? | Notes                        |
|---------------|------------------|-------------|---------------|-----------------------------|
| Aliasing      | b = a            | ❌ No        | ❌ No         | Same object (dangerous)     |
| Shallow Clone | b = a[:] / a.copy() | ✅ Yes    | ❌ No         | Works for 1-level lists     |
| Deep Clone    | copy.deepcopy(a) | ✅ Yes       | ✅ Yes        | Full independent structure  |
"""

# -------------------------------------------------
# 💡 Best Practices:
"""
1. Jab list ya dict ko kisi aur variable ko dena ho – use .copy() ya [:]
2. Agar nested list ho – use deepcopy() to avoid internal linkage
3. Avoid direct aliasing unless intentionally sharing the object
"""
```

```python
# 🔰 How Nested Lists Are Stored in Memory (Referential Array)

"""
📌 Concept: Referential Array
    ➤ Python mein list ek **referential array** hoti hai, jisme original values ke **memory addresses** store hote hain.
    ➤ C++ jaise languages mein arrays **contiguous memory** mein store hote hain, lekin Python mein har element alag memory location pe hota hai.

👉 Iska Matlab:
    - Python list khud sirf references rakhti hai un values ke jo kahin aur memory mein store hoti hain.
    - Ye dynamic aur flexible structure allow karta hai — mixed types, variable size, nesting.
"""

# ✅ C++: Contiguous Memory Example
"""
int arr[] = {1, 2, 3};
// Memory: [1][2][3] (Contiguous memory blocks)
"""

# ✅ Python: Referential Storage Example
arr = [1, 2, [3, 4]]
# Memory: [ref1][ref2][ref3]
# ref1 -> 1, ref2 -> 2, ref3 -> [3, 4] (Each value stored separately in memory)

print("ID of arr[0]:", id(arr[0]))      # Address of 1
print("ID of arr[1]:", id(arr[1]))      # Address of 2
print("ID of arr[2]:", id(arr[2]))      # Address of nested list [3, 4]

# ✅ Python Lists ke Features:

"""
✅ Dynamic Size:
    ➤ List kabhi bhi grow/shrink ho sakti hai without fixed size constraint.

✅ Mixed Types:
    ➤ Ek hi list mein multiple data types store ho sakte hain:
      Example: [1, "hello", 3.14, [4, 5]]

✅ Nesting:
    ➤ Lists ke andar bhi lists store ki ja sakti hain:
      Example: [[1, 2], [3, 4], [5, 6]]

✅ Flexibility:
    ➤ Is structure ki wajah se Python lists bohot flexible aur powerful hoti hain compared to static arrays.
"""


""" 
2D lists mein bhi almost aisa hi concept use hota h 
bas ek list khi joi ban gyi h 
uska first block adress flat wali list mein daal denge 
"""
```

```python
# ✅ How Strings Are Stored in Memory

"""
📌 Explanation:

➤ Python mein strings **immutable** hoti hain — ek baar ban gayi toh usmein 
   koi change directly nahi kar sakte.

➤ Internally, Python strings ko **referential array** ke concept se store karta hai,
   lekin ye **dynamic array** jaise resize nahi hoti.

➤ Jab bhi hum koi new string banate hain, toh Python uske characters ke liye 
   ek new memory block allocate karta hai — har character ka ek fixed size hota hai.

➤ Immutable hone ki wajah se strings:
    - Memory-efficient hote hain (kyunki koi extra capacity nahi chahiye).
    - Thread-safe hote hain.
    - Hashable hote hain — isliye dictionary keys ya set elements ban sakte hain.

🔸 Isliye strings ka memory footprint kam hota hai, 
   aur wo tuple jaisi compact storage strategy follow karte hain.
"""

# ✅ Demo Code to Visualize Storage and IDs

s1 = "hello"
s2 = "hello"
s3 = "hello world"

print("ID of s1:", id(s1))
print("ID of s2:", id(s2))     # s1 aur s2 ka ID same ho sakta hai (string interning)
print("ID of s3:", id(s3))     # different hoga

# Let's check character-wise memory reference
print("ID of s1[0]:", id(s1[0]))   # 'h'
print("ID of s1[1]:", id(s1[1]))   # 'e'

# Memory Size
import sys
print("Memory size of s1:", sys.getsizeof(s1), "bytes")
print("Memory size of s3:", sys.getsizeof(s3), "bytes")
```

```python
# ✅ Why Tuple Takes Less Memory Than List

"""
📌 Explanation:

➤ Lists Python mein **referential array** ki tarah store hoti hain,
   jismein har element ka memory address list ke andar hota hai.

➤ Saath hi, lists **dynamic array** ka concept bhi follow karti hain —
   yani wo grow/shrink ho sakti hain. Is dynamic resizing ke liye 
   Python ko extra memory allocation karni padti hai.

🔸 Isliye list extra space occupy karti hai, even when elements same hoon.

➤ Tuples on the other hand:
   - **Immutable** hoti hain.
   - Unmein dynamic resize ka overhead nahi hota.
   - Aur fixed size hone ki wajah se unki memory packing tight hoti hai.

✅ Strings bhi referential array jaise store hoti hain, 
   lekin unmein dynamic resizing ka feature nahi hota — 
   isliye wo tuple ki tarah lightweight hoti hain.
"""

# ✅ Demo Code to Compare Memory Size

import sys

list_obj = [1, 2, 3, 4, 5]
tuple_obj = (1, 2, 3, 4, 5)

print("Size of List   :", sys.getsizeof(list_obj), "bytes")     # zyda hoti hai
print("Size of Tuple  :", sys.getsizeof(tuple_obj), "bytes")    # kam hoti hai

# Output on most systems:
# Size of List   : 96 bytes
# Size of Tuple  : 80 bytes
```

```python
# ✅ How Set Index Position is Decided in Python

"""
📌 Concept: Hashing and Bucket Indexing

➤ Python ke `set` aur `dict` dono **hash table** structure use karte hain.

➤ Jab koi element set mein add hota hai, toh:
    1. `hash()` function us element ka **hash value** generate karta hai.
    2. Us hash value ko internally kisi **bucket index** mein map kiya jata hai — 
       ye index hota hai (hash % table_size).
    3. Table size grow/shrink hone pe hash position change bhi ho sakta hai.

🔸 Important:
    - Set mein elements ka order fixed nahi hota (Python 3.7+ mein insertion order preserved dikhta hai, but not guaranteed for all versions).
    - Index position purely hash value pe depend karta hai, not actual value.

✅ Isliye, jab bhi tum ek hi values ka set baar-baar run karoge, 
   unka **print order ya memory index** different ho sakta hai (depending on runtime + hash).
"""

# ✅ Demo Code: Hash value aur memory bucket samajhne ke liye

# Element ke hash values
print("Hash of 10:", hash(10))
print("Hash of 'apple':", hash("apple"))
print("Hash of (1, 2):", hash((1, 2)))      # Immutable elements hi allowed hain

# Set Creation
s = {"apple", "banana", "cherry"}
for item in s:
    print(f"Item: {item}, Hash: {hash(item)}")

# Same set baar baar run karne pe order change ho sakta hai
s2 = {"banana", "cherry", "apple"}
print("Set:", s2)
```

# String Data Types in SQL

## Types of String Data Types

String data types are used to store **text / character data** in SQL databases.

Below are the commonly used **string data types** across SQL systems.

---

## 1️⃣ CHAR  
- Fixed-length string  
- Pads extra spaces if data is shorter  
- Faster access but wastes space  

---

## 2️⃣ VARCHAR / VARCHAR2  
- Variable-length string  
- Uses only required space  
- Most commonly used string type  

---

## 3️⃣ TEXT Family (MySQL Specific)

The `TEXT` type in MySQL has **four variants**, based on storage size.

---

### 🔹 TINYTEXT  
- Stores very small text  
- Maximum size: **255 characters**  
- Similar size limit as VARCHAR(255), but stored differently  

Use case:
- Short descriptions
- Small comments

---

### 🔹 TEXT  
- Stores medium-sized text  
- Maximum size: **65,535 characters (~64 KB)**  

Use case:
- Blog content
- User comments
- Descriptions

---

### 🔹 MEDIUMTEXT  
- Stores large text  
- Maximum size: **16,777,215 characters (~16 MB)**  

Use case:
- Large articles
- JSON data
- Logs

---

### 🔹 LONGTEXT  
- Stores extremely large text  
- Maximum size: **4,294,967,295 characters (~4 GB)**  

Use case:
- Full books
- Huge logs
- AI / ML generated content

---

## 4️⃣ NCHAR  
- Fixed-length Unicode string  
- Supports multiple languages  
- Uses more storage than CHAR  

---

## 5️⃣ NVARCHAR  
- Variable-length Unicode string  
- Supports multilingual data  
- More flexible than NCHAR  

---

## 6️⃣ ENUM  
- Stores predefined set of string values  
- Only allowed values can be stored  
- Improves data consistency  
- Common in MySQL  

---

# Important Notes About TEXT Types

- TEXT types **cannot have default values** (in most databases)
- TEXT columns are stored separately from the main row
- TEXT is slower than VARCHAR for searching
- TEXT columns cannot always be indexed fully

---

## Summary Table

| Data Type | Max Size | Unicode | Use Case |
|----------|----------|---------|----------|
| CHAR | Fixed | ❌ | Fixed-size strings |
| VARCHAR | Variable | ❌ | General text |
| TINYTEXT | 255 chars | ❌ | Small text |
| TEXT | 64 KB | ❌ | Medium text |
| MEDIUMTEXT | 16 MB | ❌ | Large text |
| LONGTEXT | 4 GB | ❌ | Very large text |
| NCHAR | Fixed | ✅ | Fixed Unicode text |
| NVARCHAR | Variable | ✅ | Multilingual text |
| ENUM | Fixed Set | ❌ | Limited values |

---

# Final Tip

- Use `VARCHAR` for most cases  
- Use `TEXT / MEDIUMTEXT / LONGTEXT` **only when data size is unknown or very large**  
- Avoid `LONGTEXT` unless absolutely necessary (performance impact)

👉 Choosing the right string type improves **performance, storage efficiency, and scalability**.

# Wildcard in SQL

A **wildcard** is a special character used with the `LIKE` operator 
to search for patterns in string (text) data.

Wildcards are mainly used in **pattern matching**.

---

# Common Wildcards

## 1️⃣ %  (Percent)

- Represents **zero, one, or multiple characters**
- Most commonly used wildcard

Examples of meaning:

'A%'  → Starts with A  
'%A'  → Ends with A  
'%A%' → Contains A anywhere  

---

## 2️⃣ _  (Underscore)

- Represents **exactly one character**
- Used when you want fixed-length matching

Examples of meaning:

'A_'  → A followed by exactly one character  
'_A'  → Any one character followed by A  
'A__' → A followed by exactly two characters  

---

# Example of Wildcard Usage

Suppose we have a column called `name`:

Data:
- Aman
- Akash
- Rahul
- Riya
- Ram

Search Patterns:

'A%'  
→ Aman, Akash  

'%a'  
→ Riya (if case insensitive system)  

'R__'  
→ Ram  

'%ha%'  
→ Rahul  

---

# LIKE Operator

The `LIKE` operator is used with wildcards 
to perform pattern-based searching.

It works only with string (text) data.

---

# How LIKE Operator Works

1. It compares each row's column value with the given pattern.
2. If the value matches the pattern → it is returned.
3. If it does not match → it is ignored.
4. Matching behavior may be case-sensitive or case-insensitive depending on the database.

---


# String Functions in SQL

String functions are used to **manipulate, modify, analyze, and format text data** in SQL.

Below are the most commonly used string functions with short descriptions.

---

## 1️⃣ UPPER()
Converts all characters of a string to **uppercase**.

Use case: Standardizing text for comparison.

---

## 2️⃣ LOWER()
Converts all characters of a string to **lowercase**.

Use case: Case-insensitive searching.

---

## 3️⃣ LENGTH() / LEN()
Returns the **number of characters** in a string.

- `LENGTH()` → MySQL, PostgreSQL  
- `LEN()` → SQL Server  

---

## 4️⃣ TRIM()
Removes **leading and trailing spaces** from a string.

Helps clean messy user input.

---

## 5️⃣ LTRIM()
Removes spaces from the **left side** of a string.

---

## 6️⃣ RTRIM()
Removes spaces from the **right side** of a string.

---

## 7️⃣ CONCAT()
Joins two or more strings together.

Used for combining names, addresses, etc.

---

## 8️⃣ SUBSTRING() / SUBSTR()
Extracts a **portion of a string**.

Used when you need specific characters from a string.

---

## 9️⃣ REPLACE()
Replaces a specific part of a string with another value.

Useful for cleaning or modifying text.

---

## 🔟 INSTR() / POSITION()
Returns the **position of a substring** inside a string.

---

## 1️⃣1️⃣ LEFT()
Returns specified number of characters from the **left side**.

---

## 1️⃣2️⃣ RIGHT()
Returns specified number of characters from the **right side**.

---

## 1️⃣3️⃣ CHAR_LENGTH()
Returns the number of characters (important for Unicode).

---

## 1️⃣4️⃣ REVERSE()
Reverses the string.

---

## 1️⃣5️⃣ LPAD()
Pads the left side of a string with specific characters.

---

## 1️⃣6️⃣ RPAD()
Pads the right side of a string with specific characters.

---

## 1️⃣7️⃣ FORMAT()
Formats numbers as strings (adds commas, decimals, etc.).

---

# Summary

String functions help to:

- Change case (UPPER, LOWER)
- Measure length (LENGTH, LEN)
- Clean spaces (TRIM, LTRIM, RTRIM)
- Extract text (SUBSTRING, LEFT, RIGHT)
- Modify text (REPLACE)
- Combine text (CONCAT)
- Locate text (INSTR, POSITION)
- Pad or format text (LPAD, RPAD, FORMAT)

👉 These functions are essential for **data cleaning, reporting, and transformation** in SQL.

## Joins in SQL (with Table Examples)

**SQL JOIN** is used to combine rows from two or more tables based on a related column between them.

---

## Example Tables

### Table: `employees`

| emp_id | name   | dept_id |
|------:|--------|--------:|
| 1     | Rahul  | 101     |
| 2     | Aditi  | 102     |
| 3     | Aman   | 103     |
| 4     | Neha   | NULL    |

### Table: `departments`

| dept_id | dept_name |
|--------:|-----------|
| 101     | HR        |
| 102     | IT        |
| 104     | Finance   |

---

## 1. INNER JOIN

Returns **only matching records** from both tables.

### Result Condition  
`employees.dept_id = departments.dept_id`

### Output

| emp_id | name  | dept_name |
|------:|-------|-----------|
| 1     | Rahul | HR        |
| 2     | Aditi | IT        |

✔ Common data only  
✖ Aman, Neha, Finance excluded

---

## 2. LEFT JOIN (LEFT OUTER JOIN)

Returns **all records from left table** and matching records from right table.

### Output

| emp_id | name  | dept_name |
|------:|-------|-----------|
| 1     | Rahul | HR        |
| 2     | Aditi | IT        |
| 3     | Aman  | NULL      |
| 4     | Neha  | NULL      |

✔ All employees included  
✖ Missing department shown as `NULL`

---

## 3. RIGHT JOIN (RIGHT OUTER JOIN)

Returns **all records from right table** and matching records from left table.

### Output

| emp_id | name  | dept_name |
|------:|-------|-----------|
| 1     | Rahul | HR        |
| 2     | Aditi | IT        |
| NULL  | NULL  | Finance   |

✔ All departments included  
✖ Employees missing shown as `NULL`

---

## 4. FULL JOIN (FULL OUTER JOIN)

Returns **all records from both tables**.

### Output

| emp_id | name  | dept_name |
|------:|-------|-----------|
| 1     | Rahul | HR        |
| 2     | Aditi | IT        |
| 3     | Aman  | NULL      |
| 4     | Neha  | NULL      |
| NULL  | NULL  | Finance   |

✔ Nothing is lost  
✔ Non-matching data filled with `NULL`

---

## 5. CROSS JOIN

Returns **Cartesian product**  
Each row of first table × each row of second table

### If:
- employees = 4 rows  
- departments = 3 rows  

➡ Result = **12 rows**

⚠️ Rarely used, can be very large

---

## Summary Table

| JOIN Type   | Keeps Left | Keeps Right | Matches Only |
|------------|------------|-------------|--------------|
| INNER JOIN | ❌         | ❌          | ✅           |
| LEFT JOIN  | ✅         | ❌          | ❌           |
| RIGHT JOIN | ❌         | ✅          | ❌           |
| FULL JOIN  | ✅         | ✅          | ❌           |
| CROSS JOIN | N/A        | N/A         | ❌           |

---

### Interview Tip
> INNER → common  
> LEFT → all left  
> RIGHT → all right  
> FULL → everything  
> CROSS → combinations


## Difference Between OUTER JOIN and CROSS JOIN in SQL

OUTER JOIN and CROSS JOIN are both used to combine tables, but **their purpose and results are completely different**.

---

## 1. OUTER JOIN

### Meaning
An **OUTER JOIN** returns **matching records** plus **non-matching records** from one or both tables.

### Types of OUTER JOIN
- LEFT OUTER JOIN
- RIGHT OUTER JOIN
- FULL OUTER JOIN

### Key Characteristics
- Based on a **join condition**
- Uses a **common column**
- Missing matches are filled with `NULL`
- Controls which table’s data must be preserved

### Example Result Behavior
- LEFT JOIN → keeps all rows from left table
- RIGHT JOIN → keeps all rows from right table
- FULL JOIN → keeps all rows from both tables

### Use Case
- When you want **complete data**, even if matching rows don’t exist

---

## 2. CROSS JOIN

### Meaning
A **CROSS JOIN** returns the **Cartesian product** of both tables.

### Key Characteristics
- **No join condition**
- Every row of Table A combines with every row of Table B
- Output size = (rows in A × rows in B)
- Can generate very large results

### Example Result Behavior
If:
- Table A → 3 rows
- Table B → 4 rows

Result → **12 rows**

### Use Case
- Creating all possible combinations
- Matrix-like data
- Test data generation

---

## Core Differences Table

| Feature            | OUTER JOIN                         | CROSS JOIN                     |
|--------------------|-----------------------------------|--------------------------------|
| Join condition     | Required                           | Not required                   |
| Matching logic     | Matches + non-matches              | All combinations               |
| NULL values        | Yes (for missing matches)          | No (unless data already NULL)  |
| Result size        | Depends on matching rows           | RowsA × RowsB                  |
| Performance impact | Moderate                           | Can be very heavy              |
| Practical use      | Real-world relational queries      | Special-case scenarios         |

---

## Easy Way to Remember

- **OUTER JOIN** → "Keep missing data"
- **CROSS JOIN** → "Multiply rows"

---

## Interview One-Liner

> OUTER JOIN preserves unmatched rows using NULLs,  
> while CROSS JOIN creates all possible row combinations without any condition.


## SQL Set Operations

**SQL Set Operations** are used to combine the results of two or more `SELECT` queries into a **single result set**.

⚠️ Important:  
- Each `SELECT` must have the **same number of columns**
- Columns must have **compatible data types**
- Column order must be the **same**

---

## Common SQL Set Operations

1. UNION  
2. UNION ALL  
3. INTERSECT  
4. EXCEPT (or MINUS)

---

## Example Tables

### Table: `students_2023`

| id | name   |
|----|--------|
| 1  | Rahul  |
| 2  | Aditi  |
| 3  | Aman   |

### Table: `students_2024`

| id | name   |
|----|--------|
| 3  | Aman   |
| 4  | Neha   |
| 5  | Kunal  |

---

## 1. UNION

### Description
- Combines results
- **Removes duplicate rows**
- Performs implicit `DISTINCT`

### Result

| id | name   |
|----|--------|
| 1  | Rahul  |
| 2  | Aditi  |
| 3  | Aman   |
| 4  | Neha   |
| 5  | Kunal  |

✔ No duplicates  
❌ Slightly slower

---

## 2. UNION ALL

### Description
- Combines results
- **Keeps duplicate rows**
- Faster than UNION

### Result

| id | name   |
|----|--------|
| 1  | Rahul  |
| 2  | Aditi  |
| 3  | Aman   |
| 3  | Aman   |
| 4  | Neha   |
| 5  | Kunal  |

✔ Keeps duplicates  
✔ Better performance

---

## 3. INTERSECT

### Description
- Returns **only common rows**

### Result

| id | name |
|----|------|
| 3  | Aman |

✔ Common data only

---

## 4. EXCEPT / MINUS

### Description
- Returns rows from first query **not present** in second
- `EXCEPT` → MySQL / PostgreSQL  
- `MINUS` → Oracle

### Result (`students_2023 - students_2024`)

| id | name  |
|----|-------|
| 1  | Rahul |
| 2  | Aditi |

✔ Difference of sets

---

## Comparison Table

| Operation   | Removes Duplicates | Keeps Duplicates | Common Rows | Difference |
|------------|--------------------|------------------|------------|------------|
| UNION      | ✅                 | ❌               | ❌         | ❌         |
| UNION ALL  | ❌                 | ✅               | ❌         | ❌         |
| INTERSECT  | ✅                 | ❌               | ✅         | ❌         |
| EXCEPT     | ✅                 | ❌               | ❌         | ✅         |

---

## Interview Tip

- UNION → Merge unique
- UNION ALL → Merge all
- INTERSECT → Common
- EXCEPT → A − B

---

## Summary

- Set operations work on **query results**, not tables
- Structure of queries must match
- Very useful in reporting and analytics

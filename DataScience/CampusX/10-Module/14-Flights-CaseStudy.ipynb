alter table casestudy.flights 
add column duration_mins integer,
add column arrival datetime;

SELECT 
    duration,
    
    SUBSTRING_INDEX(duration, ' ', 1) AS first_part,
    
    CASE 
        WHEN SUBSTRING_INDEX(duration, ' ', -1) = SUBSTRING_INDEX(duration, ' ', 1) 
            THEN 0
        ELSE REPLACE(SUBSTRING_INDEX(duration, ' ', -1), 'm', '')
    END AS mins

FROM flights;


-- always give if exists
SELECT * FROM casestudy.flights;

UPDATE casestudy.flights
SET duration_mins = 
    CASE
        -- case 1: only minutes (e.g., '50m')
        WHEN duration NOT LIKE '%h%' 
            THEN CAST(REPLACE(duration, 'm', '') AS UNSIGNED)

        -- case 2: only hours (e.g., '5h')
        WHEN duration NOT LIKE '%m%' 
            THEN CAST(REPLACE(duration, 'h', '') AS UNSIGNED) * 60

        -- case 3: both hours + minutes (e.g., '2h 50m')
        ELSE 
            CAST(REPLACE(SUBSTRING_INDEX(duration,' ',1),'h','') AS UNSIGNED) * 60 +
            CAST(REPLACE(SUBSTRING_INDEX(duration,' ',-1),'m','') AS UNSIGNED)
    END;

    
-- Step 2: Calculate arrival time
UPDATE casestudy.flights
SET arrival = DATE_ADD(departure, INTERVAL duration_mins MINUTE);

# Zomato Case Study – SQL Solutions (Q1 to Q9)

## Q1. 
SELECT COUNT(*) 
FROM casestudy.zomato_order_details;


## Q2
SELECT *
FROM casestudy.zomato_users
ORDER BY RAND()
LIMIT 5;



## Q3
SELECT *
FROM casestudy.zomato_orders
WHERE restaurant_rating IS NULL;



## Q4
UPDATE casestudy.zomato_orders
SET restaurant_rating = 0
WHERE restaurant_rating IS NULL;


## Q6
SELECT u.user_id,
       u.name,
       COUNT(*) AS total_orders
FROM casestudy.zomato_orders o
JOIN casestudy.zomato_users u
ON o.user_id = u.user_id
GROUP BY u.user_id, u.name;



## Q7
SELECT r.r_id,
       r.r_name,
       COUNT(*) AS menu_items
FROM casestudy.zomato_menu m
JOIN casestudy.zomato_restaurants r
ON m.r_id = r.r_id
GROUP BY r.r_id, r.r_name;



## Q8
SELECT f.f_id,
       f.f_name,
       COUNT(*) AS order_count
FROM casestudy.zomato_order_details d
JOIN casestudy.zomato_food f
ON d.f_id = f.f_id
GROUP BY f.f_id, f.f_name
ORDER BY order_count DESC
LIMIT 1;


## Q9
SELECT r.r_id,
       r.r_name,
       SUM(o.amount) AS revenue
FROM casestudy.zomato_orders o
JOIN casestudy.zomato_restaurants r
ON o.r_id = r.r_id
WHERE MONTHNAME(DATE(o.date)) = 'July'
GROUP BY r.r_id, r.r_name
ORDER BY revenue DESC
LIMIT 1;



# 10

group by r_id
having t_amt > 1500;

# 12
SELECT * FROM casestudy.zomato_orders
where date between `2022-05-10` and `2022-07-10` and user_id =1

SELECT o.order_id, f.f_name, o.date
FROM casestudy.zomato_orders o
JOIN casestudy.zomato_order_details d
ON o.order_id = d.order_id
JOIN casestudy.zomato_food f
ON d.f_id = f.f_id
WHERE o.user_id = 1
AND o.date BETWEEN '2022-05-10' AND '2022-07-10';


# 13th method -1
SELECT d.f_id, count(*) as cnt
 FROM casestudy.zomato_orders o 
join casestudy.zomato_order_details d
on o.order_id = d.order_id
group by d.f_id
order by cnt desc limit 1



# 14th
SELECT r_id, AVG(price) AS avg_price
FROM casestudy.zomato_menu
GROUP BY r_id
ORDER BY avg_price ASC
LIMIT 1;



# 15th 
SELECT d.partner_id,
       COUNT(*) * 100 + AVG(o.delivery_rating) * 1000 AS salary
FROM casestudy.zomato_delivery_partner d
JOIN casestudy.zomato_orders o
ON d.partner_id = o.partner_id
GROUP BY d.partner_id;




## Q16. 

SELECT 
    CORR(delivery_time, delivery_rating) AS corr_value
FROM casestudy.zomato_orders;



# 17th
SELECT r.r_id,
       r.r_name
FROM casestudy.zomato_menu m
JOIN casestudy.zomato_food f
ON m.f_id = f.f_id
JOIN casestudy.zomato_restaurants r
ON m.r_id = r.r_id
GROUP BY r.r_id, r.r_name
HAVING MIN(f.type) = 'Veg'
   AND MAX(f.type) = 'Veg';




# 18th 
SELECT u.user_id,
       MAX(o.amount) AS maximum,
       MIN(o.amount) AS minimum,
       AVG(o.amount) AS avg_amt
FROM casestudy.zomato_users u
JOIN casestudy.zomato_orders o
ON u.user_id = o.user_id
GROUP BY u.user_id;

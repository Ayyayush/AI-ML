# Database Design

## What is Database Designing?

Database Designing ka matlab hota hai **data ko properly organize karna** taaki:
- data easily store ho sake
- efficiently retrieve (nikala) ja sake
- redundancy (duplicate data) kam ho
- performance fast rahe

Simple words me:
👉 Database design = **data ko smart tareeke se structure karna**

---

## Real Life Example

Socho tum ek college system bana rahe ho:

Agar tum sab data ek hi table me daal doge:
- student name
- course
- teacher
- marks

Toh bohot repetition hoga 😵

Isliye hum alag tables banate hain:
- Students
- Courses
- Teachers
- Marks

👉 Isko hi kehte hain **proper database design**

---

## Why Database Designing is Important?

- ✅ Data duplication kam hota hai  
- ✅ Data consistency maintain hoti hai  
- ✅ Query fast chalti hai  
- ✅ Data easy to manage hota hai  
- ✅ Future me scale karna easy hota hai  

---

# SQL Datatypes

## What are SQL Datatypes?

SQL Datatypes define karte hain ki **column me kis type ka data store hoga**  
👉 jaise number, text, date, etc.

Simple Hinglish:  
**Datatype = data ka type (kaisa data store karna hai)**

---

## 1. Numeric Datatypes (Numbers store karne ke liye)

### a) INT  
👉 Whole numbers (no decimal)  
Example: age INT  

storage- 4 bytes  
range- signed / unsigned  
max- 2147483647 (signed), 4294967295 (unsigned)  
min- -2147483648 (signed), 0 (unsigned)  

---

### b) SMALLINT  
👉 Chhote range ke integers  
Example: count SMALLINT  

storage- 2 bytes  
range- signed / unsigned  
max- 32767 (signed), 65535 (unsigned)  
min- -32768 (signed), 0 (unsigned)  

---

### c) TINYINT  
👉 Very small integers  

storage- 1 byte  
range- signed / unsigned  
max- 127 (signed), 255 (unsigned)  
min- -128 (signed), 0 (unsigned)  

---

### d) MEDIUMINT  
👉 Medium size integer  

storage- 3 bytes  
range- signed / unsigned  
max- 8388607 (signed), 16777215 (unsigned)  
min- -8388608 (signed), 0 (unsigned)  

---

### e) BIGINT  
👉 Bahut bade numbers ke liye  
Example: population BIGINT  

storage- 8 bytes  
range- signed / unsigned  
max- 9223372036854775807 (signed)  
min- -9223372036854775808 (signed)  

---

### f) DECIMAL / NUMERIC  
👉 Exact decimal values (finance ke liye best)  
Example: salary DECIMAL(10,2)  

storage- depends on precision  
range- exact values  
max- depends on precision  
min- depends on precision  

---

### g) FLOAT / DOUBLE  
👉 Approximate decimal values  
⚠️ precision utni accurate nahi hoti jitni DECIMAL me hoti hai  
👉 ek limit ke baad rounding hone lagti hai  

Example: price FLOAT  

storage- FLOAT(4 bytes), DOUBLE(8 bytes)  
range- very large approximate values  
max- depends on type  
min- depends on type  

---



## 2. String Datatypes (Text store karne ke liye)

## 3. Date & Time Datatypes

## 4. Boolean Datatype

# Database Normalization

## What is Database Normalization?

Database Normalization ek process hai jisme hum data ko **properly organize** karte hain taaki:
- duplicate data (redundancy) kam ho
- data consistent rahe
- update karna easy ho

Simple Hinglish:  
**Normalization = data ko clean aur structured format me todna (break karna)**

---

## Why don't we store all data in a single table?

Agar hum saara data ek hi table me store karte hain to problems aati hain:

### ❌ 1. Data Redundancy (Duplicate Data)
Same data baar-baar repeat hota hai  
👉 storage waste + inconsistency ka risk  

---

### ❌ 2. Data Inconsistency
Agar ek jagah update kiya aur dusri jagah nahi kiya  
👉 data mismatch ho jata hai  

---

### ❌ 3. Difficult Updates
Ek value change karne ke liye multiple rows update karni padti hain  

---

### ❌ 4. Poor Performance
Table bada ho jata hai → queries slow ho jati hain  

---

### ❌ 5. Maintenance Problem
Database manage karna mushkil ho jata hai  

---

👉 Isliye hum data ko multiple tables me todte hain (normalize karte hain)

---

# Types of Anomalies

👉 Jab database properly design nahi hota, tab anomalies (problems) aati hain  

---

## 1. Insertion Anomaly

👉 Jab hum naya data insert nahi kar paate bina unnecessary data ke  

### Example:
Agar student table me course bhi store ho raha hai  
Aur course assign nahi hua  
👉 student ko insert hi nahi kar paoge  

---

## 2. Update Anomaly

👉 Same data multiple jagah store ho  
Aur update sab jagah nahi hua  

### Example:
Teacher ka naam 10 rows me stored hai  
1 jagah update kiya, baaki jagah nahi  
👉 inconsistency  

---

## 3. Deletion Anomaly

👉 Kisi data ko delete karte waqt important info bhi delete ho jata hai  

### Example:
Agar ek student delete kiya  
Aur uske saath course info bhi delete ho gaya  
👉 data loss  

---

## Conclusion

👉 Normalization ka main goal:
- redundancy kam karna  
- consistency maintain karna  
- anomalies avoid karna  

Simple line:  
**"Normalization = clean, efficient aur safe database design"**

# 1NF (First Normal Form)

## Rule

👉 Table me:
- koi repeating groups nahi hone chahiye  
- har column me **atomic (single) value** honi chahiye  
- ek cell me multiple values nahi honi chahiye  
- order does not matter how data is tored 
- uniqye name for evyr / attribute or column 


---

## Simple Hinglish

**1NF = ek cell me sirf ek hi value honi chahiye**

---

## Example (Not in 1NF)

Student Table:

| id | name | subjects        |
|----|------|----------------|
| 1  | Aman | DBMS, OS       |

👉 Problem: subjects column me multiple values hain ❌  

---

## Convert to 1NF

| id | name | subject |
|----|------|--------|
| 1  | Aman | DBMS   |
| 1  | Aman | OS     |

---








# 2NF (Second Normal Form)

## Rule

👉 Table:
- pehle 1NF me hona chahiye  
- aur **partial dependency nahi honi chahiye**  

👉 Non-key attributes should depend on **whole primary key**

---

## Simple Hinglish

**2NF = half dependency nahi honi chahiye (poore key par depend kare)**

partial deendcny occurs when a non ke attriute is oly dependent on a part of primary keyu 
---

## Example (Not in 2NF)

| student_id | course_id | student_name |
|------------|----------|--------------|

👉 Primary key = (student_id, course_id)  
👉 student_name sirf student_id pe depend karta hai ❌  

---

## Convert to 2NF

Split into 2 tables:

### Students Table
| student_id | student_name |

### Courses Table
| course_id |

### Enrollment Table
| student_id | course_id |

---

# 3NF (Third Normal Form)

## Rule

👉 Table:
- 2NF me hona chahiye  
- aur **transitive dependency nahi honi chahiye**  

👉 Non-key attribute kisi aur non-key attribute pe depend nahi hona chahiye  

---

## Simple Hinglish

**3NF = indirect dependency nahi honi chahiye**

---

## Example (Not in 3NF)

| student_id | dept_id | dept_name |

👉 dept_name depend karta hai dept_id pe  
👉 aur dept_id depend karta hai student_id pe  

👉 indirect dependency ❌  

---

## Convert to 3NF

Split into:

### Students Table
| student_id | dept_id |

### Department Table
| dept_id | dept_name |

---

# How to Normalize a Denormalized Data

## Step by Step Process

### Step 1: Convert to 1NF
👉 Multiple values ko alag rows me tod do  

---

### Step 2: Convert to 2NF
👉 Partial dependencies identify karo  
👉 unnecessary columns ko alag table me daalo  

---

### Step 3: Convert to 3NF
👉 Transitive dependencies hatao  
👉 indirect relationships ko separate tables me convert karo  

---

## Final Understanding

👉 Normalization ka flow:

Denormalized Data  
→ 1NF (atomic values)  
→ 2NF (no partial dependency)  
→ 3NF (no transitive dependency)

---

## Conclusion

👉 Normalization se:
- data clean hota hai  
- duplication kam hota hai  
- anomalies avoid hoti hain  

Simple line:  
**"Normalize karo = smart database banao"**

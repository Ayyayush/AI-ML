### Window Functions

- Used to perform **calculations across a set of rows related to the current row**.
- Does **not collapse rows** like GROUP BY.
- Commonly used for **ranking, running totals, moving averages**, etc.
- Operates using a **window (partition)** defined by `OVER()`.

**Examples:**  
ROW_NUMBER, RANK, DENSE_RANK, SUM() OVER(), AVG() OVER()


'Window Functions aise SQL functions hote hain jo current row ke around related rows par calculation karte hain,
without rows ko collapse kiye (unlike GROUP BY).'

# FRAMES (Window Function Frames)

## What is a Frame?

A **frame** defines which rows are considered when a window function 
(like SUM, AVG, LAST_VALUE, NTH_VALUE, etc.) is calculated for the current row.

👉 In simple words:  
**Frame = Subset of rows inside the window**

When working with window functions:

- PARTITION → Divides data into groups  
- ORDER → Sorts rows inside each partition  
- FRAME → Selects which rows (around the current row) are used for calculation  

The frame works only after ordering is applied.

---

# Why Frames Are Important

Without a frame, the database might use a default behavior.
Frames allow you to control:

- Running totals
- Moving averages
- Cumulative sums
- Sliding window calculations
- Lagging/leading calculations
- Financial rolling metrics

---

# Frame Boundaries

Frames are defined using boundary keywords.

Common boundaries:

- UNBOUNDED PRECEDING → From first row in partition
- N PRECEDING → N rows before current row
- CURRENT ROW → Only current row
- N FOLLOWING → N rows after current row
- UNBOUNDED FOLLOWING → Till last row in partition

---



# Types of Frames

There are three main types of frames:

1. ROWS
2. RANGE
3. GROUPS (available in modern SQL engines like PostgreSQL)

---



## 1️⃣ ROWS Frame

### Definition
ROWS defines the frame using **physical row positions** relative to the current row.

It counts actual rows.

### Key Characteristics

- Based on row numbers
- Ignores value similarity
- Each row is treated independently
- Works even if ORDER BY column has duplicates
- Most predictable behavior

### Example Concept

If frame says:
- 2 preceding rows

Then it means:
- Take 2 rows above current row in sorted order

Even if values are same or different — it does not matter.

---

## 2️⃣ RANGE Frame

### Definition
RANGE defines the frame using a **logical value range**, not row position.

It works based on ORDER BY column values.

### Key Characteristics

- Based on value range
- Groups equal ORDER BY values automatically
- Includes all duplicate values
- Can behave differently than ROWS
- Depends on data values, not row count

### Important Difference from ROWS

If multiple rows have the same ORDER BY value:

- ROWS → Treats them separately
- RANGE → Treats them as a group automatically

---

## 3️⃣ GROUPS Frame

### Definition
GROUPS works based on **groups of equal ORDER BY values**.

It moves group-by-group instead of row-by-row.

### Key Characteristics

- Based on value groups
- Each group = rows with same ORDER BY value
- Steps through groups instead of rows
- More structured than RANGE

---



# Types of Frame Shapes

Frames can be:

1. Cumulative Frame  
   From beginning till current row  

2. Sliding Frame  
   Fixed number of rows before and/or after  

3. Forward-Looking Frame  
   Current row to future rows  

4. Entire Partition Frame  
   Whole partition

---

# Default Frame Behavior

If ORDER BY is used but frame is not specified:

Most databases default to:

RANGE BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW

This means:
- Cumulative calculation
- Includes duplicates automatically

Important: Default behavior may differ across SQL engines.

---

# Comparison Summary

| Feature | ROWS | RANGE | GROUPS |
|----------|--------|--------|----------|
| Based On | Physical rows | Value range | Value groups |
| Duplicate Handling | Treated separately | Automatically included | Included as groups |
| Predictability | High | Medium | High |
| Movement Type | Row-by-row | Value-based | Group-by-group |

---

# Mental Model

Think of it like this:

Window = Full ordered dataset  
Frame = Moving spotlight over part of that dataset  

As the current row changes, the spotlight (frame) moves.

---

# Final Understanding

Window Function =  
Partition → Organizes data  
Order → Sorts data  
Frame → Decides which rows are visible to current row  

👉 Frame controls the exact portion of ordered data 
used in the calculation for each row.

### Demo Window Function Queries

**1. ROW_NUMBER**
```sql
SELECT name, department, salary,
       ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS rn
FROM employees;
```



# 2. RANK
```sql
SELECT name, department, salary,
       RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS rnk
FROM employees;
```



# 3. DENSE_RANK
```sql
SELECT name, department, salary,
       DENSE_RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS drnk
FROM employees;
```



 # 4. Running Total
```sql
SELECT date, amount,
       SUM(amount) OVER (ORDER BY date) AS running_total
FROM sales;
```



# 5. Moving Average
```sql
SELECT date, amount,
       AVG(amount) OVER (ORDER BY date ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) AS moving_avg
FROM sales;
```



# 6. Difference from Average
```sql
SELECT name, salary,
       salary - AVG(salary) OVER () AS diff_from_avg
FROM employees;
```

# beginner questions solutions 
```sql

SELECT student_id, name, class, marks,
       ROW_NUMBER() OVER (PARTITION BY class ORDER BY marks DESC) AS rn
FROM students;


SELECT student_id, name, class, subject, marks,
       RANK() OVER (PARTITION BY subject ORDER BY marks DESC) AS ranking
FROM students;


SELECT student_id, name, class, subject, marks,
       DENSE_RANK() OVER (PARTITION BY subject ORDER BY marks DESC) AS dense_ranking
FROM students;


SELECT student_id, name, class, marks,
       AVG(marks) OVER (PARTITION BY class) AS avg_marks
FROM students;


SELECT student_id, name, class, marks,
       SUM(marks) OVER (PARTITION BY class) AS class_total
FROM students;



```

```sql
SELECT student_id, name, class, subject, marks,
       LAG(marks) OVER (PARTITION BY class ORDER BY marks) AS prev_marks
FROM students;



SELECT student_id, name, class, subject, marks,
       LEAD(marks) OVER (PARTITION BY subject ORDER BY marks) AS next_marks
FROM students;



SELECT student_id, name, subject, marks,
       marks - AVG(marks) OVER (PARTITION BY subject) AS diff_from_avg
FROM students;



SELECT *
FROM (
    SELECT student_id, name, subject, marks,
           DENSE_RANK() OVER (PARTITION BY subject ORDER BY marks DESC) AS rnk
    FROM students
) t
WHERE rnk <= 2;

```

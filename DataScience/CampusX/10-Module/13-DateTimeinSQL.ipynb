# Temporal Datatypes in SQL

SQL me temporal datatypes ka use date aur time related values store karne ke liye hota hai.
Ye mainly 4 types ke hote hain:

---

## 1. DATE

* Sirf date store karta hai (year, month, day)
* Format: YYYY-MM-DD
* Time store nahi hota
  👉 Example: 2026-03-21

---

## 2. TIME

* Sirf time store karta hai (hour, minute, second)
* Date store nahi hota
  👉 Example: 14:30:45

---

## 3. DATETIME

* Date + time dono store karta hai
* Timezone support nahi hota
  👉 Example: 2026-03-21 14:30:45

---

## 4. TIMESTAMP

* Date + time store karta hai
* Timezone dependent hota hai (auto convert ho sakta hai)
  👉 Example: 2026-03-21 14:30:45

---

## Quick Summary

| Type      | Stores      | Timezone |
| --------- | ----------- | -------- |
| DATE      | Date only   | ❌        |
| TIME      | Time only   | ❌        |
| DATETIME  | Date + Time | ❌        |
| TIMESTAMP | Date + Time | ✅        |


# Datetime Functions

SQL me datetime functions ka use date aur time ko manipulate (change, calculate, extract) karne ke liye hota hai.

---

## Current Date & Time Functions

### 1. CURDATE()

* Current date return karta hai
  👉 Example: CURDATE() → 2026-03-21

### 2. CURTIME()

* Current time return karta hai
  👉 Example: CURTIME() → 14:30:45

### 3. NOW()

* Current date + time dono return karta hai
  👉 Example: NOW() → 2026-03-21 14:30:45

---

# Extraction Functions

Ye functions date/time se specific part nikalne ke liye use hote hain.

## 1. YEAR()

* Year nikalta hai
  👉 Example: YEAR('2026-03-21') → 2026

## 2. MONTH()

* Month number nikalta hai
  👉 Example: MONTH('2026-03-21') → 3

## 3. DAY()

* Day of month nikalta hai
  👉 Example: DAY('2026-03-21') → 21

## 4. HOUR()

* Hour nikalta hai
  👉 Example: HOUR('14:30:45') → 14

## 5. MINUTE()

* Minute nikalta hai
  👉 Example: MINUTE('14:30:45') → 30

## 6. SECOND()

* Second nikalta hai
  👉 Example: SECOND('14:30:45') → 45

## 7. DAYNAME()

* Week ka day ka naam deta hai
  👉 Example: DAYNAME('2026-03-21') → Saturday

## 8. DAYOFWEEK()

* Week me day ka number deta hai (1 = Sunday)
  👉 Example: DAYOFWEEK('2026-03-21') → 7

## 9. DAYOFMONTH()

* Month ka day number deta hai
  👉 Example: DAYOFMONTH('2026-03-21') → 21

## 10. DAYOFYEAR()

* Year ka kaunsa din hai batata hai
  👉 Example: DAYOFYEAR('2026-03-21') → 80

## 11. WEEK()

* Year ka week number batata hai
  👉 Example: WEEK('2026-03-21') → 12

## 12. WEEKDAY()

* Weekday number deta hai (0 = Monday)
  👉 Example: WEEKDAY('2026-03-21') → 5

## 13. WEEKOFYEAR()

* Year ka week number deta hai (similar to WEEK)
  👉 Example: WEEKOFYEAR('2026-03-21') → 12

## 14. MONTHNAME()

* Month ka naam deta hai
  👉 Example: MONTHNAME('2026-03-21') → March

## 15. QUARTER()

* Year ka quarter batata hai (1–4)
  👉 Example: QUARTER('2026-03-21') → 1

## 16. LAST_DAY()

* Month ka last date deta hai
  👉 Example: LAST_DAY('2026-03-21') → 2026-03-31

---

## Quick Summary

* Current functions → CURDATE(), CURTIME(), NOW()
* Extraction functions → date/time ke alag-alag parts nikalte hain
* Extra useful functions → DAYOFWEEK, WEEKDAY, QUARTER, LAST_DAY
* Mostly reporting, analytics aur filtering me use hote hain


```sql
SELECT 
    *,  

    DATE(start_time)            AS only_date,  
    TIME(start_time)            AS only_time,  

    MONTH(start_time)           AS month_num,  
    MONTHNAME(start_time)       AS month_name,  

    DAY(start_time)             AS day_of_month,  
    DAYNAME(start_time)         AS day_name,  

    DAYOFWEEK(start_time)       AS day_of_week,  
    DAYOFYEAR(start_time)       AS day_of_year,  

    WEEKOFYEAR(start_time)      AS week_of_year,  
    QUARTER(start_time)         AS quarter,  

    HOUR(start_time)            AS hour,  
    MINUTE(start_time)          AS minute,  
    SECOND(start_time)          AS second,  

    LAST_DAY(start_time)        AS last_day_of_month  

FROM campusx.uber_rides;   
```

# DATE_FORMAT(date, format)

DATE_FORMAT() ka use SQL me date ko custom format me convert karne ke liye hota hai.

* Agar koi bhi argument NULL ho → result bhi NULL aayega
* Format string me `%` lagana compulsory hota hai

👉 Example:
DATE_FORMAT('2026-03-21', '%d-%m-%Y') → 21-03-2026

---

## Important Note

* Ye specifiers sirf DATE_FORMAT() me nahi, balki in functions me bhi use hote hain:

  * STR_TO_DATE()
  * TIME_FORMAT()
  * UNIX_TIMESTAMP()

---

# Common Specifiers (Easy & Important)

## Day Related

* %a → Short weekday (Sun..Sat)
* %W → Full weekday (Sunday..Saturday)
* %d → Day (00–31)
* %e → Day (0–31)
* %D → Day with suffix (1st, 2nd, 3rd)
* %j → Day of year (001–366)
* %w → Day of week (0=Sunday)

---

## Month Related

* %b → Short month (Jan..Dec)
* %M → Full month (January..December)
* %m → Month (00–12)
* %c → Month (0–12)

---

## Year Related

* %Y → Full year (2026)
* %y → Short year (26)

---

## Time Related

* %H → Hour (00–23)
* %h / %I → Hour (01–12)
* %i → Minutes (00–59)
* %s / %S → Seconds (00–59)
* %p → AM/PM
* %r → 12-hour time
* %T → 24-hour time

---

## Week Related

* %U → Week (Sunday start)
* %u → Week (Monday start)
* %V / %v → Week number (01–53)
* %X / %x → Year for week

---

## Extra Useful

* %f → Microseconds
* %% → Print % symbol
* %x → Any character

---

## Quick Summary

* DATE_FORMAT() → date ko readable/custom format me convert karta hai
* `%` ke through formatting control hota hai
* Different specifiers → day, month, year, time, week extract karte hain
* Reporting aur UI display ke liye bahut useful hota hai

# Type Conversion in SQL

SQL me type conversion ka use ek datatype ko dusre datatype me convert karne ke liye hota hai.

Ye mainly 2 types ke hote hain:

---

## 1. Implicit Conversion (Automatic)

* Ye automatically DBMS khud karta hai
* User ko explicitly convert karne ki zarurat nahi hoti
* Jab different datatypes ek saath use hote hain, system khud convert karta hai

👉 Example:
Agar number aur string compare ho rahe ho, DBMS khud convert kar dega

👉 Example:
'10' + 5 → 15  (string automatically number me convert ho gaya)

---

## 2. Explicit Conversion (Manual)

* Ye user khud karta hai
* Jab hume control chahiye hota hai conversion pe

### Common Functions:

### CAST()

* General purpose conversion
  👉 Example:
  CAST('123' AS INT) → 123

### CONVERT()

* Mostly same kaam karta hai (DB specific)
  👉 Example:
  CONVERT(INT, '123') → 123

---

## Quick Summary

* Implicit → automatic conversion (no control)
* Explicit → manual conversion (full control)
* CAST() & CONVERT() → sabse important functions


# Type Conversion in SQL (Datetime Examples)

## 1. Implicit Conversion (Automatic)

* Jab SQL khud automatically datatype convert kar deta hai
* Mostly tab hota hai jab string ko date ke saath use karte hain

👉 Example 1:

```sql
SELECT * FROM orders 
WHERE order_date = '2026-03-21';
```

✔️ Yaha `'2026-03-21'` (string) automatically DATE me convert ho gaya

👉 Example 2:

```sql
SELECT '2026-03-21' + INTERVAL 1 DAY;
```

✔️ String automatically DATE ban gaya aur 1 day add ho gaya → `2026-03-22`

---

## 2. Explicit Conversion (Manual)

* Jab hum khud datatype convert karte hain
* Jab format different ho ya control chahiye

### Using CAST()

👉 Example:

```sql
SELECT CAST('2026-03-21' AS DATE);
```

✔️ String ko DATE me convert kiya

---

### Using STR_TO_DATE()

👉 Example:

```sql
SELECT STR_TO_DATE('21-03-2026', '%d-%m-%Y');
```

✔️ Custom format string ko DATE me convert kiya

---

### Using CONVERT()

👉 Example:

```sql
SELECT CONVERT('2026-03-21', DATE);
```

✔️ String → DATE conversion

---

## Quick Summary

* Implicit → SQL khud convert karta hai (easy but less control)
* Explicit → CAST(), CONVERT(), STR_TO_DATE() use karke manually convert karte hain
* Datetime conversion mostly string ↔ date/time ke beech hota hai


# Datetime Arithmetic and Their Examples

SQL me datetime arithmetic ka use date aur time ke beech difference nikalne ya unhe add/subtract karne ke liye hota hai.

---

## 1. DATEDIFF()

* Do dates ke beech ka difference (days me) nikalta hai

👉 Example:
DATEDIFF('2026-03-21', '2026-03-18') → 3

✔️ Sirf **days** me difference deta hai

---

## 2. TIMEDIFF()

* Do time (ya datetime) ke beech ka difference nikalta hai

👉 Example:
TIMEDIFF('14:30:45', '12:15:30') → 02:15:15

✔️ Result hh:mm:ss format me aata hai

---

## 3. DATE_ADD()

* Date me days/months/years add karta hai

👉 Example:
DATE_ADD('2026-03-21', INTERVAL 5 DAY) → 2026-03-26

👉 Example:
DATE_ADD('2026-03-21', INTERVAL 1 MONTH) → 2026-04-21

---

## 4. DATE_SUB()

* Date se days/months/years subtract karta hai

👉 Example:
DATE_SUB('2026-03-21', INTERVAL 5 DAY) → 2026-03-16

---

## 5. ADDTIME()

* Time ya datetime me time add karta hai

👉 Example:
ADDTIME('10:00:00', '02:30:00') → 12:30:00

---

## 6. SUBTIME()

* Time ya datetime se time subtract karta hai

👉 Example:
SUBTIME('10:00:00', '02:30:00') → 07:30:00

---

## Quick Summary

* DATEDIFF() → dates ke beech difference (days)
* TIMEDIFF() → time/datetime ka difference
* DATE_ADD() / DATE_SUB() → date me interval add/subtract
* ADDTIME() / SUBTIME() → time me addition/subtraction


# TIMESTAMP vs DATETIME in SQL (Detailed Comparison)

Dono date + time store karte hain, but inke internal behavior aur use-case kaafi different hote hain.

---

## 1. Range

### DATETIME

* Range: `1000-01-01 00:00:00` → `9999-12-31 23:59:59`
  ✔️ Wide range (past + future dono cover karta hai)

### TIMESTAMP

* Range: `1970-01-01 00:00:01` → `2038-01-19 03:14:07`
  ❌ Limited range (Unix timestamp limit)

---

## 2. Storage Format

### DATETIME

* Value as-it-is store hoti hai (no conversion)
  ✔️ Human-readable format internally

### TIMESTAMP

* Internally **Unix timestamp (seconds since 1970)** me store hota hai
  ✔️ Compact storage

---

## 3. Behavior in Insertion / Update

### DATETIME

* Jo value insert karoge wahi store hogi
* Timezone ka koi effect nahi
* Auto update by default nahi hota

👉 Example:
Insert: '2026-03-21 10:00:00'
✔️ Same value retrieve hogi

---

### TIMESTAMP

* Insert hote waqt → UTC me convert hota hai
* Retrieve hote waqt → system timezone ke hisaab se convert hota hai

👉 Example:
Insert (India): '2026-03-21 10:00:00'
Retrieve (US): different time dikh sakta hai

✔️ Timezone aware behavior

---

## 4. Precision

### DATETIME

* Default precision: seconds
* Fractional seconds support (DATETIME(3), DATETIME(6))

### TIMESTAMP

* Same fractional precision support karta hai
* TIMESTAMP(6) → microseconds tak

✔️ Dono almost same precision de sakte hain

---

## 5. Auto Update Feature

### DATETIME

* Auto update nahi hota by default
* Manually handle karna padta hai

---

### TIMESTAMP

* Auto-update support karta hai

👉 Example:
created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP ON UPDATE CURRENT_TIMESTAMP

✔️ Insert/update pe automatic change ho jata hai

---

## Final Comparison Table

| Feature           | DATETIME ❌ | TIMESTAMP ✅ |
| ----------------- | ---------- | ----------- |
| Range             | Wide       | Limited     |
| Storage           | Direct     | Unix format |
| Timezone Handling | ❌ No       | ✅ Yes       |
| Insert Behavior   | Same value | Converted   |
| Precision         | Same       | Same        |
| Auto Update       | ❌ No       | ✅ Yes       |

---



## Final Summary (Interview Ready)

* DATETIME → fixed value, no timezone, wide range
* TIMESTAMP → timezone aware, auto-update, limited range
* Logging & tracking → TIMESTAMP best
* General storage → DATETIME better


```sql
CREATE TABLE posts (
    user_id     INT,
    content     TEXT,
    created_at  TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    updated_at  TIMESTAMP DEFAULT CURRENT_TIMESTAMP 
                ON UPDATE CURRENT_TIMESTAMP
);

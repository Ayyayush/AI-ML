# Ranking Window Function

# A window function used to assign ranks to rows within a partition
# Ranking is done based on a specific column (like marks, salary, score)

# Common Ranking Functions:
# 1. RANK()        → Skips ranks when there are ties
# 2. DENSE_RANK()  → Does NOT skip ranks when there are ties
# 3. ROW_NUMBER()  → Gives unique sequential numbers (no ties)

# Used with:
# OVER(PARTITION BY column ORDER BY column)

# Example Use Case:
# Ranking students based on marks within each class

# cumulative sum# Cumulative Sum

# Calculates the running total of values in a column
# Each value is the sum of the current value and all previous values
# Order of data matters for cumulative sum

# Commonly used to:
# - Track running totals (sales, revenue, score)
# - Analyze growth over time
# - Monitor progress

# Examples:
# SQL      → SUM(column) OVER(ORDER BY column)
# Pandas   → df['column'].cumsum()

# Cumulative Average

# Calculates the running average of values in a column
# Each value is the average of all values up to that point
# Order of data matters for cumulative average

# Commonly used to:
# - Track performance trends over time
# - Smooth out fluctuations in data
# - Monitor long-term behavior

# Examples:
# SQL      → AVG(column) OVER(ORDER BY column)
# Pandas   → df['column'].expanding().mean()

# Running Average (Moving Average)

# Calculates the average of a fixed number of recent values
# Unlike cumulative average, it does NOT include all previous values
# Uses a sliding window of size 'k'

# Helps to:
# - Smooth short-term fluctuations
# - Identify trends in time-series data
# - Reduce noise in data

# Examples:
# SQL      → AVG(column) OVER(ORDER BY column ROWS BETWEEN k-1 PRECEDING AND CURRENT ROW)
# Pandas   → df['column'].rolling(window=k).mean()

# Percent of Total

# Calculates how much each value contributes to the overall total
# Expressed as a percentage (%)
# Helps compare individual values against the whole

# Commonly used to:
# - Analyze contribution (sales by product, revenue by region)
# - Identify dominant categories
# - Create pie charts or stacked charts

# Examples:
# SQL      → (value / SUM(value)) * 100
# Pandas   → (df['column'] / df['column'].sum()) * 100

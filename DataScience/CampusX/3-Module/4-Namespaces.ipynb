# Namespaces in Python
A **namespace** ek aisi "space" hoti hai jo names (identifiers — variables, functions, classes, etc.) ko store karti hai.  
Programmatically bolen to: *identifiers ka ek dictionary* jahan key name (string) aur value us name ka object hota hai.

Example simple sentence mein:  
`x = 10` — yahan `x` ek name hai jo kisi namespace (jaise module/global) mein `10` se map hota hai.

---

# Types of Namespaces
Python mein kuch common namespaces hote hain:

- **Built-in Namespace**  
  Python interpreter start hote hi load hota hai. Isme built-in functions aur exceptions hote hain (`len`, `print`, `Exception`, ...).

- **Global (Module) Namespace**  
  Har Python file/module ka apna namespace hota hai — jo module-level variables aur functions rakhta hai. Jab aap `import` karte ho ya top-level pe kuch define karte ho, woh global namespace mein chala jaata hai.

- **Enclosing (Nonlocal) Namespace**  
  Agar aap nested functions use karte ho to outer function ki local namespace inner function ke liye enclosing namespace hoti hai.

- **Local Namespace**  
  Har function call apna local namespace create karta hai (function ke arguments aur local variables wahan store hote hain). Function return hone par ye namespace typically destroy ho jaata hai.

> Note: Ye "types" conceptual hain — implementation wise har namespace ek mapping (jaise dictionary) hai jo names ko objects se map karta hai.

---

# Scope and LEGB Rule
**Scope** ka matlab hota hai: "kahan se koi name accessible hai". Python mein name lookup ka order **LEGB** rule follow karta hai:
it is a textual region of python program where namespace is directly accessible


- **L — Local** : Current function/local scope
- **E — Enclosing** : Enclosing function scopes (for nested functions)
- **G — Global (module)** : Module level scope
- **B — Built-in** : Built-in names provided by Python


Matlab jab Python koi name resolve karta hai, to pehle Local check karta hai, agar nahi mila to Enclosing, fir Global, aur akhir mein Built-in.




````python
# local and global
a=2  # it is global

def temp():
    b=3   # it is in local scope 
    print(b)
    
temp()
print(a)
````

````python
# local and global same name 
a=2              # it is global

def temp():
    a=3         # it is in local scope toh name samne bhi h toh koi dikkat nhi kyunki alag scope h 
    print(a)
    
temp()
print(a)
````

````python

## Example — basic LEGB demo
# demo.py    (module / global scope)
x = 'global x'

def outer():
    x = 'enclosing x'
    def inner():
        x = 'local x'
        print('inner:', x)   # local -> prints 'local x'
    inner()
    print('outer:', x)       # enclosing -> prints 'enclosing x'

outer()
print('module:', x)          # global -> prints 'global x'
````

````python
# ---------------------------------------------
# LOCAL vs GLOBAL VARIABLES (LEGB Rule Example)
# ---------------------------------------------
# L -> Local
# E -> Enclosing
# G -> Global
# B -> Built-in
# Python searches variables in this order
# ---------------------------------------------

# Global variable
a = 2                          # Defined outside any function (GLOBAL scope)


def temp():
    # No local variable named 'a' is defined here
    # Python first searches in LOCAL scope → not found
    # Then searches in GLOBAL scope → found
    print(a)                   # Accesses global variable 'a'


# Function call
temp()                         # Output: 2

# Accessing global variable directly
print(a)                       # Output: 2


# ---------------------------------------------
# IMPORTANT NOTE
# ---------------------------------------------
# If a variable is NOT found in local scope,
# Python looks for it in the GLOBAL scope.
# This is a simple demonstration of the LEGB rule.
````

````python
# ---------------------------------------------
# LOCAL vs GLOBAL (Editing Global Variable)
# ---------------------------------------------
# This example shows that:
# - Global variables can be READ inside a function
# - Global variables CANNOT be MODIFIED directly
#   inside a function without using the 'global' keyword
# ---------------------------------------------

# Global variable
a = 2


def temp():
    # Python assumes 'a' is a LOCAL variable
    # because we are trying to modify it (a += 1)
    a += 1                     # ❌ Error occurs here
    print(a)


# Function call
temp()

# Print global variable
print(a)
````

````python
# ---------------------------------------------
# LOCAL vs GLOBAL → Modifying Global Variable
# ---------------------------------------------
# This example demonstrates how a global variable
# can be MODIFIED inside a function using the
# 'global' keyword.
# ---------------------------------------------

# Global variable
a = 2                          # Defined outside function (GLOBAL scope)


def temp():
    # By declaring 'a' as global, we tell Python
    # that we want to refer to the GLOBAL variable,
    # not create a new local one
    global a

    a += 1                     # Modifying global variable
    print(a)                   # Prints updated value of 'a'


# Function call
temp()                         # Output inside function → 3

# Accessing global variable after modification
print(a)                       # Output → 3


# ---------------------------------------------
# IMPORTANT NOTES
# ---------------------------------------------
# - Global variables can be READ inside a function
# - To WRITE / MODIFY a global variable, we MUST
#   use the 'global' keyword
# - Using global variables is generally NOT
#   recommended because it can lead to
#   unexpected side effects and hard-to-debug code
````

````python
# ---------------------------------------------
# LOCAL vs GLOBAL → Overwriting Global Variable
# ---------------------------------------------
# This example shows that a GLOBAL variable
# can be CREATED / OVERWRITTEN inside a function
# using the 'global' keyword.
# ---------------------------------------------

# Global variable
a = 2                          # Initial global value


def temp():
    # Declaring 'a' as global means:
    # - We are NOT creating a local variable
    # - We are directly accessing the global 'a'
    global a

    a = 1                      # Overwriting the global variable
    print(a)                   # Prints updated global value


# Function call
temp()                         # Output inside function → 1

# Printing global variable after function execution
print(a)                       # Output → 1


# ---------------------------------------------
# IMPORTANT NOTES
# ---------------------------------------------
# - 'global' allows creation or overwriting
#   of global variables from inside a function
# - This affects the entire program state
# - Such practice is NOT recommended
#   because it can cause unexpected behavior
#   and makes code difficult to debug
````

````python
# ---------------------------------------------
# LOCAL vs GLOBAL → Function Parameter is LOCAL
# ---------------------------------------------
# Function parameters are ALWAYS local
# to the function in which they are defined.
# ---------------------------------------------

def temp(z):
    # 'z' is a LOCAL variable
    # It exists ONLY inside this function
    print(z)                  # Prints value passed to the function


# Global variable
a = 5

# Function call
temp(5)                       # Output → 5

# Accessing global variable
print(a)                      # Output → 5

# Trying to access function parameter outside
print(z)                      # ❌ ERROR
````

# Built-in Scope in Python
**Built-in scope** wo namespace hota hai jo Python interpreter ke start hote hi automatically load hota hai.  
Is scope mein Python ke **built-in functions**, **exceptions**, aur kuch **built-in constants** hote hain jo aap directly use kar sakte ho bina `import` kiye.

Example — aap directly `len()`, `print()`, `sum()`, `max()`, `min()` ya exceptions jaise `ValueError`, `TypeError` use kar sakte ho kyunki ye sab built-in scope ka hissa hain.

⚡ **Important Points**:
- Built-in scope har Python program ke liye by default available hota hai.
- Ye scope **LEGB rule** ke last step me check hota hai (Local → Enclosing → Global → **Built-in**).
- Built-in names Python ke ek special module `builtins` me store hote hain.

---


````python
# ---------------------------------------------
# How to See All the Built-in Names in Python
# ---------------------------------------------
# Python provides many built-in functions,
# constants, and exceptions that are always
# available without importing anything.
# ---------------------------------------------


# ---------------------------------------------
# 1. Using dir(__builtins__)
# ---------------------------------------------
# __builtins__ contains all built-in names.
# dir() returns a list of all attributes.

print(dir(__builtins__))



# ---------------------------------------------
# 2. Using builtins module
# ---------------------------------------------
# builtins module explicitly contains
# all Python built-in names.

import builtins
print(dir(builtins))



# ---------------------------------------------
# 3. Filtering Built-ins
# ---------------------------------------------
# We can filter built-ins to view only
# specific types like functions or constants.


import builtins

# ---------------------------------------------
# Filter only built-in functions
# ---------------------------------------------
# callable() checks whether an object
# can be called like a function

functions = [
    name for name in dir(builtins)
    if callable(getattr(builtins, name))
]


print("Built-in Functions:")
print(functions)


# ---------------------------------------------
# Filter only built-in constants
# ---------------------------------------------
# Built-in constants are usually
# written in UPPERCASE (e.g., True, False, None)

constants = [
    name for name in dir(builtins)
    if name.isupper()
]

print("Built-in Constants:")
print(constants)
````

# Renaming built in's in Python

 Built-ins ko rename karna technically possible hai — lekin aksar **unsafe** aur **confusing** hota hai. Yaha pe examples, warnings, aur best practices diye gaye hain.

---

## 1. Kya matlab hai "renaming a built-in"?

Python ke **built-in functions** (jaise `len`, `sum`, `open`, `list`, `str`) global names hote hain jo har program me available rehte hain. "Renaming" ka matlab hai ki hum unko **ek naye naam** (alias) ke through access karein, ya un existing naam ko override/shadow kar dein.

```python
# alias bana ke built-in use karna
orig_len = len
print(orig_len([1,2,3]))  # 3


# lekin direct override karna (shadowing):
len = lambda x: 42
print(len([1,2,3]))  # 42  -> built-in behaviour lost
```



## 2. Builtins ko alias karna (safe-ish)

Agar tumhe temporary reference chahiye to original behaviour ko preserve karne ke liye, toh alias bana sakte ho.

```python
import builtins

# safe alias 
my_open = builtins.open
f = my_open('file.txt', 'r')
```

Is tarah se `open` ki original functionality preserved rahegi, aur tum `open` ko override kar sako.



## 3. Shadowing / Overriding built-ins (bad idea)

Global scope me directly `len = ...` ya `list = ...` jaisa kuch karna recommended nahi hai:

* Dusre developers confuse honge.
* Kuch libraries ya built-in operations break ho sakte hain.
* Bugs dhundhna mushkil ho jata hai.

```python
list = (1,2,3)   # ab list ek tuple hai, not the built-in type
x = list([4,5,6])  # TypeError: 'tuple' object is not callable
```



## 4. Kaise check karein agar koi name built-in ko shadow kar raha hai?

```python
import builtins

def is_shadowing(name):
    return name in globals() and getattr(builtins, name, None) is not globals()[name]

# usage
len = 5
print(is_shadowing('len'))  # True
```

## 5. Builtins module use karna (recommended for explicit access)

`builtins` module se direct access lena most explicit aur safe approach hai.

```python
import builtins

# agar kisi reason se len override ho gaya hai:
len = 100
print(len)                  # 100
print(builtins.len([1,2]))  # 2  -> original function
```

## 6. Temporarily replace a built-in (context manager pattern)

Agar testing me temporary replacement chahiye ho, toh context manager bana lo.

```python
import builtins
from contextlib import contextmanager

@contextmanager
def replace_builtin(name, new):
    orig = getattr(builtins, name)
    setattr(builtins, name, new)
    try:
        yield
    finally:
        setattr(builtins, name, orig)

# usage
with replace_builtin('open', lambda *a, **k: None):
    # inside this block builtins.open replaced
    pass
# outside block original restored
```

## 7. Using `__builtins__` (CPython specific caveat)

Har module ke globals me `__builtins__` hota hai — lekin iska use recommended nahi hai for renaming. Iska behaviour interpreter/implementation ke hisaab se vary kar sakta hai.

```python
# rarely recommended
print(__builtins__.get('len'))
```

## 8. Restore a shadowed built-in

Agar galti se override ho gaya, toh `builtins` se restore kar sakte ho.

```python
import builtins
len = 42
# restore
len = builtins.len
```

## 9. Best practices / Recommendations (Hinglish)

* **Avoid** direct renaming/shadowing of built-ins. Agar zarurat ho toh:

  * Use a different variable name (e.g., `count` instead of `len`).
  * Ya `import builtins` aur `builtins.len` use karo when you need the original.
* Tests ke liye temporary replace karne ke liye context manager ya monkeypatching tools (pytest fixtures) use karo.
* Agar library bana rahe ho, user-facing API me built-ins ko override mat karo.

## 10. Kaise dekhen saare built-ins? (quick)

```python
import builtins
print(sorted(dir(builtins)))
```

````python
# len() ko bina import kiye call kar sakte ho
data = [1, 2, 3]
print(len(data))   # Output: 3

# Agar same name ka local variable banate ho, to built-in shadow ho jaata hai
len = 100
print(len)         # Output: 100
# print(len(data)) # ❌ Error — 'int' object is not callable

del len            # shadowed name hata diya
print(len(data))   # Output: 3 (built-in wapas accessible)
````

````python
# ---------------------------------------------
# ENCLOSING SCOPE (LEGB Rule Example)
# ---------------------------------------------
# Enclosing scope refers to variables defined
# in an outer function but accessed inside
# an inner (nested) function.
# ---------------------------------------------

def outer():
    a = 3                         # 'a' is in ENCLOSING scope for inner()

    def inner():
        print('inner function')

        # This assignment creates a NEW LOCAL variable 'a'
        # It does NOT modify the enclosing 'a'
        a = 4                     # Local variable inside inner()
        print(a)                  # Prints local 'a' (4)

    inner()                        # Call inner function
    print('outer function')       # No change to enclosing 'a'


# Global variable
a = 1

# Function call
outer()

print('main program')             # Execution continues in global scope
````

````python
# ---------------------------------------------
# NONLOCAL KEYWORD (Enclosing Scope Modification)
# ---------------------------------------------
# nonlocal is used to modify a variable that
# belongs to the ENCLOSING scope (outer function),
# not the local or global scope.
# ---------------------------------------------

def outer():
    a = 1                          # 'a' belongs to ENCLOSING scope

    def inner():
        # nonlocal tells Python that 'a' is NOT local
        # and should be taken from the enclosing scope
        nonlocal a

        a += 1                     # Modifies enclosing variable
        print('inner', a)          # Prints updated value

    inner()                        # Call inner function
    print('outer', a)              # Shows updated enclosing value


# Function call
outer()

# Global scope execution
print('main program')
````

````python
# ---------------------------------------------
# GLOBAL vs ENCLOSING vs NONLOCAL EXAMPLE
# ---------------------------------------------
# This example demonstrates:
# - How enclosing variables work
# - How nonlocal modifies enclosing scope
# - How global modifies global scope
# ---------------------------------------------

count = 0                     # GLOBAL variable


def outer():
    count = 5                 # ENCLOSING variable for inner()

    def inner():
        # nonlocal tells Python that 'count'
        # belongs to the ENCLOSING scope (outer),
        # not local and not global
        nonlocal count

        count += 1            # Modifies enclosing 'count'
        print('inner count:', count)

    inner()                   # Call inner function
    print('after inner, outer count:', count)   # Enclosing value updated


# Call outer function
outer()

# Global 'count' remains unchanged
print('global count:', count)


# ---------------------------------------------
# USING GLOBAL KEYWORD
# ---------------------------------------------

def modify_global():
    # global tells Python that we want to
    # modify the GLOBAL variable 'count'
    global count

    count += 10               # Modifies global 'count'


# Call function that modifies global variable
modify_global()

# Global value is now changed
print('global after modify:', count)
````

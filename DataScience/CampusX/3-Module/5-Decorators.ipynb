# Decorators in Python  

**Definition:**  
Decorators in Python are special functions that allow you to modify or extend the behavior of other functions or methods **without changing their actual code**.  
They are often used for logging, access control, performance measurement, and more.  

A decorator **wraps** another function and can execute code before or after the wrapped function runs.  
They are applied using the `@decorator_name` syntax above the function definition.

---

# Types of Decorators  

1. **Function Decorators**  
   - Used to modify the behavior of normal functions.  
   - Most common type.  

2. **Class Decorators**  
   - Used to modify or extend the behavior of classes.  

3. **Built-in Decorators**  
   - Examples: `@staticmethod`, `@classmethod`, `@property`  

4. **Parameterized Decorators**  
   - Accept arguments to customize their behavior.  

---


```python
# ---------------------------------------------
# DECORATOR: MULTIPLE ARGUMENTS
# ---------------------------------------------

def my_decorator(func):
    def wrapper(*args):
        print("Arguments received:")
        for arg in args:
            print(arg)
        func(*args)                   # Call original function
    return wrapper


@my_decorator
def hello(a, b, c, d):
    print("i am ayush")


hello(1, 2, 3, 4)
```

```python
# ---------------------------------------------
# FUNCTION DECORATOR EXAMPLE
# ---------------------------------------------
# A decorator is a function that:
# - takes another function as input
# - adds extra behavior
# - returns a new function
# without modifying the original function code
# ---------------------------------------------

def my_decorator(func):
    """
    This is the DECORATOR function.
    It receives another function as argument.
    """

    def wrapper():
        """
        This is the WRAPPER function.
        It adds extra code before and after
        the original function call.
        """
        print("Before the function runs")   # Code before original function
        func()                              # Call original function
        print("After the function runs")    # Code after original function

    return wrapper                          # Return modified function


@my_decorator                               # Decorator applied to say_hello()
def say_hello():
    """
    Original function that prints a message.
    """
    print("Hello, World!")


# Function call
say_hello()
```

```python
# ---------------------------------------------
# FUNCTIONS ARE FIRST-CLASS CITIZENS IN PYTHON
# ---------------------------------------------
# In Python, functions are treated like normal
# objects. This means:
# 1. Functions can be assigned to variables
# 2. Functions can be passed as arguments
# 3. Functions can be returned from other functions
# ---------------------------------------------


# ---------------------------------------------
# 1. FUNCTION ASSIGNED TO A VARIABLE
# ---------------------------------------------

def func():                            # Defining a simple function
    print('hello')

a = func                               # Assigning function reference to variable 'a'
a()                                    # Calls func() → prints "hello"


# ---------------------------------------------
# 2. DELETING ORIGINAL FUNCTION NAME
# ---------------------------------------------

del func                               # Deletes name 'func' from namespace

# func()                               # ❌ ERROR: NameError (func no longer exists)

a()                                    # ✅ Works because 'a' still references the function


# ---------------------------------------------
# 3. PASSING FUNCTION AS ARGUMENT
# ---------------------------------------------

def modify(func, num):
    """
    Takes a function and a number,
    applies the function to the number,
    and returns the result.
    """
    return func(num)


def square(num):
    return num ** 2


# Function call: passing 'square' as argument
result = modify(square, 2)
print(result)                          # Output → 4
```

```python
# ---------------------------------------------
# EXAMPLE 2: BEHAVIOR BEHIND DECORATORS
# ---------------------------------------------
# This example demonstrates how decorators work
# internally by passing a function as an argument
# to another function.
# ---------------------------------------------

def modify2(func, num):
    """
    This function accepts:
    - func → another function
    - num  → an argument for that function

    It then calls the passed function with 'num'
    and returns the result.
    """
    return func(num)                      # Executes the function 'func'


def square(num):
    """
    Function that returns square of a number.
    """
    return num ** 2


# ---------------------------------------------
# FUNCTION CALL
# ---------------------------------------------
# Passing 'square' function as an argument
# to modify2 along with value 2

result = modify2(square, 2)

# Display the result
print(result)                             # Output → 4
```

```python
# ---------------------------------------------
# SIMPLE DECORATOR EXAMPLE (WITHOUT @ SYNTAX)
# ---------------------------------------------
# This example demonstrates:
# - How decorators work internally
# - How functions are passed and wrapped
# - How CLOSURE is used
# ---------------------------------------------

def my_decorator(func):
    # 'my_decorator' is a function that takes
    # another function as an argument

    def wrapper():
        # 'wrapper' is a NESTED function
        # It wraps the original function behavior

        print('********')      # Code BEFORE original function call
        func()                 # Call the original function
        print('********')      # Code AFTER original function call

    return wrapper             # Return the wrapper function


# ---------------------------------------------
# ORIGINAL FUNCTIONS
# ---------------------------------------------

def hello():
    # Simple function
    print('hello')


def display():
    # Another simple function
    print('hello ayush')


# ---------------------------------------------
# APPLYING DECORATOR MANUALLY
# ---------------------------------------------

a = my_decorator(hello)
# 'hello' is passed to my_decorator
# 'a' now holds reference to 'wrapper'

a()                             # Calls wrapper(), not hello()


b = my_decorator(display)
# 'display' is passed to my_decorator
# 'b' now holds reference to 'wrapper'

b()                             # Calls wrapper(), not display()
```

```python
# ---------------------------------------------
# BETTER DECORATOR SYNTAX (Using @ symbol)
# ---------------------------------------------
# The @ syntax is just a cleaner and more
# readable way to apply decorators.
# ---------------------------------------------

def my_decorator(func):
    # Decorator function that receives
    # another function as argument

    def wrapper():
        # Wrapper function adds extra behavior
        print('***********************')   # Code BEFORE original function
        func()                           # Call original function
        print('***********************')   # Code AFTER original function

    return wrapper                       # Return wrapped function


@my_decorator                             # Better syntax for decoration
def hello():
    # Original function
    print('hello')


# Function call
hello()                                  # Actually calls wrapper()
```

```python
# ---------------------------------------------
# MEANINGFUL DECORATOR EXAMPLE: TIMER DECORATOR
# ---------------------------------------------
# This decorator measures the execution time
# of any function it decorates.
# ---------------------------------------------

import time                          # Used to measure time


def timer(func):
    """
    Decorator function that calculates
    the time taken by another function
    to execute.
    """

    def wrapper(*args):
        # *args allows wrapper to accept
        # any number of positional arguments

        start = time.time()          # Record start time

        func(*args)                  # Call the original function

        end = time.time()            # Record end time

        # Print execution time with function name
        print(
            'time taken by',
            func.__name__,
            end - start,
            'secs'
        )

    return wrapper                   # Return wrapped function


# ---------------------------------------------
# FUNCTIONS USING @timer DECORATOR
# ---------------------------------------------

@timer
def hello():
    print('hello world')
    time.sleep(2)                    # Simulates slow task


@timer
def square(num):
    time.sleep(1)                    # Simulates computation delay
    print(num ** 2)


@timer
def power(a, b):
    print(a ** b)


# ---------------------------------------------
# FUNCTION CALLS
# ---------------------------------------------

hello()
square(2)
power(2, 3)
```

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)   
```

```python
import json

# Python objects
l = [1, 2, 3, 4]

d = {
    'name': 'ayush',
    'age': 33
}

t = (1, 2, 3, 4)          # Tuple (JSON me list ban jaayega)

# Sabko ek hi dictionary me wrap kiya
data = {
    'list_data': l,
    'dict_data': d,
    'tuple_data': t       # Automatically list me convert hoga
}

# ---------------- SERIALIZATION ----------------
with open('demo.json', 'w') as f:
    json.dump(data, f, indent=4)

# ---------------- DESERIALIZATION ----------------
with open('demo.json', 'r') as f:
    new_data = json.load(f)

# Output
print(new_data)
print(type(new_data))
```

```python
# SERIALIZING AND DESERIALIZING CUSTOM OBJECTS

import json     # ✅ Import json globally (class ke andar nahi)

# ✅ Define your custom class
class Person:
    def __init__(self, fname, lname, age, gender):
        self.fname = fname                          # First name
        self.lname = lname                          # Last name
        self.age = age                              # Age
        self.gender = gender                        # Gender

    # ✅ Human-readable format mein convert karne ke liye string representation override
    def __str__(self):
        return f"-{self.fname} {self.lname} age - {self.age} gender - {self.gender}"

# ✅ Custom serializer function
def show_object(obj):
    if isinstance(obj, Person):                     # Agar object Person class ka hai
        return {
            'fname': obj.fname,
            'lname': obj.lname,
            'age': obj.age,
            'gender': obj.gender
        }
    raise TypeError("Object of type Person expected")

# ✅ Create an object
person = Person("Nitish", "Singh", 33, "male")

# ✅ Serialize the object into JSON
with open('demo.json', 'w') as f:
    json.dump(person, f, default=show_object, indent=4)   # default=show_object se bataya ki kaise convert karna hai

# ✅ Deserialize (JSON -> Python object)
with open('demo.json', 'r') as f:
    data = json.load(f)

# ✅ Convert dict back to Person object manually
new_person = Person(**data)

print(new_person)     # Output: -Nitish Singh age - 33 gender - male
```

# Pickling 
# ----------------------------------------------
# Pickling ka matlab hota hai: Python object ko bytes (binary format) mein convert karna
# Taaki hum us object ko kisi file mein store kar sakein ya network ke through bhej sakein
# Yeh ek tarika hai data ko persist karne ka (save for later use)


```python
import pickle     # ✅ Python ka built-in module for pickling

# ✅ Example object
data = {
    'name': 'Nitish',
    'age': 33,
    'skills': ['Python', 'React', 'SQL']
}


# ✅ Pickling: Save the object into a binary file
with open('data.pkl', 'wb') as f:              # wb = write binary
    pickle.dump(data, f)                       # data ko binary format mein convert karke file mein store kiya

print("✅ Object has been pickled and saved into 'data.pkl'")


```

```python


# Unpickling
# ----------------------------------------------
# Unpickling ka matlab hota hai: Bytes (binary) file ko wapas Python object mein convert karna
# Jab hume apne stored ya received object ko wapas Python mein use karna ho

# ✅ Unpickling: Load the object from the binary file
with open('data.pkl', 'rb') as f:              # rb = read binary
    loaded_data = pickle.load(f)               # Binary file se Python object wapas ban gaya

print("✅ Unpickled object:", loaded_data)     # Output dikha raha hai original object
```

```python
# -------------------------------
# Person class define ki
# -------------------------------
class Person:
    def __init__(self, name, age):
        self.name = name                  # Person ka name store hota hai
        self.age = age                    # Person ki age store hoti hai

    def display_info(self):
        # Object ki information print karta hai
        print('Hi my name is', self.name, 'and I am', self.age, 'years old')


# -------------------------------
# Person class ka object banaya
# -------------------------------
p = Person('nitish', 33)


# -------------------------------
# pickle module import kiya
# -------------------------------
import pickle


# -------------------------------
# Object ko binary file me store karna (Serialization)
# -------------------------------
with open('person.pkl', 'wb') as f:
    # pickle.dump() Python object ko
    # binary format me convert karke file me store karta hai
    pickle.dump(p, f)
```

# Pickle vs JSON
# -------------------------------------------------------

# ✅ 1. Format:
# Pickle  -> Converts object into binary format (non-human-readable)
# JSON    -> Converts object into text format (human-readable)

# ✅ 2. Language Support:
# Pickle  -> Python-specific only (can't be used with Java, C++, etc.)
# JSON    -> Language-independent (can be shared between Python, JavaScript, Java, etc.)

# ✅ 3. Supported Data Types:
# Pickle  -> Supports almost all Python objects (sets, tuples, classes, etc.)
# JSON    -> Only supports basic data types (dict, list, str, int, float, bool, None)

# ✅ 4. File Size:
# Pickle  -> Usually more compact and faster to serialize/deserialize
# JSON    -> Bigger file size due to text format

# ✅ 5. Security:
# Pickle  -> ❌ Unsafe to load data from untrusted sources (can execute malicious code)
# JSON    -> ✅ Much safer to load as it's just text and doesn’t execute code

# ✅ 6. Readability:
# Pickle  -> ❌ Not human-readable
# JSON    -> ✅ Easily readable and editable

# ✅ 7. Use Cases:
# Pickle  -> Best for Python-only applications, internal storage
# JSON    -> Best for APIs, data sharing across platforms, web applications

# ✅ 8. Speed:
# Pickle  -> Generally faster for complex Python objects
# JSON    -> Slower but more interoperable

# ✅ 9. Syntax Example:

import pickle, json

# Example Python object
data = {'name': 'Nitish', 'skills': ['Python', 'SQL'], 'age': 33}

# Pickle Serialization
with open('data.pkl', 'wb') as f:
    pickle.dump(data, f)

# JSON Serialization
with open('data.json', 'w') as f:
    json.dump(data, f, indent=4)

# -------------------------------------------------------
# In Summary:

# ➤ Use Pickle:
#   ✔ When working only within Python
#   ✔ When you need to store complex objects
#   ✘ Never use with untrusted sources

# ➤ Use JSON:
#   ✔ When you want cross-language compatibility
#   ✔ When sharing data between server and web/client
#   ✔ When readability is important

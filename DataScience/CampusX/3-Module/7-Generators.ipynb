## What is a Generator

A **Generator** in Python is a special type of function used to create **iterators** in a simple and memory-efficient way.

- Generators produce values **one at a time**
- They do **not store all values in memory**
- They use the `yield` keyword instead of `return`

---

## Key Characteristics

- Automatically implements the iterator protocol
- Remembers its state between iterations
- Stops execution when there are no more values
- Raises `StopIteration` when finished

---

## How Generator Works

- Function execution pauses at `yield`
- Value is returned to the caller
- On next iteration, execution resumes **from where it stopped**

---

## Generator vs Normal Function

| Normal Function | Generator |
|-----------------|-----------|
| Uses `return`   | Uses `yield` |
| Ends execution | Pauses execution |
| Returns once   | Returns multiple values |
| Stores values  | Generates values on demand |

---

## Why Use Generators?

- Memory efficient
- Ideal for large datasets
- Cleaner and readable code
- Faster startup time

---

## Summary

- A generator is an **iterator created using a function**
- Uses `yield` to produce values
- Generates values **lazily**
- Best for handling large or infinite sequences


```python
# iterable
class mera_range:
    
    def __init__(self,start,end):
        self.start = start
        self.end = end
        
    def __iter__(self):
        return mera_iterator(self)
    

# iterator
class mera_iterator:
    
    def __init__(self,iterable_obj):
        self.iterable = iterable_obj
    
    def __iter__(self):
        return self
    
    def __next__(self):
        
        if self.iterable.start >= self.iterable.end:
            raise StopIteration
        
        current = self.iterable.start
        self.iterable.start+=1
        return current
```

## A Simple Example

```python
def gen_demo():
    # This is a generator function
    # Presence of 'yield' makes it a generator, not a normal function

    yield "first statement"     # Execution pauses here and value is returned
    yield "second statement"    # Resumes from here on next request
    yield "third statement"     # Resumes again and yields last value


# Creating a generator object
# NOTE: Function body does NOT execute yet
gen = gen_demo()


# print(next(gen))
# print(next(gen))

# Iterating over generator using for loop
# for loop internally uses iter() and next()
for i in gen:
    print(i)
```

## Example 2

```python
def square(num):
    # Generator function that yields square of numbers
    # from 1 to num (inclusive)
    
    for i in range(1, num + 1):
        yield i ** 2          # Pause here and return square


# Creating generator object
# Function body does NOT execute yet
gen = square(10)


# Using next() manually
print(next(gen))   # 1^2 = 1
print(next(gen))   # 2^2 = 4
print(next(gen))   # 3^2 = 9


# Continuing iteration using for loop
# Generator resumes from where it was paused
for i in gen:
    print(i)
```

## Range Function using Generator

```python
def mera_range(start, end):
    # This is a generator function
    # It yields numbers from start to end-1 (like range)
    
    for i in range(start, end):
        yield i            # Pause execution and return i


# Using the generator
for i in mera_range(15, 26):
    print(i)
```

## Generator Expression

```python
# LIST COMPREHENSION
# This creates a LIST that stores squares of numbers from 1 to 100
# ALL values are computed and stored in memory immediately
L = [i**2 for i in range(1, 101)]


# GENERATOR EXPRESSION
# This creates a GENERATOR object
# Values are NOT stored in memory
# Squares are generated one-by-one when needed
gen = (i**2 for i in range(1, 101))


# Iterating over generator
# for loop internally uses iter() and next()
for i in gen:
    print(i)
```

## Practical Example

```python
import os
import cv2

def image_data_reader(folder_path):
    """
    Generator function that reads images from a folder
    and yields one image array at a time
    """

    # Loop through all files in the given folder
    for file in os.listdir(folder_path):
        
        # Create full path of the image file
        full_path = os.path.join(folder_path, file)
        
        # Read image using OpenCV (returns NumPy array)
        img_array = cv2.imread(full_path)
        
        # Yield one image at a time (lazy loading)
        yield img_array


# ----------------------------
# Creating generator object
# ----------------------------

gen = image_data_reader('C:/Users/91842/emotion-detector/train/Sad')


# Fetching images one by one using next()
img1 = next(gen)
img2 = next(gen)
img3 = next(gen)
img4 = next(gen)
```

## Benefits of Using a Generator

Generators provide an efficient and elegant way to work with data in Python, especially when dealing with large or continuous sequences.

---

## 1. Memory Efficient

- Generates values **one at a time**
- Does not store the entire sequence in memory
- Ideal for large datasets or infinite streams

---

## 2. Faster Startup Time

- Execution starts immediately
- Values are produced only when needed
- No upfront computation cost

---

## 3. Automatic Iterator Behavior

- Automatically implements `__iter__()` and `__next__()`
- Can be used directly in `for` loops

---

## 4. State Preservation

- Remembers local variables between yields
- Execution resumes from the last `yield`

---

## 5. Cleaner and Readable Code

- Replaces complex iterator classes
- Less boilerplate code
- Easier to understand logic

---

## 6. Handles Infinite Sequences

- Can generate endless data safely
- Stops only when you decide

---

## 7. Better Performance for Streaming Data

- Processes data lazily
- Works well with file reading, logs, network streams

---

#### 3. Representing Infinite Streams

```python
def all_even():
    n = 0
    while True:
        yield n
        n += 2
```

```python
even_num_gen = all_even()
next(even_num_gen)
next(even_num_gen)
```

#### 4. Chaining Generators

```python
def fibonacci_numbers(nums):
    x, y = 0, 1
    for _ in range(nums):
        x, y = y, x+y
        yield x

def square(nums):
    for num in nums:
        yield num**2

print(sum(square(fibonacci_numbers(10))))
```

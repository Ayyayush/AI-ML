```python
# Numpy

import numpy as np

## create array using numpy
## create a 1D array 
arr1=np.array([1,2,3,4,5])
print(arr1)
print(type(arr1))

print(arr1.shape)
arr1.mean()

# reshaping to one row and 5 columns
arr1.reshape(1,5)
```

```python
np.ones((3,4))
np.zeros((3,4))

# dimension based 
np.zeros_like((1))
np.zeros_like((1,2))
np.zeros_like((3,4))
np.zeros_like((1,2,3))


# identity matrix methods
np.eye(3)
np.identity(4)
```

```python
# numpy vectorized operations
arr4=np.array([1,2,3,4,5])
arr5=np.array([10,20,30,40,50])


print(arr4+arr5)
print(arr4-arr5)
print(arr4*arr5)
print(arr4/arr5)
print(arr4+arr5)
```

```python
# Universal functions
# functions that  apply to  the entire  array

print(np.sqrt(arr4))
print(np.exp(arr4))
print(np.log(arr4))
print(np.argmax(arr4))
print(np.max(arr4))
print(np.absolute(arr4))

# this will give error
# print(np.dtype(arr4))
```

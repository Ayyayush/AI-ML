```python
# Dataframes


# creating from dictioanry of dictionaries
data={
    'name':['ayush','aryan'],
    'age':[12,15]
}
df=pd.DataFrame(data)
print(df)




# from a list of dictionaries
data=[
    {'name':'ayush','age' : 15},
    {'name':'aryan','age':12}
]

df=pd.DataFrame(data)
df
```

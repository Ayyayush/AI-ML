```python
# Tuples
# they are ordered collection of items that are immutable
```

```python
# creating a tuple
empty_tuple=()
lst=list()
tpl=tuple()
print(type(tpl))
print(type(lst))
```

```python
numbers=tuple([1,2,3,4,5,6])
numbers

# tuple can be converted to tuple, tuple can be converted to list

mixed_tuple=(1,"hello","ayush",5.6)
```

```python
# accessing tuple elements 
numbers[0]
```

```python
# tuple operations
concat=numbers+mixed_tuple
concat
```

```python
new_mixed=mixed_tuple*3
new_mixed
```

```python
# immutable nature of tuples
```

```python
# tuple methods
numbers.index(3)
numbers.count(1)
```

```python
# packing and unpacking tuples
packed=1,"ayush",5.6
print(packed)

a,b,c=packed
print(a,b,c)
```

```python
# unpacking with *
first,*middle,last=numbers
print(first,middle,last)
```

```python
# nested tuples
tpl=((1,2),(3,4))
tpl
tpl[0][0]
tpl[-1][0]
tpl[::-1]

# it will cancel out
tpl[::-1][::-1]


# iterating over nested tuple
for sub_tuple in tpl:
    for item in sub_tuple:
        print(item)

```

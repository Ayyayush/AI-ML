```python
# map() example

numbers = [1, 2, 3, 4, 5, 6]

# Lambda function to square a number
squared_numbers = lambda x: x * x

print(squared_numbers(2))      # Output: 4

# Apply lambda to every element of the list
result = list(map(squared_numbers, numbers))

print(result)                  # Output: [1, 4, 9, 16, 25, 36]
```

```python
# accessing dictioanry elements
# first method
print(student['name'])
# this will give error
# print(student['last_name'])


# using default egt methiod
print(student.get('name'))
print(student.get('last_name'))
print(student.get('last_name', "Not Available"))
```

```python
# dictioanry Methods
print(student.keys())
print(student.values())


items=student.items()
items
```

```python
# Deep Copy  pointing  to memeory locaiton
student_copy=student
print(student)
print(student_copy)

student['age']=23
print(student)
print(student_copy)



```

```python
# Shallow copy

shallow_student=student.copy()
print(student)
print(shallow_student)
student['address']="jsr"
print(student)
print(shallow_student)
```

```python
# iterating over dictionaries
for key in student.keys():
    print(key)
    
for items in student.items():
    print(items)
    
for key,value in student.items():
    print(key,value)
    
    
```

## 🔹 How `open()` Works Internally

### 1️⃣ File Location
- Python first checks if the file exists in the given path.
- If the file exists → it is opened.
- If the file does not exist:
  - Read mode (`r`) → error occurs.
  - Write (`w`) or Append (`a`) mode → new file is created.

---

### 2️⃣ Mode-Based Permissions
The mode decides **what you are allowed to do** with the file.

| Mode | Description |
|-----|-------------|
| `r` | Read only |
| `w` | Write (overwrites existing data) |
| `a` | Append (adds data at the end) |
| `r+` | Read and Write |
| `w+` | Write and Read |
| `a+` | Append and Read |

---

### 3️⃣ File Pointer (Cursor)
- A pointer is set to track where reading or writing will happen.
- In read mode → pointer starts at the beginning.
- In write mode → pointer starts at the beginning and old data is erased.
- In append mode → pointer starts at the end of the file.

---

### 4️⃣ File Object Creation
- `open()` returns a **file object**.
- This object is used to interact with the file.

---

## 🔹 File Object Capabilities
A file object allows:
- Reading data from a file
- Writing data into a file
- Closing the file safely

---

## 🔹 Write Mode (`w`) Behavior
- Creates a new file if it does not exist.
- Deletes existing content if the file already exists.
- New data replaces old data completely.

---

## 🔹 Writing Multiple Lines
- Files do not move to the next line automatically.
- New lines must be added manually.
- Each new write continues from the current pointer position.

---

## 🔹 Importance of Closing a File
Closing a file is very important because:
- Data is safely saved.
- System memory is released.
- File corruption is prevented.
- No further operations are allowed after closing.

---

## 🔹 Best Practice
Using automatic file handling is recommended because:
- The file closes automatically.
- Reduces chances of errors.
- Makes code cleaner and safer.

---

## 🔹 Common Mistakes
- Forgetting to close the file.
- Using write mode when append was required.
- Trying to read from a closed file.
- Expecting automatic new lines while writing.

---


```python
# Reading a whole file 
with open('example.txt','r') as file:
    content=file.read()
    print(content)
    
```

```python
with open('example.txt','r+') as file:
    for line in file:
      print(line.strip())
```

```python
with open('example.txt','w+') as file:
   file.write("hello ayush how are you\n")
   content=file.read(5)
   print(content)
   # ye bhi isliye kuchb show nhi kr rha kyunki cursor end mein ja chuka h
```

```python
# ✅ Creating a list of strings — each ending with '\n' to represent a new line
L = ['hello\n', 'hi\n', 'how are you\n', 'I am fine']

# ✅ Opening a file named 'sample.txt' in write mode ('w')
# If the file doesn't exist, it will be created. If it exists, its content will be overwritten.
f = open('sample.txt', 'w')

# ✅ Writing all the strings from the list into the file
# Each string will be written on a separate line due to '\n'
f.writelines(L)

# ✅ Closing the file to ensure all data is saved properly and file is released
f.close()
```

```python
# Binary files
data=b'\x00\x01\x02\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
with open('example.bin', 'rb') as file:
    content=file.read()
    print(content)
```

```python
# --------------------------------------------
# 📖 TELL & SEEK Mechanism in File Handling
# --------------------------------------------

"""
📌 tell() ek mechanism hai jo yeh batata hai ki file pointer (cursor) abhi kis position par hai.
Yani, ab tak kitne characters read ho chuke hain.
"""

"""
📌 seek(offset) ka use karke hum file pointer ko kisi bhi specific position par bhej sakte hain.
Yeh cursor ko move karta hai, jaise rewind karna ya aage jump karna.
"""

with open('sample.txt', 'r') as f:                  # File ko read mode mein open karo
    print(f.read(10))                               # Pehle 10 characters read karo
    print(f.tell())                                 # Cursor ab kis position par hai, yeh print karo

    f.seek(0)                                       # Cursor ko wapas starting (0th position) pe le jao
    print(f.read(10))                               # Dobara pehle 10 characters read karo
    
    print(f.tell())
```

```python
# writing and then reading a file
with open('example.txt', 'w+') as file:
    file.write('Hello World\n')
    file.write('This is a new line')
    
     # moving the curosr to beginning so that we can read it  
    file.seek(0)
    print(file.read())
    
```
